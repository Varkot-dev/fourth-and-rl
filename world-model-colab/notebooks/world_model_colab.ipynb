# NFL world model on Colab: next-play transformer as a fourth-down simulator

**What this is.** A causal transformer is trained on real NFL play-by-play (nflverse, via
`nflreadpy`): training seasons 2010 to 2021, validation 2022 to 2023, test 2024 to 2025. It
reads a game's records so far and predicts the next play's transition. It is then used as a
simulator: from any game state it plays the rest of the game many times.

**What it answers.**
1. Does the model predict the next play better than a frequency baseline (per-field NLL)?
2. Do simulated games look like real games (points, plays, ties, home win share)?
3. Is win probability by simulation calibrated, compared with nflverse's own `home_wp` and
   `vegas_home_wp` on the same states (log loss, Brier, game-bootstrap intervals)?
4. On test-season fourth downs, what does the simulator recommend (go, field goal, punt), and
   how often does that match the real coach? Plus one constructed state: up 4, 4:00 left,
   fourth and 2 at the offense's own 40.

**How long it takes.** About 55 minutes on a T4, L4 or A100 GPU. The run sizes itself to the
GPU it finds and to a fixed time budget; each stage checks the clock and scales down or skips
(with the reason recorded), so it always ends with a results file.

**Choosing a GPU.** Runtime > Change runtime type > Hardware accelerator: GPU (T4 is fine;
L4 or A100 run larger samples). Then Runtime > Run all.

**If you are disconnected.** Reconnect and Run all again. With Google Drive mounted, every
finished stage and the training checkpoint are kept in
`MyDrive/fourth_and_rl_world_model`; finished stages are skipped and training resumes from
its checkpoint. Each configuration preregisters its test plan in
`prereg/<config hash>.json` before it reads the test seasons, and scores them once.

**What to bring back.** The last cell downloads `world_model_results.zip`. Unzip it into the
repo under `results/world_model/` on the branch `claude/world-model-colab`.


### What changed since the first run

The first run (Tesla T4, 18 November 2026) finished every stage, but its training ran on the
CPU and its fourth-down planning was cut to 67 decisions, 56 of them too close to call. This
version changes five things.

- **Training runs on the GPU.** The first run never moved the model to the GPU, so its 20
  minutes of training happened on the CPU. Training now moves the model to the GPU before the
  optimizer is built, keeps the training data on the GPU as ready-made tensors, and uses mixed
  precision (float16 with loss scaling on a T4, bfloat16 on newer GPUs) and a fused AdamW
  optimizer. Training should be many times faster; the training log shows the device,
  records per second and how each epoch's time splits between batching, the forward pass,
  the backward pass, validation and checkpointing.
- **Faster rollouts.** Each simulated play now encodes only the new record and reuses a
  key/value cache of everything before it, and the real game so far is stored once for all
  rollouts of a state. The preflight speed test measures this path at the batch size the
  planning stage uses, and scales the run down only if the GPU is less than half as fast as
  the preset assumes, never below 200 fourth downs. Sample sizes are larger: on a T4, 600
  win-probability states with 256 rollouts each, and 300 fourth downs with 512 rollouts per
  action.
- **Action-support guards and a field-goal check.** The planner no longer simulates field
  goals longer than the longest made field goal in the training seasons plus 3 yards, or
  punts from inside the opponent's 30. Every recommendation reports how often coaches made
  that call in comparable training situations; a call made fewer than 10 times is flagged as
  low support, and agreement with the coaches is reported separately for well-supported and
  low-support calls. A new figure, `fg_support.png`, compares the simulator's field-goal make
  rate and punt net yards with the training seasons. These rules were designed after the
  first run's test-season results had been seen, so they are a post hoc change and are
  labelled as such in the results.
- **Preregistration per model.** Each configuration writes its own
  `prereg/<config hash>.json` before it reads the test seasons, and scores them once. The
  first run's preregistration is kept. `results.json` lists every preregistration in the
  work folder, so it is visible that the test seasons have now been scored by more than one
  model.
- **A clean restart in the same Drive folder.** The preset saved by the first run is
  recognised as out of date and replaced automatically. The first run's results, figures and
  time ledger are moved to `archive/` in the Drive folder, so nothing from it is overwritten
  and the new run gets the full time budget. The downloaded data is reused.


### Stopping and resuming

- Press the stop button at any time. The current stage saves what it has (training keeps a
  checkpoint after every epoch; evaluation and planning save each finished batch of states),
  `results.json` is written with every stage's status, and the cell prints
  "Stopped safely at <stage>. Run all again to continue from here."
- Run all again later, even after a disconnect or with an updated notebook: finished stages
  are skipped and training continues from its checkpoint. Stages are keyed on the
  configuration and the data, not on the code, so an edited notebook still resumes.
- Keep Google Drive mounted, and do not delete the `fourth_and_rl_world_model` folder in
  Drive: that folder is the saved progress.
- To train tonight and evaluate tomorrow, set `STOP_AFTER = "train"` in the next cell, Run
  all, and tomorrow set it back to `None` and Run all again.

```
%%writefile /content/wm_code/schema.py
"""The data contract for the NFL world model. Every other module follows this file.

A game is a sequence of RECORDS, one per snap or kick in regulation (overtime is cut).
Record t holds the pre-snap STATE and the ACTION taken. The model reads records 0..t and
predicts the TRANSITION to record t+1: a few "delta" fields (points scored, seconds
elapsed, timeouts used) plus the next record's categorical fields and its action.
Arithmetic bookkeeping (score, clock, timeouts) is done by `apply_transition`, never by
the model, so those three can never become inconsistent.

All values are small non-negative integers ("codes"). Each field has its own vocabulary
of size `size`; code c of field f is embedded with that field's own embedding table.
Perspective: possession and score are stored from the HOME team's side; yard line is from
the side of the team with the ball (yardline_100, 1 = opponent's 1, 99 = own 1).
"""

from __future__ import annotations

from dataclasses import dataclass

import numpy as np

# ---------------------------------------------------------------- categorical values
PHASES = ("PAD", "KICKOFF", "PLAY", "TRY", "END")  # TRY = extra point or two-point try
PHASE = {name: i for i, name in enumerate(PHASES)}

POSS = ("PAD", "HOME", "AWAY")
POSS_CODE = {name: i for i, name in enumerate(POSS)}

ACTIONS = (
    "PAD",
    "RUN",
    "PASS",
    "PUNT",
    "FG",
    "KNEEL",
    "SPIKE",
    "KICKOFF",
    "ONSIDE",
    "XP",
    "TWO_PT",
    "NOPLAY",  # play wiped out by a penalty whose run/pass type is unknown
)
ACTION = {name: i for i, name in enumerate(ACTIONS)}
GO_ACTIONS = (ACTION["RUN"], ACTION["PASS"])  # "go for it" on fourth down = run or pass

# Points change on one transition, HOME minus AWAY. Code 0 is padding.
SCORE_DELTAS = (0, -8, -7, -6, -3, -2, -1, 0, 1, 2, 3, 6, 7, 8)  # index 0 unused (pad)
SCORE_DELTA_CODE = {v: i for i, v in enumerate(SCORE_DELTAS) if i > 0}

# ---------------------------------------------------------------- numeric ranges
MAX_YDSTOGO = 30  # ydstogo is clipped to 1..30; code = ydstogo (0 = not a down)
MAX_ELAPSED = 60  # seconds between records, clipped to 0..60; code = seconds + 1 (0 = pad)
CLOCK_BIN = 5  # clock embedding uses 5-second bins of game seconds remaining
GAME_SECONDS = 3600
HALF_SECONDS = 1800
SCORE_CLIP = 35  # score differential clipped to -35..35; code = diff + 36 (0 = pad)
N_TIMEOUTS = 3
SPREAD_CLIP = 21  # pregame home spread (home favoured by x) clipped, rounded to integers

# ---------------------------------------------------------------- field tables
STATE_FIELDS = (
    # name, vocabulary size
    ("phase", len(PHASES)),
    ("poss", len(POSS)),
    ("down", 5),  # 0 = not a down (kickoff or try), 1..4
    ("ydstogo", MAX_YDSTOGO + 1),  # 0 = not a down
    ("yardline", 100),  # 1..99; 0 = pad
    ("clock", GAME_SECONDS // CLOCK_BIN + 2),  # bin of seconds remaining, +1 offset, 0 = pad
    ("score", 2 * SCORE_CLIP + 2),  # home minus away + 36
    ("to_home", N_TIMEOUTS + 2),  # timeouts + 1, 0 = pad
    ("to_away", N_TIMEOUTS + 2),
    ("action", len(ACTIONS)),
)
STATE_INDEX = {name: i for i, (name, _) in enumerate(STATE_FIELDS)}

# The transition from record t to record t+1, in the order the decoder samples it.
TARGET_FIELDS = (
    ("phase", len(PHASES)),  # END means the game (regulation) is over
    ("poss", len(POSS)),
    ("score_delta", len(SCORE_DELTAS)),
    ("elapsed", MAX_ELAPSED + 2),
    ("to_used_home", 4),  # 0, 1, 2 used (code = used + 1; 0 = pad)
    ("to_used_away", 4),
    ("down", 5),
    ("ydstogo", MAX_YDSTOGO + 1),
    ("yardline", 100),
    ("action", len(ACTIONS)),
)
TARGET_INDEX = {name: i for i, (name, _) in enumerate(TARGET_FIELDS)}

HEADER_FIELDS = (
    ("spread", 2 * SPREAD_CLIP + 2),  # home spread + 22 (0 = pad)
    ("opening_receiver", len(POSS)),  # HOME or AWAY received the opening kickoff
)

# Raw (un-binned) per-record columns kept next to the codes, used for bookkeeping.
RAW_COLUMNS = ("clock_seconds", "score_diff", "timeouts_home", "timeouts_away")


@dataclass
class Split:
    """One split (train, validation or test) in array form.

    states   int16 [N, len(STATE_FIELDS)]   record codes
    raw      int32 [N, len(RAW_COLUMNS)]    unbinned clock seconds, score diff, timeouts
    targets  int16 [N, len(TARGET_FIELDS)]  transition from record i to record i+1; for the
                                            last record of a game phase = END and every
                                            other target field = 0 (masked in the loss)
    offsets  int64 [G + 1]                  records of game g are offsets[g]:offsets[g+1]
    header   int16 [G, len(HEADER_FIELDS)]
    final_diff int16 [G]                    home minus away at the end of regulation
    game_ids list[str]                      nflverse game_id per game
    seasons  int16 [G]
    play_ids int64 [N]                      nflverse play_id per record (for audits)
    """

    states: np.ndarray
    raw: np.ndarray
    targets: np.ndarray
    offsets: np.ndarray
    header: np.ndarray
    final_diff: np.ndarray
    game_ids: list[str]
    seasons: np.ndarray
    play_ids: np.ndarray


# ---------------------------------------------------------------- encoders
def clock_code(seconds: np.ndarray) -> np.ndarray:
    s = np.clip(np.asarray(seconds), 0, GAME_SECONDS)
    return (s // CLOCK_BIN + 1).astype(np.int16)


def score_code(diff: np.ndarray) -> np.ndarray:
    return (np.clip(np.asarray(diff), -SCORE_CLIP, SCORE_CLIP) + SCORE_CLIP + 1).astype(np.int16)


def timeout_code(n: np.ndarray) -> np.ndarray:
    return (np.clip(np.asarray(n), 0, N_TIMEOUTS) + 1).astype(np.int16)


def elapsed_code(seconds: np.ndarray) -> np.ndarray:
    return (np.clip(np.asarray(seconds), 0, MAX_ELAPSED) + 1).astype(np.int16)


def spread_code(spread: np.ndarray) -> np.ndarray:
    s = np.rint(np.clip(np.asarray(spread, dtype=float), -SPREAD_CLIP, SPREAD_CLIP))
    return (s + SPREAD_CLIP + 1).astype(np.int16)


def score_delta_value(code: np.ndarray) -> np.ndarray:
    """Points (home minus away) for a score-delta code; pad code 0 maps to 0."""
    table = np.asarray(SCORE_DELTAS, dtype=np.int32)
    return table[np.asarray(code)]


def apply_transition(raw: np.ndarray, target: np.ndarray) -> np.ndarray:
    """Bookkeeping: the next record's raw columns from the current raw and a transition.

    raw [B, 4] (clock_seconds, score_diff, timeouts_home, timeouts_away); target [B, T].
    Clock falls by the elapsed seconds (never below 0, and never across halftime: a
    transition that crosses 1800 seconds stops at 1800, where both teams get 3 timeouts
    back). Score adds the delta. Timeouts subtract the used counts, floored at 0.
    """
    raw = np.asarray(raw, dtype=np.int32)
    t = np.asarray(target)
    elapsed = np.clip(t[:, TARGET_INDEX["elapsed"]].astype(np.int32) - 1, 0, MAX_ELAPSED)
    clock = raw[:, 0]
    new_clock = np.maximum(clock - elapsed, 0)
    crossed = (clock > HALF_SECONDS) & (new_clock <= HALF_SECONDS)
    new_clock = np.where(crossed, HALF_SECONDS, new_clock)
    score = raw[:, 1] + score_delta_value(t[:, TARGET_INDEX["score_delta"]])
    used_h = np.clip(t[:, TARGET_INDEX["to_used_home"]].astype(np.int32) - 1, 0, 2)
    used_a = np.clip(t[:, TARGET_INDEX["to_used_away"]].astype(np.int32) - 1, 0, 2)
    to_h = np.where(crossed, N_TIMEOUTS, np.maximum(raw[:, 2] - used_h, 0))
    to_a = np.where(crossed, N_TIMEOUTS, np.maximum(raw[:, 3] - used_a, 0))
    return np.stack([new_clock, score, to_h, to_a], axis=1).astype(np.int32)


def next_state_codes(raw_next: np.ndarray, target: np.ndarray) -> np.ndarray:
    """Record codes for record t+1 from its raw columns and the sampled transition."""
    t = np.asarray(target)
    out = np.zeros((len(t), len(STATE_FIELDS)), dtype=np.int16)
    for name in ("phase", "poss", "down", "ydstogo", "yardline", "action"):
        out[:, STATE_INDEX[name]] = t[:, TARGET_INDEX[name]]
    out[:, STATE_INDEX["clock"]] = clock_code(raw_next[:, 0])
    out[:, STATE_INDEX["score"]] = score_code(raw_next[:, 1])
    out[:, STATE_INDEX["to_home"]] = timeout_code(raw_next[:, 2])
    out[:, STATE_INDEX["to_away"]] = timeout_code(raw_next[:, 3])
    return out


def outcome(final_diff: np.ndarray) -> np.ndarray:
    """Home result at the end of regulation: 1 win, 0.5 tie, 0 loss."""
    d = np.asarray(final_diff)
    return np.where(d > 0, 1.0, np.where(d < 0, 0.0, 0.5))
```

```
%%writefile /content/wm_code/records.py
"""Turn nflverse play-by-play into world-model records (see schema.py for the contract).

Standalone: imports only numpy, polars, nflreadpy and the stdlib.

nflverse column facts this module relies on (checked on 2022-2023 data, nflreadpy 0.1.5):
- Kickoffs: `posteam` is the RECEIVING team and `yardline_100` is the kick spot measured
  from the receiver's side (35 for a normal kickoff, 20 for a free kick after a safety).
- `total_home_score` / `total_away_score` are POST-play; `posteam_score` is pre-play. The
  pre-play score of a row is therefore the previous row's totals (0 at game start).
- `home_timeouts_remaining` / `away_timeouts_remaining` are POST-play (the timeout row
  itself, and a play row followed by a timeout, already show the decrement). Pre-play
  timeouts are the previous row's values within the same half (3 at the start of a half).
- Two-point tries: `two_point_attempt == 1` with play_type run or pass and null down.
  Penalised tries are play_type no_play with null down.
- Onside kicks: desc contains "onside" (`own_kickoff_recovery` is the result, not the
  attempt).
- Schedules: `spread_line` > 0 means the HOME team is favoured by that many points.
"""

from __future__ import annotations

import numpy as np
import polars as pl
import schema as S

MIN_SEASON, MAX_SEASON = 2010, 2025  # 2026 is the project's sealed holdout: never load it

RECORD_PLAY_TYPES = (
    "run", "pass", "punt", "field_goal", "kickoff", "extra_point", "qb_kneel", "qb_spike",
    "no_play",
)  # fmt: skip

NEEDED_COLUMNS = (
    "game_id", "play_id", "season", "season_type", "home_team", "away_team", "posteam",
    "order_sequence", "desc", "qtr", "game_half", "down", "ydstogo", "yardline_100",
    "game_seconds_remaining", "total_home_score", "total_away_score",
    "home_timeouts_remaining", "away_timeouts_remaining", "play_type", "play_type_nfl",
    "rush", "pass", "two_point_attempt", "home_score", "away_score",
)  # fmt: skip


def _check_seasons(seasons: list[int]) -> None:
    if not seasons:
        raise ValueError("no seasons given")
    bad = [s for s in seasons if s < MIN_SEASON or s > MAX_SEASON]
    if bad:
        raise ValueError(
            f"seasons {bad} outside {MIN_SEASON}..{MAX_SEASON}; 2026 is a sealed holdout"
        )


def load_pbp(seasons: list[int]) -> pl.DataFrame:
    """Raw nflverse play-by-play for the given seasons (2010..2025 only)."""
    seasons = [int(s) for s in seasons]
    _check_seasons(seasons)
    import nflreadpy as nfl

    return nfl.load_pbp(seasons).filter(pl.col("season").is_in(seasons))


def load_schedules(seasons: list[int]) -> pl.DataFrame:
    """nflverse schedules (game_id, spread_line, ...) for the given seasons."""
    seasons = [int(s) for s in seasons]
    _check_seasons(seasons)
    import nflreadpy as nfl

    # nflverse publishes one schedules file for all seasons; keep only the requested ones
    return nfl.load_schedules(seasons).filter(pl.col("season").is_in(seasons))


# ---------------------------------------------------------------- building
def _prepare(pbp: pl.DataFrame) -> pl.DataFrame:
    """Cast the needed columns and add pre-play score and timeouts (full frame)."""
    missing = [c for c in NEEDED_COLUMNS if c not in pbp.columns]
    if missing:
        raise ValueError(f"play-by-play is missing columns {missing}")
    f = pl.Float64
    df = (
        pbp.select(NEEDED_COLUMNS)
        .with_columns(
            pl.col("desc").fill_null(""),
            pl.col("play_type_nfl").fill_null(""),
            *[
                pl.col(c).cast(f)
                for c in (
                    "qtr",
                    "down",
                    "ydstogo",
                    "yardline_100",
                    "game_seconds_remaining",
                    "total_home_score",
                    "total_away_score",
                    "home_timeouts_remaining",
                    "away_timeouts_remaining",
                    "rush",
                    "pass",
                    "two_point_attempt",
                    "order_sequence",
                    "home_score",
                    "away_score",
                )
            ],  # fmt: skip
        )
        .with_columns(pl.col("order_sequence").fill_null(pl.col("play_id").cast(f)))
        .sort("game_id", "order_sequence", "play_id")
    )
    g = "game_id"
    df = df.with_columns(
        pl.col("total_home_score").forward_fill().over(g).fill_null(0),
        pl.col("total_away_score").forward_fill().over(g).fill_null(0),
        pl.col("home_timeouts_remaining").forward_fill().over(g, "game_half"),
        pl.col("away_timeouts_remaining").forward_fill().over(g, "game_half"),
    )
    return df.with_columns(
        pl.col("total_home_score").shift(1).over(g).fill_null(0).alias("pre_home"),
        pl.col("total_away_score").shift(1).over(g).fill_null(0).alias("pre_away"),
        pl.col("home_timeouts_remaining").shift(1).over(g, "game_half").fill_null(3)
        .clip(0, 3).alias("pre_to_home"),
        pl.col("away_timeouts_remaining").shift(1).over(g, "game_half").fill_null(3)
        .clip(0, 3).alias("pre_to_away"),
    )  # fmt: skip


def _drop(df: pl.DataFrame, keep: pl.Expr, reason: str, dropped: dict) -> pl.DataFrame:
    out = df.filter(keep.fill_null(False))
    dropped[reason] = dropped.get(reason, 0) + df.height - out.height
    return out


def _classify(df: pl.DataFrame) -> pl.DataFrame:
    """Add phase and action codes."""
    P, A = S.PHASE, S.ACTION
    pt = pl.col("play_type")
    desc = pl.col("desc")
    two_pt = pl.col("two_point_attempt") == 1
    noplay_nodown = (pt == "no_play") & pl.col("down").is_null()
    noplay_kick = noplay_nodown & desc.str.contains("(?i) kicks ")
    noplay_try = noplay_nodown & ~noplay_kick
    onside = desc.str.contains("(?i)onside")
    phase = (
        pl.when(two_pt | (pt == "extra_point") | noplay_try).then(P["TRY"])
        .when((pt == "kickoff") | noplay_kick).then(P["KICKOFF"])
        .otherwise(P["PLAY"])
    )  # fmt: skip
    action = (
        pl.when(two_pt).then(A["TWO_PT"])
        .when(pt == "extra_point").then(A["XP"])
        .when(((pt == "kickoff") | noplay_kick) & onside).then(A["ONSIDE"])
        .when((pt == "kickoff") | noplay_kick).then(A["KICKOFF"])
        .when(noplay_try & desc.str.contains("(?i)two-point")).then(A["TWO_PT"])
        .when(noplay_try & desc.str.contains("(?i)kick formation|extra point")).then(A["XP"])
        .when(noplay_try).then(A["NOPLAY"])
        .when(pt == "punt").then(A["PUNT"])
        .when(pt == "field_goal").then(A["FG"])
        .when(pt == "qb_kneel").then(A["KNEEL"])
        .when(pt == "qb_spike").then(A["SPIKE"])
        # dropbacks (incl. sacks and scrambles) are PASS, designed runs RUN
        .when(pl.col("pass") == 1).then(A["PASS"])
        .when(pl.col("rush") == 1).then(A["RUN"])
        .when(pt == "pass").then(A["PASS"])
        .when(pt == "run").then(A["RUN"])
        # a punt or field goal wiped out by a penalty is still the punt / FG decision
        .when((pt == "no_play") & desc.str.contains(" punts ")).then(A["PUNT"])
        .when((pt == "no_play") & desc.str.contains("(?i)field goal is")).then(A["FG"])
        .otherwise(A["NOPLAY"])
    )  # fmt: skip
    return df.with_columns(
        phase.cast(pl.Int16).alias("phase"),
        action.cast(pl.Int16).alias("action"),
        noplay_try.alias("_noplay_try"),
    )


def _select_records(df: pl.DataFrame, dropped: dict, repairs: dict) -> pl.DataFrame:
    pt = pl.col("play_type")
    desc = pl.col("desc")
    df = _drop(df, pl.col("qtr") <= 4, "overtime_or_null_qtr", dropped)
    df = _drop(df, pt.is_in(RECORD_PLAY_TYPES), "not_a_snap_or_kick", dropped)
    df = _drop(df, pl.col("posteam").is_not_null(), "null_posteam", dropped)
    timeout_only = (pt == "no_play") & (
        (pl.col("play_type_nfl") == "TIMEOUT") | desc.str.contains(r"^(\(\d+:\d+\) )?Timeout #")
    )
    df = _drop(df, ~timeout_only, "timeout_only", dropped)
    admin = desc.str.contains("(?i)two-minute warning|end of quarter|end quarter|end of half")
    df = _drop(df, ~(admin & (pt == "no_play")), "two_minute_or_end_of_quarter", dropped)
    df = _drop(df, pl.col("game_seconds_remaining").is_not_null(), "null_clock", dropped)
    df = _drop(df, pl.col("yardline_100").is_not_null(), "null_yardline", dropped)
    df = _classify(df)
    is_play = pl.col("phase") == S.PHASE["PLAY"]
    df = _drop(df, ~is_play | pl.col("down").is_between(1, 4), "play_without_down", dropped)
    df = _drop(df, ~is_play | pl.col("ydstogo").is_not_null(), "play_without_ydstogo", dropped)
    repairs["noplay_null_down_as_try"] = int(df["_noplay_try"].sum())
    yl = pl.col("yardline_100")
    repairs["yardline_clipped_1_99"] = int(df.select(((yl < 1) | (yl > 99)).sum()).item())
    ytg = pl.col("ydstogo")
    repairs["ydstogo_clipped_1_30"] = int(
        df.filter(is_play).select(((ytg < 1) | (ytg > S.MAX_YDSTOGO)).sum()).item()
    )
    clk = pl.col("game_seconds_remaining")
    repairs["clock_clipped_0_3600"] = int(df.select(((clk < 0) | (clk > 3600)).sum()).item())
    # schema.apply_transition resets timeouts when the clock reaches 1800, so a
    # first-half snap at 0:00 of Q2 (try after a TD at the gun, untimed down) would look
    # like the start of the second half. Store it at 1801 s: same clock code (bin 361).
    h1_at_half = (pl.col("game_half") == "Half1") & (clk <= S.HALF_SECONDS)
    repairs["first_half_clock_1800_set_to_1801"] = int(df.select(h1_at_half.sum()).item())
    df = df.with_columns(
        pl.when(h1_at_half).then(S.HALF_SECONDS + 1).otherwise(clk).alias("game_seconds_remaining")
    )
    poss = (
        pl.when(pl.col("posteam") == pl.col("home_team")).then(S.POSS_CODE["HOME"])
        .when(pl.col("posteam") == pl.col("away_team")).then(S.POSS_CODE["AWAY"])
        .otherwise(None)
    )  # fmt: skip
    df = df.with_columns(poss.cast(pl.Int16).alias("poss"))
    df = _drop(df, pl.col("poss").is_not_null(), "posteam_not_home_or_away", dropped)
    return df.with_columns(
        pl.when(is_play).then(pl.col("down")).otherwise(0).cast(pl.Int16).alias("down_c"),
        pl.when(is_play).then(ytg.clip(1, S.MAX_YDSTOGO)).otherwise(0).cast(pl.Int16)
        .alias("ytg_c"),
        yl.clip(1, 99).cast(pl.Int16).alias("yl_c"),
        clk.clip(0, S.GAME_SECONDS).cast(pl.Int32).alias("clock_s"),
        (pl.col("pre_home") - pl.col("pre_away")).cast(pl.Int32).alias("diff"),
        pl.col("pre_to_home").cast(pl.Int32).alias("to_h"),
        pl.col("pre_to_away").cast(pl.Int32).alias("to_a"),
    )  # fmt: skip


# target fields that are 0 on a game's last record (its END target)
END_ZERO_FIELDS = tuple(
    S.TARGET_INDEX[f] for f, _ in S.TARGET_FIELDS if f not in ("phase", "score_delta", "elapsed")
)


def _nearest_delta(d: np.ndarray) -> np.ndarray:
    legal = np.asarray(S.SCORE_DELTAS[1:])
    return legal[np.abs(d[:, None] - legal[None, :]).argmin(axis=1)]


def build_split(
    pbp: pl.DataFrame,
    schedules: pl.DataFrame,
    seasons: list[int],
    season_types: tuple[str, ...] = ("REG", "POST"),
    illegal_delta: str = "drop_game",
) -> tuple[S.Split, dict]:
    """Records for the given seasons as a schema.Split, plus a report of counts.

    illegal_delta: what to do when a transition's points change is not in
    schema.SCORE_DELTAS: "drop_game" (default) drops the whole game, "nearest" maps it to
    the nearest legal value (the bookkeeping check then fails on that pair).
    """
    seasons = [int(s) for s in seasons]
    _check_seasons(seasons)
    dropped: dict[str, int] = {}
    repairs: dict[str, int] = {}
    soft: dict[str, int] = {}
    raw = pbp.filter(
        pl.col("season").is_in(seasons) & pl.col("season_type").is_in(list(season_types))
    )
    df = _prepare(raw)
    rows_in = df.height

    # end-of-regulation score: post-play totals of the last regulation row
    reg_end = (
        df.filter(pl.col("qtr") <= 4)
        .group_by("game_id", maintain_order=True)
        .agg(
            (pl.col("total_home_score").last() - pl.col("total_away_score").last())
            .cast(pl.Int32).alias("final_diff"),
            (pl.col("home_score").first() - pl.col("away_score").first()).alias("box_diff"),
            (pl.col("qtr").max() <= 4).alias("no_ot_rows"),
        )
    )  # fmt: skip
    ot_games = set(df.filter(pl.col("qtr") >= 5)["game_id"].unique().to_list())
    reg_end = reg_end.with_columns(pl.col("game_id").is_in(list(ot_games)).alias("ot"))

    rec = _select_records(df, dropped, repairs)

    # ---- transitions (numpy over the sorted record table)
    gid = rec["game_id"].to_numpy()
    n = len(gid)
    starts = np.flatnonzero(np.r_[True, gid[1:] != gid[:-1]]) if n else np.zeros(0, int)
    last = np.zeros(n, bool)
    if n:
        last[np.r_[starts[1:] - 1, n - 1]] = True
    nxt = np.minimum(np.arange(n) + 1, max(n - 1, 0))

    phase = rec["phase"].to_numpy().astype(np.int16)
    poss = rec["poss"].to_numpy().astype(np.int16)
    down = rec["down_c"].to_numpy().astype(np.int16)
    ytg = rec["ytg_c"].to_numpy().astype(np.int16)
    yl = rec["yl_c"].to_numpy().astype(np.int16)
    action = rec["action"].to_numpy().astype(np.int16)
    clock = rec["clock_s"].to_numpy().astype(np.int32)
    diff = rec["diff"].to_numpy().astype(np.int32)
    to_h = rec["to_h"].to_numpy().astype(np.int32)
    to_a = rec["to_a"].to_numpy().astype(np.int32)
    half = rec["game_half"].to_numpy()
    within = ~last

    # points on a game's final record (FG as time expires, TD or try at 0:00) go in its
    # END target: final regulation diff minus the last record's pre-play diff
    fd_map = dict(reg_end.select("game_id", "final_diff").iter_rows())
    end_fd = np.array([fd_map[g] for g in gid[last]], dtype=np.int32)
    delta = diff[nxt] - diff
    delta[last] = end_fd - diff[last]
    soft["games_scoring_on_last_record"] = int((delta[last] != 0).sum())
    legal = np.isin(delta, S.SCORE_DELTAS[1:])
    bad_games = set(gid[~legal].tolist())
    soft["score_delta_illegal_pairs"] = int((~legal).sum())
    if illegal_delta == "nearest":
        soft["score_delta_mapped"] = int((~legal).sum())
        delta = np.where(legal, delta, _nearest_delta(delta))
        bad_games = set()
    elapsed = clock - clock[nxt]
    elapsed[last] = clock[last]  # regulation ends at 0 seconds
    soft["elapsed_negative"] = int((within & (elapsed < 0)).sum())
    soft["elapsed_over_60"] = int((within & (elapsed > S.MAX_ELAPSED)).sum())
    soft["elapsed_over_45"] = int((within & (elapsed > 45)).sum())
    crossing = within & (half == "Half1") & (half[nxt] == "Half2")
    soft["halftime_crossings"] = int(crossing.sum())
    used_h = np.where(crossing, 0, np.maximum(0, to_h - to_h[nxt]))
    used_a = np.where(crossing, 0, np.maximum(0, to_a - to_a[nxt]))
    soft["timeout_increase_mid_half"] = int(
        (within & ~crossing & ((to_h[nxt] > to_h) | (to_a[nxt] > to_a))).sum()
    )
    soft["timeouts_used_over_2"] = int((within & ((used_h > 2) | (used_a > 2))).sum())

    T = S.TARGET_INDEX
    targets = np.zeros((n, len(S.TARGET_FIELDS)), np.int16)
    targets[:, T["phase"]] = np.where(last, S.PHASE["END"], phase[nxt])
    targets[:, T["poss"]] = poss[nxt]
    targets[:, T["score_delta"]] = [S.SCORE_DELTA_CODE.get(int(d), 0) for d in delta]
    targets[:, T["elapsed"]] = S.elapsed_code(elapsed)
    targets[:, T["to_used_home"]] = np.clip(used_h, 0, 2) + 1
    targets[:, T["to_used_away"]] = np.clip(used_a, 0, 2) + 1
    targets[:, T["down"]] = down[nxt]
    targets[:, T["ydstogo"]] = ytg[nxt]
    targets[:, T["yardline"]] = yl[nxt]
    targets[:, T["action"]] = action[nxt]
    # END target: phase, score_delta and elapsed are real; every other field is 0 (masked)
    targets[np.ix_(last, list(END_ZERO_FIELDS))] = 0

    SI = S.STATE_INDEX
    states = np.zeros((n, len(S.STATE_FIELDS)), np.int16)
    states[:, SI["phase"]] = phase
    states[:, SI["poss"]] = poss
    states[:, SI["down"]] = down
    states[:, SI["ydstogo"]] = ytg
    states[:, SI["yardline"]] = yl
    states[:, SI["clock"]] = S.clock_code(clock)
    states[:, SI["score"]] = S.score_code(diff)
    states[:, SI["to_home"]] = S.timeout_code(to_h)
    states[:, SI["to_away"]] = S.timeout_code(to_a)
    states[:, SI["action"]] = action
    raw_cols = np.stack([clock, diff, to_h, to_a], axis=1).astype(np.int32)
    play_ids = rec["play_id"].cast(pl.Float64).to_numpy().astype(np.int64)
    rec_season = rec["season"].to_numpy()

    # ---- drop games with illegal score deltas
    keep = ~np.isin(gid, list(bad_games)) if bad_games else np.ones(n, bool)
    dropped["games_with_illegal_score_delta"] = len(bad_games)
    dropped["records_in_dropped_games"] = int((~keep).sum())
    states, raw_cols, targets = states[keep], raw_cols[keep], targets[keep]
    play_ids, gid, rec_season = play_ids[keep], gid[keep], rec_season[keep]
    m = len(gid)
    starts = np.flatnonzero(np.r_[True, gid[1:] != gid[:-1]]) if m else np.zeros(0, int)
    offsets = np.r_[starts, m].astype(np.int64)
    game_ids = [str(x) for x in gid[starts]]
    seasons_arr = rec_season[starts].astype(np.int16)

    # ---- header
    games = pl.DataFrame({"game_id": game_ids, "_g": np.arange(len(game_ids))})
    sched = schedules.select("game_id", pl.col("spread_line").cast(pl.Float64))
    games = games.join(sched, on="game_id", how="left").sort("_g")
    spread = games["spread_line"].to_numpy()
    soft["games_missing_spread"] = int(np.isnan(spread).sum()) if len(spread) else 0
    spread = np.nan_to_num(spread, nan=0.0)
    header = np.zeros((len(game_ids), len(S.HEADER_FIELDS)), np.int16)
    header[:, 0] = S.spread_code(spread)
    opening = np.zeros(len(game_ids), np.int16)
    soft["games_first_record_not_kickoff"] = 0
    for g in range(len(game_ids)):
        s, e = offsets[g], offsets[g + 1]
        ko = np.flatnonzero(states[s:e, SI["phase"]] == S.PHASE["KICKOFF"])
        if len(ko) == 0 or ko[0] != 0:
            soft["games_first_record_not_kickoff"] += 1
        opening[g] = states[s + ko[0], SI["poss"]] if len(ko) else states[s, SI["poss"]]
    header[:, 1] = opening

    # ---- final regulation score
    fin = games.join(reg_end, on="game_id", how="left").sort("_g")
    final_diff = fin["final_diff"].to_numpy().astype(np.int16)
    no_ot = ~fin["ot"].to_numpy().astype(bool)
    box = fin["box_diff"].to_numpy()
    soft["final_diff_vs_box_mismatch_non_ot"] = int((no_ot & (final_diff != box)).sum())

    split = S.Split(
        states=states,
        raw=raw_cols,
        targets=targets,
        offsets=offsets,
        header=header,
        final_diff=final_diff,
        game_ids=game_ids,
        seasons=seasons_arr,
        play_ids=play_ids,
    )
    sizes = np.diff(offsets)
    report = {
        "seasons": seasons,
        "rows_in": rows_in,
        "rows_kept": int(m),
        "dropped": dropped,
        "repairs": repairs,
        "soft": soft,
        "games": len(game_ids),
        "overtime_games_cut": int(np.sum(~no_ot)),
        "records_per_game": {
            "min": int(sizes.min()) if len(sizes) else 0,
            "mean": round(float(sizes.mean()), 1) if len(sizes) else 0.0,
            "max": int(sizes.max()) if len(sizes) else 0,
        },
    }
    return split, report


# ---------------------------------------------------------------- validation
def _pair_mask(split: S.Split) -> np.ndarray:
    """True for record i when i+1 is in the same game."""
    n = len(split.states)
    m = np.ones(n, bool)
    if n:
        m[split.offsets[1:] - 1] = False
    return m


def transition_mismatches(split: S.Split) -> dict:
    """Within-game pairs where apply_transition does not reproduce raw[i+1], by reason."""
    pair = _pair_mask(split)
    idx = np.flatnonzero(pair)
    pred = S.apply_transition(split.raw[idx], split.targets[idx])
    true = split.raw[idx + 1]
    bad = pred != true
    out = {"pairs": int(len(idx)), "mismatched_pairs": int(bad.any(axis=1).sum())}
    for j, name in enumerate(S.RAW_COLUMNS):
        out[f"mismatch_{name}"] = int(bad[:, j].sum())
    any_bad = bad.any(axis=1)
    el = split.raw[idx, 0] - split.raw[idx + 1, 0]
    half_at_1800 = (
        (split.raw[idx, 0] == S.HALF_SECONDS)
        & (split.states[idx + 1, S.STATE_INDEX["phase"]] == S.PHASE["KICKOFF"])
        & (split.raw[idx + 1, 0] == S.HALF_SECONDS)
    )
    reasons = {
        "reason_elapsed_over_60": any_bad & (el > S.MAX_ELAPSED),
        "reason_elapsed_negative": any_bad & (el < 0),
        "reason_untimed_down_at_1800_before_halftime": any_bad & half_at_1800,
    }
    explained = np.zeros(len(idx), bool)
    for k, v in reasons.items():
        out[k] = int((v & ~explained).sum())
        explained |= v
    out["reason_other"] = int((any_bad & ~explained).sum())
    return out


def validate_split(split: S.Split, min_transition_match: float = 0.99) -> dict:
    """Invariant checks. Returns check -> violations; raises AssertionError on hard ones."""
    st, tg, off = split.states, split.targets, split.offsets
    n, G = len(st), len(split.game_ids)
    hard: dict[str, int] = {}
    for j, (name, size) in enumerate(S.STATE_FIELDS):
        hard[f"state_{name}_out_of_vocab"] = int(((st[:, j] < 0) | (st[:, j] >= size)).sum())
    for j, (name, size) in enumerate(S.TARGET_FIELDS):
        hard[f"target_{name}_out_of_vocab"] = int(((tg[:, j] < 0) | (tg[:, j] >= size)).sum())
    for j, (name, size) in enumerate(S.HEADER_FIELDS):
        col = split.header[:, j]
        hard[f"header_{name}_out_of_vocab"] = int(((col <= 0) | (col >= size)).sum())
    SI, T = S.STATE_INDEX, S.TARGET_INDEX
    never_zero = ("phase", "poss", "yardline", "clock", "score", "to_home", "to_away", "action")
    hard["state_pad_codes"] = int(sum((st[:, SI[f]] == 0).sum() for f in never_zero))
    hard["state_phase_end_or_pad"] = int(np.isin(st[:, SI["phase"]], [0, S.PHASE["END"]]).sum())
    is_play = st[:, SI["phase"]] == S.PHASE["PLAY"]
    hard["down_zero_iff_not_play"] = int(((st[:, SI["down"]] == 0) == is_play).sum())
    hard["ydstogo_zero_iff_not_play"] = int(((st[:, SI["ydstogo"]] == 0) == is_play).sum())
    hard["offsets_bad"] = int(
        len(off) != G + 1 or (G >= 0 and (off[0] != 0 or off[-1] != n))
        or bool(np.any(np.diff(off) <= 0))
    )  # fmt: skip
    if hard["offsets_bad"]:
        raise AssertionError("validate_split failed: offsets do not cover the records")
    pair = _pair_mask(split)
    is_end = tg[:, T["phase"]] == S.PHASE["END"]
    hard["end_not_exactly_last"] = int((is_end != ~pair).sum())
    end_tg = tg[~pair]
    hard["last_target_nonzero_fields"] = int((end_tg[:, list(END_ZERO_FIELDS)] != 0).sum())
    hard["end_target_pad_score_or_elapsed"] = int(
        ((end_tg[:, T["score_delta"]] == 0) | (end_tg[:, T["elapsed"]] == 0)).sum()
    )
    # the last record's score plus its END score delta is the final regulation score
    end_score = split.raw[~pair, 1] + S.score_delta_value(end_tg[:, T["score_delta"]])
    fd_bad = end_score != split.final_diff.astype(np.int32)
    hard["final_diff_ne_last_score_plus_end_delta"] = int(fd_bad.sum())
    pad_tg = tg[pair][
        :,
        [
            T[f]
            for f in (
                "phase",
                "poss",
                "score_delta",
                "elapsed",
                "to_used_home",
                "to_used_away",
                "yardline",
                "action",
            )
        ],
    ]
    hard["target_pad_codes"] = int((pad_tg == 0).sum())
    # target categorical fields equal the next record's fields (exact by construction)
    idx = np.flatnonzero(pair)
    nxt = S.next_state_codes(split.raw[idx + 1], tg[idx])
    hard["target_ne_next_state"] = int((nxt != st[idx + 1]).any(axis=1).sum())
    hard["duplicate_game_ids"] = G - len(set(split.game_ids))
    tm = transition_mismatches(split)
    frac = 1 - tm["mismatched_pairs"] / max(tm["pairs"], 1)
    hard["apply_transition_below_threshold"] = int(frac < min_transition_match)
    report = {**hard, **{f"info_{k}": v for k, v in tm.items()}}
    report["info_apply_transition_match_fraction"] = round(frac, 5)
    report["info_final_diff_mismatch_by_season"] = {
        int(y): int(fd_bad[split.seasons == y].sum()) for y in np.unique(split.seasons)
    }
    report["info_games_scoring_on_last_record"] = int(
        (S.score_delta_value(end_tg[:, T["score_delta"]]) != 0).sum()
    )
    sd = tg[pair, T["score_delta"]]
    report["soft_elapsed_over_45"] = int((tg[pair, T["elapsed"]] - 1 > 45).sum())
    report["soft_nonzero_score_deltas"] = int((S.score_delta_value(sd) != 0).sum())
    failed = {k: v for k, v in hard.items() if v}
    if failed:
        raise AssertionError(f"validate_split failed: {failed}")
    return report


def validate_disjoint(splits: dict[str, S.Split]) -> dict:
    """No game id in two splits. Raises AssertionError if any is shared."""
    seen: dict[str, str] = {}
    shared = 0
    for name, sp in splits.items():
        for g in sp.game_ids:
            if g in seen and seen[g] != name:
                shared += 1
            seen[g] = name
    if shared:
        raise AssertionError(f"{shared} game ids appear in more than one split")
    return {"shared_game_ids": 0}


# ---------------------------------------------------------------- inspection
def _game_of(split: S.Split, i: int) -> int:
    return int(np.searchsorted(split.offsets, i, side="right") - 1)


def decode_record(split: S.Split, i: int) -> dict:
    """Human-readable state, action and transition target of record i."""
    st, tg, rw = split.states[i], split.targets[i], split.raw[i]
    SI, T = S.STATE_INDEX, S.TARGET_INDEX
    g = _game_of(split, i)
    is_end = int(tg[T["phase"]]) == S.PHASE["END"]
    target = {
        "phase": S.PHASES[tg[T["phase"]]],
        "score_delta": int(S.SCORE_DELTAS[tg[T["score_delta"]]]),
        "elapsed": int(tg[T["elapsed"]]) - 1,
    }
    if not is_end:
        target.update(
            poss=S.POSS[tg[T["poss"]]],
            score_delta=int(S.SCORE_DELTAS[tg[T["score_delta"]]]),
            elapsed=int(tg[T["elapsed"]]) - 1,
            to_used_home=int(tg[T["to_used_home"]]) - 1,
            to_used_away=int(tg[T["to_used_away"]]) - 1,
            down=int(tg[T["down"]]),
            ydstogo=int(tg[T["ydstogo"]]),
            yardline=int(tg[T["yardline"]]),
            action=S.ACTIONS[tg[T["action"]]],
        )
    return {
        "game_id": split.game_ids[g],
        "play_id": int(split.play_ids[i]),
        "record_in_game": int(i - split.offsets[g]),
        "phase": S.PHASES[st[SI["phase"]]],
        "poss": S.POSS[st[SI["poss"]]],
        "down": int(st[SI["down"]]),
        "ydstogo": int(st[SI["ydstogo"]]),
        "yardline_100": int(st[SI["yardline"]]),
        "clock_seconds": int(rw[0]),
        "score_diff_home": int(rw[1]),
        "timeouts_home": int(rw[2]),
        "timeouts_away": int(rw[3]),
        "action": S.ACTIONS[st[SI["action"]]],
        "target": target,
    }


def fourth_down_index(split: S.Split) -> np.ndarray:
    """Records that are real fourth-down decisions: PLAY, down 4, RUN/PASS/PUNT/FG."""
    st = split.states
    SI, A = S.STATE_INDEX, S.ACTION
    ok = (
        (st[:, SI["phase"]] == S.PHASE["PLAY"])
        & (st[:, SI["down"]] == 4)
        & np.isin(st[:, SI["action"]], [A["RUN"], A["PASS"], A["PUNT"], A["FG"]])
    )
    return np.flatnonzero(ok)
```

```
%%writefile /content/wm_code/model.py
"""A causal transformer "world model" over game records, with an autoregressive field decoder.

Sequence per game: position 0 is the header token (sum of header field embeddings), then one
token per record (sum of that record's STATE_FIELDS embeddings). Every token also gets a
learned positional embedding. A pre-LayerNorm causal transformer turns record t into h_t.

The FIELD DECODER predicts the transition t -> t+1 one field at a time, in TARGET_FIELDS order:
    logits_j = MLP_j([h_t ; e_0(y_0) + ... + e_{j-1}(y_{j-1})])
where e_k is an embedding table for target field k and y_k is the true code (training, teacher
forcing) or the sampled code (generation). Each head outputs only its own field's vocabulary.

Two execution paths each, with the same parameters and the same math:
- decoder: in training every true field is known, so all ten heads run in one batched pass
  (`decode_logits_padded`); `decode_logits` is the one-head-at-a-time reference. Sampling is
  sequential by nature (`sample_transition`).
- encoder: `encode` runs the whole window; `encode_step` adds ONE new record per row using a
  key/value cache of the earlier positions (incremental decoding for rollouts). Causal
  attention makes the two identical: position p only ever attends to positions 0..p.

Standalone: imports only torch, numpy and the sibling `schema` module.
"""

from __future__ import annotations

import numpy as np
import schema as S
import torch
import torch.nn.functional as F
from torch import nn

DEFAULT_CFG: dict = {
    # model
    "d_model": 256,
    "n_layers": 4,
    "n_heads": 4,
    "d_ff": 1024,
    "dec_hidden": 256,
    "dropout": 0.1,
    "max_len": 256,  # records of context; longer games keep their LAST max_len records
    # training
    "seed": 0,
    "epochs": 30,
    "batch_games": 32,
    "eval_batch_games": 128,
    "parallel_decoder": True,  # training loss: all field heads in one pass (same math)
    "data_on_device": "auto",  # keep the pre-built split tensors on the GPU when they fit
    "train_mem_frac": 0.6,  # share of free GPU memory a training step may plan to use
    "timing_sync_every": 8,  # CUDA: synchronize every k-th step to split time by phase
    "lr": 3e-4,
    "min_lr_frac": 0.1,
    "weight_decay": 0.1,
    "betas": (0.9, 0.95),
    "warmup_steps": 300,
    "grad_clip": 1.0,
    "patience": 4,
    "stop_after_epoch": None,  # stop cleanly after this many epochs (schedule unchanged)
    # simulation
    "temperature": 1.0,
    "max_steps": 400,  # transitions per rollout before it is counted as truncated
    "sim_batch": 1024,  # rollouts per full-window transformer pass (prefill, fallback)
    "sample_batch": 16384,  # rows per sampling call (one call per simulated play)
    "sim_max_live": 8192,  # rollouts held at once (queries are processed in groups)
    "mask_pad_codes": True,  # never sample code 0 for fields where 0 means padding
    "score_on_end": True,  # add the sampled score delta of the END transition (final play)
    "sort_by_length": True,  # simulator: encode rows sorted by context length (less padding)
    "kv_cache": True,  # simulator: incremental decoding (False = recompute the window each play)
    "kv_cache_mem_frac": 0.4,  # CUDA: share of total GPU memory the key/value cache may use
    "kv_cache_cpu_gb": 4.0,  # CPU: memory the key/value cache may use
}

# Target fields where code 0 is a real value ("not a down"), not padding.
ZERO_IS_VALUE = frozenset({"down", "ydstogo"})
TARGET_NAMES = tuple(name for name, _ in S.TARGET_FIELDS)
PHASE_T = S.TARGET_INDEX["phase"]
# Fields that carry a real value on a game's LAST record (target phase END): the final play's
# points and seconds. Every other field is padding there: it never counts in the loss, and its
# decoder input is forced to code 0 both in training and in sampling, so the score_delta and
# elapsed heads see the same prefix (END, poss=0) in both.
END_FIELDS = frozenset({"phase", "score_delta", "elapsed"})
NOT_END_T = tuple(j for j, n in enumerate(TARGET_NAMES) if n not in END_FIELDS)
TARGET_SIZES = tuple(n for _, n in S.TARGET_FIELDS)
VMAX = max(TARGET_SIZES)  # padded vocabulary width of the batched decoder


def merge_cfg(cfg: dict | None) -> dict:
    out = dict(DEFAULT_CFG)
    out.update(cfg or {})
    return out


class Block(nn.Module):
    """Pre-LayerNorm transformer block with causal scaled-dot-product attention."""

    def __init__(self, d: int, n_heads: int, d_ff: int, p: float) -> None:
        super().__init__()
        assert d % n_heads == 0, "d_model must be divisible by n_heads"
        self.n_heads = n_heads
        self.p = p
        self.ln1 = nn.LayerNorm(d)
        self.qkv = nn.Linear(d, 3 * d)
        self.proj = nn.Linear(d, d)
        self.ln2 = nn.LayerNorm(d)
        self.ff = nn.Sequential(nn.Linear(d, d_ff), nn.GELU(), nn.Linear(d_ff, d))
        self.drop = nn.Dropout(p)

    def forward(self, x: torch.Tensor, return_kv: bool = False):
        """x [b, n, d] -> [b, n, d]; with return_kv also this layer's keys and values, each
        [b, heads, n, d_head] (what a key/value cache stores)."""
        b, n, d = x.shape
        qkv = self.qkv(self.ln1(x)).view(b, n, 3, self.n_heads, d // self.n_heads)
        q, k, v = qkv.permute(2, 0, 3, 1, 4)  # each [b, heads, n, d_head]
        a = F.scaled_dot_product_attention(
            q, k, v, dropout_p=self.p if self.training else 0.0, is_causal=True
        )
        x = x + self.drop(self.proj(a.transpose(1, 2).reshape(b, n, d)))
        x = x + self.drop(self.ff(self.ln2(x)))
        return (x, k, v) if return_kv else x

    def step(
        self,
        x: torch.Tensor,
        k_cache: torch.Tensor,
        v_cache: torch.Tensor,
        pos: torch.Tensor,
        n_keys: int,
        prefix: tuple | None = None,
    ) -> torch.Tensor:
        """One new token per row (inference only).

        x [R, d] is the new token; k_cache, v_cache [R, heads, C, d_head] are each row's own
        cache, written in place at index pos [R]. Without prefix the token attends to cache
        indices 0..pos[r] (its whole history), exactly what the causal mask lets that position
        see in `forward`; n_keys >= max(pos) + 1.

        With prefix = (k_pre, v_pre [U, heads, P, d_head], cell [R], groups, bias [R, P + n_keys])
        the history is split: keys and values shared by a group of rows (the real game so far,
        identical for every rollout of a query) are stored ONCE per group, and the row's own
        cache holds only what it simulated. The token attends to both in one softmax (see
        `cascade_attention`), the same function as attending to the concatenated history."""
        r, d = x.shape
        qkv = self.qkv(self.ln1(x)).view(r, 3, self.n_heads, d // self.n_heads)
        q, k, v = qkv.unbind(1)  # each [R, heads, d_head]
        rows = torch.arange(r, device=x.device)
        k_cache[rows, :, pos] = k.to(k_cache.dtype)
        v_cache[rows, :, pos] = v.to(v_cache.dtype)
        if prefix is None:
            keys = torch.arange(n_keys, device=x.device)
            allowed = (keys.unsqueeze(0) <= pos.unsqueeze(1))[:, None, None, :]  # [R,1,1,n]
            a = F.scaled_dot_product_attention(
                q.unsqueeze(2).to(k_cache.dtype),
                k_cache[:, :, :n_keys],
                v_cache[:, :, :n_keys],
                attn_mask=allowed,
            )
        else:
            k_pre, v_pre, cell, groups, bias = prefix
            a = cascade_attention(
                q, k_cache[:, :, :n_keys], v_cache[:, :, :n_keys], k_pre, v_pre, cell, groups, bias
            )
        x = x + self.proj(a.reshape(r, d))
        return x + self.ff(self.ln2(x))


def cascade_attention(
    q: torch.Tensor,
    k_own: torch.Tensor,
    v_own: torch.Tensor,
    k_pre: torch.Tensor,
    v_pre: torch.Tensor,
    cell: torch.Tensor,
    groups: int,
    bias: torch.Tensor,
) -> torch.Tensor:
    """Attention of one query per row over a shared prefix plus the row's own keys.

    q [R, heads, dh]; k_own, v_own [R, heads, n, dh]; k_pre, v_pre [U, heads, P, dh] shared by
    the rows of a group; cell [R] = u * groups + (row's place in group u), so the queries of
    a group form one [groups, dh] matrix and the prefix keys are read once per group (one
    batched matrix product) instead of once per row; bias [R, P + n] is 0 for a key the row
    may see and -inf otherwise. Returns softmax([q k_pre ; q k_own] / sqrt(dh) + bias) @
    [v_pre ; v_own], [R, heads, dh]: one softmax over both parts, so exactly attention over
    the concatenated keys (up to the order of floating-point additions)."""
    r, h, dh = q.shape
    u, _, p, _ = k_pre.shape
    qd = q.to(k_pre.dtype)
    grid = qd.new_zeros(u * groups, h, dh)
    grid[cell] = qd
    grid = grid.view(u, groups, h, dh).transpose(1, 2)  # [U, heads, groups, dh]
    s_pre = torch.matmul(grid, k_pre.transpose(-1, -2))  # [U, heads, groups, P]
    s_pre = s_pre.transpose(1, 2).reshape(u * groups, h, p)[cell]  # [R, heads, P]
    s_own = torch.matmul(qd.unsqueeze(2), k_own.transpose(-1, -2)).squeeze(2)  # [R, heads, n]
    scores = torch.cat([s_pre, s_own], dim=-1).float() * dh**-0.5 + bias.unsqueeze(1)
    w = torch.softmax(scores, dim=-1).to(v_pre.dtype)
    w_grid = w.new_zeros(u * groups, h, p)
    w_grid[cell] = w[..., :p]
    o_pre = torch.matmul(w_grid.view(u, groups, h, p).transpose(1, 2), v_pre)  # [U,h,groups,dh]
    o_pre = o_pre.transpose(1, 2).reshape(u * groups, h, dh)[cell]
    o_own = torch.matmul(w[..., p:].unsqueeze(2), v_own).squeeze(2)
    return o_pre + o_own


class WorldModel(nn.Module):
    def __init__(self, cfg: dict | None = None) -> None:
        super().__init__()
        self.cfg = merge_cfg(cfg)
        c = self.cfg
        d = c["d_model"]
        self.max_len = int(c["max_len"])
        self.state_emb = nn.ModuleList(nn.Embedding(n, d) for _, n in S.STATE_FIELDS)
        self.header_emb = nn.ModuleList(nn.Embedding(n, d) for _, n in S.HEADER_FIELDS)
        self.pos_emb = nn.Embedding(self.max_len + 1, d)  # 0 = header, 1..max_len = records
        self.drop = nn.Dropout(c["dropout"])
        self.blocks = nn.ModuleList(
            Block(d, c["n_heads"], c["d_ff"], c["dropout"]) for _ in range(c["n_layers"])
        )
        self.ln_f = nn.LayerNorm(d)
        # embeddings of target fields 0..T-2 (the last field is never a decoder input)
        self.tgt_emb = nn.ModuleList(nn.Embedding(n, d) for _, n in S.TARGET_FIELDS[:-1])
        h = c["dec_hidden"]
        self.heads = nn.ModuleList(
            nn.Sequential(nn.Linear(2 * d, h), nn.GELU(), nn.Linear(h, n))
            for _, n in S.TARGET_FIELDS
        )
        self.apply(self._init)
        # Constants of the batched decoder and loss, kept on the model's device so that no
        # step copies anything from the host (a host-to-GPU copy of pageable memory waits for
        # the GPU). Not saved in checkpoints.
        pad = torch.full((len(TARGET_SIZES), VMAX), float("-inf"))
        for j, n in enumerate(TARGET_SIZES):
            pad[j, :n] = 0.0
        self.register_buffer("_logit_pad", pad, persistent=False)
        no_pad = pad.clone()  # sampling with mask_pad: never code 0 where 0 means padding
        for j, name in enumerate(TARGET_NAMES):
            if name not in ZERO_IS_VALUE:
                no_pad[j, 0] = float("-inf")
        self.register_buffer("_logit_no_pad", no_pad, persistent=False)
        flags = {
            "_is_phase": [j == PHASE_T for j in range(len(TARGET_NAMES))],
            "_zero_is_value": [n in ZERO_IS_VALUE for n in TARGET_NAMES],
            "_end_field": [n in END_FIELDS for n in TARGET_NAMES],
        }
        for name, v in flags.items():
            self.register_buffer(name, torch.tensor(v, dtype=torch.bool), persistent=False)

    @staticmethod
    def _init(m: nn.Module) -> None:
        if isinstance(m, nn.Linear):
            nn.init.normal_(m.weight, std=0.02)
            nn.init.zeros_(m.bias)
        elif isinstance(m, nn.Embedding):
            nn.init.normal_(m.weight, std=0.02)

    def count_params(self) -> int:
        return sum(p.numel() for p in self.parameters())

    # ------------------------------------------------------------------ encoder
    def embed_records(self, states: torch.Tensor) -> torch.Tensor:
        """states [..., S] -> sum of the record's field embeddings [..., d]."""
        return sum(emb(states[..., k]) for k, emb in enumerate(self.state_emb))

    def encode(self, header: torch.Tensor, states: torch.Tensor, return_kv: bool = False):
        """header [B, H], states [B, L, S] (L <= max_len, right-padded) -> h [B, L, d].

        With return_kv, also every layer's (keys, values), each [B, heads, L + 1, d_head]
        (position 0 is the header token): the prefill of a key/value cache."""
        b, n, _ = states.shape
        if n > self.max_len:
            raise ValueError(f"{n} records exceed max_len={self.max_len}; cut to the last ones")
        x0 = sum(emb(header[:, k]) for k, emb in enumerate(self.header_emb)).unsqueeze(1)
        xr = self.embed_records(states)
        x = torch.cat([x0, xr], dim=1) + self.pos_emb.weight[: n + 1].unsqueeze(0)
        x = self.drop(x)
        kv = []
        for blk in self.blocks:
            if return_kv:
                x, k, v = blk(x, return_kv=True)
                kv.append((k, v))
            else:
                x = blk(x)
        h = self.ln_f(x)[:, 1:]
        return (h, kv) if return_kv else h

    def encode_step(
        self,
        states: torch.Tensor,
        pos: torch.Tensor,
        k_cache: torch.Tensor,
        v_cache: torch.Tensor,
        n_keys: int,
        cache_pos: torch.Tensor | None = None,
        prefix: tuple | None = None,
    ) -> torch.Tensor:
        """Incremental encode (inference only): one new record per row.

        states [R, S] codes of the new record, pos [R] its position in the window (1..max_len;
        0 is the header), k_cache / v_cache [layers, R, heads, C, d_head] the rows' own caches;
        this record's keys and values are written at cache_pos (default pos). Without prefix
        the own cache holds positions 0..pos-1. With prefix (per layer: k_pre, v_pre
        [layers, U, heads, P, d_head], cell, groups, bias; see `Block.step`) the shared part
        of the history is in k_pre / v_pre and the own cache starts after it. Returns h [R, d],
        equal (to float rounding) to `encode` of the whole window at that position."""
        if self.training:
            raise RuntimeError("encode_step is for inference; call model.eval() first")
        cache_pos = pos if cache_pos is None else cache_pos
        x = self.embed_records(states) + self.pos_emb.weight[pos]
        for layer, blk in enumerate(self.blocks):
            pre = None
            if prefix is not None:
                k_pre, v_pre, cell, groups, bias = prefix
                pre = (k_pre[layer], v_pre[layer], cell, groups, bias)
            x = blk.step(x, k_cache[layer], v_cache[layer], cache_pos, n_keys, pre)
        return self.ln_f(x)

    def kv_bytes_per_row(self, element_bytes: int) -> int:
        """Memory of one rollout's key/value cache over a full window: layers x 2 x
        (max_len + 1) x d_model x element_bytes."""
        c = self.cfg
        return int(c["n_layers"]) * 2 * (self.max_len + 1) * int(c["d_model"]) * element_bytes

    # ------------------------------------------------------------------ decoder
    def _prefix_sums(self, targets: torch.Tensor) -> torch.Tensor:
        """targets [..., T] -> s [..., T, d], s_j = sum of embeddings of fields 0..j-1."""
        e = torch.stack([emb(targets[..., k]) for k, emb in enumerate(self.tgt_emb)], dim=-2)
        c = torch.cumsum(e, dim=-2)
        return torch.cat([torch.zeros_like(c[..., :1, :]), c], dim=-2)

    def decode_logits(self, h: torch.Tensor, targets: torch.Tensor) -> list[torch.Tensor]:
        """Teacher-forced logits for every target field, one head at a time (the reference
        for `decode_logits_padded`). h [..., d], targets [..., T]."""
        s = self._prefix_sums(targets)
        return [head(torch.cat([h, s[..., j, :]], dim=-1)) for j, head in enumerate(self.heads)]

    def decode_logits_padded(self, h: torch.Tensor, targets: torch.Tensor) -> torch.Tensor:
        """All ten heads in one batched pass (teacher forcing: every true field is known).

        h [N, d], targets [N, T] -> logits [T, N, VMAX] float32; columns past a field's own
        vocabulary are -inf, so a softmax over the row equals the softmax over that field's
        vocabulary. Head j's first layer W_j [hidden, 2d] acts on [h ; s_j], which is split
        as W_j[:, :d] h + W_j[:, d:] s_j: one matrix product for the h part of all heads and
        one batched product for the prefix-sum parts. Same parameters and same function as
        `decode_logits`; only the order of floating-point additions differs."""
        d = h.shape[-1]
        s = self._prefix_sums(targets)  # [N, T, d]
        w1 = torch.stack([hd[0].weight for hd in self.heads])  # [T, hidden, 2d]
        b1 = torch.stack([hd[0].bias for hd in self.heads])  # [T, hidden]
        t, hid, _ = w1.shape
        zh = torch.matmul(h, w1[:, :, :d].reshape(t * hid, d).t())  # [N, T * hidden]
        zh = zh.view(-1, t, hid).transpose(0, 1)  # [T, N, hidden]
        zs = torch.bmm(s.transpose(0, 1), w1[:, :, d:].transpose(1, 2))  # [T, N, hidden]
        z = F.gelu(zh + zs + b1.to(zh.dtype).unsqueeze(1))
        w2 = torch.stack(
            [F.pad(hd[2].weight, (0, 0, 0, VMAX - hd[2].weight.shape[0])) for hd in self.heads]
        )  # [T, VMAX, hidden]
        b2 = torch.stack([F.pad(hd[2].bias, (0, VMAX - hd[2].bias.shape[0])) for hd in self.heads])
        logits = torch.baddbmm(b2.unsqueeze(1).to(z.dtype), z, w2.transpose(1, 2).to(z.dtype))
        return logits.float() + self._logit_pad.unsqueeze(1)

    def field_logits(self, h: torch.Tensor, prefix: torch.Tensor, j: int) -> torch.Tensor:
        """Logits of field j given h [R, d] and the codes of fields 0..j-1 in prefix [R, >=j]."""
        s = torch.zeros_like(h)
        for k in range(j):
            s = s + self.tgt_emb[k](prefix[:, k])
        return self.heads[j](torch.cat([h, s], dim=-1))

    @torch.no_grad()
    def sample_transition(
        self,
        h: torch.Tensor,
        temperature: float = 1.0,
        generator: torch.Generator | None = None,
        mask_pad: bool = True,
    ) -> torch.Tensor:
        """Sample one transition per row of h [R, d] -> codes [R, T] (long).

        The heads run in float32 even inside an autocast region: bf16 logits are rounded to
        about 1/64 of their size, which would bias the sampled probabilities by around 1%.
        On rows whose sampled phase is END, fields outside END_FIELDS are set to code 0
        (the value they have on every END row in training).

        Sampling is Gumbel-max: argmax(logits / temperature + g) with g = -log(-log(u)), u
        uniform, is a draw from softmax(logits / temperature). The noise for all fields is
        drawn in one call, and the h half of every head's first layer is one matrix product
        (only the prefix-sum half depends on the fields sampled before), so a call launches
        about 11 small kernels per field.
        """
        r = h.shape[0]
        t = len(TARGET_NAMES)
        out = torch.zeros(r, t, dtype=torch.long, device=h.device)
        with torch.autocast(h.device.type, enabled=False):
            h = h.float()
            d = h.shape[-1]
            w1 = torch.stack([hd[0].weight for hd in self.heads])  # [T, hidden, 2d]
            hid = w1.shape[1]
            b1 = torch.stack([hd[0].bias for hd in self.heads]).reshape(-1)
            zh = torch.addmm(b1, h, w1[:, :, :d].reshape(t * hid, d).t())  # [R, T * hidden]
            bias = self._logit_no_pad if mask_pad else self._logit_pad
            if temperature > 0:
                u = torch.rand((r, t, VMAX), generator=generator, device=h.device)
                noise = -torch.log(-torch.log(u.clamp_(1e-10, 1.0 - 1e-7))) + bias
            s = torch.zeros_like(h)
            is_end = None
            for j, head in enumerate(self.heads):
                n = TARGET_SIZES[j]
                z = torch.addmm(zh[:, j * hid : (j + 1) * hid], s, w1[j, :, d:].t())
                logits = F.linear(F.gelu(z), head[2].weight, head[2].bias)
                if temperature <= 0:
                    code = (logits + bias[j, :n]).argmax(-1)
                else:
                    code = (logits / temperature + noise[:, j, :n]).argmax(-1)
                if j == PHASE_T:
                    is_end = code == S.PHASE["END"]
                elif TARGET_NAMES[j] not in END_FIELDS:
                    code = code.masked_fill(is_end, 0)
                out[:, j] = code
                if j < len(self.tgt_emb):
                    s = s + self.tgt_emb[j](code)
        return out

    # ------------------------------------------------------------------ loss
    @staticmethod
    def end_prefix(targets: torch.Tensor, last: torch.Tensor) -> torch.Tensor:
        """Targets with every field outside END_FIELDS set to 0 on last records."""
        tg = targets.clone()
        tg[..., list(NOT_END_T)] = tg[..., list(NOT_END_T)] * (~last).unsqueeze(-1).to(tg.dtype)
        return tg

    @staticmethod
    def target_masks(targets: torch.Tensor, mask: torch.Tensor, last: torch.Tensor) -> list:
        """Per field, which targets count in the loss ([B, L] bool each).

        A target counts when its record is real and its code is not padding. Code 0 is
        padding for every field except down and ydstogo ("not a down"). On a game's last
        record only END_FIELDS count: phase (END) always, score_delta and elapsed when
        records.py filled them (code > 0).
        """
        out = []
        for j, name in enumerate(TARGET_NAMES):
            y = targets[..., j]
            if j == PHASE_T:
                m = mask
            elif name in ZERO_IS_VALUE:
                m = mask & ~last
            else:
                m = mask & (y != 0)
                if name not in END_FIELDS:
                    m = m & ~last
            out.append(m)
        return out

    def end_prefix_fast(self, targets: torch.Tensor, last: torch.Tensor) -> torch.Tensor:
        """`end_prefix` without list indexing (which copies an index list to the GPU and so
        waits for it): a broadcast multiply by a keep mask held on the device."""
        keep = (~last).unsqueeze(-1) | self._end_field
        return targets * keep.to(targets.dtype)

    def target_mask_tensor(
        self, targets: torch.Tensor, mask: torch.Tensor, last: torch.Tensor
    ) -> torch.Tensor:
        """`target_masks` for all fields at once: [..., T] bool, the same rule."""
        not_last = (~last).unsqueeze(-1)
        other = (targets != 0) & (self._end_field | not_last)
        rule = torch.where(self._zero_is_value, not_last, other) | self._is_phase
        return mask.unsqueeze(-1) & rule

    def loss(
        self, batch: dict, denom: torch.Tensor | None = None, parallel: bool | None = None
    ) -> tuple[torch.Tensor, dict]:
        """Sum over fields of the mean cross-entropy per target (see target_masks).

        batch: header [B,H], states [B,L,S], targets [B,L,T], mask [B,L] (real record),
        last [B,L] (last record of its game). denom [T], if given, replaces the per-field
        target counts in the means (used to split one batch into exact micro-batches).
        parallel (default cfg["parallel_decoder"]): all heads in one batched pass and one
        cross-entropy call; False runs the one-head-at-a-time reference. Same loss either way.
        Returns (total, {"fields": {name: mean NLL tensor}, "sums": [T], "counts": [T]}).
        """
        if parallel is None:
            parallel = bool(self.cfg.get("parallel_decoder", True))
        h = self.encode(batch["header"], batch["states"])
        mask, last = batch["mask"], batch["last"]
        if parallel:
            return self._loss_parallel(h, batch["targets"], mask, last, denom)
        tg = self.end_prefix(batch["targets"], last)
        logits = self.decode_logits(h, tg)
        masks = self.target_masks(tg, mask, last)
        total = h.new_zeros((), dtype=torch.float32)
        fields, sums, counts = {}, [], []
        for j, (name, _) in enumerate(S.TARGET_FIELDS):
            y = tg[..., j]
            m = masks[j]
            ce = F.cross_entropy(logits[j].float().flatten(0, 1), y.flatten(), reduction="none")
            ce = ce.view_as(y) * m
            n = m.sum()
            sm = ce.sum()
            d = n.clamp(min=1) if denom is None else denom[j].clamp(min=1)
            mean = sm / d
            fields[name] = mean
            sums.append(sm.detach())
            counts.append(n.detach())
            total = total + mean
        return total, {"fields": fields, "sums": torch.stack(sums), "counts": torch.stack(counts)}

    def _loss_parallel(self, h, targets, mask, last, denom):
        t = len(TARGET_NAMES)
        tg = self.end_prefix_fast(targets, last)
        tf = tg.reshape(-1, t)
        logits = self.decode_logits_padded(h.reshape(-1, h.shape[-1]), tf)  # [T, N, VMAX]
        y = tf.t()  # [T, N]
        ce = F.cross_entropy(logits.reshape(-1, VMAX), y.reshape(-1), reduction="none")
        m = self.target_mask_tensor(tg, mask, last).reshape(-1, t).t()  # [T, N] bool
        sums = (ce.view(t, -1) * m).sum(1)
        counts = m.sum(1)
        d = counts.clamp(min=1) if denom is None else denom.clamp(min=1)
        means = sums / d
        fields = {name: means[j] for j, name in enumerate(TARGET_NAMES)}
        return means.sum(), {"fields": fields, "sums": sums.detach(), "counts": counts.detach()}


def window_start(game_start: int, i: int, max_len: int) -> int:
    """First record of the context window that ends at record i (inclusive)."""
    return max(game_start, i + 1 - max_len)


def set_seed(seed: int) -> None:
    np.random.seed(seed % (2**32))
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)


def device_of(model: nn.Module) -> torch.device:
    return next(model.parameters()).device


def param_summary(model: WorldModel) -> str:
    n = model.count_params()
    return f"{n:,} parameters ({n / 1e6:.2f} M), {4 * n / 2**20:.1f} MiB in float32"


__all__ = [
    "DEFAULT_CFG",
    "ZERO_IS_VALUE",
    "END_FIELDS",
    "WorldModel",
    "merge_cfg",
    "window_start",
    "set_seed",
    "device_of",
    "param_summary",
]
```

```
%%writefile /content/wm_code/train.py
"""Batching, training loop (AdamW, warmup + cosine, AMP, resume, early stopping) and NLL.

Speed notes (see the first Colab run in results/world_model/colab_t4_run_summary.md, which
trained on the CPU because the model was never moved to the GPU):
- `train_model` moves the model to cfg["device"].
- Each split is pre-built ONCE as padded int16 tensors (games x window x fields) and kept on
  the GPU when it fits; a batch is an index operation (`tensor_batch`), not a Python loop
  (`collate`, kept as the reference).
- No per-step host synchronisation: losses are summed on the device and read once per epoch,
  record counts come from host-side lengths, and on CUDA the optimizer is fused AdamW, which
  takes the fp16 GradScaler's inf check on the device instead of calling .item().
- Every epoch's wall time is split into batch construction, host-to-device, forward,
  backward plus optimizer, validation and checkpointing, logged and kept in the history.
"""

from __future__ import annotations

import copy
import math
import os
import time
from collections.abc import Callable, Iterator
from dataclasses import dataclass

import numpy as np
import schema as S
import torch
from model import TARGET_NAMES, VMAX, WorldModel, device_of, merge_cfg


# ---------------------------------------------------------------------- batching (reference)
def collate(split: S.Split, games: np.ndarray, max_len: int) -> dict:
    """Pad the given games into one batch, one game per row, each cut to its last max_len."""
    starts = split.offsets[games]
    ends = split.offsets[games + 1]
    starts = np.maximum(starts, ends - max_len)
    lens = (ends - starts).astype(np.int64)
    b, n = len(games), int(lens.max())
    states = np.zeros((b, n, len(S.STATE_FIELDS)), dtype=np.int64)
    targets = np.zeros((b, n, len(S.TARGET_FIELDS)), dtype=np.int64)
    mask = np.zeros((b, n), dtype=bool)
    last = np.zeros((b, n), dtype=bool)
    for r, (a, e, k) in enumerate(zip(starts, ends, lens, strict=True)):
        states[r, :k] = split.states[a:e]
        targets[r, :k] = split.targets[a:e]
        mask[r, :k] = True
        last[r, k - 1] = True
    return {
        "header": torch.from_numpy(split.header[games].astype(np.int64)),
        "states": torch.from_numpy(states),
        "targets": torch.from_numpy(targets),
        "mask": torch.from_numpy(mask),
        "last": torch.from_numpy(last),
        "games": torch.from_numpy(np.asarray(games, dtype=np.int64)),
    }


def make_batches(
    split: S.Split,
    batch_games: int,
    max_len: int,
    rng: np.random.Generator | None,
) -> Iterator[dict]:
    """Yield padded batches of whole games. rng=None keeps the split's order (evaluation)."""
    n_games = len(split.offsets) - 1
    order = np.arange(n_games) if rng is None else rng.permutation(n_games)
    for b in range(0, n_games, batch_games):
        yield collate(split, order[b : b + batch_games], max_len)


def to_device(batch: dict, device: torch.device) -> dict:
    nb = device.type == "cuda"
    return {k: v.to(device, non_blocking=nb) for k, v in batch.items()}


# ---------------------------------------------------------------------- pre-built tensors
@dataclass
class TensorSplit:
    """A split built once as padded tensors, one row per game, each game cut to its LAST
    max_len records (as `collate` does). A batch is then an index operation."""

    states: torch.Tensor  # [G, W, S] int16, right-padded with 0 (W = longest kept game)
    targets: torch.Tensor  # [G, W, T] int16
    header: torch.Tensor  # [G, H] int16
    lens: torch.Tensor  # [G] int64, records kept per game
    lens_np: np.ndarray  # host copy: batch widths and record counts need no device sync
    device: torch.device

    def nbytes(self) -> int:
        return sum(
            t.numel() * t.element_size()
            for t in (self.states, self.targets, self.header, self.lens)
        )


def prebuild_split(split: S.Split, max_len: int, device: torch.device | str = "cpu") -> TensorSplit:
    device = torch.device(device)
    starts = np.asarray(split.offsets[:-1], dtype=np.int64)
    ends = np.asarray(split.offsets[1:], dtype=np.int64)
    starts = np.maximum(starts, ends - max_len)
    lens = (ends - starts).astype(np.int64)
    w = max(1, int(lens.max()) if len(lens) else 1)
    ar = np.arange(w)
    valid = ar[None, :] < lens[:, None]
    src = np.where(valid, starts[:, None] + ar[None, :], 0)
    states = (split.states[src] * valid[..., None]).astype(np.int16)
    targets = (split.targets[src] * valid[..., None]).astype(np.int16)
    pin = device.type == "cpu" and torch.cuda.is_available()

    def put(a: np.ndarray) -> torch.Tensor:
        t = torch.from_numpy(np.ascontiguousarray(a))
        return t.pin_memory() if pin else t.to(device)

    return TensorSplit(
        states=put(states),
        targets=put(targets),
        header=put(split.header.astype(np.int16)),
        lens=put(lens),
        lens_np=lens,
        device=device,
    )


def _gather(ts: TensorSplit, games_np: np.ndarray, games_t: torch.Tensor) -> dict:
    n = max(1, int(ts.lens_np[games_np].max()))
    return {
        "header": ts.header[games_t],
        "states": ts.states[games_t, :n],
        "targets": ts.targets[games_t, :n],
        "lens": ts.lens[games_t],
        "games": games_t,
    }


def _finish(raw: dict) -> dict:
    lens = raw["lens"]
    ar = torch.arange(raw["states"].shape[1], device=lens.device)
    return {
        "header": raw["header"].long(),
        "states": raw["states"].long(),
        "targets": raw["targets"].long(),
        "mask": ar.unsqueeze(0) < lens.unsqueeze(1),
        "last": ar.unsqueeze(0) == (lens - 1).unsqueeze(1),
        "games": raw["games"].long(),
    }


def tensor_batch(ts: TensorSplit, games, device: torch.device | None = None) -> dict:
    """The batch `collate(split, games, max_len)` builds, from the pre-built tensors."""
    games_np = np.asarray(games, dtype=np.int64)
    raw = _gather(ts, games_np, torch.as_tensor(games_np, device=ts.device))
    if device is not None and torch.device(device) != ts.device:
        raw = to_device(raw, torch.device(device))
    return _finish(raw)


def _storage_device(cfg: dict, device: torch.device, n_bytes: int) -> torch.device:
    """Keep the pre-built tensors on the GPU when they take under a quarter of free memory
    (the full training split is about 30 MB)."""
    if device.type != "cuda":
        return device
    want = cfg.get("data_on_device", "auto")
    if want is False or want == "never":
        return torch.device("cpu")
    if want == "auto":
        free, _ = torch.cuda.mem_get_info(device)
        if n_bytes > 0.25 * free:
            return torch.device("cpu")
    return device


def _split_bytes(split: S.Split, max_len: int) -> int:
    lens = np.minimum(np.diff(split.offsets), max_len)
    w = int(lens.max()) if len(lens) else 1
    return len(lens) * w * (len(S.STATE_FIELDS) + len(S.TARGET_FIELDS)) * 2


# ---------------------------------------------------------------------- timing
DURATION_KEYS = (
    "batch_s",
    "h2d_s",
    "forward_s",
    "backward_opt_s",
    "train_loop_s",
    "validation_s",
    "checkpoint_s",
    "epoch_s",
)


class PhaseTimer:
    """Wall-clock split of a loop into named phases.

    On CPU every phase is timed directly. On CUDA kernels run asynchronously, so a phase's
    CPU time is not its GPU time: every `sync_every`-th iteration (starting with the
    second, after warm-up) is synchronised at each phase boundary, and the shares measured
    on those iterations split the loop's measured total. The other iterations run without
    any synchronisation."""

    def __init__(self, device: torch.device, sync_every: int = 8) -> None:
        self.cuda = device.type == "cuda"
        self.sync_every = max(1, int(sync_every))
        self.direct: dict[str, float] = {}
        self.sampled: dict[str, float] = {}
        self.iters = 0
        self.sampled_iters = 0
        self._sync = False
        self._t = time.perf_counter()

    def start(self) -> None:
        self.iters += 1
        self._sync = self.cuda and self.iters % self.sync_every == 2 % self.sync_every
        if self._sync:
            torch.cuda.synchronize()
            self.sampled_iters += 1
        self._t = time.perf_counter()

    def mark(self, name: str) -> None:
        if self._sync:
            torch.cuda.synchronize()
        t = time.perf_counter()
        dt = t - self._t
        self.direct[name] = self.direct.get(name, 0.0) + dt
        if self._sync:
            self.sampled[name] = self.sampled.get(name, 0.0) + dt
        self._t = t

    def split(self, total_s: float) -> dict[str, float]:
        """Seconds per phase: measured on CPU; on CUDA, total_s split by the shares of the
        synchronised iterations (or the CPU-side times if none was sampled)."""
        src = self.sampled if (self.cuda and self.sampled_iters) else self.direct
        tot = sum(src.values())
        if not self.cuda or tot <= 0:
            return dict(self.direct)
        return {k: total_s * v / tot for k, v in src.items()}

    def method(self) -> str:
        if not self.cuda:
            return "measured per phase (CPU, synchronous)"
        return (
            f"total measured; split by phase from {self.sampled_iters} of {self.iters} "
            "steps synchronised at every phase boundary"
        )


# ---------------------------------------------------------------------- precision
def amp_settings(device: torch.device) -> tuple[bool, torch.dtype, bool]:
    """(enabled, dtype, use_grad_scaler). bf16 only on GPUs with native support (sm80+)."""
    if device.type != "cuda":
        return False, torch.float32, False
    major, _ = torch.cuda.get_device_capability(device)
    if major >= 8:
        return True, torch.bfloat16, False
    return True, torch.float16, True


def train_bytes_per_token(cfg: dict, amp: bool) -> float:
    """Rough activation memory of one padded token in a training step (forward activations
    kept for backward, with a 1.5x margin for temporaries and allocator slack)."""
    d, layers, ff, hid = cfg["d_model"], cfg["n_layers"], cfg["d_ff"], cfg["dec_hidden"]
    a = 2 if amp else 4
    t = len(S.TARGET_FIELDS)
    enc = layers * (d * (16 + 8 * a) + ff * 3 * a)  # layer norms in fp32, the rest in AMP
    dec = t * (d * (4 + a) + hid * 3 * a + VMAX * 12)  # prefix sums, hidden, padded logits
    emb = 4 * d * (len(S.STATE_FIELDS) + 4)
    return 1.5 * (enc + dec + emb)


def _planned_micro(cfg: dict, batch_games: int, width: int, device: torch.device, amp: bool) -> int:
    """Micro-batches per step so one micro-batch fits in cfg["train_mem_frac"] of free GPU
    memory (1 on CPU). The out-of-memory fallback still doubles it if this is too low."""
    if device.type != "cuda":
        return 1
    free, _ = torch.cuda.mem_get_info(device)
    need = batch_games * width * train_bytes_per_token(cfg, amp)
    budget = float(cfg.get("train_mem_frac", 0.6)) * free
    k = 1
    while k < batch_games and need / k > budget:
        k *= 2
    return k


# ---------------------------------------------------------------------- evaluation
@torch.no_grad()
def evaluate_nll(
    model: WorldModel,
    split: S.Split,
    cfg: dict | None = None,
    prebuilt: TensorSplit | None = None,
) -> dict:
    """Per-field mean NLL (nats per target) and total NLL per record over a split.

    Games are batched by length (less padding); sums stay on the device until the end."""
    c = merge_cfg({**model.cfg, **(cfg or {})})
    device = device_of(model)
    enabled, dtype, _ = amp_settings(device)
    was_training = model.training
    model.eval()
    ts = prebuilt
    if ts is None:
        ts = prebuild_split(
            split, model.max_len, _storage_device(c, device, _split_bytes(split, model.max_len))
        )
    t = len(S.TARGET_FIELDS)
    sums = torch.zeros(t, dtype=torch.float64, device=device)
    counts = torch.zeros(t, dtype=torch.float64, device=device)
    order = np.argsort(ts.lens_np, kind="stable")
    order_t = torch.as_tensor(order, device=ts.device)
    bg = max(1, int(c["eval_batch_games"]))
    for b in range(0, len(order), bg):
        raw = _gather(ts, order[b : b + bg], order_t[b : b + bg])
        if ts.device != device:
            raw = to_device(raw, device)
        with torch.autocast(device.type, dtype=dtype, enabled=enabled):
            _, info = model.loss(_finish(raw))
        sums += info["sums"].double()
        counts += info["counts"].double()
    model.train(was_training)
    sums, counts = sums.cpu(), counts.cpu()
    n_records = int(ts.lens_np.sum())
    fields = {n: float(sums[j] / max(counts[j], 1.0)) for j, n in enumerate(TARGET_NAMES)}
    return {
        "total": float(sums.sum() / max(n_records, 1)),  # nats per record, all fields
        "sum_of_field_means": float(sum(fields.values())),
        "fields": fields,
        "n_records": n_records,
    }


# ---------------------------------------------------------------------- training
def _lr_factor(step: int, progress: float, cfg: dict) -> float:
    """Linear warmup over warmup_steps, then cosine from 1 to min_lr_frac over all epochs."""
    warm = min(1.0, (step + 1) / max(1, cfg["warmup_steps"]))
    p = min(1.0, max(0.0, progress / max(1, cfg["epochs"])))
    floor = cfg["min_lr_frac"]
    return warm * (floor + (1 - floor) * 0.5 * (1 + math.cos(math.pi * p)))


def _param_groups(model: torch.nn.Module, weight_decay: float) -> list[dict]:
    decay, no_decay = [], []
    for name, p in model.named_parameters():
        is_matrix = p.dim() >= 2 and "emb" not in name
        (decay if is_matrix else no_decay).append(p)
    return [
        {"params": decay, "weight_decay": weight_decay},
        {"params": no_decay, "weight_decay": 0.0},
    ]


def _match_fused(opt: torch.optim.Optimizer, fused: bool) -> None:
    """After loading an optimizer state saved with the other setting: fused AdamW keeps its
    step counters on the parameter's device, the unfused one on the CPU."""
    for group in opt.param_groups:
        group["fused"] = True if fused else None
        if fused:
            group["foreach"] = None
        for p in group["params"]:
            st = opt.state.get(p)
            if st and torch.is_tensor(st.get("step")):
                dev = p.device if fused else torch.device("cpu")
                st["step"] = st["step"].to(device=dev, dtype=torch.float32)


def _save(path: str, payload: dict) -> None:
    tmp = f"{path}.tmp"
    torch.save(payload, tmp)
    os.replace(tmp, path)


def _is_oom(err: BaseException) -> bool:
    oom = getattr(torch, "OutOfMemoryError", None) or getattr(torch.cuda, "OutOfMemoryError", None)
    return (oom is not None and isinstance(err, oom)) or "out of memory" in str(err).lower()


def load_checkpoint(path: str, cfg_override: dict | None = None, device: str | None = None):
    """Rebuild the BEST model stored at path. Returns (model, checkpoint dict)."""
    ck = torch.load(path, map_location="cpu", weights_only=False)
    model = WorldModel({**ck["cfg"], **(cfg_override or {})})
    model.load_state_dict(ck["model"])
    if device is not None:
        model.to(device)
    model.eval()
    return model, ck


def _micro_slices(batch: dict, k: int, lens: np.ndarray | None = None) -> list[dict]:
    """Split a batch into k row groups, each trimmed to its own longest game (the causal
    encoder makes right padding irrelevant, so the results do not change). lens (host
    copy of the rows' record counts) avoids reading the mask back from the GPU."""
    b = batch["mask"].shape[0]
    out = []
    for rows in np.array_split(np.arange(b), min(k, b)):
        if len(rows) == 0:
            continue
        sl = slice(int(rows[0]), int(rows[-1]) + 1)
        if lens is not None:
            n = max(1, int(np.asarray(lens)[sl].max()))
        else:
            n = max(1, int(batch["mask"][sl].sum(1).max()))
        part = {}
        for key, v in batch.items():
            v = v[sl]
            if key in ("states", "targets", "mask", "last"):
                v = v[:, :n]
            part[key] = v
        out.append(part)
    return out


def _full_counts(model: WorldModel, batch: dict) -> torch.Tensor:
    """Per-field target counts of a whole batch (the denominators of its loss)."""
    tg = model.end_prefix_fast(batch["targets"], batch["last"])
    m = model.target_mask_tensor(tg, batch["mask"], batch["last"])
    return m.reshape(-1, m.shape[-1]).sum(0).float()


def _fresh_scaler(scaler: torch.amp.GradScaler, enabled: bool) -> torch.amp.GradScaler:
    """A scaler with the same scale and growth tracker but no per-step state (an OOM between
    unscale_ and update would otherwise make the next unscale_ raise)."""
    new = torch.amp.GradScaler("cuda", enabled=enabled)
    if enabled:
        new.load_state_dict(scaler.state_dict())
    return new


def make_optimizer(
    model: WorldModel, c: dict, device: torch.device, log: Callable[[str], None] = print
) -> tuple[torch.optim.Optimizer, bool]:
    """AdamW; fused on CUDA (one kernel for all parameters, and with the fp16 GradScaler the
    inf check stays on the device). Some torch builds reject fused=True (older versions,
    or parameters it cannot handle); then the default implementation is used. Returns
    (optimizer, fused)."""
    groups = _param_groups(model, c["weight_decay"])
    kw = {"lr": c["lr"], "betas": tuple(c["betas"])}
    if device.type == "cuda":
        try:
            return torch.optim.AdamW(groups, fused=True, **kw), True
        except (RuntimeError, TypeError, ValueError) as err:
            log(f"fused AdamW is not available here ({type(err).__name__}: {err}); "
                "using the default AdamW")
            groups = _param_groups(model, c["weight_decay"])
    return torch.optim.AdamW(groups, **kw), False


def _place(model: WorldModel, want: str | None, log: Callable[[str], None]) -> torch.device:
    """Move the model to cfg["device"] (falls back to the CPU when CUDA is missing)."""
    if want:
        dev = torch.device(want)
        if dev.type == "cuda" and not torch.cuda.is_available():
            log("cfg asks for cuda but no GPU is available; training on the CPU")
            dev = torch.device("cpu")
        model.to(dev)
    return device_of(model)


def train_model(
    model: WorldModel,
    train_split: S.Split,
    val_split: S.Split,
    cfg: dict,
    ckpt_path: str,
    deadline_ts: float,
    log: Callable[[str], None] = print,
) -> dict:
    """Train with AdamW, warmup + cosine, gradient clipping and AMP on CUDA.

    The model is moved to cfg["device"] when given. One checkpoint file at ckpt_path holds:
    "model" = best weights so far (by validation NLL), "current_model" + optimizer + scaler +
    RNG states + epoch + step + position inside the epoch (to resume exactly), cfg, history.
    If ckpt_path exists, training resumes from it. Validation NLL is evaluated after every
    epoch; training stops on early stopping (patience), at cfg["epochs"], at
    cfg["stop_after_epoch"], or cleanly at the first batch after time.time() > deadline_ts
    (a mid-epoch stop saves the position and resumes there). On out-of-memory the SAME step
    is retried with the batch split into twice as many micro-batches whose gradients are
    accumulated with the full batch's denominators, so the data order, batch composition and
    schedule do not change; on CUDA the starting number of micro-batches is planned from
    free memory, so a batch that cannot fit never raises. Each history entry has a "timing"
    dict: seconds of batch construction, host-to-device, forward, backward plus optimizer,
    validation and checkpointing (the checkpoint of an epoch is written before its own
    duration is known, so a resumed history has None there for the epoch before the stop).
    At return the model holds the best weights (or, if no epoch finished, the current ones).
    """
    c = merge_cfg({**model.cfg, **cfg})
    device = _place(model, c.get("device"), log)
    enabled, amp_dtype, use_scaler = amp_settings(device)
    torch.manual_seed(c["seed"])
    opt, fused = make_optimizer(model, c, device, log)  # after the model is on its device
    scaler = torch.amp.GradScaler("cuda", enabled=use_scaler)
    state = {
        "epoch": 0,
        "step": 0,
        "best_val": float("inf"),
        "best_epoch": -1,
        "bad_epochs": 0,
        "history": [],
        "batch_games": int(c["batch_games"]),
        "micro": 1,  # micro-batches per step (grows on out-of-memory)
        "batch_in_epoch": 0,  # batches of the current epoch already done (mid-epoch resume)
        "epoch_loss_sum": 0.0,
        "epoch_loss_n": 0,
        "finished": False,
    }
    best_weights = copy.deepcopy({k: v.detach().cpu() for k, v in model.state_dict().items()})

    if ckpt_path and os.path.exists(ckpt_path):
        ck = torch.load(ckpt_path, map_location="cpu", weights_only=False)
        model.load_state_dict(ck["current_model"])
        opt.load_state_dict(ck["optimizer"])
        _match_fused(opt, fused)
        if ck.get("scaler") and use_scaler:
            scaler.load_state_dict(ck["scaler"])
        best_weights = ck["model"]
        for k in state:
            if k in ck:
                state[k] = ck[k]
        torch.set_rng_state(ck["torch_rng"])
        if device.type == "cuda" and ck.get("cuda_rng") is not None:
            torch.cuda.set_rng_state_all(ck["cuda_rng"])
        log(
            f"resumed from {ckpt_path}: epoch {state['epoch']} batch {state['batch_in_epoch']}, "
            f"step {state['step']}, best val {state['best_val']:.4f}"
        )

    t_prep = time.perf_counter()
    store = _storage_device(
        c, device, _split_bytes(train_split, model.max_len) + _split_bytes(val_split, model.max_len)
    )
    ts_train = prebuild_split(train_split, model.max_len, store)
    ts_val = prebuild_split(val_split, model.max_len, store)
    width = int(ts_train.lens_np.max()) if len(ts_train.lens_np) else 1
    planned = _planned_micro(c, state["batch_games"], width, device, enabled)
    if planned > state["micro"]:
        log(
            f"planning {planned} micro-batches per step from free GPU memory "
            f"(about {train_bytes_per_token(c, enabled) / 1024:.0f} KiB per padded token)"
        )
        state["micro"] = planned
    prep_s = time.perf_counter() - t_prep
    gpu = torch.cuda.get_device_name(device) if device.type == "cuda" else "cpu"
    log(
        f"training on {device} ({gpu}); data tensors on {store} "
        f"({(ts_train.nbytes() + ts_val.nbytes()) / 2**20:.1f} MiB, built in {prep_s:.1f}s); "
        f"AMP {amp_dtype if enabled else 'off'}"
        f"{' with GradScaler' if use_scaler else ''}; {'fused ' if fused else ''}AdamW; "
        f"{state['batch_games']} games per step, {state['micro']} micro-batch(es)"
    )

    def checkpoint() -> None:
        if not ckpt_path:
            return
        _save(
            ckpt_path,
            {
                "model": best_weights,
                "current_model": {k: v.detach().cpu() for k, v in model.state_dict().items()},
                "optimizer": opt.state_dict(),
                "scaler": scaler.state_dict() if use_scaler else None,
                "cfg": model.cfg,
                "torch_rng": torch.get_rng_state(),
                "cuda_rng": torch.cuda.get_rng_state_all() if device.type == "cuda" else None,
                **state,
            },
        )

    def autocast():
        return torch.autocast(device.type, dtype=amp_dtype, enabled=enabled)

    def train_step(batch: dict, timer: PhaseTimer, lens: np.ndarray) -> torch.Tensor:
        """One optimizer step on batch, split into state["micro"] micro-batches. Returns the
        loss as a device tensor (no synchronisation)."""
        nonlocal scaler
        while True:
            k = int(state["micro"])
            try:
                opt.zero_grad(set_to_none=True)
                if k == 1:
                    with autocast():
                        total, _ = model.loss(batch)
                    timer.mark("forward")
                    scaler.scale(total).backward()
                    value = total.detach()
                    del total
                else:
                    denom = _full_counts(model, batch)
                    value = torch.zeros((), dtype=torch.float32, device=device)
                    for part in _micro_slices(batch, k, lens):
                        with autocast():
                            part_total, _ = model.loss(part, denom=denom)
                        timer.mark("forward")
                        scaler.scale(part_total).backward()
                        value = value + part_total.detach()
                        del part_total
                        timer.mark("backward_opt")
                scaler.unscale_(opt)
                torch.nn.utils.clip_grad_norm_(model.parameters(), c["grad_clip"])
                scaler.step(opt)
                scaler.update()
                timer.mark("backward_opt")
                return value
            except RuntimeError as err:  # torch.OutOfMemoryError subclasses RuntimeError
                rows = int(batch["mask"].shape[0])
                if not _is_oom(err) or k >= rows:
                    raise
            # outside the except block, so the traceback (and its activations) is released
            opt.zero_grad(set_to_none=True)
            scaler = _fresh_scaler(scaler, use_scaler)
            if device.type == "cuda":
                torch.cuda.empty_cache()
            state["micro"] = 2 * k
            log(f"out of memory with {k} micro-batch(es) per step; retrying with {2 * k}")

    stopped = "finished" if state["finished"] else "epochs"
    n_games = len(train_split.offsets) - 1
    records_seen, train_time = 0, 0.0
    sync_every = int(c.get("timing_sync_every", 8))
    while not state["finished"] and state["epoch"] < c["epochs"]:
        epoch = state["epoch"]
        if c["stop_after_epoch"] is not None and epoch >= c["stop_after_epoch"]:
            stopped = "stop_after_epoch"
            break
        if time.time() > deadline_ts:
            stopped = "deadline"
            break
        model.train()
        if device.type == "cuda":
            torch.cuda.reset_peak_memory_stats(device)
        bg = state["batch_games"]
        n_batches = math.ceil(n_games / bg)
        rng = np.random.default_rng([c["seed"], epoch])
        order = rng.permutation(n_games)  # the same batches as make_batches(train, bg, ., rng)
        t0 = time.time()
        loop_t0 = time.perf_counter()
        timer = PhaseTimer(device, sync_every)
        order_t = torch.as_tensor(order, device=ts_train.device)
        loss_acc = torch.full(
            (), float(state["epoch_loss_sum"]), dtype=torch.float64, device=device
        )
        skip = int(state["batch_in_epoch"])  # done before a mid-epoch stop
        hit_deadline = False
        epoch_records = epoch_steps = 0
        for bi in range(skip, n_batches):
            if time.time() > deadline_ts:
                hit_deadline = True
                break
            timer.start()
            sel = order[bi * bg : (bi + 1) * bg]
            raw = _gather(ts_train, sel, order_t[bi * bg : (bi + 1) * bg])
            timer.mark("batch")
            if ts_train.device != device:
                raw = to_device(raw, device)
                timer.mark("h2d")
            batch = _finish(raw)
            factor = _lr_factor(state["step"], epoch + bi / n_batches, c)
            for g in opt.param_groups:
                g["lr"] = c["lr"] * factor
            timer.mark("batch")
            value = train_step(batch, timer, ts_train.lens_np[sel])
            loss_acc += value
            state["step"] += 1
            state["batch_in_epoch"] = bi + 1
            state["epoch_loss_n"] += 1
            n_rec = int(ts_train.lens_np[sel].sum())
            records_seen += n_rec
            epoch_records += n_rec
            epoch_steps += 1
        state["epoch_loss_sum"] = float(loss_acc)  # the one read-back of the epoch
        loop_s = time.perf_counter() - loop_t0
        train_time += time.time() - t0
        phases = timer.split(loop_s)
        timing = {
            "batch_s": phases.get("batch", 0.0),
            "h2d_s": phases.get("h2d", 0.0),
            "forward_s": phases.get("forward", 0.0),
            "backward_opt_s": phases.get("backward_opt", 0.0),
            "train_loop_s": loop_s,
            "steps": epoch_steps,
            "train_records": epoch_records,
            "train_records_per_s": epoch_records / loop_s if loop_s > 0 else float("nan"),
            "micro_batches": int(state["micro"]),
            "data_on": str(ts_train.device),
            "method": timer.method(),
        }
        if hit_deadline:
            stopped = "deadline"
            if state["best_epoch"] < 0:  # no validated epoch yet: keep what was learned
                best_weights = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
            checkpoint()
            log(
                f"deadline reached during epoch {epoch}; saved current state "
                f"({epoch_steps} steps, {timing['train_records_per_s']:.0f} records/s)"
            )
            break
        t_val = time.perf_counter()
        val = evaluate_nll(model, val_split, c, prebuilt=ts_val)
        timing["validation_s"] = time.perf_counter() - t_val
        train_loss = state["epoch_loss_sum"] / max(state["epoch_loss_n"], 1)
        improved = val["total"] < state["best_val"]
        if improved:
            state["best_val"] = val["total"]
            state["best_epoch"] = epoch
            state["bad_epochs"] = 0
            best_weights = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
        else:
            state["bad_epochs"] += 1
        if device.type == "cuda":
            timing["peak_mem_gb"] = torch.cuda.max_memory_allocated(device) / 2**30
        timing["checkpoint_s"] = None
        state["history"].append(
            {
                "epoch": epoch,
                "train_loss": train_loss,
                "val_total": val["total"],
                "val_fields": val["fields"],
                "lr": opt.param_groups[0]["lr"],
                "seconds": time.time() - t0,
                "batch_games": bg,
                "micro": int(state["micro"]),
                "timing": timing,
            }
        )
        state["epoch"] = epoch + 1
        state["batch_in_epoch"] = 0
        state["epoch_loss_sum"], state["epoch_loss_n"] = 0.0, 0
        if state["bad_epochs"] >= c["patience"]:
            state["finished"] = True
            stopped = "early_stopping"
        elif state["epoch"] >= c["epochs"]:
            state["finished"] = True
        t_ck = time.perf_counter()
        checkpoint()
        timing["checkpoint_s"] = time.perf_counter() - t_ck
        timing["epoch_s"] = time.time() - t0
        log(
            f"epoch {epoch}: train {train_loss:.4f}  val {val['total']:.4f} nats/record"
            f"{'  (best)' if improved else ''}  {time.time() - t0:.0f}s | "
            f"{timing['train_records_per_s']:,.0f} records/s; batch {timing['batch_s']:.1f}s, "
            f"h2d {timing['h2d_s']:.1f}s, forward {timing['forward_s']:.1f}s, backward+opt "
            f"{timing['backward_opt_s']:.1f}s, validation {timing['validation_s']:.1f}s, "
            f"checkpoint {timing['checkpoint_s']:.1f}s"
        )
    model.load_state_dict(best_weights)
    totals: dict[str, float] = {}
    for h in state["history"]:
        for k in DURATION_KEYS:
            v = (h.get("timing") or {}).get(k)
            if isinstance(v, (int, float)):
                totals[k] = totals.get(k, 0.0) + float(v)
    return {
        "stopped": stopped,
        "epochs_done": state["epoch"],
        "best_epoch": state["best_epoch"],
        "best_val_nll": state["best_val"],
        "batch_games": state["batch_games"],
        "micro_batches": int(state["micro"]),
        "history": state["history"],
        "train_records_per_s": records_seen / train_time if train_time > 0 else float("nan"),
        "device": str(device),
        "amp": str(amp_dtype) if enabled else None,
        "grad_scaler": bool(use_scaler),
        "optimizer": "fused AdamW" if fused else "AdamW",
        "data_on": str(ts_train.device),
        "prebuild_s": prep_s,
        "timing_totals_s": totals,
    }
```

```
%%writefile /content/wm_code/simulate.py
"""Use the trained world model as a game simulator.

A rollout starts from a real record (its game's header and records up to and including it),
optionally with that record's action forced, and samples transitions until the game ends.
Bookkeeping (clock, score, timeouts) is done by `schema.apply_transition`; the model only
samples the transition fields. The model's own learned action distribution plays both teams
for every record after the start, so the result is "what happens if this team makes this call
and then everyone behaves like an average NFL team in the training data".
"""

from __future__ import annotations

import time

import numpy as np
import pandas as pd
import schema as S
import torch
from model import TARGET_NAMES, WorldModel, device_of, merge_cfg, window_start
from train import PhaseTimer, _is_oom, amp_settings

T = len(S.TARGET_FIELDS)
TI = S.TARGET_INDEX
SI = S.STATE_INDEX
PH = S.PHASE
A = S.ACTION
GO = "GO"

# Which actions may follow which phase; anything else is counted as inconsistent.
_ALLOWED = np.zeros((len(S.PHASES), len(S.ACTIONS)), dtype=bool)
for _a in ("RUN", "PASS", "PUNT", "FG", "KNEEL", "SPIKE", "NOPLAY"):
    _ALLOWED[PH["PLAY"], A[_a]] = True
for _a in ("KICKOFF", "ONSIDE", "NOPLAY"):
    _ALLOWED[PH["KICKOFF"], A[_a]] = True
for _a in ("XP", "TWO_PT", "NOPLAY"):
    _ALLOWED[PH["TRY"], A[_a]] = True

COUNTER_KEYS = (
    "rollouts",
    "transitions",
    "ended_by_end",
    "ended_by_clock",
    "truncated_max_steps",
    "inconsistent",  # transitions with at least one problem below
    "bad_phase_pad",
    "bad_poss_pad",
    "bad_down",  # down not 1..4 on PLAY, or not 0 off PLAY
    "bad_ydstogo",
    "bad_yardline_zero",
    "bad_action_for_phase",
)


def check_transitions(tg: np.ndarray) -> dict:
    """Count inconsistent sampled transitions (rows whose phase is not END)."""
    tg = np.asarray(tg)
    ph = tg[:, TI["phase"]]
    live = ph != PH["END"]
    play = ph == PH["PLAY"]
    down, togo = tg[:, TI["down"]], tg[:, TI["ydstogo"]]
    bad = {
        "bad_phase_pad": live & (ph == PH["PAD"]),
        "bad_poss_pad": live & (tg[:, TI["poss"]] == S.POSS_CODE["PAD"]),
        "bad_down": live & np.where(play, (down < 1) | (down > 4), down != 0),
        "bad_ydstogo": live & np.where(play, togo == 0, togo != 0),
        "bad_yardline_zero": live & (tg[:, TI["yardline"]] == 0),
        "bad_action_for_phase": live & ~_ALLOWED[ph, tg[:, TI["action"]]],
    }
    out = {k: int(v.sum()) for k, v in bad.items()}
    out["inconsistent"] = int(np.logical_or.reduce(list(bad.values())).sum())
    return out


def _torch_generator(rng: np.random.Generator, device: torch.device) -> torch.Generator:
    g = torch.Generator(device=device)
    g.manual_seed(int(rng.integers(0, 2**62)))
    return g


def _game_of(split: S.Split, idx: np.ndarray) -> np.ndarray:
    return np.searchsorted(split.offsets, idx, side="right") - 1


def _window_buffer(split: S.Split, idx: np.ndarray, max_len: int) -> tuple[np.ndarray, np.ndarray]:
    """Context windows ending at each record in idx -> (buf [Q, max_len, S], lens [Q])."""
    games = _game_of(split, idx)
    buf = np.zeros((len(idx), max_len, len(S.STATE_FIELDS)), dtype=np.int64)
    lens = np.zeros(len(idx), dtype=np.int64)
    for q, (g, i) in enumerate(zip(games, idx, strict=True)):
        a = window_start(int(split.offsets[g]), int(i), max_len)
        k = int(i) - a + 1
        buf[q, :k] = split.states[a : i + 1]
        lens[q] = k
    return buf, lens


@torch.no_grad()
def _last_hidden(
    model: WorldModel, header: torch.Tensor, buf: torch.Tensor, lens: np.ndarray, chunk: int
) -> torch.Tensor:
    device = buf.device
    enabled, dtype, _ = amp_settings(device)
    outs = []
    for a in range(0, len(lens), chunk):
        sl = slice(a, a + chunk)
        n = int(lens[sl].max())
        with torch.autocast(device.type, dtype=dtype, enabled=enabled):
            h = model.encode(header[sl], buf[sl, :n])
        ln = torch.as_tensor(lens[sl] - 1, device=device)
        outs.append(h[torch.arange(len(ln), device=device), ln].float())
    return torch.cat(outs)


@torch.no_grad()
def _go_probs(model: WorldModel, split: S.Split, idx: np.ndarray, cfg: dict) -> np.ndarray:
    """P(RUN), P(PASS) for record i, renormalised over the two, from the model's prediction
    of record i's action at record i-1 (given the true other fields of record i)."""
    device = device_of(model)
    out = np.full((len(idx), 2), 0.5)
    games = _game_of(split, idx)
    has_prev = idx > split.offsets[games]
    if not has_prev.any():
        return out
    prev = idx[has_prev] - 1
    buf, lens = _window_buffer(split, prev, model.max_len)
    header = torch.as_tensor(split.header[games[has_prev]].astype(np.int64), device=device)
    h = _last_hidden(model, header, torch.as_tensor(buf, device=device), lens, cfg["sim_batch"])
    prefix = torch.as_tensor(split.targets[prev].astype(np.int64), device=device)
    logits = model.field_logits(h, prefix, TI["action"]).float()
    p = torch.softmax(logits, -1)[:, list(S.GO_ACTIONS)].cpu().numpy().astype(np.float64)
    tot = p.sum(1, keepdims=True)
    ok = tot[:, 0] > 0
    out_prev = np.full_like(p, 0.5)
    out_prev[ok] = p[ok] / tot[ok]
    out[has_prev] = out_prev
    return out


def cache_rows(model: WorldModel, cfg: dict) -> int:
    """Rollouts whose key/value caches fit in memory at once.

    One rollout's cache is layers x 2 x (max_len + 1) x d_model elements (2 bytes under AMP
    on a GPU, 4 on the CPU). On CUDA the budget is cfg["kv_cache_mem_frac"] of the device's
    TOTAL memory (not free memory, so the number is the same on every call and the random
    draws are reproducible), lowered only if that is more than 70% of what is actually
    available now (`torch.cuda.mem_get_info` free memory plus what PyTorch holds reserved
    but unused), for example when another process shares the GPU; on the CPU it is
    cfg["kv_cache_cpu_gb"]. Rounded down to a multiple of 256 above 256."""
    cfg = merge_cfg({**model.cfg, **(cfg or {})})
    device = device_of(model)
    enabled, dtype, _ = amp_settings(device)
    per_row = model.kv_bytes_per_row(2 if enabled else 4)
    if device.type == "cuda":
        total = torch.cuda.get_device_properties(device).total_memory
        budget = float(cfg.get("kv_cache_mem_frac", 0.4)) * total
        free, _ = torch.cuda.mem_get_info(device)
        idle = torch.cuda.memory_reserved(device) - torch.cuda.memory_allocated(device)
        budget = min(budget, 0.7 * (free + max(idle, 0)))
    else:
        budget = float(cfg.get("kv_cache_cpu_gb", 4.0)) * 2**30
    rows = max(1, int(budget // per_row))
    return rows - rows % 256 if rows >= 512 else rows


def live_rows(model: WorldModel, cfg: dict) -> int:
    """Rollouts simulated together: cfg["sim_max_live"], lowered to what the key/value cache
    fits. Both encoder paths use the same groups, so they make the same random draws."""
    cfg = merge_cfg({**model.cfg, **(cfg or {})})
    return max(1, min(int(cfg["sim_max_live"]), cache_rows(model, cfg)))


def _to_dev(a, device: torch.device) -> torch.Tensor:
    """A host array on the device. On CUDA through pinned memory with non_blocking=True: a
    plain copy from pageable memory makes PyTorch synchronise the stream, so the host would
    wait for every queued kernel at each of the small index copies of a simulated play. The
    pinned staging copy also means later in-place edits of the array cannot race the copy."""
    if device.type != "cuda":
        return torch.as_tensor(a, device=device)
    t = torch.from_numpy(np.ascontiguousarray(a))
    return t.pin_memory().to(device, non_blocking=True)


class _Book:
    """Per-row bookkeeping of one group of rollouts (host arrays)."""

    def __init__(self, lens: np.ndarray, raw: np.ndarray) -> None:
        r = len(lens)
        self.lens = lens.astype(np.int64).copy()
        self.raw = np.asarray(raw, dtype=np.int32).copy()
        self.active = np.ones(r, dtype=bool)
        self.steps = np.zeros(r, dtype=np.int64)
        self.final = self.raw[:, 1].astype(np.int64).copy()
        self.appended = np.zeros(r, dtype=np.int64)
        self.points = np.zeros(r, dtype=np.int64)
        self.kickoffs = np.zeros(r, dtype=np.int64)
        self.cnt = dict.fromkeys(COUNTER_KEYS, 0)
        self.cnt["rollouts"] = r

    def advance(self, act: np.ndarray, tg: np.ndarray, cfg: dict, max_steps: int):
        """Apply the transitions tg [len(act), T] sampled for rows act: counters, clock,
        score, timeouts, stopping. Returns (rows that continue, their new record codes)."""
        self.steps[act] += 1
        self.cnt["transitions"] += len(act)
        for k, v in check_transitions(tg).items():
            self.cnt[k] += v
        phase = tg[:, TI["phase"]]
        is_end = phase == PH["END"]
        raw_next = S.apply_transition(self.raw[act], tg)
        delta = S.score_delta_value(tg[:, TI["score_delta"]]).astype(np.int64)
        if not cfg["score_on_end"]:
            delta = np.where(is_end, 0, delta)
            raw_next[is_end, 1] = self.raw[act[is_end], 1]
        self.points[act] += np.abs(delta)
        clock0 = ~is_end & (raw_next[:, 0] <= 0) & (phase != PH["TRY"])
        trunc = ~is_end & ~clock0 & (self.steps[act] >= max_steps)
        cont = ~(is_end | clock0 | trunc)
        self.cnt["ended_by_end"] += int(is_end.sum())
        self.cnt["ended_by_clock"] += int(clock0.sum())
        self.cnt["truncated_max_steps"] += int(trunc.sum())
        self.final[act] = raw_next[:, 1]
        self.raw[act] = raw_next
        self.active[act[~cont]] = False
        rows = act[cont]
        codes = np.zeros((0, len(S.STATE_FIELDS)), dtype=np.int64)
        if len(rows):
            codes = S.next_state_codes(raw_next[cont], tg[cont]).astype(np.int64)
            self.appended[rows] += 1
            self.kickoffs[rows] += codes[:, SI["phase"]] == PH["KICKOFF"]
        return rows, codes

    def append(self, buf_t: torch.Tensor, rows: np.ndarray, codes: np.ndarray,
               max_len: int) -> np.ndarray:
        """Write each continuing row's new record into its window buffer, dropping the oldest
        record of a full window. Returns the rows whose window was shifted."""
        device = buf_t.device
        full = rows[self.lens[rows] >= max_len]
        if len(full):
            ft = _to_dev(full, device)
            buf_t[ft, :-1] = buf_t[ft, 1:].clone()
        pos = np.minimum(self.lens[rows], max_len - 1)
        buf_t[_to_dev(rows, device), _to_dev(pos, device)] = _to_dev(codes, device)
        self.lens[rows] = np.minimum(self.lens[rows] + 1, max_len)
        return full

    def result(self, track: bool) -> dict:
        out = {"final_diff": self.final, "steps": self.steps, "counters": self.cnt}
        if track:
            out.update({"appended": self.appended, "points": self.points,
                        "kickoffs": self.kickoffs})
        return out


def _sample_rows(model, h_rows: torch.Tensor, chunk: int, temp: float, gen, mask_pad: bool):
    """Sampled transitions for h_rows [R, d], in chunks of cfg["sample_batch"] rows (one
    call per play in practice; the chunking fixes the order of the random draws, and both
    encoder paths use it the same way)."""
    parts = [model.sample_transition(h_rows[a : a + chunk], temp, gen, mask_pad)
             for a in range(0, len(h_rows), chunk)]
    return torch.cat(parts).cpu().numpy()


def _timing(timer: PhaseTimer, loop_s: float, **extra) -> dict:
    out = {f"{k}_s": v for k, v in timer.split(loop_s).items()}
    out.update(loop_s=loop_s, iterations=timer.iters, method=timer.method(), **extra)
    return out


@torch.no_grad()
def _run_recompute(model, header, buf, lens, raw, cfg, deadline_ts, gen, track=False,
                   prefix_key=None) -> dict | None:
    """Reference path: every simulated play re-encodes each row's whole window."""
    device = device_of(model)
    max_len, max_steps, chunk = model.max_len, int(cfg["max_steps"]), int(cfg["sim_batch"])
    temp, mask_pad = float(cfg["temperature"]), bool(cfg["mask_pad_codes"])
    sample_batch = int(cfg.get("sample_batch", 16384))
    sort_len = bool(cfg.get("sort_by_length", True))
    bk = _Book(lens, raw)
    buf_t = torch.as_tensor(buf, device=device)
    head_t = torch.as_tensor(header.astype(np.int64), device=device)
    enabled, dtype, _ = amp_settings(device)
    timer = PhaseTimer(device, int(cfg.get("timing_sync_every", 8)))
    t0 = time.perf_counter()
    while bk.active.any():
        if time.time() > deadline_ts:
            return None
        timer.start()
        act = np.flatnonzero(bk.active)
        # Encode in chunks; with sort_by_length the chunks hold rows of similar context
        # length, so less padding is computed. The causal encoder ignores right padding, and
        # sampling below uses the original row order and chunks, so the draws are unchanged.
        enc_order = np.argsort(bk.lens[act], kind="stable") if sort_len else np.arange(len(act))
        hl_all = torch.empty(len(act), model.cfg["d_model"], device=device)
        for a in range(0, len(act), chunk):
            pos_rows = enc_order[a : a + chunk]
            rows = act[pos_rows]
            rows_t = torch.as_tensor(rows, device=device)
            n = int(bk.lens[rows].max())
            with torch.autocast(device.type, dtype=dtype, enabled=enabled):
                h = model.encode(head_t[rows_t], buf_t[rows_t, :n])
            ln = torch.as_tensor(bk.lens[rows] - 1, device=device)
            hl_all[torch.as_tensor(pos_rows, device=device)] = h[
                torch.arange(len(rows), device=device), ln
            ].float()
            del h
        timer.mark("encode")
        tg = _sample_rows(model, hl_all, sample_batch, temp, gen, mask_pad)
        timer.mark("decode")
        rows, codes = bk.advance(act, tg, cfg, max_steps)
        if len(rows):
            bk.append(buf_t, rows, codes, max_len)
        timer.mark("bookkeeping")
    out = bk.result(track)
    out["timing"] = _timing(timer, time.perf_counter() - t0, path="recompute")
    return out


@torch.no_grad()
def _run_cached(model, header, buf, lens, raw, cfg, deadline_ts, gen, track=False,
                prefix_key=None) -> dict | None:
    """Incremental decoding: each simulated play encodes ONE new record per row.

    Cache layout. Rows with the same prefix_key (all rollouts of one query) have the same
    header and the same records except the last one (whose action may be forced or sampled
    per row). Those shared positions are encoded once per key and stored once (the prefix
    cache); each row's own cache holds its last real record and every record it simulates.
    Attention reads the shared part once per key (`model.cascade_attention`), so its cost
    grows with the simulated plays, not with the real game so far.

    A row whose window is full (max_len records) and must drop its oldest record changes
    every position embedding, so no cache can be rolled exactly: from then on that row
    re-encodes its window each play (the reference path). Rows that finish leave dead slots
    in the own cache, which is compacted once a quarter is dead. Sampling uses the same rows
    and order as `_run_recompute`, so with equal hidden states the draws are identical."""
    device = device_of(model)
    max_len, max_steps, chunk = model.max_len, int(cfg["max_steps"]), int(cfg["sim_batch"])
    temp, mask_pad = float(cfg["temperature"]), bool(cfg["mask_pad_codes"])
    sample_batch = int(cfg.get("sample_batch", 16384))
    sort_len = bool(cfg.get("sort_by_length", True))
    enabled, dtype, _ = amp_settings(device)
    c = model.cfg
    d, nl, nh = int(c["d_model"]), int(c["n_layers"]), int(c["n_heads"])
    dh = d // nh
    r = len(lens)
    bk = _Book(lens, raw)
    buf_t = torch.as_tensor(buf, device=device)
    head_t = torch.as_tensor(header.astype(np.int64), device=device)
    timer = PhaseTimer(device, int(cfg.get("timing_sync_every", 8)))
    t0 = time.perf_counter()
    timer.start()

    def autocast():
        return torch.autocast(device.type, dtype=dtype, enabled=enabled)

    def dev(a):
        return _to_dev(a, device)

    # ---- groups of rows that share a prefix, and each row's cell in the [groups] grid
    key = np.arange(r) if prefix_key is None else np.asarray(prefix_key)
    _, first, base = np.unique(key, return_index=True, return_inverse=True)
    base = np.asarray(base).reshape(-1)
    n_base = len(first)
    counts = np.bincount(base, minlength=n_base)
    groups = int(counts.max())
    by_base = np.argsort(base, kind="stable")
    within = np.empty(r, dtype=np.int64)
    within[by_base] = np.arange(r) - np.repeat(np.cumsum(counts) - counts, counts)
    plen = bk.lens[first]  # shared positions per key: header + all records but the last
    p_max, own_cap = int(plen.max()), max_len + 1 - int(plen.min())

    # ---- caches. Zero-filled, not empty: a masked key gets attention weight 0, but 0 x NaN
    # from uninitialised memory would still be NaN in the weighted sum of values.
    cache_dtype = dtype if enabled else torch.float32
    kp = torch.zeros(nl, n_base, nh, p_max, dh, dtype=cache_dtype, device=device)
    vp = torch.zeros_like(kp)
    ko = torch.zeros(nl, r, nh, own_cap, dh, dtype=cache_dtype, device=device)
    vo = torch.zeros_like(ko)
    u_order = np.argsort(plen, kind="stable") if sort_len else np.arange(n_base)
    for a in range(0, n_base, chunk):
        us = u_order[a : a + chunk]
        src_t = dev(first[us])
        n = int(plen[us].max()) - 1  # shared records (the header is the extra position)
        with autocast():
            _, kv = model.encode(head_t[src_t], buf_t[src_t, :n], return_kv=True)
        for layer, (k, v) in enumerate(kv):
            kp[layer, dev(us), :, : n + 1] = k.to(cache_dtype)
            vp[layer, dev(us), :, : n + 1] = v.to(cache_dtype)
        del kv

    # ---- slots of the own cache (compacted as rows finish)
    slot = np.arange(r)  # slot of each row, -1 once it has left the cache
    slot_row = np.arange(r)  # row in each slot, -1 when the slot is dead
    slot_cell = base * groups + within
    slot_plen = plen[base].copy()
    slot_pos = bk.lens.copy()  # window position of each slot's newest token (header = 0)
    m = r
    p_ar = torch.arange(p_max, device=device)

    def step(codes_m: torch.Tensor) -> torch.Tensor:
        """encode_step for all m slots; their newest tokens are at slot_pos[:m]."""
        own = slot_pos[:m] - slot_plen[:m]
        n_own = int(own.max()) + 1
        own_t, plen_t = dev(own), dev(slot_plen[:m])
        o_ar = torch.arange(n_own, device=device)
        ok = torch.cat([p_ar.unsqueeze(0) < plen_t.unsqueeze(1),
                        o_ar.unsqueeze(0) <= own_t.unsqueeze(1)], dim=1)
        bias = torch.zeros(ok.shape, device=device).masked_fill_(~ok, float("-inf"))
        with autocast():
            h = model.encode_step(codes_m, dev(slot_pos[:m]), ko[:, :m], vo[:, :m], n_own,
                                  cache_pos=own_t,
                                  prefix=(kp, vp, dev(slot_cell[:m]), groups, bias))
        return h.float()

    # each row's last real record: own index 0, at window position lens
    h_last = step(buf_t[torch.arange(r, device=device), dev(bk.lens - 1)])
    timer.mark("prefill")
    n_fallback = n_compact = 0
    first_iter = True  # the prefill is part of the first iteration's timing
    while bk.active.any():
        if time.time() > deadline_ts:
            return None
        if not first_iter:
            timer.start()
        first_iter = False
        act = np.flatnonzero(bk.active)
        tg = _sample_rows(model, h_last[dev(act)], sample_batch, temp, gen, mask_pad)
        timer.mark("decode")
        rows, codes = bk.advance(act, tg, cfg, max_steps)
        stopped = act[~bk.active[act]]
        full = bk.append(buf_t, rows, codes, max_len) if len(rows) else rows
        for gone in (stopped, full):  # finished rows and rows whose window shifted
            live = gone[slot[gone] >= 0]
            slot_row[slot[live]] = -1
            slot[live] = -1
        timer.mark("bookkeeping")
        if len(full):  # window shifted: re-encode the whole window (exact)
            n_fallback += len(full)
            for a in range(0, len(full), chunk):
                fr_t = dev(full[a : a + chunk])
                with autocast():
                    h = model.encode(head_t[fr_t], buf_t[fr_t, :max_len])
                h_last[fr_t] = h[:, -1].float()
                del h
        keep = slot[rows] >= 0
        cached = rows[keep]
        if len(cached):
            s = slot[cached]
            slot_pos[s] = bk.lens[cached]
            s_t = dev(s)
            new = torch.zeros(m, len(S.STATE_FIELDS), dtype=torch.long, device=device)
            new[s_t] = dev(codes[keep])
            h_last[dev(cached)] = step(new)[s_t]
        dead = int((slot_row[:m] < 0).sum())
        if dead and dead >= m // 4:  # compact: live slots to the front, in place
            alive = np.flatnonzero(slot_row[:m] >= 0)
            k_alive = len(alive)
            if k_alive:
                a_t = dev(alive)
                for layer in range(nl):
                    ko[layer, :k_alive] = ko[layer, :m].index_select(0, a_t)
                    vo[layer, :k_alive] = vo[layer, :m].index_select(0, a_t)
            moved = slot_row[alive]
            slot[moved] = np.arange(k_alive)
            for arr in (slot_cell, slot_plen, slot_pos):
                arr[:k_alive] = arr[alive]
            slot_row[:k_alive] = moved
            slot_row[k_alive:m] = -1
            m = k_alive
            n_compact += 1
        timer.mark("encode")
    out = bk.result(track)
    elem = 2 if enabled else 4
    out["timing"] = _timing(
        timer, time.perf_counter() - t0, path="kv_cache", rows=r, shared_prefixes=n_base,
        fallback_row_steps=n_fallback, compactions=n_compact,
        cache_mib=(ko.numel() + kp.numel()) * elem * 2 / 2**20,
        cache_mib_per_rollout=model.kv_bytes_per_row(elem) / 2**20,
    )
    return out


def _merge_timing(parts: list[dict]) -> dict:
    out: dict = {}
    for t in parts:
        for k, v in (t or {}).items():
            if isinstance(v, (int, float)) and not isinstance(v, bool):
                out[k] = out.get(k, 0) + v
            else:
                out.setdefault(k, v)
    if parts and "cache_mib_per_rollout" in out:
        out["cache_mib_per_rollout"] = parts[0]["cache_mib_per_rollout"]
    return out


@torch.no_grad()
def _run(
    model: WorldModel,
    header: np.ndarray,
    buf: np.ndarray,
    lens: np.ndarray,
    raw: np.ndarray,
    cfg: dict,
    deadline_ts: float,
    gen: torch.Generator,
    track: bool = False,
    prefix_key: np.ndarray | None = None,
) -> dict | None:
    """Roll every row to the end of its game. Returns None if the deadline passes first.

    cfg["kv_cache"] picks the encoder path (True: incremental decoding; False: re-encode the
    window every play). Rows with the same prefix_key must have the same header and the same
    records except the last (the rollouts of one query); the cache stores that part once.
    Rows are simulated in groups of `live_rows` (memory), one group at a time. out["timing"]
    splits the wall time into prefill, encode, decode (sampling) and bookkeeping."""
    run_group = _run_cached if cfg.get("kv_cache", True) else _run_recompute
    lim = live_rows(model, cfg)
    r = len(lens)
    parts = []
    oom_retries = 0
    a = 0
    while a < r:
        sl = slice(a, min(r, a + lim))
        pk = None if prefix_key is None else np.asarray(prefix_key)[sl]
        oom = False
        try:
            out = run_group(model, header[sl], buf[sl], lens[sl], raw[sl], cfg, deadline_ts,
                            gen, track, pk)
        except RuntimeError as err:  # torch.OutOfMemoryError subclasses RuntimeError
            if not _is_oom(err) or sl.stop - sl.start <= 1:
                raise
            oom = True
        if oom:
            # Out of GPU memory: retry this group in halves (outside the except block, so
            # the traceback and its tensors are released first). The generator has moved
            # on, so the draws differ from an uninterrupted run (still valid samples).
            if device_of(model).type == "cuda":
                torch.cuda.empty_cache()
            lim = max(1, (sl.stop - sl.start) // 2)
            oom_retries += 1
            continue
        if out is None:
            return None
        parts.append(out)
        a = sl.stop
    if len(parts) == 1:
        merged = parts[0]
    else:
        merged = {k: np.concatenate([p[k] for p in parts]) for k in parts[0]
                  if isinstance(parts[0][k], np.ndarray)}
        merged["counters"] = {k: sum(p["counters"][k] for p in parts) for k in COUNTER_KEYS}
        merged["timing"] = _merge_timing([p["timing"] for p in parts])
    merged["timing"]["oom_retries"] = oom_retries
    _accumulate_timing(merged["timing"])
    return merged


ROLLOUT_TIMING: dict = {}


def _accumulate_timing(t: dict) -> None:
    for k, v in t.items():
        if k == "cache_mib_per_rollout":
            ROLLOUT_TIMING[k] = v
        elif isinstance(v, (int, float)) and not isinstance(v, bool):
            ROLLOUT_TIMING[k] = ROLLOUT_TIMING.get(k, 0) + v


def rollout_timing(reset: bool = False) -> dict:
    """Seconds by phase summed over every rollout since the last reset (this process)."""
    out = dict(ROLLOUT_TIMING)
    if reset:
        ROLLOUT_TIMING.clear()
    return out


def _resolve_action(forced_action) -> int | str | None:
    if forced_action is None or forced_action == GO:
        return forced_action
    if isinstance(forced_action, str):
        return A[forced_action]
    code = int(forced_action)
    if not 0 < code < len(S.ACTIONS):
        raise ValueError(f"action code {code} is outside the action vocabulary")
    return code


def rollout(
    model: WorldModel,
    split: S.Split,
    query_idx,
    forced_action,
    n_rollouts: int,
    cfg: dict | None,
    deadline_ts: float,
    rng: np.random.Generator,
) -> dict:
    """Home win probability by simulation from each query record.

    forced_action: None (keep the real action), an action code or name, or "GO" (record i's
    action is RUN or PASS, sampled from the model's own probabilities renormalised over the two).
    Queries are processed in groups of up to `live_rows` rollouts (cfg["sim_max_live"],
    lowered to what the key/value cache fits); if the deadline passes, the unfinished
    queries come back as NaN with done=False. res["timing"] splits the time by phase.
    """
    c = merge_cfg({**model.cfg, **(cfg or {})})
    model.eval()
    device = device_of(model)
    idx = np.atleast_1d(np.asarray(query_idx, dtype=np.int64))
    nq, n = len(idx), int(n_rollouts)
    fa = _resolve_action(forced_action)
    res = {
        "query_idx": idx,
        "home_wp": np.full(nq, np.nan),
        "home_wp_se": np.full(nq, np.nan),
        "mean_steps": np.full(nq, np.nan),
        "done": np.zeros(nq, dtype=bool),
        "forced_action": forced_action,
        "counters": dict.fromkeys(COUNTER_KEYS, 0),
        "start_actions": {},  # action of the query record, over all rollouts
    }
    gen = _torch_generator(rng, device)
    games = _game_of(split, idx)
    per_group = max(1, live_rows(model, c) // max(1, n))
    timings = []
    t0 = time.time()
    for a in range(0, nq, per_group):
        if time.time() > deadline_ts:
            break
        q = np.arange(a, min(nq, a + per_group))
        buf, lens = _window_buffer(split, idx[q], model.max_len)
        buf = np.repeat(buf, n, axis=0)
        lens = np.repeat(lens, n)
        if fa == GO:
            p = _go_probs(model, split, idx[q], c)
            u = rng.random(len(q) * n)
            pick = np.where(u < np.repeat(p[:, 0], n), S.GO_ACTIONS[0], S.GO_ACTIONS[1])
            buf[np.arange(len(lens)), lens - 1, SI["action"]] = pick
        elif fa is not None:
            buf[np.arange(len(lens)), lens - 1, SI["action"]] = fa
        starts = buf[np.arange(len(lens)), lens - 1, SI["action"]]
        for code, k in zip(*np.unique(starts, return_counts=True), strict=True):
            name = _action_name(code)
            res["start_actions"][name] = res["start_actions"].get(name, 0) + int(k)
        header = np.repeat(split.header[games[q]], n, axis=0)
        raw = np.repeat(split.raw[idx[q]], n, axis=0)
        # The rollouts of one query share the header and every record but the last (whose
        # action may differ), so the key/value cache stores that history once per query.
        prefix_key = np.repeat(np.arange(len(q)), n)
        out = _run(model, header, buf, lens, raw, c, deadline_ts, gen, prefix_key=prefix_key)
        if out is None:
            break
        timings.append(out["timing"])
        o = S.outcome(out["final_diff"]).reshape(len(q), n)
        res["home_wp"][q] = o.mean(1)
        res["home_wp_se"][q] = o.std(1, ddof=1) / np.sqrt(n) if n > 1 else np.nan
        res["mean_steps"][q] = out["steps"].reshape(len(q), n).mean(1)
        res["done"][q] = True
        for k, v in out["counters"].items():
            res["counters"][k] += v
    res["seconds"] = time.time() - t0
    res["timing"] = _merge_timing(timings)
    return res


def _action_name(code: int) -> str:
    return S.ACTIONS[int(code)]


def _choice_of(code: int) -> str:
    if int(code) in S.GO_ACTIONS:
        return GO
    return _action_name(code)


def plan_fourth_downs(
    model: WorldModel,
    split: S.Split,
    idx,
    n_rollouts: int,
    cfg: dict | None,
    deadline_ts: float,
    rng: np.random.Generator,
    action_mask: np.ndarray | None = None,
) -> pd.DataFrame:
    """For every fourth-down PLAY record in idx: win probability of GO, FG and PUNT for the
    team with the ball, standard errors, best action, margin and a "too close" flag (margin
    under 2 standard errors of the difference between the best and second-best action).

    action_mask: optional bool [len(idx), 3] for GO, FG, PUNT, aligned with idx as passed.
    A False entry means that action is not simulated: its wp and se are NaN and its done
    flag stays False. "best" is chosen among the allowed actions only, and a state is done
    when every allowed action is done. Without a mask every action is simulated, with the
    same rollout calls and random draws as before the argument existed."""
    idx = np.atleast_1d(np.asarray(idx, dtype=np.int64))
    st = split.states[idx]
    keep = (st[:, SI["down"]] == 4) & (st[:, SI["phase"]] == PH["PLAY"])
    if action_mask is None:
        allowed = np.ones((int(keep.sum()), 3), dtype=bool)
    else:
        am = np.asarray(action_mask, dtype=bool)
        if am.shape != (len(idx), 3):
            raise ValueError(f"action_mask has shape {am.shape}, expected ({len(idx)}, 3)")
        allowed = am[keep]
    idx = idx[keep]
    poss = split.states[idx, SI["poss"]]
    home_has_ball = poss == S.POSS_CODE["HOME"]
    choices = (GO, "FG", "PUNT")
    games = _game_of(split, idx)
    df = pd.DataFrame(
        {
            "idx": idx,
            "game_id": [split.game_ids[g] for g in games],
            "season": split.seasons[games],
            "poss": np.where(home_has_ball, "HOME", "AWAY"),
            "ydstogo": split.states[idx, SI["ydstogo"]],
            "yardline_100": split.states[idx, SI["yardline"]],
            "clock_seconds": split.raw[idx, 0],
            "score_diff_poss": np.where(home_has_ball, 1, -1) * split.raw[idx, 1],
        }
    )
    # The three actions are simulated chunk by chunk of states (one rollout group each), so
    # a deadline leaves whole states finished rather than GO for all and PUNT for none.
    c = merge_cfg({**model.cfg, **(cfg or {})})
    per = max(1, int(c["sim_max_live"]) // max(1, int(n_rollouts)))
    counters = {ch: dict.fromkeys(COUNTER_KEYS, 0) for ch in choices}
    for ch in choices:
        key = ch.lower()
        df[f"wp_{key}"] = np.nan
        df[f"se_{key}"] = np.nan
        df[f"done_{key}"] = False
    for a in range(0, len(idx), per):
        sl = slice(a, min(len(idx), a + per))
        for j, ch in enumerate(choices):
            ok = allowed[sl, j]
            if not ok.any():
                continue  # masked for every state in this chunk: nothing simulated
            r = rollout(model, split, idx[sl][ok], ch, n_rollouts, cfg, deadline_ts, rng)
            key = ch.lower()
            rows_ok = df.index[sl][ok]
            hb = home_has_ball[sl][ok]
            df.loc[rows_ok, f"wp_{key}"] = np.where(hb, r["home_wp"], 1.0 - r["home_wp"])
            df.loc[rows_ok, f"se_{key}"] = r["home_wp_se"]
            df.loc[rows_ok, f"done_{key}"] = r["done"]
            for k, v in r["counters"].items():
                counters[ch][k] += v
    wps = df[[f"wp_{c.lower()}" for c in choices]].to_numpy()
    ses = df[[f"se_{c.lower()}" for c in choices]].to_numpy()
    sim_done = df[[f"done_{c.lower()}" for c in choices]].to_numpy().astype(bool)
    done = (sim_done | ~allowed).all(1) & allowed.any(1)
    for j, ch in enumerate(choices):
        df[f"allowed_{ch.lower()}"] = allowed[:, j]
    order = np.argsort(-np.nan_to_num(wps, nan=-1.0), axis=1)
    rows = np.arange(len(df))
    b, s2 = order[:, 0], order[:, 1] if len(choices) > 1 else order[:, 0]
    margin = wps[rows, b] - wps[rows, s2]
    margin_se = np.sqrt(ses[rows, b] ** 2 + ses[rows, s2] ** 2)
    df["best"] = np.where(done, np.array(choices, dtype=object)[b], None)
    df["margin"] = np.where(done, margin, np.nan)
    df["margin_se"] = np.where(done, margin_se, np.nan)
    df["too_close"] = done & (margin < 2 * margin_se)
    real = split.states[idx, SI["action"]]
    df["real_action"] = [_action_name(a) for a in real]
    df["real_choice"] = [_choice_of(a) for a in real]
    df["agrees"] = done & (df["best"] == df["real_choice"])
    df["done"] = done
    df.attrs["counters"] = counters
    return df


def wp_by_simulation(
    model: WorldModel,
    split: S.Split,
    idx,
    n_rollouts: int,
    cfg: dict | None,
    deadline_ts: float,
    rng: np.random.Generator,
) -> pd.DataFrame:
    """Simulated home win probability from each record with its real action, plus the actual
    home result (1/0.5/0) for calibration."""
    idx = np.atleast_1d(np.asarray(idx, dtype=np.int64))
    r = rollout(model, split, idx, None, n_rollouts, cfg, deadline_ts, rng)
    games = _game_of(split, idx)
    df = pd.DataFrame(
        {
            "idx": idx,
            "game_id": [split.game_ids[g] for g in games],
            "season": split.seasons[games],
            "phase": [S.PHASES[p] for p in split.states[idx, SI["phase"]]],
            "poss": [S.POSS[p] for p in split.states[idx, SI["poss"]]],
            "clock_seconds": split.raw[idx, 0],
            "score_diff": split.raw[idx, 1],
            "home_wp_sim": r["home_wp"],
            "home_wp_se": r["home_wp_se"],
            "mean_steps": r["mean_steps"],
            "home_result": S.outcome(split.final_diff[games]),
            "done": r["done"],
        }
    )
    # run.py saves attrs["counters"] per chunk, so the phase timing travels with it.
    df.attrs["counters"] = {**r["counters"], "timing_s": r["timing"]}
    df.attrs["timing"] = r["timing"]
    return df


def real_game_stats(split: S.Split, games: np.ndarray | None = None) -> dict:
    """League statistics of real games, defined the same way as for simulated games."""
    g_all = np.arange(len(split.offsets) - 1) if games is None else np.asarray(games)
    n_rec = np.diff(split.offsets)[g_all]
    pts_rec = np.abs(S.score_delta_value(split.targets[:, TI["score_delta"]]))
    kick_rec = split.states[:, SI["phase"]] == PH["KICKOFF"]
    seg = split.offsets[:-1]
    pts = np.add.reduceat(pts_rec, seg)[g_all] if len(pts_rec) else np.zeros(len(g_all))
    kicks = np.add.reduceat(kick_rec.astype(np.int64), seg)[g_all]
    fd = split.final_diff[g_all].astype(np.int64)
    last = split.offsets[1:][g_all] - 1
    chain_diff = split.raw[last, 1].astype(np.int64)  # score before the final transition
    return {
        "records_per_game": float(n_rec.mean()),
        "points_per_game": float(pts.mean()),  # all transitions that carry a score delta
        "final_margin_sd": float(fd.std()),
        "tie_share": float((fd == 0).mean()),
        "kickoffs_per_game": float(kicks.mean()),
        "home_win_share": float(S.outcome(fd).mean()),
        "final_play_points_share": float((fd != chain_diff).mean()),
    }


def simulate_games(
    model: WorldModel,
    split: S.Split,
    n_games: int,
    cfg: dict | None,
    deadline_ts: float,
    rng: np.random.Generator,
) -> dict:
    """Simulate whole games from the header and opening record of sampled real games and
    compare league statistics with the same real games."""
    c = merge_cfg({**model.cfg, **(cfg or {})})
    model.eval()
    device = device_of(model)
    n_real = len(split.offsets) - 1
    games = rng.choice(n_real, size=n_games, replace=n_games > n_real)
    starts = split.offsets[games]
    gen = _torch_generator(rng, device)
    per = live_rows(model, c)
    parts = []
    for a in range(0, n_games, per):
        g = games[a : a + per]
        s = starts[a : a + per]
        buf, lens = _window_buffer(split, s, model.max_len)
        out = _run(model, split.header[g], buf, lens, split.raw[s], c, deadline_ts, gen, track=True)
        if out is None:
            break
        parts.append(out)
    if not parts:
        return {"n_games": 0, "sim": {}, "real": real_game_stats(split, games), "counters": {}}
    cat = {k: np.concatenate([p[k] for p in parts]) for k in ("final_diff", "appended", "points")}
    cat["kickoffs"] = np.concatenate([p["kickoffs"] for p in parts])
    done = len(cat["final_diff"])
    first_kick = (split.states[starts[:done], SI["phase"]] == PH["KICKOFF"]).astype(np.int64)
    fd = cat["final_diff"]
    sim = {
        "records_per_game": float((cat["appended"] + 1).mean()),
        "points_per_game": float(cat["points"].mean()),
        "final_margin_sd": float(fd.std()),
        "tie_share": float((fd == 0).mean()),
        "kickoffs_per_game": float((cat["kickoffs"] + first_kick).mean()),
        "home_win_share": float(S.outcome(fd).mean()),
    }
    counters = dict.fromkeys(COUNTER_KEYS, 0)
    for p in parts:
        for k, v in p["counters"].items():
            counters[k] += v
    real = real_game_stats(split, games[:done])
    table = pd.DataFrame({"real": pd.Series(real), "simulated": pd.Series(sim)}, index=list(real))
    return {"n_games": done, "sim": sim, "real": real, "table": table, "counters": counters,
            "timing": _merge_timing([p["timing"] for p in parts])}


__all__ = [
    "COUNTER_KEYS",
    "GO",
    "TARGET_NAMES",
    "cache_rows",
    "check_transitions",
    "live_rows",
    "plan_fourth_downs",
    "real_game_stats",
    "rollout",
    "rollout_timing",
    "simulate_games",
    "wp_by_simulation",
]
```

```
%%writefile /content/wm_code/support.py
"""Action support for the fourth-down planner: what the training data can and cannot vouch for.

The world model learned from what coaches did. For a call coaches almost never make from a
state (a 70-yard field goal, a punt from the opponent's 25, going for it on fourth and 15 from
your own 10) it has seen almost no examples, and a simulator can imagine good outcomes there
without anything in the data to correct it. This is the distribution-shift problem of offline
reinforcement learning on logged data: a planner that maximises over actions seeks out exactly
the actions whose values are least constrained by data. This module makes that measurable and
guards against it:

1. Support tables, from the TRAINING split only: for every fourth-down state (yard line, yards
   to go, optionally score and time buckets), how often coaches chose GO, FG and PUNT, summed
   over comparable states (yard line within +-3, yards to go in the same bucket). Plus the
   empirical field-goal make rate by yard line and the empirical punt net yards by yard line.
2. Hard rules, applied through `simulate.plan_fourth_downs(..., action_mask=...)`: no field
   goal beyond the longest made field goal in the training split plus 3 yards; no punt inside
   the opponent's 30; going for it is always allowed.
3. Soft flags reported with every recommendation: how many comparable training fourth downs
   there were, how many times coaches made the recommended call there and its share; "low
   support" when coaches made that call fewer than 10 times in comparable training states.
4. A diagnostic: the world model's simulated field-goal make rate (share of sampled first
   transitions after a forced field goal that score 3 for the kicking team) against the
   empirical make rate at the same distance, and the same for punt net yards.

Standalone: numpy, pandas and the sibling `schema` module; torch only inside the sampling
functions.
"""

from __future__ import annotations

import hashlib
import time
from dataclasses import dataclass, field
from typing import Any

import numpy as np
import pandas as pd
import schema as S

SI, TI = S.STATE_INDEX, S.TARGET_INDEX
PH, A = S.PHASE, S.ACTION
CHOICES = ("GO", "FG", "PUNT")  # column order of every [.., 3] array and of action_mask

# Field goal distance = yardline_100 + 18 (10 yards of end zone plus the snap and hold, about
# 8 yards behind the line of scrimmage). This is the usual relation in nflverse's
# kick_distance; a few attempts are recorded a yard shorter or longer.
KICK_DISTANCE_OFFSET = 18

# Yards-to-go buckets for "comparable" states. Short yardage is kept exact (1, 2, 3) because
# the go rate changes fast there; long yardage is pooled because it is rare.
YTG_BUCKETS = ((1, 1), (2, 2), (3, 3), (4, 5), (6, 7), (8, 10), (11, 15), (16, 30))
# Optional context buckets: score from the offense's side, and game clock.
SCORE_BUCKETS = ("down 9+", "down 1-8", "tied", "up 1-8", "up 9+")
TIME_BUCKETS = ("first half", "second half, over 5 min left", "last 5 min")

DEFAULTS: dict[str, Any] = {
    "version": 1,
    "fg_margin_yards": 3,  # field goals allowed up to the longest made one plus this
    "punt_min_yardline": 30,  # punts allowed only from yardline_100 >= this
    "min_support": 10,  # "low support" below this many comparable training calls
    "yardline_window": 3,  # comparable states: yard line within +-3
    "diag_states_per_band": 12,  # diagnostic: test fourth downs per distance band
    "diag_draws": 256,  # diagnostic: sampled first transitions per state
    "fg_band_yards": 5,  # diagnostic field-goal bands, in yards of distance
    "fg_band_max_yardline": 60,  # ... up to a 78-yard attempt
    "punt_band_yards": 10,
    "punt_band_min_yardline": 20,
}


def config(plan_cfg: dict | None = None) -> dict:
    """The support settings: DEFAULTS updated with plan_cfg["support"] (if any)."""
    over = dict((plan_cfg or {}).get("support") or {})
    unknown = set(over) - set(DEFAULTS)
    if unknown:
        raise ValueError(f"unknown support settings {sorted(unknown)}")
    return {**DEFAULTS, **over}


# ---------------------------------------------------------------- small helpers
def poss_sign(poss: np.ndarray) -> np.ndarray:
    """+1 when HOME has the ball, -1 when AWAY has it (score deltas are home minus away)."""
    return np.where(np.asarray(poss) == S.POSS_CODE["HOME"], 1, -1)


def choice_index(actions: np.ndarray) -> np.ndarray:
    """0 GO (RUN or PASS), 1 FG, 2 PUNT, -1 anything else."""
    a = np.asarray(actions)
    return np.select(
        [np.isin(a, S.GO_ACTIONS), a == A["FG"], a == A["PUNT"]], [0, 1, 2], default=-1
    )


_YTG_TO_BUCKET = np.zeros(S.MAX_YDSTOGO + 1, dtype=np.int64)
for _k, (_lo, _hi) in enumerate(YTG_BUCKETS):
    _YTG_TO_BUCKET[_lo : _hi + 1] = _k


def ytg_bucket(ytg: np.ndarray) -> np.ndarray:
    return _YTG_TO_BUCKET[np.clip(np.asarray(ytg, dtype=np.int64), 1, S.MAX_YDSTOGO)]


def score_bucket(diff_offense: np.ndarray) -> np.ndarray:
    d = np.asarray(diff_offense)
    return np.select([d <= -9, d < 0, d == 0, d <= 8], [0, 1, 2, 3], default=4)


def time_bucket(clock_seconds: np.ndarray) -> np.ndarray:
    c = np.asarray(clock_seconds)
    return np.select([c > S.HALF_SECONDS, c > 300], [0, 1], default=2)


def fourth_down_records(split: S.Split) -> np.ndarray:
    """Real fourth-down decisions: PLAY records on fourth down whose action is GO, FG or PUNT
    (the same rule as records.fourth_down_index)."""
    st = split.states
    ok = (
        (st[:, SI["phase"]] == PH["PLAY"])
        & (st[:, SI["down"]] == 4)
        & (choice_index(st[:, SI["action"]]) >= 0)
    )
    return np.flatnonzero(ok)


def fg_made(poss: np.ndarray, targets: np.ndarray) -> np.ndarray:
    """True where the transition scores 3 points for the team that had the ball.
    poss [...]; targets [..., T] (a trailing sample axis broadcasts against poss[..., None])."""
    tg = np.asarray(targets)
    pts = S.score_delta_value(tg[..., TI["score_delta"]])
    sign = poss_sign(poss)
    while sign.ndim < pts.ndim:
        sign = sign[..., None]
    return pts == 3 * sign


def punt_net_yards(yardline: np.ndarray, poss: np.ndarray, targets: np.ndarray) -> np.ndarray:
    """Net yards of a punt from the next record: how much closer to the opponent's goal line
    the ball ends up, from the punting team's side. NaN when the next record is not a PLAY
    (return touchdown, safety, end of game), which is counted separately as "other".

    Possession changed: the receiver's yard line y is 100 - y from the punter's side, so
    net = yardline - (100 - y). Possession kept (a muff recovered, a penalty re-kick):
    net = yardline - y.
    """
    tg = np.asarray(targets)
    yl, poss = np.asarray(yardline), np.asarray(poss)
    while yl.ndim < tg.ndim - 1:
        yl, poss = yl[..., None], poss[..., None]
    nxt_yl = tg[..., TI["yardline"]].astype(np.float64)
    kept = tg[..., TI["poss"]] == poss
    after = np.where(kept, nxt_yl, 100.0 - nxt_yl)
    net = yl - after
    return np.where(tg[..., TI["phase"]] == PH["PLAY"], net, np.nan)


# ---------------------------------------------------------------- support tables
def _box_smooth(counts: np.ndarray, window: int) -> np.ndarray:
    """Sum counts [100, 31, ...] over comparable states: yard line within +-window (1..99)
    and yards to go in the same YTG_BUCKETS bucket. Same shape out."""
    rest = counts.shape[2:]
    by_bucket = np.zeros((100, len(YTG_BUCKETS), *rest), dtype=np.int64)
    for k, (lo, hi) in enumerate(YTG_BUCKETS):
        by_bucket[:, k] = counts[:, lo : hi + 1].sum(axis=1)
    sm = np.zeros_like(by_bucket)
    for yl in range(1, 100):
        sm[yl] = by_bucket[max(1, yl - window) : min(99, yl + window) + 1].sum(axis=0)
    out = np.zeros_like(counts, dtype=np.int64)
    out[:, 1:] = sm[:, _YTG_TO_BUCKET[1:]]
    return out


@dataclass
class Support:
    """Training-split tables. Index 0 of every yard-line axis is unused (yard lines 1..99)."""

    counts: np.ndarray  # [100, 31, 3] fourth-down calls by yard line, yards to go
    ctx_counts: np.ndarray  # [100, 31, 5, 3, 3] ... by score bucket and time bucket too
    smooth: np.ndarray  # counts summed over comparable states
    ctx_smooth: np.ndarray
    fg_attempts: np.ndarray  # [100] field-goal attempts by yard line (PLAY, any down)
    fg_makes: np.ndarray  # [100]
    punts: np.ndarray  # [100] punts by yard line
    punts_next_play: np.ndarray  # [100] punts followed by a PLAY record (net yards defined)
    punt_net_sum: np.ndarray  # [100]
    cfg: dict
    meta: dict = field(default_factory=dict)

    def to_json(self) -> dict:
        """Compact tables for results.json (comparable-state calls by yard line and
        yards-to-go bucket; kicking tables by yard line)."""
        by_bucket = np.zeros((100, len(YTG_BUCKETS), 3), dtype=np.int64)
        for k, (lo, hi) in enumerate(YTG_BUCKETS):
            by_bucket[:, k] = self.counts[:, lo : hi + 1].sum(axis=1)
        with np.errstate(invalid="ignore", divide="ignore"):
            make = self.fg_makes / self.fg_attempts
            net = self.punt_net_sum / self.punts_next_play
        return {
            **self.meta,
            "choices": list(CHOICES),
            "ytg_buckets": [list(b) for b in YTG_BUCKETS],
            "calls_by_yardline_and_ytg_bucket_raw": by_bucket[1:].tolist(),
            "yardlines": list(range(1, 100)),
            "fg_attempts_by_yardline": self.fg_attempts[1:].tolist(),
            "fg_makes_by_yardline": self.fg_makes[1:].tolist(),
            "fg_make_rate_by_yardline": [None if not np.isfinite(v) else round(float(v), 4)
                                         for v in make[1:]],
            "punts_by_yardline": self.punts[1:].tolist(),
            "punt_mean_net_yards_by_yardline": [None if not np.isfinite(v) else
                                                round(float(v), 2) for v in net[1:]],
            "kick_distance": f"field goal distance = yardline_100 + {KICK_DISTANCE_OFFSET}",
        }


def build_support(train: S.Split, cfg: dict | None = None) -> Support:
    """Support tables from one split. Call it with the TRAINING split only: these tables
    stand for what the simulator could have learned from."""
    c = config({"support": dict(cfg or {})})
    st, tg, raw = train.states, train.targets, train.raw
    phase, yl = st[:, SI["phase"]], st[:, SI["yardline"]].astype(np.int64)
    ytg, act, poss = st[:, SI["ydstogo"]].astype(np.int64), st[:, SI["action"]], st[:, SI["poss"]]
    fd = fourth_down_records(train)
    ch = choice_index(act[fd])
    counts = np.zeros((100, S.MAX_YDSTOGO + 1, 3), dtype=np.int64)
    np.add.at(counts, (yl[fd], ytg[fd], ch), 1)
    sb = score_bucket(poss_sign(poss[fd]) * raw[fd, 1])
    tb = time_bucket(raw[fd, 0])
    ctx = np.zeros((100, S.MAX_YDSTOGO + 1, len(SCORE_BUCKETS), len(TIME_BUCKETS), 3),
                   dtype=np.int64)
    np.add.at(ctx, (yl[fd], ytg[fd], sb, tb, ch), 1)

    play = phase == PH["PLAY"]
    fg = np.flatnonzero(play & (act == A["FG"]))
    made = fg_made(poss[fg], tg[fg])
    fg_att = np.bincount(yl[fg], minlength=100)[:100]
    fg_mk = np.bincount(yl[fg], weights=made, minlength=100)[:100].astype(np.int64)
    pu = np.flatnonzero(play & (act == A["PUNT"]))
    net = punt_net_yards(yl[pu], poss[pu], tg[pu])
    ok = np.isfinite(net)
    punts = np.bincount(yl[pu], minlength=100)[:100]
    p_play = np.bincount(yl[pu][ok], minlength=100)[:100]
    p_sum = np.bincount(yl[pu][ok], weights=net[ok], minlength=100)[:100]
    h = hashlib.sha256("\x1f".join(train.game_ids).encode()).hexdigest()[:16]
    meta = {
        "split_games": len(train.game_ids),
        "split_seasons": sorted({int(s) for s in np.asarray(train.seasons)}),
        "split_game_ids_hash": h,
        "fourth_downs": int(len(fd)),
        "fourth_down_calls": dict(zip(CHOICES, counts.sum(axis=(0, 1)).tolist(), strict=True)),
        "fg_attempts": int(len(fg)),
        "fg_makes": int(made.sum()),
        "punts": int(len(pu)),
    }
    w = int(c["yardline_window"])
    return Support(counts=counts, ctx_counts=ctx, smooth=_box_smooth(counts, w),
                   ctx_smooth=_box_smooth(ctx, w), fg_attempts=fg_att, fg_makes=fg_mk,
                   punts=punts, punts_next_play=p_play, punt_net_sum=p_sum, cfg=c, meta=meta)


# ---------------------------------------------------------------- hard rules
def hard_rules(sup: Support) -> dict:
    """Rules no planner should violate, with the numbers behind them.

    - Field goal: allowed only from yardline_100 <= longest made field goal in the training
      split + fg_margin_yards (3). Beyond that the simulator has never seen a make.
    - Punt: allowed only from yardline_100 >= punt_min_yardline (30): no punt inside the
      opponent's 30, where a punt gains almost nothing and coaches essentially never punt.
    - Go: always allowed (every fourth down has a real snap).
    """
    c = sup.cfg
    made_at = np.flatnonzero(sup.fg_makes > 0)
    longest = int(made_at.max()) if len(made_at) else None
    margin = int(c["fg_margin_yards"])
    fg_max = (longest if longest is not None else 0) + margin
    punt_min = int(c["punt_min_yardline"])
    return {
        "go": "always allowed",
        "fg_max_yardline": fg_max,
        "fg_max_distance_yards": fg_max + KICK_DISTANCE_OFFSET,
        "longest_made_fg_yardline_train": longest,
        "longest_made_fg_distance_yards_train": None if longest is None
        else longest + KICK_DISTANCE_OFFSET,
        "fg_margin_yards": margin,
        "punt_min_yardline": punt_min,
        "train_fourth_down_fg_beyond_limit": int(sup.counts[fg_max + 1 :, :, 1].sum()),
        "train_fourth_down_punts_inside_limit": int(sup.counts[:punt_min, :, 2].sum()),
        "train_fourth_downs": int(sup.counts.sum()),
        "text": (f"field goals only from the opponent's {fg_max} or closer (a "
                 f"{fg_max + KICK_DISTANCE_OFFSET}-yard attempt: the longest made in the "
                 f"training split plus {margin} yards); no punt inside the opponent's "
                 f"{punt_min}; going for it is always allowed"),
    }


def action_mask(split: S.Split, idx, rules: dict) -> np.ndarray:
    """[len(idx), 3] bool for GO, FG, PUNT (True = allowed), from the hard rules."""
    idx = np.atleast_1d(np.asarray(idx, dtype=np.int64))
    yl = split.states[idx, SI["yardline"]].astype(np.int64)
    m = np.ones((len(idx), 3), dtype=bool)
    m[:, 1] = yl <= int(rules["fg_max_yardline"])
    m[:, 2] = yl >= int(rules["punt_min_yardline"])
    return m


def masked_actions(split: S.Split, idx, rules: dict, label: str = "test") -> list[dict]:
    """One entry per masked (state, action): which state, which action, which rule."""
    idx = np.atleast_1d(np.asarray(idx, dtype=np.int64))
    m = action_mask(split, idx, rules)
    games = np.searchsorted(split.offsets, idx, side="right") - 1
    out = []
    for q, j in zip(*np.nonzero(~m), strict=True):
        i = int(idx[q])
        yl = int(split.states[i, SI["yardline"]])
        if CHOICES[j] == "FG":
            rule = (f"field goal from yardline {yl} ({yl + KICK_DISTANCE_OFFSET} yards) beyond "
                    f"the limit {rules['fg_max_yardline']} "
                    f"({rules['fg_max_distance_yards']} yards)")
        else:
            rule = f"punt from yardline {yl}, inside the opponent's {rules['punt_min_yardline']}"
        out.append({"set": label, "idx": i, "game_id": split.game_ids[int(games[q])],
                    "yardline_100": yl, "ydstogo": int(split.states[i, SI["ydstogo"]]),
                    "real_action": S.ACTIONS[int(split.states[i, SI["action"]])],
                    "masked": CHOICES[j], "rule": rule})
    return out


# ---------------------------------------------------------------- soft flags
def lookup(sup: Support, split: S.Split, idx) -> pd.DataFrame:
    """Comparable-state call counts for each record in idx (training-split tables)."""
    idx = np.atleast_1d(np.asarray(idx, dtype=np.int64))
    st, raw = split.states[idx], split.raw[idx]
    yl = st[:, SI["yardline"]].astype(np.int64)
    ytg = np.clip(st[:, SI["ydstogo"]].astype(np.int64), 1, S.MAX_YDSTOGO)
    base = sup.smooth[yl, ytg]
    sb = score_bucket(poss_sign(st[:, SI["poss"]]) * raw[:, 1])
    tb = time_bucket(raw[:, 0])
    ctx = sup.ctx_smooth[yl, ytg, sb, tb]
    out = {"sup_n": base.sum(1), "ctx_n": ctx.sum(1),
           "ctx_bucket": [f"{SCORE_BUCKETS[s]}, {TIME_BUCKETS[t]}" for s, t in zip(sb, tb,
                                                                                strict=True)]}
    for j, ch in enumerate(CHOICES):
        out[f"sup_{ch.lower()}"] = base[:, j]
        out[f"ctx_{ch.lower()}"] = ctx[:, j]
    return pd.DataFrame(out)


def add_support_columns(df: pd.DataFrame, sup: Support, split: S.Split,
                        rules: dict) -> pd.DataFrame:
    """Soft flags next to every recommendation in a plan_fourth_downs frame.

    sup_n: comparable training fourth downs (yard line +-3, same yards-to-go bucket);
    sup_count_best / sup_share_best: how often coaches made the recommended call there;
    low_support: sup_count_best < min_support; ctx_*: the same within the state's score
    and time bucket (narrower, reported only). allowed_*: the hard-rule mask.
    """
    df = df.copy()
    if len(df) == 0:
        for k in ("sup_n", "sup_count_best", "sup_share_best", "low_support"):
            df[k] = []
        return df
    idx = df["idx"].to_numpy(dtype=np.int64)
    lk = lookup(sup, split, idx)
    for k in lk.columns:
        df[k] = lk[k].to_numpy()
    m = action_mask(split, idx, rules)
    for j, ch in enumerate(CHOICES):
        df[f"allowed_{ch.lower()}"] = m[:, j]
    best = df["best"].astype(object).where(df["best"].notna(), None)
    jbest = np.array([CHOICES.index(b) if b in CHOICES else -1 for b in best])
    has = jbest >= 0
    rows = np.arange(len(df))
    base = lk[[f"sup_{c.lower()}" for c in CHOICES]].to_numpy()
    ctx = lk[[f"ctx_{c.lower()}" for c in CHOICES]].to_numpy()
    cnt = np.where(has, base[rows, np.maximum(jbest, 0)], -1)
    ccnt = np.where(has, ctx[rows, np.maximum(jbest, 0)], -1)
    with np.errstate(invalid="ignore", divide="ignore"):
        share = np.where(has & (lk["sup_n"] > 0), cnt / lk["sup_n"], np.nan)
        cshare = np.where(has & (lk["ctx_n"] > 0), ccnt / lk["ctx_n"], np.nan)
    df["sup_count_best"] = np.where(has, cnt, np.nan)
    df["sup_share_best"] = share
    df["ctx_count_best"] = np.where(has, ccnt, np.nan)
    df["ctx_share_best"] = cshare
    df["low_support"] = has & (cnt < int(sup.cfg["min_support"]))
    real = df["real_choice"] if "real_choice" in df.columns else None
    if real is not None:
        jr = np.array([CHOICES.index(r) if r in CHOICES else -1 for r in real])
        df["sup_share_real"] = np.where(
            (jr >= 0) & (lk["sup_n"] > 0), base[rows, np.maximum(jr, 0)] / np.maximum(
                lk["sup_n"].to_numpy(), 1), np.nan)
    return df


def empirical_fg_rate(sup: Support, yardline: np.ndarray, window: int = 2) -> tuple:
    """Training make rate and attempts within +-window yard lines of each yard line."""
    yl = np.atleast_1d(np.asarray(yardline, dtype=np.int64))
    att = np.array([sup.fg_attempts[max(1, y - window) : y + window + 1].sum() for y in yl])
    mk = np.array([sup.fg_makes[max(1, y - window) : y + window + 1].sum() for y in yl])
    with np.errstate(invalid="ignore", divide="ignore"):
        return np.where(att > 0, mk / np.maximum(att, 1), np.nan), att


# ---------------------------------------------------------------- one-step sampling
def _device(model):
    try:
        return next(model.parameters()).device
    except (AttributeError, StopIteration, TypeError):
        import torch

        return torch.device("cpu")


def first_transitions(model, split: S.Split, idx, action, n_draws: int, cfg: dict | None,
                      rng: np.random.Generator, chunk: int = 64,
                      deadline_ts: float | None = None) -> np.ndarray:
    """Sample n_draws first transitions from each record in idx with its action forced.

    This is exactly the first step of a rollout from that record (`simulate.rollout` encodes
    the context window ending at the record, with the forced action, and samples the
    transition from the last position with `model.sample_transition`), without playing on to
    the end of the game, so it costs one encoder pass per state. The encoder runs in float32
    here; rollouts on a GPU may encode under autocast, which changes only rounding. Returns
    codes [len(idx), n_draws, len(TARGET_FIELDS)] (int16).
    """
    import torch

    idx = np.atleast_1d(np.asarray(idx, dtype=np.int64))
    code = A[action] if isinstance(action, str) else int(action)
    c = {**(getattr(model, "cfg", None) or {}), **(cfg or {})}
    temp, mask_pad = float(c.get("temperature", 1.0)), bool(c.get("mask_pad_codes", True))
    max_len = int(getattr(model, "max_len", c.get("max_len", 256)))
    dev = _device(model)
    gen = torch.Generator(device=dev)
    gen.manual_seed(int(rng.integers(0, 2**62)))
    games = np.searchsorted(split.offsets, idx, side="right") - 1
    out = np.zeros((len(idx), int(n_draws), len(S.TARGET_FIELDS)), dtype=np.int16)
    if hasattr(model, "eval"):
        model.eval()
    with torch.no_grad():
        for a in range(0, len(idx), chunk):
            if deadline_ts is not None and time.time() > deadline_ts:
                raise TimeoutError("support diagnostic ran out of time")
            q = np.arange(a, min(len(idx), a + chunk))
            starts = [max(int(split.offsets[g]), int(i) + 1 - max_len)
                      for g, i in zip(games[q], idx[q], strict=True)]
            lens = np.array([int(i) - s + 1 for s, i in zip(starts, idx[q], strict=True)])
            buf = np.zeros((len(q), int(lens.max()), len(S.STATE_FIELDS)), dtype=np.int64)
            for r, (s, i) in enumerate(zip(starts, idx[q], strict=True)):
                buf[r, : lens[r]] = split.states[s : int(i) + 1]
                buf[r, lens[r] - 1, SI["action"]] = code
            header = torch.as_tensor(split.header[games[q]].astype(np.int64), device=dev)
            h = model.encode(header, torch.as_tensor(buf, device=dev))
            last = h[torch.arange(len(q), device=dev), torch.as_tensor(lens - 1, device=dev)]
            rep = last.float().repeat_interleave(int(n_draws), dim=0)
            codes = model.sample_transition(rep, temp, gen, mask_pad)
            out[q] = codes.reshape(len(q), int(n_draws), -1).cpu().numpy().astype(np.int16)
    return out


def sim_fg_make(model, split: S.Split, idx, n_draws: int, cfg: dict | None,
                rng: np.random.Generator, deadline_ts: float | None = None) -> np.ndarray:
    """Per-state simulated make rate of a forced field goal (share of first transitions that
    score 3 for the kicking team). [len(idx)]."""
    idx = np.atleast_1d(np.asarray(idx, dtype=np.int64))
    if len(idx) == 0:
        return np.zeros(0)
    codes = first_transitions(model, split, idx, "FG", n_draws, cfg, rng,
                              deadline_ts=deadline_ts)
    return fg_made(split.states[idx, SI["poss"]], codes).mean(axis=1)


def _bands(lo: int, hi: int, width: int) -> list[tuple[int, int]]:
    return [(a, min(a + width - 1, hi)) for a in range(lo, hi + 1, width)]


def _sample_band(pool: np.ndarray, yl: np.ndarray, band: tuple[int, int], k: int,
                 rng: np.random.Generator) -> np.ndarray:
    cand = pool[(yl >= band[0]) & (yl <= band[1])]
    if len(cand) == 0:
        return cand
    return np.sort(rng.choice(cand, size=min(k, len(cand)), replace=False))


def _mean_se(per_state: np.ndarray, n_draws: int) -> tuple[float | None, float | None]:
    v = per_state[np.isfinite(per_state)]
    if len(v) == 0:
        return None, None
    if len(v) == 1:
        p = float(v[0])
        return p, float(np.sqrt(max(p * (1 - p), 1e-12) / n_draws))
    return float(v.mean()), float(v.std(ddof=1) / np.sqrt(len(v)))


def sim_vs_empirical(model, split: S.Split, sup: Support, rules: dict, cfg: dict | None,
                     rng: np.random.Generator, pool_idx=None,
                     deadline_ts: float | None = None) -> dict:
    """Simulated versus empirical field-goal make rate by distance, and punt net yards by
    yard line.

    States: for each band, up to diag_states_per_band real fourth downs of `split` (the
    test split in the plan stage) whose yard line is in the band, whatever the coach did;
    the field goal or punt is then forced. Most long-range states are real punt situations,
    so this probes the simulator exactly where coaches do not kick. Empirical numbers come
    from the training-split tables in `sup` (all field-goal attempts on PLAY records, any
    down; all punts).
    """
    c = sup.cfg
    k, n = int(c["diag_states_per_band"]), int(c["diag_draws"])
    pool = fourth_down_records(split) if pool_idx is None else np.asarray(pool_idx, np.int64)
    pool_yl = split.states[pool, SI["yardline"]].astype(np.int64)
    out: dict[str, Any] = {
        "states_per_band": k, "draws_per_state": n,
        "kick_distance": f"yardline_100 + {KICK_DISTANCE_OFFSET}",
        "method": "first transition after the forced action, sampled draws_per_state times "
                  "from each state's own context (the first step of a rollout); simulated "
                  "rate = mean over states of the per-state share",
        "caveat": "empirical rates are for kicks and punts coaches chose to attempt; long "
                  "attempts are selected (strong kickers, good conditions, but also "
                  "end-of-half heaves), so a gap is not all simulator error",
    }
    fg_rows = []
    fg_states = []
    for band in _bands(1, int(c["fg_band_max_yardline"]), int(c["fg_band_yards"])):
        s = _sample_band(pool, pool_yl, band, k, rng)
        fg_states.append(s)
        att = int(sup.fg_attempts[band[0] : band[1] + 1].sum())
        mk = int(sup.fg_makes[band[0] : band[1] + 1].sum())
        fg_rows.append({"yardline": list(band),
                        "distance_yards": [band[0] + KICK_DISTANCE_OFFSET,
                                           band[1] + KICK_DISTANCE_OFFSET],
                        "train_attempts": att, "train_makes": mk,
                        "empirical_make_rate": mk / att if att else None,
                        "beyond_fg_limit": band[0] > int(rules["fg_max_yardline"]),
                        "n_states": int(len(s)), "state_idx": s.tolist()})
    all_fg = np.concatenate(fg_states) if fg_states else np.zeros(0, np.int64)
    per = sim_fg_make(model, split, all_fg, n, cfg, rng, deadline_ts)
    a = 0
    for row, s in zip(fg_rows, fg_states, strict=True):
        m, se = _mean_se(per[a : a + len(s)], n)
        a += len(s)
        row["sim_make_rate"], row["sim_se"] = m, se
        e = row["empirical_make_rate"]
        row["sim_minus_empirical"] = None if m is None or e is None else m - e
    out["fg"] = fg_rows

    punt_rows, punt_states = [], []
    for band in _bands(int(c["punt_band_min_yardline"]), 99, int(c["punt_band_yards"])):
        s = _sample_band(pool, pool_yl, band, k, rng)
        punt_states.append(s)
        sl = slice(band[0], band[1] + 1)
        npl, tot = int(sup.punts_next_play[sl].sum()), int(sup.punts[sl].sum())
        punt_rows.append({"yardline": list(band), "train_punts": tot,
                          "empirical_net_yards": float(sup.punt_net_sum[sl].sum() / npl)
                          if npl else None,
                          "empirical_other_share": (tot - npl) / tot if tot else None,
                          "inside_punt_limit": band[0] < int(rules["punt_min_yardline"]),
                          "n_states": int(len(s)), "state_idx": s.tolist()})
    all_p = np.concatenate(punt_states) if punt_states else np.zeros(0, np.int64)
    if len(all_p):
        codes = first_transitions(model, split, all_p, "PUNT", n, cfg, rng,
                                  deadline_ts=deadline_ts)
        net = punt_net_yards(split.states[all_p, SI["yardline"]],
                             split.states[all_p, SI["poss"]], codes)
    else:
        net = np.zeros((0, n))
    a = 0
    for row, s in zip(punt_rows, punt_states, strict=True):
        part = net[a : a + len(s)]
        a += len(s)
        with np.errstate(invalid="ignore"):
            per_state = np.array([np.nanmean(r) if np.isfinite(r).any() else np.nan
                                  for r in part])
        m, se = _mean_se(per_state, n)
        row["sim_net_yards"], row["sim_se"] = m, se
        row["sim_other_share"] = float(np.mean(~np.isfinite(part))) if part.size else None
        e = row["empirical_net_yards"]
        row["sim_minus_empirical"] = None if m is None or e is None else m - e
    out["punt"] = punt_rows
    return out


def fg_gap_summary(diag: dict, min_attempts: int = 10) -> dict:
    """Headline numbers of the field-goal diagnostic.

    Bands with at least min_attempts training attempts give a usable empirical rate: their
    attempt-weighted mean gap (simulated minus training) and the largest gap. Bands with
    fewer (thin data, inside the hard limit) and bands beyond the limit are summarised by
    the simulator's mean make rate there, since the data cannot say what the rate is.
    """
    rows = [r for r in diag.get("fg", []) if r.get("sim_make_rate") is not None]
    data = [r for r in rows if (r["train_attempts"] or 0) >= min_attempts]
    thin = [r for r in rows if (r["train_attempts"] or 0) < min_attempts
            and not r["beyond_fg_limit"]]
    beyond = [r for r in rows if r["beyond_fg_limit"]]
    worst = max(data, key=lambda r: r["sim_minus_empirical"], default=None)
    w = np.array([r["train_attempts"] for r in data], dtype=float)
    g = np.array([r["sim_minus_empirical"] for r in data], dtype=float)

    def mean_sim(rs: list) -> float | None:
        return float(np.mean([r["sim_make_rate"] for r in rs])) if rs else None

    return {
        "min_attempts": min_attempts,
        "bands_with_data": len(data),
        "weighted_mean_gap_with_data": float((w * g).sum() / w.sum()) if len(data) else None,
        "largest_gap_band": None if worst is None else worst["distance_yards"],
        "largest_gap": None if worst is None else worst["sim_minus_empirical"],
        "largest_gap_train_attempts": None if worst is None else worst["train_attempts"],
        "thin_bands": [r["distance_yards"] for r in thin],
        "sim_make_rate_thin_bands": mean_sim(thin),
        "sim_make_rate_beyond_limit": mean_sim(beyond),
    }


__all__ = [
    "CHOICES",
    "DEFAULTS",
    "KICK_DISTANCE_OFFSET",
    "Support",
    "action_mask",
    "add_support_columns",
    "build_support",
    "config",
    "empirical_fg_rate",
    "fg_gap_summary",
    "fg_made",
    "first_transitions",
    "hard_rules",
    "lookup",
    "masked_actions",
    "punt_net_yards",
    "sim_fg_make",
    "sim_vs_empirical",
]
```

```
%%writefile /content/wm_code/toy.py
"""A tiny synthetic `Split` for tests: short, rule-based games whose codes and transitions are
valid under schema.py and consistent with `schema.apply_transition`. Not football-realistic.
Conventions follow the real records: a kickoff record's poss is the RECEIVING team, and the
last record's target is END plus the final play's score_delta and elapsed."""

from __future__ import annotations

import numpy as np
import schema as S

PH, A, P = S.PHASE, S.ACTION, S.POSS_CODE


def _policy(phase: int, down: int, togo: int, yl: int, rng: np.random.Generator) -> int:
    if phase == PH["KICKOFF"]:
        return A["KICKOFF"]
    if phase == PH["TRY"]:
        return A["XP"] if rng.random() < 0.9 else A["TWO_PT"]
    if down == 4:
        if rng.random() < 0.2 or (togo <= 2 and yl < 60):
            return A["RUN"] if rng.random() < 0.5 else A["PASS"]
        return A["FG"] if yl <= 37 else A["PUNT"]
    return A["RUN"] if rng.random() < 0.45 else A["PASS"]


def _step(phase, poss, down, togo, yl, action, rng):
    """Next (phase, poss, down, togo, yardline), points for the team with the ball, elapsed."""
    other = 3 - poss
    if phase == PH["KICKOFF"]:  # poss of a kickoff record = receiving team
        return (PH["PLAY"], poss, 1, 10, 75), 0, int(rng.integers(4, 9))
    if phase == PH["TRY"]:
        pts = (
            (1 if rng.random() < 0.94 else 0)
            if action == A["XP"]
            else (2 if rng.random() < 0.48 else 0)
        )
        return (PH["KICKOFF"], other, 0, 0, 65), pts, 0
    el = int(rng.integers(5, 41))
    if action == A["FG"]:
        if rng.random() < max(0.2, 1.0 - yl / 50):
            return (PH["KICKOFF"], other, 0, 0, 65), 3, 5
        return (PH["PLAY"], other, 1, 10, int(np.clip(100 - (yl + 7), 1, 99))), 0, 5
    if action == A["PUNT"]:
        land = yl - 40
        yl2 = 80 if land <= 0 else int(np.clip(100 - land, 1, 99))
        return (PH["PLAY"], other, 1, 10, yl2), 0, 8
    gain = int(rng.integers(-3, 16))
    if gain >= yl:
        return (PH["TRY"], poss, 0, 0, 15), 6, el
    yl2 = int(np.clip(yl - gain, 1, 99))
    if gain >= togo:
        return (PH["PLAY"], poss, 1, min(10, yl2), yl2), 0, el
    if down == 4:
        return (PH["PLAY"], other, 1, 10, 100 - yl2), 0, el
    return (PH["PLAY"], poss, down + 1, int(np.clip(togo - gain, 1, S.MAX_YDSTOGO)), yl2), 0, el


def make_toy_split(
    n_games: int, rng: np.random.Generator, start_clock: int = 600, first_game_id: int = 0
) -> S.Split:
    """n_games short games (regulation clock starts at start_clock seconds)."""
    states, raws, targets, offsets, header, final = [], [], [], [0], [], []
    for _ in range(n_games):
        receiver = int(rng.integers(1, 3))
        header.append([S.spread_code(rng.integers(-7, 8)), receiver])
        cat = (PH["KICKOFF"], receiver, 0, 0, 65)
        raw = np.array([start_clock, 0, 3, 3], dtype=np.int32)
        action = A["KICKOFF"]
        while True:
            phase, poss, down, togo, yl = cat
            nxt, pts, el = _step(phase, poss, down, togo, yl, action, rng)
            sign = 1 if poss == P["HOME"] else -1
            used_h = int(rng.random() < 0.03 and raw[2] > 0)
            used_a = int(rng.random() < 0.03 and raw[3] > 0)
            n_action = _policy(nxt[0], nxt[2], nxt[3], nxt[4], rng)
            tg = np.zeros(len(S.TARGET_FIELDS), dtype=np.int16)
            for name, v in zip(("phase", "poss", "down", "ydstogo", "yardline"), nxt, strict=True):
                tg[S.TARGET_INDEX[name]] = v
            tg[S.TARGET_INDEX["score_delta"]] = S.SCORE_DELTA_CODE[sign * pts]
            tg[S.TARGET_INDEX["elapsed"]] = S.elapsed_code(el)
            tg[S.TARGET_INDEX["to_used_home"]] = used_h + 1
            tg[S.TARGET_INDEX["to_used_away"]] = used_a + 1
            tg[S.TARGET_INDEX["action"]] = n_action
            raw_next = S.apply_transition(raw[None], tg[None])[0]
            end = raw_next[0] == 0 and nxt[0] != PH["TRY"]
            rec = np.zeros(len(S.STATE_FIELDS), dtype=np.int16)
            for name, v in zip(("phase", "poss", "down", "ydstogo", "yardline"), cat, strict=True):
                rec[S.STATE_INDEX[name]] = v
            rec[S.STATE_INDEX["clock"]] = S.clock_code(raw[0])
            rec[S.STATE_INDEX["score"]] = S.score_code(raw[1])
            rec[S.STATE_INDEX["to_home"]] = S.timeout_code(raw[2])
            rec[S.STATE_INDEX["to_away"]] = S.timeout_code(raw[3])
            rec[S.STATE_INDEX["action"]] = action
            states.append(rec)
            raws.append(raw.copy())
            if end:
                last_tg = np.zeros_like(tg)
                last_tg[S.TARGET_INDEX["phase"]] = PH["END"]
                for name in ("score_delta", "elapsed"):
                    last_tg[S.TARGET_INDEX[name]] = tg[S.TARGET_INDEX[name]]
                targets.append(last_tg)
                final.append(int(raw_next[1]))  # includes the final play's points
                break
            targets.append(tg)
            cat, raw, action = nxt, raw_next, n_action
        offsets.append(len(states))
    n = len(states)
    return S.Split(
        states=np.asarray(states, dtype=np.int16),
        raw=np.asarray(raws, dtype=np.int32),
        targets=np.asarray(targets, dtype=np.int16),
        offsets=np.asarray(offsets, dtype=np.int64),
        header=np.asarray(header, dtype=np.int16),
        final_diff=np.asarray(final, dtype=np.int16),
        game_ids=[f"toy_{first_game_id + g}" for g in range(n_games)],
        seasons=np.full(n_games, 2000, dtype=np.int16),
        play_ids=np.arange(n, dtype=np.int64),
    )
```

```
%%writefile /content/wm_code/run.py
"""Runner for the NFL world-model experiment, built to finish inside one Colab GPU session.

Six stages, each idempotent. A finished stage writes a done-marker holding its config hash,
so re-running everything skips finished work and training resumes from its checkpoint:

  1 data     download play-by-play, build and validate train/validation/test splits
  2 train    fit the world model (checkpointed, deadline-bound)
  3 eval     validation only: per-field NLL vs a frequency baseline, simulated league
             stats vs real, win probability by simulation vs nflverse on the same states
  4 test     write prereg/<config hash>.json (this configuration's exact plan) FIRST, then
             score the test seasons once per configuration; a second scoring of the same
             configuration is refused unless allow_rescore is set
  5 plan     fourth-down recommendations on sampled test fourth downs (under the same
             preregistration), one constructed hypothetical state, and agreement with the
             real coaches
  6 package  results.json (every number), figures, a zip to download

A global deadline guard gives every stage a deadline; stages scale down (fewer states or
rollouts) or are skipped with a recorded reason, and packaging always runs, so a run
always ends with a results file. Numbers from nflverse (`home_wp`, `vegas_home_wp`) are
kept in their own labelled fields and never blended with ours.
"""

from __future__ import annotations

import copy
import hashlib
import importlib
import json
import os
import platform
import sys
import time
import traceback
import zipfile
from pathlib import Path
from typing import Any

import numpy as np
import schema as S

STEPS_PER_ROLLOUT = 90  # mean transitions from a random state to the end of regulation
MAX_SEASON = 2025  # never download a season after this one (2026 is in progress)
SEED = 20261118
STAGES = ("data", "train", "eval", "test", "plan", "package")
UPSTREAM = {
    "data": (),
    "train": ("data",),
    "eval": ("train",),
    "test": ("train",),
    "plan": ("train",),
    "package": (),
}
# Module sources recorded per stage (metadata and warnings only; hashes never use them,
# except that a changed records/schema makes the data stage rebuild in memory and compare).
STAGE_SOURCES = {
    "data": ["schema", "records"],
    "train": ["model", "train"],
    "eval": ["simulate"],
    "test": ["simulate"],
    "plan": ["simulate", "support"],
    "package": [],
}
HERE = Path(__file__).resolve().parent


# ---------------------------------------------------------------- presets
_FULL_SEASONS = {
    "train": list(range(2010, 2022)),
    "val": [2022, 2023],
    "test": [2024, 2025],
}


def _gpu_preset(name: str, model: dict, train: dict, eval_: dict, plan: dict, assumed: dict):
    return {
        "name": name,
        "device": "cuda",
        "seasons": copy.deepcopy(_FULL_SEASONS),
        "model": model,
        "train": train,
        "eval": eval_,
        "test": dict(eval_),
        "plan": plan,
        "budget_min": 55.0,
        # Per-stage ceilings in minutes. They sum to more than the budget on purpose: a
        # stage that finishes early hands its time on; the global deadline is the hard cap.
        # Plan is the last compute stage, so its ceiling is high: it may use whatever the
        # earlier stages left (training on the GPU usually takes well under its 20 minutes).
        "stage_min": {"data": 8, "train": 20, "eval": 9, "test": 7, "plan": 20, "package": 2},
        # Minimum seconds a stage needs to be worth starting; below it, it is skipped.
        "min_start_s": {"data": 60, "train": 60, "eval": 45, "test": 45, "plan": 45,
                        "package": 0},
        "reserve_package_s": 90,
        "assumed": assumed,
    }


PRESETS: dict[str, dict] = {
    "smoke": {
        "name": "smoke",
        "device": "cpu",
        "seasons": {"train": [2021], "val": [2022], "test": [2023]},
        "model": {"d_model": 64, "n_layers": 2, "n_heads": 4, "d_ff": 256, "dec_hidden": 64,
                  "max_len": 256, "dropout": 0.1},
        "train": {"batch_games": 16, "lr": 1e-3, "weight_decay": 0.01, "epochs": 4,
                  "patience": 2, "warmup_steps": 50, "max_train_games": None,
                  "sim_batch": 256, "sim_max_live": 1024, "score_on_end": True},
        "eval": {"wp_states": 16, "wp_rollouts": 16, "sim_games": 4, "bootstrap": 200,
                 "nll_max_games": 60},
        "test": {"wp_states": 16, "wp_rollouts": 16, "sim_games": 0, "bootstrap": 200,
                 "nll_max_games": 60},
        "plan": {"n_fourth": 8, "n_rollouts": 16, "hypo_rollouts": 64,
                 "support": {"diag_states_per_band": 4, "diag_draws": 64}},

        "budget_min": 12.0,
        "stage_min": {"data": 5, "train": 2.5, "eval": 2, "test": 1.5, "plan": 1.5,
                      "package": 1},
        "min_start_s": {"data": 20, "train": 10, "eval": 10, "test": 10, "plan": 10,
                        "package": 0},
        "reserve_package_s": 20,
        "assumed": {"train_rec_per_s": 3_000, "rollout_units_per_s": 2_000},
    },
    "t4": _gpu_preset(
        "t4",
        model={"d_model": 256, "n_layers": 6, "n_heads": 8, "d_ff": 1024, "dec_hidden": 256,
               "max_len": 256, "dropout": 0.1},
        train={"batch_games": 64, "lr": 8.5e-4, "weight_decay": 0.1, "epochs": 40,
               "patience": 4, "warmup_steps": 300, "max_train_games": None,
               "sim_batch": 1024, "sim_max_live": 8192, "score_on_end": True},
        eval_={"wp_states": 600, "wp_rollouts": 256, "sim_games": 128, "bootstrap": 1000,
               "nll_max_games": None},
        plan={"n_fourth": 300, "n_rollouts": 512, "hypo_rollouts": 4096},
        assumed={"train_rec_per_s": 60_000, "rollout_units_per_s": 40_000},
    ),
    "l4": _gpu_preset(
        "l4",
        model={"d_model": 320, "n_layers": 6, "n_heads": 8, "d_ff": 1280, "dec_hidden": 320,
               "max_len": 256, "dropout": 0.1},
        train={"batch_games": 64, "lr": 7e-4, "weight_decay": 0.1, "epochs": 50,
               "patience": 4, "warmup_steps": 300, "max_train_games": None,
               "sim_batch": 2048, "sim_max_live": 16384, "score_on_end": True},
        eval_={"wp_states": 1000, "wp_rollouts": 384, "sim_games": 256, "bootstrap": 1000,
               "nll_max_games": None},
        plan={"n_fourth": 400, "n_rollouts": 512, "hypo_rollouts": 8192},
        assumed={"train_rec_per_s": 120_000, "rollout_units_per_s": 80_000},
    ),
    "a100": _gpu_preset(
        "a100",
        model={"d_model": 384, "n_layers": 8, "n_heads": 8, "d_ff": 1536, "dec_hidden": 384,
               "max_len": 256, "dropout": 0.1},
        train={"batch_games": 128, "lr": 7e-4, "weight_decay": 0.1, "epochs": 60,
               "patience": 5, "warmup_steps": 300, "max_train_games": None,
               "sim_batch": 4096, "sim_max_live": 32768, "score_on_end": True},

        eval_={"wp_states": 1500, "wp_rollouts": 512, "sim_games": 512, "bootstrap": 1000,
               "nll_max_games": None},
        plan={"n_fourth": 600, "n_rollouts": 1024, "hypo_rollouts": 16384},
        assumed={"train_rec_per_s": 300_000, "rollout_units_per_s": 200_000},
    ),
}
DOWNGRADE = {"a100": "l4", "l4": "t4", "t4": "t4", "smoke": "smoke"}
# Bump when what a preset means changes without its values changing (for example a new rule
# in how the benchmark adjusts it). preset.json stores this with a hash of the PRESETS entry;
# a saved preset whose definition differs from the code is replaced by a fresh pick.
PRESET_VERSION = "2 (2026-11-18: GPU training, KV-cache rollouts, action support)"
# The preflight benchmark scales a preset down only when the measured rollout or training
# speed is below this share of the preset's assumed speed, and never below these floors
# (or the preset's own value, if that is smaller).
BENCH_SLOW_SHARE = 0.5
BENCH_FLOORS = {"n_fourth": 200, "wp_states": 300, "sim_games": 32, "hypo_rollouts": 1024}


def preset_definition(name: str) -> dict:
    """What identifies a preset's definition: its name, PRESET_VERSION and a hash of its
    PRESETS entry (before any benchmark adjustment)."""
    return {"name": name, "version": PRESET_VERSION, "hash": config_hash(PRESETS[name])}


def pick_preset(device_name: str | None = None, mem_gb: float | None = None) -> str:
    """Preset key from the CUDA device name and memory; "smoke" when there is no GPU."""
    if device_name is None:
        try:
            import torch

            if not torch.cuda.is_available():
                return "smoke"
            props = torch.cuda.get_device_properties(0)
            device_name, mem_gb = props.name, props.total_memory / 2**30
        except Exception:
            return "smoke"
    name = device_name.upper()
    mem = mem_gb or 0.0
    if "A100" in name or "H100" in name or "H200" in name or "B200" in name:
        return "a100"
    if "L4" in name or "A10" in name or "L40" in name or "A6000" in name:
        return "l4"
    if "T4" in name or "P100" in name or "V100" in name:
        return "t4"
    if mem >= 38:
        return "a100"
    if mem >= 20:
        return "l4"
    return "t4"


def get_preset(name: str, **overrides: Any) -> dict:
    p = copy.deepcopy(PRESETS[name])
    for k, v in overrides.items():
        if isinstance(v, dict) and isinstance(p.get(k), dict):
            p[k].update(v)
        else:
            p[k] = v
    return p


def check_seasons(seasons) -> list[int]:
    seasons = [int(s) for s in seasons]
    assert seasons, "no seasons given"
    assert max(seasons) <= MAX_SEASON, f"refusing to download a season after {MAX_SEASON}"
    assert min(seasons) >= 1999, "nflverse play-by-play starts in 1999"
    return seasons


# ---------------------------------------------------------------- storage
def default_workdir() -> Path:
    if os.environ.get("WM_WORKDIR"):
        return Path(os.environ["WM_WORKDIR"])
    drive = Path("/content/drive/MyDrive")
    if drive.is_dir():
        return drive / "fourth_and_rl_world_model"
    if Path("/content").is_dir():
        return Path("/content/wm_run")
    return HERE.parent.parent / "results" / "world_model_local"


def _jsonable(x: Any) -> Any:
    if isinstance(x, dict):
        return {str(k): _jsonable(v) for k, v in x.items()}
    if isinstance(x, (list, tuple)):
        return [_jsonable(v) for v in x]
    if isinstance(x, np.ndarray):
        return _jsonable(x.tolist())
    if isinstance(x, (np.integer,)):
        return int(x)
    if isinstance(x, (np.floating, float)):
        v = float(x)
        return v if np.isfinite(v) else None
    if isinstance(x, (np.bool_,)):
        return bool(x)
    if hasattr(x, "to_dict") and callable(x.to_dict):  # pandas objects
        try:
            return _jsonable(x.to_dict(orient="list"))
        except TypeError:
            return _jsonable(x.to_dict())
    if isinstance(x, Path):
        return str(x)
    if x is None or isinstance(x, (str, int, bool)):
        return x
    return str(x)


def write_json(path: Path, obj: Any) -> None:
    path.parent.mkdir(parents=True, exist_ok=True)
    tmp = path.with_suffix(path.suffix + ".tmp")
    tmp.write_text(json.dumps(_jsonable(obj), indent=2))
    tmp.replace(path)


def read_json(path: Path) -> Any:
    return json.loads(Path(path).read_text())


def config_hash(obj: Any) -> str:
    return hashlib.sha256(json.dumps(_jsonable(obj), sort_keys=True).encode()).hexdigest()[:16]


def _source_hash(module_names) -> str:
    h = hashlib.sha256()
    for m in module_names:
        p = HERE / f"{m}.py"
        mod = sys.modules.get(m)
        if mod is not None and getattr(mod, "__file__", None):
            p = Path(mod.__file__)
        h.update(p.read_bytes() if p.exists() else m.encode())
    return h.hexdigest()[:16]


def _mod(name: str):
    """Import a sibling module lazily, so tests can put stubs in sys.modules."""
    return sys.modules.get(name) or importlib.import_module(name)


# ---------------------------------------------------------------- split utilities
_SPLIT_ARRAYS = ("states", "raw", "targets", "offsets", "header", "final_diff", "seasons",
                 "play_ids")


def save_split(path: Path, split: S.Split, **extra: np.ndarray) -> None:
    path.parent.mkdir(parents=True, exist_ok=True)
    arrays = {k: np.asarray(getattr(split, k)) for k in _SPLIT_ARRAYS}
    arrays["game_ids"] = np.asarray(list(split.game_ids), dtype=str)
    arrays.update({f"extra_{k}": np.asarray(v) for k, v in extra.items()})
    tmp = path.with_name(path.stem + ".tmp.npz")
    np.savez_compressed(tmp, **arrays)
    tmp.replace(path)


def load_split(path: Path) -> tuple[S.Split, dict[str, np.ndarray]]:
    with np.load(path, allow_pickle=False) as z:
        kw = {k: z[k] for k in _SPLIT_ARRAYS}
        kw["game_ids"] = [str(g) for g in z["game_ids"]]
        extra = {k[6:]: z[k] for k in z.files if k.startswith("extra_")}
    return S.Split(**kw), extra


def concat_splits(splits: list[S.Split]) -> S.Split:
    offs, base = [np.zeros(1, dtype=np.int64)], 0
    for s in splits:
        offs.append(np.asarray(s.offsets[1:], dtype=np.int64) + base)
        base += int(s.offsets[-1])
    return S.Split(
        states=np.concatenate([s.states for s in splits]),
        raw=np.concatenate([s.raw for s in splits]),
        targets=np.concatenate([s.targets for s in splits]),
        offsets=np.concatenate(offs),
        header=np.concatenate([s.header for s in splits]),
        final_diff=np.concatenate([s.final_diff for s in splits]),
        game_ids=[g for s in splits for g in s.game_ids],
        seasons=np.concatenate([s.seasons for s in splits]),
        play_ids=np.concatenate([s.play_ids for s in splits]),
    )


def subset_split(split: S.Split, games) -> S.Split:
    """A split holding only the listed games (in the given order)."""
    games = np.asarray(games, dtype=np.int64)
    rec = [np.arange(split.offsets[g], split.offsets[g + 1]) for g in games]
    lens = np.array([len(r) for r in rec], dtype=np.int64)
    idx = np.concatenate(rec) if rec else np.zeros(0, dtype=np.int64)
    return S.Split(
        states=split.states[idx],
        raw=split.raw[idx],
        targets=split.targets[idx],
        offsets=np.concatenate([[0], np.cumsum(lens)]).astype(np.int64),
        header=split.header[games],
        final_diff=split.final_diff[games],
        game_ids=[split.game_ids[g] for g in games],
        seasons=split.seasons[games],
        play_ids=split.play_ids[idx],
    )


def record_game(split: S.Split) -> np.ndarray:
    """Game index of every record."""
    return np.repeat(np.arange(len(split.game_ids)), np.diff(split.offsets))


def n_records(split: S.Split) -> int:
    return int(split.offsets[-1])


def synthetic_split(n_games: int = 8, seed: int = 0) -> S.Split:
    """A schema-valid toy split (drives in which the clock runs down), for tests and the
    micro-benchmark only. It is not football data."""
    rng = np.random.default_rng(seed)
    states, raws, targets, offsets, final = [], [], [], [0], []
    for _g in range(n_games):
        clock, diff, poss = S.GAME_SECONDS, 0, 1
        rows, raw = [], []
        down, togo, yl = 0, 0, 65
        phase = S.PHASE["KICKOFF"]
        while clock > 0:
            act = S.ACTION["KICKOFF"] if phase == S.PHASE["KICKOFF"] else int(
                rng.choice([S.ACTION["RUN"], S.ACTION["PASS"]]))
            if phase == S.PHASE["PLAY"] and down == 4:
                act = S.ACTION["PUNT"]
            rows.append([phase, poss, down, togo, yl, int(S.clock_code(clock)),
                         int(S.score_code(diff)), 4, 4, act])
            raw.append([clock, diff, 3, 3])
            clock = max(clock - int(rng.integers(5, 45)), 0)
            if phase == S.PHASE["KICKOFF"] or act == S.ACTION["PUNT"]:
                poss = 3 - poss
                phase, down, togo, yl = S.PHASE["PLAY"], 1, 10, int(rng.integers(50, 80))
            else:
                yl = max(1, yl - int(rng.integers(0, 8)))
                down = down + 1 if down < 4 else 1
                togo = 10
                if yl <= 2:
                    diff += 7 if poss == 1 else -7
                    phase, down, togo, yl = S.PHASE["KICKOFF"], 0, 0, 65
                    poss = 3 - poss
        states.extend(rows)
        raws.extend(raw)
        tg = np.zeros((len(rows), len(S.TARGET_FIELDS)), dtype=np.int16)
        r = np.asarray(rows)
        for name in ("phase", "poss", "down", "ydstogo", "yardline", "action"):
            tg[:-1, S.TARGET_INDEX[name]] = r[1:, S.STATE_INDEX[name]]
        tg[:-1, S.TARGET_INDEX["score_delta"]] = [
            S.SCORE_DELTA_CODE.get(int(b[1] - a[1]), S.SCORE_DELTA_CODE[0])
            for a, b in zip(raw[:-1], raw[1:], strict=True)]
        tg[:-1, S.TARGET_INDEX["elapsed"]] = S.elapsed_code(
            np.asarray(raw)[:-1, 0] - np.asarray(raw)[1:, 0])
        tg[:-1, S.TARGET_INDEX["to_used_home"]] = 1
        tg[:-1, S.TARGET_INDEX["to_used_away"]] = 1
        tg[-1, S.TARGET_INDEX["phase"]] = S.PHASE["END"]
        targets.append(tg)
        offsets.append(offsets[-1] + len(rows))
        final.append(diff)
    n = offsets[-1]
    return S.Split(
        states=np.asarray(states, dtype=np.int16),
        raw=np.asarray(raws, dtype=np.int32),
        targets=np.concatenate(targets),
        offsets=np.asarray(offsets, dtype=np.int64),
        header=np.tile(np.array([[S.SPREAD_CLIP + 1, 1]], dtype=np.int16), (n_games, 1)),
        final_diff=np.asarray(final, dtype=np.int16),
        game_ids=[f"SYN_{g:04d}" for g in range(n_games)],
        seasons=np.full(n_games, 2000, dtype=np.int16),
        play_ids=np.arange(n, dtype=np.int64),
    )


# ---------------------------------------------------------------- run context
class Run:
    """Everything a stage needs: preset, folders, the deadline guard and the log.

    Stage hashes are keyed on CONFIG and DATA CONTENT, never on source code: the train
    checkpoint and done-marker depend on the model and train config plus a content hash of
    the saved train and validation arrays, so an edited notebook resumes training as long as
    the data and config are the same. Module source hashes are recorded in each done-marker
    as metadata and produce a warning when they change.
    """

    def __init__(self, preset: dict, workdir: Path | None = None,
                 budget_min: float | None = None, allow_rescore: bool = False,
                 seed: int = SEED, quiet: bool = False, force: tuple = ()) -> None:
        self.preset = preset
        self.workdir = Path(workdir or default_workdir())
        self.workdir.mkdir(parents=True, exist_ok=True)
        for sub in ("data", "stages", "outputs", "figures", "ckpt", "partial"):
            (self.workdir / sub).mkdir(exist_ok=True)
        self.allow_rescore = allow_rescore
        self.force = set(force)
        self.seed = seed
        self.quiet = quiet
        self.session_start = time.time()
        self.ledger_path = self.workdir / "ledger.json"
        ledger = read_json(self.ledger_path) if self.ledger_path.exists() else {}
        self.consumed_before = float(ledger.get("consumed_s", 0.0))
        total = 60.0 * float(budget_min if budget_min is not None else preset["budget_min"])
        self.total_s = total
        left = total - self.consumed_before
        # Never less than the packaging reserve: packaging must always be able to run.
        self.deadline = self.session_start + max(left, preset["reserve_package_s"] + 5)
        self._models: dict[str, Any] = {}

    # time ------------------------------------------------------------
    def remaining(self) -> float:
        return self.deadline - time.time()

    def stage_deadline(self, stage: str) -> float:
        ceiling = time.time() + 60.0 * self.preset["stage_min"][stage]
        reserve = 0.0 if stage == "package" else float(self.preset["reserve_package_s"])
        return min(ceiling, self.deadline - reserve)

    def add_consumed(self, seconds: float, stage: str) -> None:
        ledger = read_json(self.ledger_path) if self.ledger_path.exists() else {}
        ledger["consumed_s"] = float(ledger.get("consumed_s", 0.0)) + float(seconds)
        ledger.setdefault("stages", []).append(
            {"stage": stage, "seconds": round(float(seconds), 1), "at": time.time()})
        write_json(self.ledger_path, ledger)

    def log(self, *msg: Any) -> None:
        line = " ".join(str(m) for m in msg)
        with open(self.workdir / "run.log", "a") as f:
            f.write(f"{time.strftime('%H:%M:%S')} {line}\n")
        if not self.quiet:
            print(line, flush=True)

    # config ----------------------------------------------------------
    def model_cfg(self) -> dict:
        """The cfg handed to model, train and simulate."""
        p = self.preset
        cfg = {**p["model"], **p["train"], "device": self.device(), "seed": self.seed}
        cfg.pop("max_train_games", None)
        return cfg

    def device(self) -> str:
        if self.preset["device"] == "cuda":
            try:
                import torch

                if torch.cuda.is_available():
                    return "cuda"
            except Exception:
                pass
        return "cpu"

    def content_hash(self, name: str) -> str:
        """Content hash of data/<name>.npz, cached by file size and mtime."""
        path = self.workdir / "data" / f"{name}.npz"
        if not path.exists():
            return "missing"
        cache_path = self.workdir / "data" / "content_hashes.json"
        cache = read_json(cache_path) if cache_path.exists() else {}
        st = path.stat()
        key = f"{st.st_size}:{st.st_mtime_ns}"
        if cache.get(name, {}).get("key") == key:
            return cache[name]["hash"]
        sp, extra = load_split(path)
        h = split_arrays_hash(sp, extra)
        cache[name] = {"key": key, "hash": h}
        write_json(cache_path, cache)
        return h

    def stage_src(self, stage: str) -> str:
        return _source_hash(STAGE_SOURCES[stage])

    def stage_hash(self, stage: str) -> str:
        p = self.preset
        parts: dict[str, Any] = {"stage": stage, "seed": self.seed}
        if stage == "data":
            parts["seasons"] = p["seasons"]
        elif stage == "train":
            parts.update(model=p["model"], train=p["train"],
                         data_train=self.content_hash("train"),
                         data_val=self.content_hash("val"))
        elif stage in ("eval", "test", "plan"):
            parts[stage] = p[stage]
            parts["up_train"] = self.stage_hash("train")
            if stage == "plan":
                # Defaults included, so a resumed run with an older saved preset still
                # recomputes the plan with the support rules.
                parts["support"] = _mod("support").config(p["plan"])
            if stage in ("test", "plan"):
                parts["data_test"] = self.content_hash("test")
        return config_hash(parts)

    def full_config_hash(self) -> str:
        return config_hash({k: self.stage_hash(k) for k in STAGES if k != "package"})

    # files -------------------------------------------------------------
    def marker(self, stage: str) -> Path:
        return self.workdir / "stages" / f"{stage}.done"

    def output_path(self, stage: str) -> Path:
        return self.workdir / "outputs" / f"{stage}.json"

    def partial_dir(self, stage: str) -> Path:
        d = self.workdir / "partial" / f"{stage}_{self.stage_hash(stage)}"
        d.mkdir(parents=True, exist_ok=True)
        return d

    def is_done(self, stage: str) -> bool:
        if stage in self.force:
            return False
        m = self.marker(stage)
        if not m.exists():
            return False
        mk = read_json(m)
        if mk.get("hash") != self.stage_hash(stage):
            return False
        # Data is the one stage whose code is re-checked: a changed records.py is rebuilt
        # in memory and compared (see stage_data).
        return not (stage == "data" and mk.get("src") != self.stage_src("data"))

    def status(self, stage: str) -> str:
        """Status of this configuration's output of a stage; "missing" when there is none
        and "other config" when the file was written under a different stage hash."""
        p = self.output_path(stage)
        if not p.exists():
            return "missing"
        out = read_json(p)
        if stage != "package" and out.get("hash") != self.stage_hash(stage):
            return "other config"
        return out.get("status", "missing")

    def current_output(self, stage: str) -> dict:
        """The stage's output file if it belongs to this configuration; otherwise a stub
        saying why not (an output from an earlier configuration is never reported as
        this run's)."""
        p = self.output_path(stage)
        if not p.exists():
            return {"status": "missing"}
        out = read_json(p)
        h = self.stage_hash(stage)
        if out.get("hash") != h:
            return {"status": "missing",
                    "reason": f"the saved output belongs to another configuration (stage hash "
                              f"{out.get('hash')}, this run {h}); not reported"}
        return out

    def split(self, name: str) -> tuple[S.Split, dict]:
        return load_split(self.workdir / "data" / f"{name}.npz")

    def ckpt_path(self) -> Path:
        return self.workdir / "ckpt" / f"model_{self.stage_hash('train')}.pt"

    def load_model(self):
        key = self.stage_hash("train")
        if key in self._models:
            return self._models[key]
        model, _ck = _mod("train").load_checkpoint(str(self.ckpt_path()), device=self.device())
        model.eval()
        self._models[key] = model
        return model


def run_stage(run: Run, stage: str, fn=None) -> dict:
    """Run one stage under the deadline guard. Finished stages are skipped by hash.

    A stage may return status "partial" (it ran out of time but saved its progress); it gets
    no done-marker, so the next run continues it. A KeyboardInterrupt is recorded as
    "interrupted" and re-raised after the stage's output file is written.
    """
    fn = fn or STAGE_FUNCS[stage]
    h = run.stage_hash(stage)
    if stage != "package" and run.is_done(stage):
        out = read_json(run.output_path(stage))
        mk = read_json(run.marker(stage))
        run.log(f"[{stage}] already done (hash {h}); skipping")
        if mk.get("src") and mk["src"] != run.stage_src(stage):
            run.log(f"[{stage}] note: {', '.join(STAGE_SOURCES[stage])} changed since this "
                    "stage finished; its results are kept (stages are keyed on config and "
                    f"data). Pass force=('{stage}',) to recompute.")
        return out
    for up in UPSTREAM[stage]:
        if not run.is_done(up):
            out = {"status": "skipped", "reason": f"upstream stage '{up}' has not finished",
                   "hash": h}
            write_json(run.output_path(stage), out)
            run.log(f"[{stage}] skipped: {out['reason']}")
            return out
    available = run.stage_deadline(stage) - time.time()
    if available < run.preset["min_start_s"][stage]:
        out = {"status": "skipped", "hash": h,
               "reason": f"time budget: {available:.0f}s left for this stage, needs "
                         f"{run.preset['min_start_s'][stage]}s"}
        write_json(run.output_path(stage), out)
        run.log(f"[{stage}] skipped: {out['reason']}")
        return out
    run.log(f"[{stage}] starting ({available / 60:.1f} min available, "
            f"{run.remaining() / 60:.1f} min left in the run)")
    t0 = time.time()
    interrupted = False
    try:
        out = fn(run, run.stage_deadline(stage))
        out.setdefault("status", "done")
    except KeyboardInterrupt:
        interrupted = True
        out = {"status": "interrupted",
               "reason": "stopped by the user; saved progress is resumed on the next run"}
    except Exception as e:  # recorded, not swallowed silently: packaging reports it
        out = {"status": "error", "error": f"{type(e).__name__}: {e}",
               "traceback": traceback.format_exc()}
        run.log(f"[{stage}] ERROR {out['error']}\n{out['traceback']}")
    out["seconds"] = round(time.time() - t0, 1)
    out["hash"] = h
    out["src"] = run.stage_src(stage)
    write_json(run.output_path(stage), out)
    if stage != "package":
        run.add_consumed(out["seconds"], stage)
    if out["status"] == "done" and stage != "package":
        # Re-hash: the data stage changes the content hashes the other stages depend on.
        write_json(run.marker(stage), {"hash": run.stage_hash(stage), "src": out["src"],
                                       "finished": time.time()})
    run.log(f"[{stage}] {out['status']} in {out['seconds']:.0f}s")
    if interrupted:
        raise KeyboardInterrupt
    return out


# ---------------------------------------------------------------- stage 1: data
def _pbp_wp_columns(pbp) -> Any:
    """game_id, play_id, home_wp, vegas_home_wp from a polars or pandas frame, as pandas."""
    cols = [c for c in ("game_id", "play_id", "home_wp", "vegas_home_wp") if c in pbp.columns]
    if hasattr(pbp, "to_pandas"):
        df = pbp.select(cols).to_pandas()
    else:
        df = pbp[cols].copy()
    for c in ("home_wp", "vegas_home_wp"):
        if c not in df.columns:
            df[c] = np.nan
    return df


def align_nflverse_wp(split: S.Split, wp_df) -> dict[str, np.ndarray]:
    """nflverse home_wp and vegas_home_wp aligned to every record (NaN when missing)."""
    import pandas as pd

    rg = record_game(split)
    keys = pd.DataFrame({
        "game_id": np.asarray(split.game_ids, dtype=object)[rg],
        "play_id": np.asarray(split.play_ids, dtype=np.float64),
    })
    wp = wp_df.copy()
    wp["play_id"] = wp["play_id"].astype(np.float64)
    wp = wp.drop_duplicates(["game_id", "play_id"])
    merged = keys.merge(wp, on=["game_id", "play_id"], how="left")
    return {
        "nflverse_home_wp": merged["home_wp"].to_numpy(dtype=np.float64),
        "nflverse_vegas_home_wp": merged["vegas_home_wp"].to_numpy(dtype=np.float64),
    }


def stage_data(run: Run, deadline: float) -> dict:
    """Per season: reuse the saved arrays when they exist. If records.py or schema.py
    changed since they were built, rebuild the season in memory and keep the saved file
    when the arrays are identical (a cosmetic change); replace it when they differ, which
    changes the data content hash and so restarts training."""
    rec = _mod("records")
    p = run.preset
    ddir = run.workdir / "data"
    src_now = run.stage_src("data")
    out: dict[str, Any] = {"splits": {}, "source_hash": src_now, "seasons_rebuilt": {}}
    built: dict[str, S.Split] = {}
    for split_name in ("train", "val", "test"):
        seasons = check_seasons(p["seasons"][split_name])
        parts, reports, wp_parts = [], {}, []
        for season in seasons:
            cache = ddir / f"season_{season}.npz"
            rep_path = ddir / f"season_{season}_report.json"
            meta_path = ddir / f"season_{season}_meta.json"
            meta = read_json(meta_path) if meta_path.exists() else {}
            if time.time() > deadline and not cache.exists():
                raise TimeoutError(f"data stage ran out of time before season {season}")
            if cache.exists() and rep_path.exists():
                sp, extra = load_split(cache)
                report = read_json(rep_path)
                if meta.get("src") != src_now:
                    run.log(f"[data] records/schema code changed: rebuilding {season} in "
                            "memory to compare")
                    new_sp, new_report, new_extra = _build_season(rec, season)
                    same = split_arrays_hash(new_sp, new_extra) == split_arrays_hash(sp, extra)
                    out["seasons_rebuilt"][season] = "identical, kept" if same else "replaced"
                    if not same:
                        sp, report, extra = new_sp, new_report, new_extra
                        save_split(cache, sp, **extra)
                        write_json(rep_path, report)
                    write_json(meta_path, {"src": src_now})
            else:
                run.log(f"[data] downloading and building season {season}")
                sp, report, extra = _build_season(rec, season)
                save_split(cache, sp, **extra)
                write_json(rep_path, report)
                write_json(meta_path, {"src": src_now})
            reports[season] = _jsonable(report)
            parts.append(sp)
            wp_parts.append(extra)
        split = concat_splits(parts)
        extra = {k: np.concatenate([w[k] for w in wp_parts]) for k in wp_parts[0]}
        validation = rec.validate_split(split)
        built[split_name] = split
        target = ddir / f"{split_name}.npz"
        new_hash = split_arrays_hash(split, extra)
        if not target.exists() or run.content_hash(split_name) != new_hash:
            save_split(target, split, **extra)  # unchanged arrays keep their file (and mtime)
        write_json(ddir / f"{split_name}_validation.json", validation)
        write_json(ddir / f"{split_name}_build_report.json", reports)
        out["splits"][split_name] = {
            "seasons": seasons,
            "games": len(split.game_ids),
            "records": n_records(split),
            "content_hash": run.content_hash(split_name),
            "nflverse_wp_coverage": float(np.mean(np.isfinite(extra["nflverse_home_wp"]))),
            "validation": _jsonable(validation),
            "build_report": reports,
        }
        run.log(f"[data] {split_name}: {len(split.game_ids)} games, "
                f"{n_records(split)} records, seasons {seasons}")
    if hasattr(rec, "validate_disjoint"):
        out["disjoint"] = _jsonable(rec.validate_disjoint(built))
    return out



# ---------------------------------------------------------------- frequency baseline
def target_mask(split: S.Split) -> np.ndarray:
    """[N, T] bool: which target fields are scored, the same rule as model.WorldModel.loss.
    Phase is always scored. Down and ydstogo (where 0 means "not a down") are scored on
    every record except a game's last; every other field is scored where its code is not
    padding (0)."""
    t = np.asarray(split.targets)
    last = np.zeros(len(t), dtype=bool)
    last[np.asarray(split.offsets[1:]) - 1] = True
    m = t != 0
    for name in ("down", "ydstogo"):
        m[:, S.TARGET_INDEX[name]] = ~last
    m[:, S.TARGET_INDEX["phase"]] = True
    return m



def fit_frequency_baseline(split: S.Split, alpha: float = 0.5) -> list[np.ndarray]:
    """Per target field, P(field | current phase, current down) with add-alpha smoothing.
    Pad code 0 gets no mass except for fields where 0 is a real value (down, ydstogo)."""
    st, tg = split.states, split.targets
    ph = st[:, S.STATE_INDEX["phase"]].astype(np.int64)
    dn = st[:, S.STATE_INDEX["down"]].astype(np.int64)
    mask = target_mask(split)
    tables = []
    for f, (name, size) in enumerate(S.TARGET_FIELDS):
        counts = np.zeros((len(S.PHASES), 5, size))
        m = mask[:, f]
        np.add.at(counts, (ph[m], dn[m], tg[m, f].astype(np.int64)), 1.0)
        prior = np.full(size, alpha)
        if name not in ("down", "ydstogo"):
            prior[0] = 0.0
        counts += prior
        tables.append(counts / counts.sum(axis=2, keepdims=True))
    return tables


def baseline_nll(tables: list[np.ndarray], split: S.Split) -> dict:
    st, tg = split.states, split.targets
    ph = st[:, S.STATE_INDEX["phase"]].astype(np.int64)
    dn = st[:, S.STATE_INDEX["down"]].astype(np.int64)
    mask = target_mask(split)
    per_field, total, n_rows = {}, 0.0, int(mask[:, 0].sum())
    for f, (name, _size) in enumerate(S.TARGET_FIELDS):
        m = mask[:, f]
        p = tables[f][ph[m], dn[m], tg[m, f].astype(np.int64)]
        nll = -np.log(np.maximum(p, 1e-12))
        per_field[name] = float(nll.mean()) if m.any() else None
        total += float(nll.sum())
    return {"per_field": per_field, "total_per_transition": total / max(n_rows, 1),
            "n_transitions": n_rows}


def _per_field(d: dict) -> dict:
    src = d.get("fields", d.get("per_field", d)) if isinstance(d, dict) else {}
    return {name: float(src[name]) for name, _ in S.TARGET_FIELDS
            if name in src and src[name] is not None}


def nll_comparison(run: Run, split: S.Split, cfg_key: str) -> dict:
    """Model per-field NLL (via train.evaluate_nll) next to the frequency baseline."""
    train_mod = _mod("train")
    tr, _ = run.split("train")
    tables = fit_frequency_baseline(tr)
    max_games = run.preset[cfg_key].get("nll_max_games")
    if max_games and len(split.game_ids) > max_games:
        split = subset_split(split, np.arange(max_games))
    base = baseline_nll(tables, split)
    model_d = train_mod.evaluate_nll(run.load_model(), split, run.model_cfg())
    model_pf = _per_field(model_d)
    return {
        "games": len(split.game_ids),
        "model": _jsonable(model_d),
        "baseline_freq_phase_down": base,
        "per_field": {name: {"model": model_pf.get(name),
                             "baseline": base["per_field"].get(name)}
                      for name, _ in S.TARGET_FIELDS},
        "note": "Model per-field NLL is conditional on the fields sampled before it in the "
                "same transition; the baseline is the marginal given phase and down. The "
                "totals per transition are the comparable numbers.",
    }


# ---------------------------------------------------------------- league stats
def real_league_stats(split: S.Split) -> dict:
    st, tg = split.states, split.targets
    g = len(split.game_ids)
    phase = st[:, S.STATE_INDEX["phase"]]
    down = st[:, S.STATE_INDEX["down"]]
    act = st[:, S.STATE_INDEX["action"]]
    pts = np.abs(S.score_delta_value(tg[:, S.TARGET_INDEX["score_delta"]]))
    fourth = (phase == S.PHASE["PLAY"]) & (down == 4)
    go = np.isin(act, S.GO_ACTIONS)
    fd = fourth & (go | (act == S.ACTION["PUNT"]) | (act == S.ACTION["FG"]))
    final = np.asarray(split.final_diff, dtype=float)
    return {
        "games": g,
        "plays_per_game": float((phase == S.PHASE["PLAY"]).sum() / g),
        "points_per_game_both_teams": float(pts.sum() / g),
        "touchdowns_per_game": float(np.isin(pts, (6, 7, 8)).sum() / g),
        "field_goals_per_game": float((pts == 3).sum() / g),
        "punts_per_game": float((act == S.ACTION["PUNT"]).sum() / g),
        "fourth_down_go_rate": float((fd & go).sum() / max(fd.sum(), 1)),
        "fourth_down_punt_rate": float((fd & (act == S.ACTION["PUNT"])).sum() / max(fd.sum(), 1)),
        "fourth_down_fg_rate": float((fd & (act == S.ACTION["FG"])).sum() / max(fd.sum(), 1)),
        "pass_rate": float((act == S.ACTION["PASS"]).sum()
                           / max(np.isin(act, S.GO_ACTIONS).sum(), 1)),
        "home_win_rate_regulation": float(np.mean(S.outcome(final))),
        "regulation_tie_rate": float(np.mean(final == 0)),
        "mean_abs_final_margin": float(np.mean(np.abs(final))),
    }


# ---------------------------------------------------------------- WP calibration
def log_loss(p: np.ndarray, y: np.ndarray) -> np.ndarray:
    p = np.clip(np.asarray(p, dtype=float), 1e-4, 1 - 1e-4)
    y = np.asarray(y, dtype=float)
    return -(y * np.log(p) + (1 - y) * np.log(1 - p))


def brier(p: np.ndarray, y: np.ndarray) -> np.ndarray:
    return (np.asarray(p, dtype=float) - np.asarray(y, dtype=float)) ** 2


def game_bootstrap(values: np.ndarray, groups: np.ndarray, n_boot: int,
                   rng: np.random.Generator) -> dict:
    """Mean of per-state values with a 95% interval from resampling whole games."""
    values = np.asarray(values, dtype=float)
    uniq, inv = np.unique(groups, return_inverse=True)
    sums = np.bincount(inv, weights=values, minlength=len(uniq))
    cnts = np.bincount(inv, minlength=len(uniq)).astype(float)
    draws = rng.integers(0, len(uniq), size=(n_boot, len(uniq)))
    boot = sums[draws].sum(axis=1) / np.maximum(cnts[draws].sum(axis=1), 1)
    lo, hi = np.percentile(boot, [2.5, 97.5])
    return {"mean": float(values.mean()), "lo95": float(lo), "hi95": float(hi),
            "n_states": int(len(values)), "n_games": int(len(uniq))}


def reliability(p: np.ndarray, y: np.ndarray, n_bins: int = 10) -> dict:
    p, y = np.asarray(p, dtype=float), np.asarray(y, dtype=float)
    b = np.clip((p * n_bins).astype(int), 0, n_bins - 1)
    out = {"bin_mean_pred": [], "bin_mean_outcome": [], "bin_count": []}
    for k in range(n_bins):
        m = b == k
        out["bin_mean_pred"].append(float(p[m].mean()) if m.any() else None)
        out["bin_mean_outcome"].append(float(y[m].mean()) if m.any() else None)
        out["bin_count"].append(int(m.sum()))
    return out


def sample_wp_states(split: S.Split, extra: dict, n: int, rng: np.random.Generator) -> np.ndarray:
    """Random PLAY records that also have both nflverse probabilities, so all three
    estimates are scored on exactly the same states."""
    phase = split.states[:, S.STATE_INDEX["phase"]]
    play = phase == S.PHASE["PLAY"]
    ok = play.copy()
    for k in ("nflverse_home_wp", "nflverse_vegas_home_wp"):
        if k in extra:
            ok &= np.isfinite(extra[k])
    cand = np.flatnonzero(ok)
    if len(cand) == 0:  # nflverse missing everywhere: score ours alone, flagged later
        cand = np.flatnonzero(play)
    n = min(n, len(cand))
    return np.sort(rng.choice(cand, size=n, replace=False)) if n else cand[:0]


def wp_calibration(run: Run, split: S.Split, extra: dict, cfg_key: str, deadline: float,
                   seed: int, pdir: Path) -> dict:
    """Win probability by simulation on a fixed random sample of PLAY records, scored next
    to nflverse's home_wp and vegas_home_wp on the same states (theirs, labelled)."""
    sim = _mod("simulate")
    c = run.preset[cfg_key]
    n_states, n_roll = int(c["wp_states"]), int(c["wp_rollouts"])
    idx = sample_wp_states(split, extra, n_states, np.random.default_rng(seed))
    model, cfg = run.load_model(), run.model_cfg()

    def fn(part, rng):
        return sim.wp_by_simulation(model, split, part, n_roll, cfg, deadline, rng)

    df, info = chunked(run, pdir, fn, idx, _chunk_size(run, n_roll, cfg_key), deadline, seed)
    rng = np.random.default_rng(seed + 1)
    used = idx[: len(df)]
    col = "home_wp_sim" if "home_wp_sim" in df.columns else "sim_wp"
    if len(df) and "idx" in df.columns:
        assert np.array_equal(df["idx"].to_numpy(), used), "simulate returned rows out of order"
    ok = np.isfinite(df[col].to_numpy(dtype=float)) if len(df) else np.zeros(0, dtype=bool)
    if "done" in df.columns:
        ok &= df["done"].to_numpy(dtype=bool)
    n_unfinished = int((~ok).sum())
    df, used = df[ok], used[ok]
    base = {"n_states": int(len(used)), "n_unfinished_dropped": n_unfinished,
            "n_rollouts_per_state": n_roll, "progress": info,
            "complete": info["complete"]}
    if len(used) == 0:
        return {**base, "metrics": {}, "paired_differences": {},
                "note": "no finished rollouts inside the time budget"}
    y = S.outcome(np.asarray(split.final_diff)[record_game(split)[used]])
    games = record_game(split)[used]
    preds = {"ours_simulation": df[col].to_numpy(dtype=float)}
    if "nflverse_home_wp" in extra:
        theirs = {"nflverse_home_wp (theirs)": extra["nflverse_home_wp"][used],
                  "nflverse_vegas_home_wp (theirs)": extra["nflverse_vegas_home_wp"][used]}
        if all(np.isfinite(v).all() for v in theirs.values()):
            preds.update(theirs)
    n_boot = int(c["bootstrap"])
    metrics, diffs = {}, {}
    for k, p in preds.items():
        metrics[k] = {
            "log_loss": game_bootstrap(log_loss(p, y), games, n_boot, rng),
            "brier": game_bootstrap(brier(p, y), games, n_boot, rng),
            "reliability": reliability(p, y),
        }
    for k, p in preds.items():
        if k == "ours_simulation":
            continue
        diffs[f"ours minus {k}"] = {
            "log_loss": game_bootstrap(log_loss(preds["ours_simulation"], y) - log_loss(p, y),
                                       games, n_boot, rng),
            "brier": game_bootstrap(brier(preds["ours_simulation"], y) - brier(p, y),
                                    games, n_boot, rng),
        }
    label_check = None
    lab = "home_result" if "home_result" in df.columns else "actual"
    if lab in df.columns:
        label_check = float(np.mean(np.isclose(df[lab].to_numpy(dtype=float), y)))
    return {
        **base,
        "mean_steps_per_rollout": float(df["mean_steps"].mean())
        if "mean_steps" in df.columns and len(df) else None,
        "outcome_label": "home result at end of regulation: 1 win, 0.5 tie, 0 loss",
        "simulate_actual_matches_label": label_check,
        "metrics": metrics,
        "paired_differences": diffs,
        "source_note": "nflverse home_wp and vegas_home_wp are nflverse's published "
                       "estimates on the same plays, not ours.",
    }


# ---------------------------------------------------------------- stage 2: train
def stage_train(run: Run, deadline: float) -> dict:
    import torch

    model_mod, train_mod = _mod("model"), _mod("train")
    torch.manual_seed(run.seed)
    np.random.seed(run.seed % 2**32)
    tr, _ = run.split("train")
    va, _ = run.split("val")
    mg = run.preset["train"].get("max_train_games")
    if mg and len(tr.game_ids) > mg:
        tr = subset_split(tr, np.arange(mg))
        va = subset_split(va, np.arange(min(len(va.game_ids), max(mg // 4, 8))))
    cfg = run.model_cfg()
    model = model_mod.WorldModel(cfg)
    ck_path = run.ckpt_path()
    fresh_reason = _check_checkpoint(ck_path, model)
    if fresh_reason:
        stale = ck_path.with_name(ck_path.name + f".stale_{int(time.time())}")
        ck_path.replace(stale)
        run.log(f"[train] checkpoint does not fit the current model ({fresh_reason}); "
                f"moved it to {stale.name} and starting fresh")
    progress = run.workdir / "outputs" / "train_progress.json"
    prev = read_json(progress) if progress.exists() else {}
    if prev.get("ckpt") != str(ck_path) or fresh_reason:
        prev = {"seconds": 0.0, "lines": [], "ckpt": str(ck_path)}
    budget_s = 60.0 * run.preset["stage_min"]["train"] - float(prev.get("seconds", 0.0))
    deadline = min(deadline, time.time() + max(budget_s, run.preset["min_start_s"]["train"]))
    t0, base = time.time(), float(prev.get("seconds", 0.0))

    def log(*msg: Any) -> None:
        # Keep the training time used current, so a disconnect mid-training does not hand
        # the resumed run a fresh training budget.
        prev["seconds"] = base + time.time() - t0
        prev["lines"] = (prev.get("lines", []) + [" ".join(str(m) for m in msg)])[-200:]
        write_json(progress, prev)
        run.log("[train]", *msg)

    run.log(f"[train] {model.count_params():,} parameters; {len(tr.game_ids)} train games, "
            f"{n_records(tr):,} records; device {cfg['device']}; checkpoint {ck_path}")
    history = train_mod.train_model(model, tr, va, cfg, str(ck_path), deadline, log)
    if not ck_path.exists():
        raise RuntimeError("train_model returned without writing a checkpoint")
    return {
        "params": int(model.count_params()),
        "train_games": len(tr.game_ids),
        "train_records": n_records(tr),
        "val_games_for_early_stopping": len(va.game_ids),
        "history": _jsonable(history),
        "train_seconds_total": round(base + time.time() - t0, 1),
        "checkpoint": str(ck_path),
        "fresh_start_reason": fresh_reason,
    }


# ---------------------------------------------------------------- stage 3: eval (val)
def stage_eval(run: Run, deadline: float) -> dict:
    sim = _mod("simulate")
    pdir = run.partial_dir("eval")
    va, extra = run.split("val")
    out: dict[str, Any] = {"split": "validation", "seasons": run.preset["seasons"]["val"]}
    out["nll"] = _cached_json(pdir / "nll.json", lambda: nll_comparison(run, va, "eval"))
    run.log(f"[eval] NLL per transition: model "
            f"{_total_nll(out['nll']['model'])}, baseline "
            f"{out['nll']['baseline_freq_phase_down']['total_per_transition']:.3f}")
    out["league_real"] = real_league_stats(va)
    n_games = int(run.preset["eval"]["sim_games"])
    # Simulated games get at most a third of the stage; the rest goes to WP calibration.
    sim_deadline = time.time() + (deadline - time.time()) / 3
    league_path = pdir / "league_sim.json"
    if n_games <= 0:
        out["league_sim"] = {"skipped": "sim_games is 0 in this preset"}
    elif league_path.exists():
        out["league_sim"] = read_json(league_path)
    else:
        res = sim.simulate_games(run.load_model(), va, n_games, run.model_cfg(), sim_deadline,
                                 np.random.default_rng(run.seed + 30))
        res.pop("table", None)  # the same numbers as "sim" and "real"
        out["league_sim"] = _jsonable(res)
        if res.get("n_games", 0) >= n_games:
            write_json(league_path, out["league_sim"])
        run.log(f"[eval] league stats over {res.get('n_games')} simulated games: "
                f"sim {res.get('sim')} real {res.get('real')}")
    out["wp"] = wp_calibration(run, va, extra, "eval", deadline, run.seed + 3, pdir / "wp")
    _log_wp(run, "eval", out["wp"])
    if not out["wp"]["complete"]:
        out["status"] = "partial"
        out["reason"] = (f"time budget: {out['wp']['n_states']} of "
                         f"{out['wp']['progress']['planned_states']} states; the next run "
                         "continues from the saved chunks")
    return out


def _total_nll(d: dict) -> str:
    for k in ("total", "total_per_transition", "nll", "mean"):
        if isinstance(d, dict) and isinstance(d.get(k), (int, float)):
            return f"{d[k]:.3f}"
    return "?"


def _log_wp(run: Run, stage: str, wp: dict) -> None:
    for k, m in wp.get("metrics", {}).items():
        ll, br = m["log_loss"], m["brier"]
        run.log(f"[{stage}] {k}: log loss {ll['mean']:.4f} [{ll['lo95']:.4f}, "
                f"{ll['hi95']:.4f}], Brier {br['mean']:.4f} on {ll['n_states']} states")


# ---------------------------------------------------------------- stage 4: prereg + test
def test_plan(run: Run) -> dict:
    p = run.preset
    sc = _mod("support").config(p["plan"])
    return {
        "config_hash": run.full_config_hash(),
        "train_hash": run.stage_hash("train"),
        "test_hash": run.stage_hash("test"),
        "test_seasons": p["seasons"]["test"],
        "seed": run.seed + 4,
        "metrics": [
            "per-field and total NLL per transition, model vs frequency baseline fitted on "
            "the training split (P(field | phase, down), add-0.5 smoothing)",
            f"win probability by simulation on {p['test']['wp_states']} random PLAY records "
            f"with both nflverse probabilities present, {p['test']['wp_rollouts']} rollouts "
            "each; log loss and Brier with a 95% game-bootstrap interval "
            f"({p['test']['bootstrap']} resamples) for ours, nflverse home_wp and nflverse "
            "vegas_home_wp on the same states, and paired differences",
            f"fourth-down plan on {p['plan']['n_fourth']} sampled test fourth downs, "
            f"{p['plan']['n_rollouts']} rollouts per action; field goals beyond the longest "
            f"made in the training split plus {sc['fg_margin_yards']} yards and punts inside "
            f"the opponent's {sc['punt_min_yardline']} are not simulated; agreement of the "
            "recommended action with the real coach on decisions not flagged too_close, also "
            "split by training support (low support: the recommended call made fewer than "
            f"{sc['min_support']} times by coaches in comparable training states)",
        ],
        "outcome_label": "home result at end of regulation: 1 win, 0.5 tie, 0 loss",
        "preset": p["name"],
        "model": p["model"],
        "rule": "each configuration (full config hash: data, model, training, evaluation, "
                "test and plan settings) preregisters its own plan in prereg/<hash>.json "
                "before it reads the test seasons, and scores them once; a second scoring of "
                "the same configuration needs allow_rescore=True and is recorded as such. "
                "Every preregistration in the work folder is kept and listed in results.json.",
    }


# ---------------------------------------------------------------- preregistration files
# prereg/<config hash>.json           the plan, written once before any test data is read,
#                                     never overwritten
# prereg/<config hash>.attempts.json  every test or plan stage start (a log; appended)
# prereg/<config hash>.scored.json    written once, when that configuration's test scoring
#                                     completed
def prereg_dir(run: Run) -> Path:
    d = run.workdir / "prereg"
    d.mkdir(exist_ok=True)
    return d


def prereg_path(run: Run, config_hash_: str | None = None) -> Path:
    return prereg_dir(run) / f"{config_hash_ or run.full_config_hash()}.json"


def _write_once(path: Path, obj: Any) -> bool:
    """Write path only if it does not exist yet. Returns True if this call wrote it."""
    if path.exists():
        return False
    write_json(path, obj)
    return True


def migrate_legacy_prereg(run: Run) -> list[str]:
    """Copy the first notebook's single prereg.json (and any earlier plans chained under
    "previous") and outputs/test_scored.json into the per-configuration files. The legacy
    files are left where they are. Returns the hashes imported."""
    imported = []
    legacy = run.workdir / "prereg.json"
    d = prereg_dir(run)
    p = read_json(legacy) if legacy.exists() else None
    while isinstance(p, dict):
        h = p.get("config_hash")
        rec = {k: v for k, v in p.items() if k != "previous"}
        rec["migrated_from"] = "prereg.json (single-file format of the first notebook)"
        if h and _write_once(d / f"{h}.json", rec):
            imported.append(h)
        p = p.get("previous")
    scored = run.workdir / "outputs" / "test_scored.json"
    if scored.exists():
        s = read_json(scored)
        if s.get("config_hash"):
            _write_once(d / f"{s['config_hash']}.scored.json",
                        {**s, "scored_at": s.get("at"),
                         "migrated_from": "outputs/test_scored.json"})
    return imported


def _migrate_and_log(run: Run, stage: str) -> None:
    for h in migrate_legacy_prereg(run):
        run.log(f"[{stage}] imported the earlier single-file preregistration of config {h} "
                "into prereg/ (prereg.json is left in place)")


def ensure_prereg(run: Run, stage: str, rescore: bool = False) -> Path:
    """This configuration's preregistration: written now if it does not exist (BEFORE the
    caller reads any test data), never overwritten. Every call is logged as an attempt."""
    _migrate_and_log(run, stage)
    plan = test_plan(run)
    path = prereg_path(run, plan["config_hash"])
    if _write_once(path, {**plan, "written_at": time.time(), "written_by_stage": stage}):
        run.log(f"[{stage}] preregistered this configuration's test plan: {path}")
    elif read_json(path).get("test_hash") != plan["test_hash"]:  # cannot happen by design
        raise RuntimeError(f"{path} does not match this configuration's test plan")
    att_path = path.with_name(f"{plan['config_hash']}.attempts.json")
    attempts = read_json(att_path) if att_path.exists() else []
    attempts.append({"stage": stage, "started_at": time.time(), "rescore": rescore})
    write_json(att_path, attempts)
    return path


def list_preregistrations(run: Run) -> list[dict]:
    """Every preregistration in the work folder, oldest first, with whether and when that
    configuration scored the test seasons."""
    migrate_legacy_prereg(run)
    current = run.full_config_hash()
    rows = []
    for f in prereg_dir(run).glob("*.json"):
        if f.name.endswith((".attempts.json", ".scored.json")):
            continue
        p = read_json(f)
        h = p.get("config_hash", f.stem)
        sc = f.with_name(f"{h}.scored.json")
        att = f.with_name(f"{h}.attempts.json")
        attempts = read_json(att) if att.exists() else p.get("attempts", [])
        scored = read_json(sc) if sc.exists() else None
        rows.append({
            "config_hash": h,
            "test_hash": p.get("test_hash"),
            "train_hash": p.get("train_hash"),
            "preset": p.get("preset"),
            "written_at": p.get("written_at"),
            "test_scored": scored is not None,
            "test_scored_at": (scored or {}).get("scored_at"),
            "attempts": len(attempts),
            "rescores": sum(bool(a.get("rescore")) for a in attempts),
            "this_run": h == current,
            "file": f"prereg/{f.name}",
            "migrated_from": p.get("migrated_from"),
        })
    return sorted(rows, key=lambda r: r["written_at"] or 0)


def _scorings_note(rows: list[dict]) -> dict:
    scored = [r for r in rows if r["test_scored"]]
    note = (f"The test seasons in this work folder have been scored by {len(scored)} "
            "configuration(s)")
    if len(scored) > 1:
        note += ("; each is a separate experiment on the same test seasons, and only the "
                 "first was chosen before any test-season result existed")
    return {"n_configurations_scored": len(scored),
            "config_hashes_scored": [r["config_hash"] for r in scored],
            "note": note + "."}


def stage_test(run: Run, deadline: float) -> dict:
    plan = test_plan(run)
    h = plan["config_hash"]
    scored_path = prereg_dir(run) / f"{h}.scored.json"
    _migrate_and_log(run, "test")
    rescore = scored_path.exists()
    if rescore and not run.allow_rescore:
        raise PermissionError(
            f"the test seasons were already scored by this configuration (config hash {h}); "
            "refusing a second scoring of the same configuration. Set allow_rescore=True to "
            "override; the rescore is recorded in prereg/.")
    path = ensure_prereg(run, "test", rescore=rescore)  # BEFORE any test data is loaded

    pdir = run.partial_dir("test")
    te, extra = run.split("test")
    out: dict[str, Any] = {"split": "test", "seasons": run.preset["seasons"]["test"],
                           "prereg": str(path), "config_hash": h,
                           "test_hash": plan["test_hash"], "rescore": rescore}
    out["nll"] = _cached_json(pdir / "nll.json", lambda: nll_comparison(run, te, "test"))
    out["league_real"] = real_league_stats(te)
    out["wp"] = wp_calibration(run, te, extra, "test", deadline, run.seed + 4, pdir / "wp")
    _log_wp(run, "test", out["wp"])
    if out["wp"]["complete"]:
        _write_once(scored_path, {"config_hash": h, "test_hash": plan["test_hash"],
                                  "scored_at": time.time()})
    else:
        out["status"] = "partial"
        out["reason"] = "time budget; the next run continues the same preregistered plan"
    return out


# ---------------------------------------------------------------- stage 5: plan
HYPOTHETICAL = {
    "description": "offense up 4, 4:00 left in the 4th quarter, 4th and 2 at its own 40",
    "lead_for_offense": 4,
    "clock_seconds": 240,
    "down": 4,
    "ydstogo": 2,
    "yardline_100": 60,
}


def build_hypothetical(split: S.Split) -> tuple[S.Split, dict]:
    """Edit the codes of a real late-game record into the hypothetical state.

    The base record is the first PLAY record, in game order of the split, with 3:20 to
    5:00 left in the game and a real down. Its game is cut after that record; score,
    clock, down, distance and yard line are overwritten (score from the offense's side),
    while possession, timeouts, the pregame spread and the earlier records stay as they
    were in the real game. The record's own action field is left as recorded; the planner
    replaces it with each candidate action.
    """
    st, raw = split.states, split.raw
    phase = st[:, S.STATE_INDEX["phase"]]
    down = st[:, S.STATE_INDEX["down"]]
    poss = st[:, S.STATE_INDEX["poss"]]
    clock = raw[:, 0]
    cand = np.flatnonzero((phase == S.PHASE["PLAY"]) & (down >= 1) & (clock >= 200)
                          & (clock <= 300) & (poss > 0))
    if len(cand) == 0:
        raise ValueError("no late-game PLAY record to edit")
    i = int(cand[0])
    g = int(record_game(split)[i])
    sub = subset_split(split, [g])
    j = i - int(split.offsets[g])
    sub = S.Split(
        states=sub.states[: j + 1].copy(), raw=sub.raw[: j + 1].copy(),
        targets=sub.targets[: j + 1].copy(),
        offsets=np.array([0, j + 1], dtype=np.int64), header=sub.header.copy(),
        final_diff=sub.final_diff.copy(), game_ids=[sub.game_ids[0] + "_HYPOTHETICAL"],
        seasons=sub.seasons.copy(), play_ids=sub.play_ids[: j + 1].copy())
    h = HYPOTHETICAL
    home_has_ball = int(sub.states[j, S.STATE_INDEX["poss"]]) == S.POSS_CODE["HOME"]
    diff = h["lead_for_offense"] if home_has_ball else -h["lead_for_offense"]
    before = {name: int(sub.states[j, S.STATE_INDEX[name]]) for name, _ in S.STATE_FIELDS}
    raw_before = [int(v) for v in sub.raw[j]]
    sub.raw[j, 0] = h["clock_seconds"]
    sub.raw[j, 1] = diff
    sub.states[j, S.STATE_INDEX["clock"]] = S.clock_code(h["clock_seconds"])
    sub.states[j, S.STATE_INDEX["score"]] = S.score_code(diff)
    sub.states[j, S.STATE_INDEX["down"]] = h["down"]
    sub.states[j, S.STATE_INDEX["ydstogo"]] = h["ydstogo"]
    sub.states[j, S.STATE_INDEX["yardline"]] = h["yardline_100"]
    sub.targets[j] = 0
    sub.targets[j, S.TARGET_INDEX["phase"]] = S.PHASE["END"]
    doc = {
        **h,
        "base_game_id": split.game_ids[g],
        "base_play_id": int(split.play_ids[i]),
        "base_record_index": i,
        "base_codes_before_edit": before,
        "base_raw_before_edit": dict(zip(S.RAW_COLUMNS, raw_before, strict=True)),
        "codes_after_edit": {name: int(sub.states[j, S.STATE_INDEX[name]])
                             for name, _ in S.STATE_FIELDS},
        "raw_after_edit": dict(zip(S.RAW_COLUMNS, [int(v) for v in sub.raw[j]], strict=True)),
        "offense": "HOME" if home_has_ball else "AWAY",
        "kept_from_real_game": "possession, timeouts, pregame spread, opening receiver, "
                               "all earlier records",
        "limitation": "earlier records are the real game's, so the history does not lead "
                      "exactly to the edited score and clock",
    }
    return sub, doc


def _agreement(df) -> dict:
    if df is None or len(df) == 0 or "best" not in df.columns:
        return {"n": 0}
    n_unfinished = 0
    if "done" in df.columns:
        n_unfinished = int((~df["done"].astype(bool)).sum())
        df = df[df["done"].astype(bool)]
    if len(df) == 0:
        return {"n": 0, "n_unfinished_dropped": n_unfinished}
    col = "real_choice" if "real_choice" in df.columns else "real_action"
    real = df[col].astype(str).str.upper()
    best = df["best"].astype(str).str.upper()
    close = df["too_close"].astype(bool) if "too_close" in df.columns else best.isna()
    agree = real == best
    confusion: dict[str, dict[str, int]] = {}
    for b, r in zip(best, real, strict=True):
        confusion.setdefault(b, {}).setdefault(r, 0)
        confusion[b][r] += 1
    return {
        "n": int(len(df)),
        "n_unfinished_dropped": n_unfinished,
        "real_column": col,
        "n_clear": int((~close).sum()),
        "agreement_all": float(agree.mean()),
        "agreement_clear": float(agree[~close].mean()) if (~close).any() else None,
        "recommended_counts": best.value_counts().to_dict(),
        "real_counts": real.value_counts().to_dict(),
        "confusion_recommended_by_real": confusion,
    }


SUPPORT_NOTE = ("The hard action rules and the support flags were added after the first Colab "
                "T4 run's test-season fourth-down results had been seen (see "
                "results/world_model/colab_t4_run_summary.md). They are a post hoc change to "
                "the planner, not part of that run's preregistered test plan.")


def _agreement_by_support(df, min_support: int) -> dict:
    """Agreement with the coach, separately for well-supported and low-support
    recommendations (finished decisions only)."""
    if df is None or len(df) == 0 or "low_support" not in df.columns:
        return {"min_support": min_support, "well_supported": {"n": 0},
                "low_support": {"n": 0}}
    d = df[df["done"].astype(bool)] if "done" in df.columns else df
    low = d["low_support"].astype(bool)
    return {
        "min_support": min_support,
        "rule": f"low support: coaches made the recommended call fewer than {min_support} "
                "times in comparable training fourth downs (yard line within 3, same "
                "yards-to-go bucket)",
        "n_low_support": int(low.sum()),
        "well_supported": _agreement(d[~low]),
        "low_support": _agreement(d[low]),
    }


def _fg_recommendations(df) -> dict:
    """The field-goal calls: the simulator's one-step make rate against the training make
    rate at that yard line, and what the coach did."""
    if df is None or len(df) == 0 or "best" not in df.columns:
        return {"n": 0}
    d = df[df["best"] == "FG"]
    if "done" in d.columns:
        d = d[d["done"].astype(bool)]
    out: dict[str, Any] = {"n": int(len(d))}
    if len(d) == 0:
        return out
    out["coach_choice_counts"] = d["real_choice"].astype(str).value_counts().to_dict()
    if "low_support" in d.columns:
        out["n_low_support"] = int(d["low_support"].astype(bool).sum())
    for k in ("sim_fg_make", "emp_fg_make"):
        if k in d.columns:
            v = d[k].to_numpy(dtype=float)
            out[f"mean_{k}"] = float(np.nanmean(v)) if np.isfinite(v).any() else None
    return out


def _support_diagnostic(run: Run, sp, model, te: S.Split, support, rules: dict, cfg: dict,
                        pool: np.ndarray, decision_idx: np.ndarray, hyp_split: S.Split,
                        hyp_idx: np.ndarray, pdir: Path, deadline: float) -> dict:
    """Simulated versus empirical make rate and punt net yards (support.sim_vs_empirical),
    plus the one-step field-goal make rate of every planned state and of the hypothetical.
    One encoder pass per state, so it is cheap; it gets at most an eighth of the time left
    and is saved once complete. A failure is recorded, never fatal to the plan."""
    path = pdir / "support_diagnostic.json"
    if path.exists():
        return read_json(path)
    dl = time.time() + max(deadline - time.time(), 0.0) / 8
    rng = np.random.default_rng(run.seed + 70)
    n = int(support.cfg["diag_draws"])
    try:
        d = sp.sim_vs_empirical(model, te, support, rules, cfg, rng, pool, dl)
        d["fg_headline"] = sp.fg_gap_summary(d, int(support.cfg["min_support"]))
        d["decision_states"] = {
            "idx": decision_idx,
            "sim_fg_make": sp.sim_fg_make(model, te, decision_idx, n, cfg, rng, dl)}
        hyp = sp.sim_fg_make(model, hyp_split, hyp_idx, n, cfg, rng, dl)
        yl = int(hyp_split.states[int(hyp_idx[0]), S.STATE_INDEX["yardline"]])
        emp, att = sp.empirical_fg_rate(support, [yl])
        d["hypothetical_fg"] = {"yardline_100": yl,
                                "distance_yards": yl + sp.KICK_DISTANCE_OFFSET,
                                "sim_make_rate": float(hyp[0]),
                                "train_attempts_within_2_yards": int(att[0]),
                                "empirical_make_rate": float(emp[0])}
    except Exception as e:  # recorded in results.json; the plan itself goes on
        run.log(f"[plan] support diagnostic failed: {type(e).__name__}: {e}")
        return {"error": f"{type(e).__name__}: {e}"}
    d = _jsonable(d)
    write_json(path, d)
    return d


def stage_plan(run: Run, deadline: float) -> dict:
    rec, sim, sp = _mod("records"), _mod("simulate"), _mod("support")
    # The plan also reads the test seasons, so it needs this configuration's preregistration
    # too (written here if the test stage was skipped or failed).
    ensure_prereg(run, "plan")
    pdir = run.partial_dir("plan")
    p = run.preset["plan"]
    scfg = sp.config(p)
    # Support tables and hard rules: from the TRAINING split only, never validation or test.
    tr, _ = run.split("train")
    support = sp.build_support(tr, scfg)
    del tr
    rules = sp.hard_rules(support)
    run.log(f"[plan] action rules: {rules['text']}")
    te, _ = run.split("test")
    model, cfg = run.load_model(), run.model_cfg()
    out: dict[str, Any] = {"split": "test", "seasons": run.preset["seasons"]["test"]}
    out["support"] = {"config": scfg, "rules": rules, "note": SUPPORT_NOTE,
                      "training_tables": support.to_json()}
    # The hypothetical first (one state, many rollouts), with up to a quarter of the time.
    hyp_split, doc = build_hypothetical(te)
    hyp_deadline = time.time() + (deadline - time.time()) / 4
    hyp_idx = np.array([n_records(hyp_split) - 1])

    def hyp_fn(part, rng):
        return sim.plan_fourth_downs(model, hyp_split, part, int(p["hypo_rollouts"]), cfg,
                                     hyp_deadline, rng,
                                     action_mask=sp.action_mask(hyp_split, part, rules))

    hdf, hinfo = chunked(run, pdir / "hypothetical", hyp_fn, hyp_idx, 1, hyp_deadline,
                         run.seed + 50)
    if len(hdf):
        hdf = sp.add_support_columns(hdf, support, hyp_split, rules)
    out["hypothetical"] = {"state": doc, "result": _jsonable(hdf), "progress": hinfo,
                           "n_rollouts_per_action": int(p["hypo_rollouts"])}
    if len(hdf):
        run.log(f"[plan] hypothetical: {hdf.to_dict(orient='records')}")
    # Sampled real fourth downs (fixed sample; chunks saved as they finish).
    idx_all = np.asarray(rec.fourth_down_index(te))
    n = min(int(p["n_fourth"]), len(idx_all))
    rng = np.random.default_rng(run.seed + 5)
    idx = np.sort(rng.choice(idx_all, size=n, replace=False)) if n else idx_all[:0]
    masked = (sp.masked_actions(hyp_split, hyp_idx, rules, "hypothetical")
              + sp.masked_actions(te, idx, rules, "test"))
    out["masked_actions"] = masked
    test_masked = [m for m in masked if m["set"] == "test"]
    out["masked_summary"] = {
        "planned_states": int(n),
        "states_with_a_masked_action": len({m["idx"] for m in test_masked}),
        "fg_masked": sum(m["masked"] == "FG" for m in test_masked),
        "punt_masked": sum(m["masked"] == "PUNT" for m in test_masked),
        "hypothetical_masked": [m["masked"] for m in masked if m["set"] == "hypothetical"],
    }
    # Simulated versus empirical kicking (cheap), before the long planning loop.
    diag = _support_diagnostic(run, sp, model, te, support, rules, cfg, idx_all, idx,
                               hyp_split, hyp_idx, pdir, deadline)
    out["support"]["diagnostic"] = diag
    if diag.get("fg_headline"):
        run.log(f"[plan] support diagnostic: {diag['fg_headline']}")

    def fn(part, rng):
        return sim.plan_fourth_downs(model, te, part, int(p["n_rollouts"]), cfg, deadline, rng,
                                     action_mask=sp.action_mask(te, part, rules))

    chunk = max(1, _chunk_size(run, 3 * int(p["n_rollouts"]), "plan"))
    df, info = chunked(run, pdir / "fourth", fn, idx, chunk, deadline, run.seed + 5)
    if len(df):
        df = sp.add_support_columns(df, support, te, rules)
        yl = te.states[df["idx"].to_numpy(dtype=np.int64), S.STATE_INDEX["yardline"]]
        emp, att = sp.empirical_fg_rate(support, yl)
        df["emp_fg_make"], df["emp_fg_attempts"] = emp, att
        ds = diag.get("decision_states") or {}
        sim_fg = dict(zip(ds.get("idx", []), ds.get("sim_fg_make", []), strict=True))
        df["sim_fg_make"] = [sim_fg.get(int(i), np.nan) for i in df["idx"]]
    out["progress"] = info
    out["n_fourth_downs_available"] = int(len(idx_all))
    out["n_rollouts_per_action"] = int(p["n_rollouts"])
    out["decisions"] = _jsonable(df)
    out["agreement"] = _agreement(df)
    out["agreement_by_support"] = _agreement_by_support(df, int(scfg["min_support"]))
    out["fg_recommendations"] = _fg_recommendations(df)
    run.log(f"[plan] {out['agreement']}")
    run.log(f"[plan] by support: well supported {out['agreement_by_support']['well_supported']}"
            f"; low support {out['agreement_by_support']['low_support']}")
    if not (info["complete"] and hinfo["complete"]):
        out["status"] = "partial"
        out["reason"] = (f"time budget: {info['reused_states'] + info['computed_states']} of "
                         f"{n} fourth downs; the next run continues from the saved chunks")
    return out


# ---------------------------------------------------------------- stage 6: package
def versions() -> dict:
    v: dict[str, Any] = {"python": platform.python_version(), "platform": platform.platform()}
    for name in ("torch", "numpy", "polars", "pandas", "nflreadpy", "matplotlib"):
        try:
            mod = importlib.import_module(name)
            v[name] = getattr(mod, "__version__", "unknown")
        except Exception:
            v[name] = None
    try:
        import torch

        v["cuda_available"] = bool(torch.cuda.is_available())
        if torch.cuda.is_available():
            props = torch.cuda.get_device_properties(0)
            v["gpu"] = props.name
            v["gpu_mem_gb"] = round(props.total_memory / 2**30, 1)
            v["cuda"] = torch.version.cuda
    except Exception:
        pass
    return v


def stage_package(run: Run, deadline: float) -> dict:
    results: dict[str, Any] = {
        "experiment": "NFL world model: next-play transformer used as a simulator",
        "preset": run.preset,
        "seed": run.seed,
        "stage_seeds": {"train": run.seed, "eval": run.seed + 3, "test": run.seed + 4,
                        "plan": run.seed + 5},
        "config_hash": run.full_config_hash(),
        "stage_hashes": {s: run.stage_hash(s) for s in STAGES},
        "versions": versions(),
        "workdir": str(run.workdir),
        "budget_s": run.total_s,
        "consumed_s_before_this_session": run.consumed_before,
        "session_seconds": round(time.time() - run.session_start, 1),
        "stages": {},
        "honesty": "Every number here was produced by this run. Fields labelled 'theirs' or "
                   "'nflverse' are nflverse's published estimates on the same plays.",
    }
    for s in STAGES:
        if s == "package":
            continue
        results["stages"][s] = run.current_output(s)
    results["benchmark"] = run.preset.get("benchmark")
    results["preset_choice"] = run.preset.get("picked")
    mine = prereg_path(run)
    if mine.exists():
        results["prereg"] = read_json(mine)
    results["preregistrations"] = list_preregistrations(run)
    results["test_scorings"] = _scorings_note(results["preregistrations"])
    figs: dict[str, Any] = {}
    try:
        figs = make_figures(run, results)
    except Exception as e:
        figs = {"error": f"{type(e).__name__}: {e}"}
    results["figures"] = figs
    write_json(run.workdir / "results.json", results)
    zpath = run.workdir / "world_model_results.zip"
    with zipfile.ZipFile(zpath, "w", zipfile.ZIP_DEFLATED) as z:
        z.write(run.workdir / "results.json", "results.json")
        for sub in ("outputs", "figures"):
            for f in sorted((run.workdir / sub).glob("*")):
                z.write(f, f"{sub}/{f.name}")
        for f in sorted((run.workdir / "data").glob("*.json")):
            z.write(f, f"data_reports/{f.name}")
        for f in sorted(prereg_dir(run).glob("*.json")):
            z.write(f, f"prereg/{f.name}")
        for f in ("prereg.json", "run.log", "ledger.json", "preset.json"):
            if (run.workdir / f).exists():
                z.write(run.workdir / f, f)
    run.log(f"[package] results at {run.workdir / 'results.json'}; zip at {zpath}")
    return {"results": str(run.workdir / "results.json"), "zip": str(zpath), "figures": figs}


# ---------------------------------------------------------------- figures
_STYLE = {
    "SURFACE": "#fcfcfb", "INK": "#0b0b0b", "INK_2": "#52514e", "MUTED": "#898781",
    "GRID": "#e1e0d9", "AXIS": "#c3c2b7", "BLUE": "#2a78d6", "ORANGE": "#eb6834",
    "AQUA": "#1baf7a", "VIOLET": "#6250d6", "NEUTRAL": "#c3c2b7",
}
_PRED_COLORS = {"ours_simulation": _STYLE["BLUE"], "nflverse_home_wp (theirs)": _STYLE["ORANGE"],
                "nflverse_vegas_home_wp (theirs)": _STYLE["VIOLET"]}
_ACTION_COLORS = {"GO": _STYLE["BLUE"], "FG": _STYLE["ORANGE"], "PUNT": _STYLE["AQUA"]}


def _apply_style(plt) -> None:
    s = _STYLE
    plt.rcParams.update({
        "figure.facecolor": s["SURFACE"], "axes.facecolor": s["SURFACE"],
        "savefig.facecolor": s["SURFACE"], "font.family": "sans-serif",
        "font.sans-serif": ["DejaVu Sans", "Arial", "Helvetica"], "font.size": 10.5,
        "text.color": s["INK"], "axes.edgecolor": s["AXIS"], "axes.labelcolor": s["INK_2"],
        "axes.titlecolor": s["INK"], "axes.titlesize": 12, "axes.titleweight": "bold",
        "axes.titlelocation": "left", "axes.spines.top": False, "axes.spines.right": False,
        "axes.grid": True, "axes.axisbelow": True, "grid.color": s["GRID"],
        "xtick.color": s["MUTED"], "ytick.color": s["MUTED"], "legend.frameon": False,
        "lines.linewidth": 2.0, "savefig.dpi": 160, "savefig.bbox": "tight",
    })


def make_figures(run: Run, results: dict) -> dict:
    import matplotlib

    matplotlib.use("Agg")
    import matplotlib.pyplot as plt

    _apply_style(plt)
    fdir = run.workdir / "figures"
    made: dict[str, str] = {}
    st = results["stages"]

    hist = (st.get("train", {}).get("history") or {}).get("history") or []
    series = {
        "train (nats per record)": [h.get("train_loss") for h in hist],
        "validation (nats per record)": [h.get("val_total") for h in hist],
    }
    if hist:
        fig, ax = plt.subplots(figsize=(7, 4))
        colors = (_STYLE["NEUTRAL"], _STYLE["BLUE"])
        for (k, v), c in zip(series.items(), colors, strict=True):
            ep = [h.get("epoch", i) + 1 for i, h in enumerate(hist)]
            ax.plot(ep, v, label=k, color=c, marker="o", ms=3)
        base = (st.get("eval", {}).get("nll") or {}).get("baseline_freq_phase_down", {})
        if base.get("total_per_transition"):
            ax.axhline(base["total_per_transition"], color=_STYLE["ORANGE"], lw=1.2, ls="--",
                       label="frequency baseline, validation")
        ax.set_xlabel("epoch")
        ax.set_ylabel("NLL, nats per record (sum over fields)")
        ax.set_title("Training curves")
        ax.legend()
        made["training_curves"] = _save(fig, fdir / "training_curves.png", plt)

    panels = [(name, st[name]["wp"]) for name in ("eval", "test")
              if st.get(name, {}).get("status") in ("done", "partial") and "wp" in st[name]]
    if panels:
        fig, axes = plt.subplots(1, len(panels), figsize=(5.5 * len(panels), 5), squeeze=False)
        for ax, (name, wp) in zip(axes[0], panels, strict=True):
            ax.plot([0, 1], [0, 1], color=_STYLE["AXIS"], lw=1, ls="--")
            for k, m in wp["metrics"].items():
                r = m["reliability"]
                xs = [a for a, b in zip(r["bin_mean_pred"], r["bin_mean_outcome"], strict=True)
                      if a is not None and b is not None]
                ys = [b for a, b in zip(r["bin_mean_pred"], r["bin_mean_outcome"], strict=True)
                      if a is not None and b is not None]
                ll = m["log_loss"]["mean"]
                ax.plot(xs, ys, marker="o", color=_PRED_COLORS.get(k, _STYLE["INK_2"]),
                        label=f"{k}, log loss {ll:.3f}")
            label = "validation" if name == "eval" else "test"
            ax.set_title(f"Home win probability calibration ({label}, {wp['n_states']} states)")
            ax.set_xlabel("predicted")
            ax.set_ylabel("observed (regulation result)")
            ax.legend(fontsize=8.5, loc="upper left")
        made["calibration"] = _save(fig, fdir / "calibration.png", plt)

    nll = st.get("eval", {}).get("nll")
    if nll:
        names = [n for n, v in nll["per_field"].items()
                 if v["model"] is not None and v["baseline"] is not None]
        if names:
            x = np.arange(len(names))
            fig, ax = plt.subplots(figsize=(8, 4))
            ax.bar(x - 0.2, [nll["per_field"][n]["baseline"] for n in names], 0.4,
                   color=_STYLE["NEUTRAL"], label="frequency baseline (phase, down)")
            ax.bar(x + 0.2, [nll["per_field"][n]["model"] for n in names], 0.4,
                   color=_STYLE["BLUE"], label="world model")
            ax.set_xticks(x, names, rotation=30, ha="right")
            ax.set_ylabel("NLL (nats), validation")
            ax.set_title("Per-field NLL: world model vs frequency baseline")
            ax.legend()
            made["nll_vs_baseline"] = _save(fig, fdir / "nll_vs_baseline.png", plt)

    plan = st.get("plan", {})
    if plan.get("status") in ("done", "partial") and plan.get("hypothetical"):
        fig, axes = plt.subplots(1, 2, figsize=(11, 4.2))
        res = plan["hypothetical"]["result"]
        acts = [("GO", "wp_go", "se_go"), ("FG", "wp_fg", "se_fg"),
                ("PUNT", "wp_punt", "se_punt")]
        ax = axes[0]
        for k, (a, w, s) in enumerate(acts):
            v = (res.get(w) or [None])[0]
            if v is None:  # masked by a hard rule, or not finished in time
                allowed = (res.get(f"allowed_{a.lower()}") or [True])[0]
                ax.text(k, 0.02, "not allowed\n(hard rule)" if allowed is False else "not run",
                        ha="center", va="bottom", fontsize=8.5, color=_STYLE["MUTED"])
                continue
            se = (res.get(s) or [0])[0]
            ax.bar(k, v, yerr=1.96 * (se or 0), color=_ACTION_COLORS[a], capsize=4)
            ax.text(k, v + 0.01, f"{v:.3f}", ha="center", va="bottom", fontsize=9)
        ax.set_xticks(range(len(acts)), [a for a, _, _ in acts])
        ax.set_ylabel("simulated win probability, offense")
        ax.set_title("Up 4, 4:00 left, 4th and 2 at own 40", fontsize=11)
        ag = plan.get("agreement", {})
        ax = axes[1]
        conf = ag.get("confusion_recommended_by_real", {})
        reals = sorted({r for v in conf.values() for r in v})
        recs = sorted(conf)
        bottom = np.zeros(len(recs))
        for r in reals:
            h = np.array([conf[b].get(r, 0) for b in recs], dtype=float)
            ax.bar(recs, h, bottom=bottom, label=f"coach: {r}",
                   color=_ACTION_COLORS.get(r, _STYLE["NEUTRAL"]))
            bottom += h
        a_all = ag.get("agreement_all")
        ax.set_title(f"Model recommendation vs coach ({ag.get('n', 0)} test 4th downs"
                     + (f", agree {a_all:.0%})" if a_all is not None else ")"), fontsize=11)
        ax.set_xlabel("model recommends")
        ax.set_ylabel("fourth downs")
        ax.legend()
        made["fourth_down"] = _save(fig, fdir / "fourth_down.png", plt)
    diag = (plan.get("support") or {}).get("diagnostic") or {}
    if plan.get("status") in ("done", "partial") and diag.get("fg"):
        made["fg_support"] = _fg_support_figure(diag, plan["support"]["rules"],
                                                fdir / "fg_support.png", plt)
    return made


def _band_series(rows: list[dict], key: str) -> np.ndarray:
    return np.array([np.nan if r.get(key) is None else float(r[key]) for r in rows])


def _fg_support_figure(diag: dict, rules: dict, path: Path, plt) -> str:
    """Simulated vs training make rate by distance (left) and punt net yards by yard line
    (right), each over the number of training attempts in the band. Shaded: masked."""
    s = _STYLE
    fig, axes = plt.subplots(2, 2, figsize=(12, 6.8), sharex="col",
                             gridspec_kw={"height_ratios": [3, 1.3], "hspace": 0.08,
                                          "wspace": 0.22})
    panels = (
        ("fg", "distance_yards", "train_attempts", "empirical_make_rate", "sim_make_rate",
         "Field goals: simulated vs training make rate", "make rate",
         "field goal distance (yards) = yard line + 18", "training\nattempts"),
        ("punt", "yardline", "train_punts", "empirical_net_yards", "sim_net_yards",
         "Punts: simulated vs training net yards", "net yards",
         "yard line of the punt (yards from the opponent's end zone)", "training\npunts"),
    )
    for col, (key, xkey, nkey, ekey, skey, title, ylab, xlab, nlab) in enumerate(panels):
        rows = diag.get(key) or []
        if not rows:
            continue
        x = np.array([(r[xkey][0] + r[xkey][1]) / 2 for r in rows])
        width = rows[0][xkey][1] - rows[0][xkey][0] + 1
        emp, sim, se = (_band_series(rows, k) for k in (ekey, skey, "sim_se"))
        ax = axes[0, col]
        if key == "fg":
            lo, hi = rules["fg_max_distance_yards"] + 0.5, x.max() + width / 2
            label = f"not allowed:\nbeyond {rules['fg_max_distance_yards']} yd"
            ax.set_ylim(0, 1.04)
        else:
            lo, hi = x.min() - width / 2, rules["punt_min_yardline"] - 0.5
            label = f"not allowed:\ninside the\nopponent's {rules['punt_min_yardline']}"
        if hi > lo:
            ax.axvspan(lo, hi, color=s["GRID"], alpha=0.7, lw=0)
            ax.text(lo + 0.6, 0.97, label, transform=ax.get_xaxis_transform(), va="top",
                    fontsize=8.5, color=s["INK_2"])
        ok = np.isfinite(emp)
        ax.plot(x[ok], emp[ok], color=s["ORANGE"], marker="o", ms=6,
                label="training seasons: what coaches' kicks did")
        ok = np.isfinite(sim)
        ax.errorbar(x[ok], sim[ok], yerr=1.96 * np.nan_to_num(se[ok]), color=s["BLUE"],
                    marker="s", ms=6, capsize=3,
                    label="world model: forced kick, simulated next play (95% interval)")
        ax.set_title(title, pad=8)
        ax.set_ylabel(ylab)
        ax = axes[1, col]
        n = np.array([r[nkey] or 0 for r in rows])
        ax.bar(x, n, width=width * 0.85, color=s["NEUTRAL"])
        for xi, ni in zip(x, n, strict=True):
            ax.text(xi, ni, f"{ni}", ha="center", va="bottom", fontsize=8, color=s["INK_2"])
        ax.set_ylim(0, max(int(n.max()), 1) * 1.35)
        ax.set_ylabel(nlab)
        ax.set_xlabel(xlab)
        ax.grid(axis="x", visible=False)
    handles, labels = axes[0, 0].get_legend_handles_labels()
    fig.legend(handles, labels, loc="lower center", bbox_to_anchor=(0.5, 0.95), ncol=2,
               fontsize=9)
    fig.text(0.01, -0.03,
             f"Simulated: up to {diag.get('states_per_band')} test-season fourth downs per band "
             f"(whatever the coach did), the kick forced, {diag.get('draws_per_state')} sampled "
             "next plays each.\nTraining: every attempt in the training seasons. Coaches pick "
             "when to kick, so long-range training rates are for selected attempts.",
             fontsize=8.5, color=s["MUTED"], va="top")
    return _save(fig, path, plt)


def _save(fig, path: Path, plt) -> str:
    fig.savefig(path)
    plt.close(fig)
    return str(path)


def split_arrays_hash(split: S.Split, extra: dict | None = None) -> str:
    """Content hash of a split's arrays (and any extra aligned arrays)."""
    h = hashlib.sha256()
    for k in _SPLIT_ARRAYS:
        a = np.ascontiguousarray(getattr(split, k))
        h.update(f"{k}{a.dtype.str}{a.shape}".encode())
        h.update(a.tobytes())
    h.update("\x1f".join(split.game_ids).encode())
    for k in sorted(extra or {}):
        a = np.ascontiguousarray(extra[k])
        h.update(f"{k}{a.dtype.str}{a.shape}".encode())
        h.update(a.tobytes())
    return h.hexdigest()[:16]


def _build_season(rec, season: int) -> tuple[S.Split, dict, dict]:
    pbp = rec.load_pbp(check_seasons([season]))
    sched = rec.load_schedules(check_seasons([season]))
    sp, report = rec.build_split(pbp, sched, [season])
    extra = align_nflverse_wp(sp, _pbp_wp_columns(pbp))
    del pbp
    return sp, report, extra


def chunked(run: Run, pdir: Path, fn, idx: np.ndarray, chunk: int, deadline: float,
            seed: int) -> tuple[Any, dict]:
    """Run fn(idx_part, rng) -> DataFrame over fixed chunks of idx, saving each finished
    chunk under pdir and reusing saved chunks on the next run. Each chunk has its own seed,
    so a resumed result is the same as an uninterrupted one. Stops before a chunk that
    would not finish by the deadline."""
    import pandas as pd

    pdir.mkdir(parents=True, exist_ok=True)
    idx = np.asarray(idx, dtype=np.int64)
    n_chunks = int(np.ceil(len(idx) / max(chunk, 1)))
    frames, counters = [], []
    reused = computed = 0
    per_state, stopped = None, None
    for k in range(n_chunks):
        part = idx[k * chunk:(k + 1) * chunk]
        f = pdir / f"chunk_{k:04d}.pkl"
        if f.exists():
            frames.append(pd.read_pickle(f))
            cf = f.with_suffix(".json")
            counters.append(read_json(cf) if cf.exists() else None)
            reused += len(part)
            continue
        left = deadline - time.time()
        if left <= 0 or (per_state is not None and left < 1.1 * per_state * len(part)):
            stopped = "time budget"
            break
        t0 = time.time()
        df = fn(part, np.random.default_rng([seed, k]))
        per_state = (time.time() - t0) / max(len(part), 1)
        complete = "done" not in df.columns or bool(df["done"].astype(bool).all())
        frames.append(df)
        counters.append(_jsonable(df.attrs.get("counters")))
        computed += len(part)
        if complete:
            tmp = f.with_suffix(".tmp")
            df.to_pickle(tmp)
            tmp.replace(f)
            write_json(f.with_suffix(".json"), counters[-1])
        else:
            stopped = "deadline reached inside a chunk"
            break
    df = pd.concat(frames, ignore_index=True) if frames else pd.DataFrame()
    info = {"planned_states": int(len(idx)), "chunk_states": int(chunk),
            "chunks": n_chunks, "reused_states": reused, "computed_states": computed,
            "stopped": stopped, "complete": stopped is None,
            "seconds_per_state": None if per_state is None else round(per_state, 4),
            "rollout_counters": counters}
    return df, info


def _chunk_size(run: Run, n_rollouts: int, key: str) -> int:
    return _chunk_states(run.preset, n_rollouts, key)


def _chunk_states(preset: dict, n_rollouts: int, key: str) -> int:
    """States per saved chunk of a stage (n_rollouts = rollouts per state per chunk call)."""
    c = preset[key]
    if c.get("chunk_states"):
        return int(c["chunk_states"])
    live = int(preset["train"].get("sim_max_live", 8192))
    return max(4, live // max(int(n_rollouts), 1))


def _cached_json(path: Path, make):
    if path.exists():
        return read_json(path)
    val = _jsonable(make())
    write_json(path, val)
    return val


def _check_checkpoint(path: Path, model) -> str | None:
    """None if there is no checkpoint or it fits the model; else why it does not."""
    if not path.exists():
        return None
    try:
        import torch

        ck = torch.load(path, map_location="cpu", weights_only=False)
        if not isinstance(ck, dict):
            return "not a checkpoint dict"
        saved = ck.get("current_model") or ck.get("model")
        if not isinstance(saved, dict):
            return None  # not ours to judge (a stub or another format)
        cur = model.state_dict()
        missing = [k for k in cur if k not in saved]
        extra = [k for k in saved if k not in cur]
        shape = [k for k in cur if k in saved and tuple(saved[k].shape) != tuple(cur[k].shape)]
        if missing or extra or shape:
            return (f"{len(missing)} missing, {len(extra)} unexpected, {len(shape)} "
                    "shape-mismatched tensors")
    except Exception as e:
        return f"unreadable: {type(e).__name__}: {e}"
    return None


def checkpoint_info(run: Run) -> dict:
    path = run.ckpt_path()
    info: dict[str, Any] = {"path": str(path), "exists": path.exists()}
    if path.exists():
        try:
            import torch

            ck = torch.load(path, map_location="cpu", weights_only=False)
            if isinstance(ck, dict):
                info["epochs_done"] = ck.get("epoch")
                info["best_val_nll"] = ck.get("best_val")
                info["finished"] = ck.get("finished")
        except Exception as e:
            info["error"] = f"{type(e).__name__}: {e}"
    return info


def status_table(run: Run) -> str:
    """Which stages are done, the checkpoint, its last epoch and best validation NLL, and
    minutes used so far."""
    lines = [f"Work folder {run.workdir}", f"{'stage':<8} {'state':<12} seconds"]
    for s in STAGES[:-1]:
        if run.is_done(s):
            state = "done"
        else:
            state = run.status(s)
            state = "not started" if state == "missing" else state
        secs = read_json(run.output_path(s)).get("seconds", "") \
            if run.output_path(s).exists() else ""
        lines.append(f"{s:<8} {state:<12} {secs}")
    if run.content_hash("train") != "missing":
        ck = checkpoint_info(run)
        if ck["exists"]:
            best = ck.get("best_val_nll")
            best_s = f"{best:.4f}" if isinstance(best, (int, float)) else "n/a"
            lines.append(f"checkpoint {ck['path']}: {ck.get('epochs_done')} epochs done, best "
                         f"validation NLL {best_s} nats/record"
                         + (", training finished" if ck.get("finished") else ""))
        else:
            lines.append(f"checkpoint {ck['path']}: none yet")
    lines.append(f"minutes used so far {run.consumed_before / 60:.1f} of "
                 f"{run.total_s / 60:.0f}; {run.remaining() / 60:.1f} left this session")
    return "\n".join(lines)


def stopped_message(stage: str) -> str:
    return f"Stopped safely at {stage}. Run all again to continue from here."


def _beyond(stage: str, stop_after: str | None) -> bool:
    if stop_after is None or stage == "package":
        return False
    if stop_after not in STAGES:
        raise ValueError(f"STOP_AFTER must be None or one of {STAGES[:-1]}")
    return STAGES.index(stage) > STAGES.index(stop_after)


def notebook_stage(run: Run, stage: str, stop_after: str | None = None) -> dict:
    """One notebook cell's worth of work. Honors STOP_AFTER; on the stop button, saves what
    exists, writes results.json and re-raises so Colab's Run all stops here."""
    if _beyond(stage, stop_after):
        msg = (f"[{stage}] not run: STOP_AFTER = {stop_after!r}. Set STOP_AFTER = None and "
               "Run all to continue.")
        print(msg)
        return {"status": "not_run", "reason": msg}
    try:
        return run_stage(run, stage)
    except KeyboardInterrupt:
        _safe_package(run)
        print(stopped_message(stage), flush=True)
        raise


def _safe_package(run: Run) -> None:
    try:
        run_stage(run, "package")
    except BaseException as e:  # never let packaging hide the original stop
        run.log(f"[package] could not package after the stop: {e}")



STAGE_FUNCS = {
    "data": stage_data,
    "train": stage_train,
    "eval": stage_eval,
    "test": stage_test,
    "plan": stage_plan,
    "package": stage_package,
}


# ---------------------------------------------------------------- preflight
def system_info() -> dict:
    info = versions()
    try:
        import shutil

        du = shutil.disk_usage(default_workdir().parent if default_workdir().parent.exists()
                               else "/")
        info["disk_free_gb"] = round(du.free / 2**30, 1)
    except Exception:
        pass
    try:
        with open("/proc/meminfo") as f:
            kb = int(next(line for line in f if line.startswith("MemTotal")).split()[1])
        info["ram_gb"] = round(kb / 2**20, 1)
    except Exception:
        pass
    return info


def _bench_states(syn: S.Split, n: int) -> np.ndarray:
    """n start records near the middle of synthetic games (cycling through the games), so
    the shared history is about as long as at a real mid-game fourth down."""
    lens = np.diff(syn.offsets)
    g = np.arange(n) % len(lens)
    return np.asarray(syn.offsets[g] + lens[g] // 2, dtype=np.int64)


def apply_benchmark(preset: dict, rec_per_s: float, units: float):
    """The benchmark's rule, as a pure function of the measured speeds. Returns (adjusted
    preset, list of changes, decision text, train speed share, rollout speed share)."""
    a = preset["assumed"]
    tr_ratio = rec_per_s / a["train_rec_per_s"]
    ro_ratio = units / a["rollout_units_per_s"]
    adjusted = copy.deepcopy(preset)
    changes: list[str] = []
    downgraded = None
    if preset["device"] == "cuda" and tr_ratio < BENCH_SLOW_SHARE \
            and DOWNGRADE[preset["name"]] != preset["name"]:
        smaller = PRESETS[DOWNGRADE[preset["name"]]]
        adjusted["model"] = copy.deepcopy(smaller["model"])
        adjusted["train"]["batch_games"] = smaller["train"]["batch_games"]
        downgraded = (f"model downgraded to the {smaller['name']} size (training at "
                      f"{tr_ratio:.0%} of the assumed speed)")
        changes.append(downgraded)
    if ro_ratio < BENCH_SLOW_SHARE:
        f = ro_ratio
        for block, key in (("eval", "wp_states"), ("eval", "sim_games"), ("test", "wp_states"),
                           ("test", "sim_games"), ("plan", "n_fourth"),
                           ("plan", "hypo_rollouts")):
            old = int(adjusted[block][key])
            new = max(int(old * f), min(old, BENCH_FLOORS[key]))
            if new != old:
                adjusted[block][key] = new
                changes.append(f"{block}.{key} {old} -> {new}")
        decision = (f"scaled down: rollouts at {units:,.0f} steps/s are {ro_ratio:.0%} of the "
                    f"{a['rollout_units_per_s']:,} the preset assumes (below the "
                    f"{BENCH_SLOW_SHARE:.0%} threshold), so sizes were multiplied by {f:.2f} "
                    f"with floors {BENCH_FLOORS}")
    else:
        decision = (f"kept the preset sizes: rollouts at {units:,.0f} steps/s are "
                    f"{ro_ratio:.0%} of the {a['rollout_units_per_s']:,} the preset assumes "
                    f"(it is scaled only below {BENCH_SLOW_SHARE:.0%})")
    if downgraded:
        decision += "; " + downgraded
    return adjusted, changes, decision, tr_ratio, ro_ratio


def micro_benchmark(preset: dict, seconds: float = 30.0) -> dict:
    """Measure training and rollout speed on this machine with the code the stages use, and
    scale the preset DOWN only if this machine is much slower than the preset assumes.

    Training: two short epochs on a synthetic split (2 x batch_games games on a GPU); the
    speed is read from the second epoch, after warm-up. Rollouts: the current rollout path
    (the key/value cache) at the batch one plan-stage call uses on this GPU (states per
    chunk x n_rollouts), from mid-game states. The model is untrained, so its END phase is
    switched off and max_steps is STEPS_PER_ROLLOUT: every rollout runs about as long as a
    real one, and the speed is rollout steps (transitions) per second.

    Rule: a size is scaled only when the measured speed is below BENCH_SLOW_SHARE of the
    preset's assumed speed, by the measured share, never below BENCH_FLOORS (or the preset's
    own value if smaller). Otherwise the preset is kept as written; every stage still runs
    against its own deadline and saves partial progress. The synthetic split is not
    football; it only fixes tensor shapes."""
    import torch

    model_mod, train_mod, sim = _mod("model"), _mod("train"), _mod("simulate")
    device = "cuda" if preset["device"] == "cuda" and torch.cuda.is_available() else "cpu"
    cfg = {**preset["model"], **preset["train"], "device": device, "seed": SEED, "epochs": 2}
    cfg.pop("max_train_games", None)
    n_games = 2 * int(cfg["batch_games"]) if device == "cuda" else 6
    syn = synthetic_split(n_games=n_games, seed=1)
    model = model_mod.WorldModel(cfg)
    tmp = Path(os.environ.get("TMPDIR", "/tmp")) / f"wm_bench_{os.getpid()}_{time.time_ns()}.pt"
    tmp.unlink(missing_ok=True)
    t0 = time.time()
    try:
        hist = train_mod.train_model(model, syn, subset_split(syn, [0, 1]), cfg, str(tmp),
                                     time.time() + seconds / 2, lambda *a: None)
    finally:
        tmp.unlink(missing_ok=True)
    train_s = time.time() - t0
    warm = [h for h in hist.get("history") or [] if h.get("epoch", 0) >= 1]
    rec_per_s = (warm[-1].get("timing") or {}).get("train_records_per_s") if warm else None
    train_basis = "second epoch (after warm-up)"
    if not rec_per_s or not np.isfinite(rec_per_s):
        rec_per_s = hist.get("train_records_per_s")
        train_basis = "all steps, including warm-up"
    if not rec_per_s or not np.isfinite(rec_per_s):
        rec_per_s = n_records(syn) / max(train_s, 1e-6)
        train_basis = "wall time of the whole call"
    rec_per_s = float(rec_per_s)

    # ---- rollouts: the current path, at one plan-stage call's batch
    if hasattr(model, "eval"):
        model.eval()
    heads = getattr(model, "heads", None)
    if heads is not None:
        with torch.no_grad():  # untrained: never end the game early by sampling END
            heads[S.TARGET_INDEX["phase"]][-1].bias[S.PHASE["END"]] = -1e4
    rcfg = {**cfg, "kv_cache": True, "max_steps": STEPS_PER_ROLLOUT}
    n_roll = int(preset["plan"]["n_rollouts"])
    live = int(sim.live_rows(model, rcfg)) if hasattr(sim, "live_rows") else n_roll
    states = max(1, min(_chunk_states(preset, 3 * n_roll, "plan"), live // max(n_roll, 1)))
    idx = _bench_states(syn, states)
    rng = np.random.default_rng(0)
    sim.rollout(model, syn, idx[:1], None, 16, {**rcfg, "max_steps": 3}, time.time() + 30,
                rng)  # warm-up: first-call kernel and allocator setup
    planned = states * n_roll * STEPS_PER_ROLLOUT
    t0 = time.time()
    res = sim.rollout(model, syn, idx, None, n_roll, rcfg, t0 + seconds / 2, rng)
    roll_s = time.time() - t0
    finished = bool(np.all(res["done"]))
    if finished:
        units = float(res["counters"]["transitions"]) / max(roll_s, 1e-6)
        speed_note = "measured"
    else:  # did not finish in its time: at most this fast
        units = planned / max(roll_s, 1e-6)
        speed_note = "upper bound (the batch did not finish in its time)"
    adjusted, changes, decision, tr_ratio, ro_ratio = apply_benchmark(preset, rec_per_s, units)

    def minutes(p: dict) -> dict:
        out = {}
        for key in ("eval", "test"):
            c = p[key]
            u = c["wp_states"] * c["wp_rollouts"] * STEPS_PER_ROLLOUT \
                + c["sim_games"] * 2 * STEPS_PER_ROLLOUT
            out[key] = round(u / max(units, 1e-6) / 60, 1)
        c = p["plan"]
        u = 3 * (c["n_fourth"] * c["n_rollouts"] + c["hypo_rollouts"]) * STEPS_PER_ROLLOUT
        out["plan"] = round(u / max(units, 1e-6) / 60, 1)
        return out

    timing = {k: (round(v, 3) if isinstance(v, float) else v)
              for k, v in (res.get("timing") or {}).items() if k != "method"}
    return {
        "device": device,
        "train_records_per_s": round(rec_per_s, 1),
        "train_speed_basis": train_basis,
        "rollout_steps_per_s": round(units, 1),
        "rollout_speed_basis": speed_note,
        "rollout_batch": {"states": int(states), "rollouts_per_state": n_roll,
                          "rows": int(states * n_roll), "live_rows_cap": live,
                          "max_steps": STEPS_PER_ROLLOUT, "path": "kv_cache",
                          "seconds": round(roll_s, 2), "timing": timing},
        "assumed": dict(preset["assumed"]),
        "train_speed_vs_assumed": round(tr_ratio, 2),
        "rollout_speed_vs_assumed": round(ro_ratio, 2),
        "threshold_share": BENCH_SLOW_SHARE,
        "floors": dict(BENCH_FLOORS),
        "decision": decision,
        "changes": changes,
        "estimated_minutes_at_measured_speed": {"preset": minutes(preset),
                                                "after_adjustment": minutes(adjusted)},
        "stage_ceilings_min": dict(preset["stage_min"]),
        "note": "rollout speed from an untrained model with END switched off on synthetic "
                "games; each stage also times its own chunks and stops at its deadline, "
                "saving what it finished",
        "adjusted_preset": adjusted,
    }


def estimate_minutes(preset: dict) -> dict:
    sm = preset["stage_min"]
    return {"budget_min": preset["budget_min"], "stage_ceilings_min": sm}


# Files of a run that a fresh preset pick starts over: moved into archive/ (results.json,
# the zip, the time ledger, preset.json, figures) or copied there (stage outputs, the legacy
# preregistration and the log, which stay in place: finished data is reused by hash).
_ARCHIVE_MOVE = ("results.json", "world_model_results.zip", "ledger.json", "preset.json")
_ARCHIVE_COPY = ("prereg.json", "run.log")


def archive_previous_run(wd: Path, reason: str) -> Path | None:
    """Keep the previous run's results before a fresh preset pick. Moves results.json, the
    zip, figures/, preset.json and ledger.json (so the time budget starts again) into
    archive/run_<time>/, and copies outputs/, prereg.json and run.log there. Data, the
    checkpoints and the per-configuration preregistrations stay where they are."""
    import shutil

    wd = Path(wd)
    present = [n for n in _ARCHIVE_MOVE + _ARCHIVE_COPY if (wd / n).exists()]
    figs = sorted((wd / "figures").glob("*")) if (wd / "figures").is_dir() else []
    outs = sorted((wd / "outputs").glob("*")) if (wd / "outputs").is_dir() else []
    if not (present or figs or outs):
        return None
    dest = wd / "archive" / f"run_{time.strftime('%Y%m%d_%H%M%S')}"
    k = 1
    while dest.exists():
        dest = dest.with_name(dest.name.split("__")[0] + f"__{k}")
        k += 1
    dest.mkdir(parents=True)
    for n in present:
        (shutil.move if n in _ARCHIVE_MOVE else shutil.copy2)(str(wd / n), str(dest / n))
    if figs:
        (dest / "figures").mkdir()
        for f in figs:
            shutil.move(str(f), str(dest / "figures" / f.name))
    if outs:
        shutil.copytree(wd / "outputs", dest / "outputs")
    write_json(dest / "ARCHIVED.json", {"archived_at": time.time(), "reason": reason})
    return dest


def prepare(preset_name: str | None = None, workdir: Path | None = None, benchmark: bool = True,
            budget_min: float | None = None, force_new_preset: bool = False,
            allow_rescore: bool = False, force: tuple = ()) -> Run:
    """Pick (or reload) the preset, benchmark it once, print the status table, return a Run.

    The (possibly benchmark-adjusted) preset is saved in the work folder with its definition
    (PRESET_VERSION and a hash of its PRESETS entry) and reused on every resume, so a
    resumed run keeps the same config hashes and skips finished stages. When the saved
    definition differs from the code (or force_new_preset), the preset is picked afresh:
    the previous run's results are archived and the time ledger starts again.
    """
    wd = Path(workdir or default_workdir())
    wd.mkdir(parents=True, exist_ok=True)
    saved = wd / "preset.json"
    notes: list[str] = []
    preset = None
    if saved.exists():
        old = read_json(saved)
        name_old = old.get("name")
        cur = preset_definition(name_old) if name_old in PRESETS else None
        if force_new_preset:
            reason = "force_new_preset was set"
        elif cur is None:
            reason = f"its preset name {name_old!r} is not in this code"
        elif old.get("definition") is None:
            reason = "it has no definition version (saved by the first notebook)"
        elif old["definition"] != cur:
            reason = (f"its definition (version {old['definition'].get('version')!r}, hash "
                      f"{old['definition'].get('hash')}) differs from the code's (version "
                      f"{cur['version']!r}, hash {cur['hash']})")
        else:
            reason = None
            preset = old
            notes.append(f"Reusing the saved preset '{name_old}' from {saved} (definition "
                         f"hash {cur['hash']})")
            want = preset_name or pick_preset()
            if want != name_old:
                notes.append(f"Note: this machine would get preset '{want}'; the saved "
                             f"'{name_old}' is kept so finished stages resume. Set "
                             "FORCE_NEW_PRESET = True to start over with a fresh pick.")
        if reason:
            dest = archive_previous_run(wd, f"fresh preset pick: {reason}")
            notes.append(f"The saved preset in {saved} is replaced: {reason}. Picking the "
                         "preset afresh"
                         + (f"; the previous run's results, figures and time ledger were "
                            f"moved to {dest}" if dest else ""))
    elif (wd / "ledger.json").exists() or (wd / "results.json").exists():
        dest = archive_previous_run(wd, "fresh preset pick: no preset.json, but an earlier "
                                        "run's ledger or results")
        notes.append(f"No saved preset, but an earlier run's files were found; moved them to "
                     f"{dest} so this run starts with the full time budget")
    if preset is None:
        name = preset_name or pick_preset()
        preset = get_preset(name)
        preset["definition"] = preset_definition(name)
        if benchmark:
            try:
                bench = micro_benchmark(preset)
                preset = bench.pop("adjusted_preset")
                preset["benchmark"] = bench
                notes.append(f"Micro-benchmark: {bench['decision']}")
                print("Micro-benchmark:", json.dumps(
                    {k: v for k, v in bench.items() if k != "rollout_batch"}, indent=1))
            except Exception as e:
                preset["benchmark"] = {"error": f"{type(e).__name__}: {e}",
                                       "decision": "kept the preset as is (benchmark failed)"}
                notes.append(f"Micro-benchmark failed, keeping the preset as is: {e}")
        preset["picked"] = {"at": time.time(), "notes": list(notes)}
        write_json(saved, preset)
    if budget_min is not None:
        preset["budget_min"] = budget_min
    run = Run(preset, wd, budget_min=budget_min, allow_rescore=allow_rescore, force=force)
    for n in notes:
        run.log(f"[preflight] {n}")
    print(status_table(run), flush=True)
    return run


def run_all(run: Run, stages=STAGES, stop_after: str | None = None) -> dict:
    """Run the stages in order (up to stop_after), then package. A KeyboardInterrupt or an
    exception in a stage still ends with results.json; an interrupt prints where it
    stopped."""
    out: dict[str, Any] = {}
    current = None
    try:
        for s in stages:
            if s == "package" or _beyond(s, stop_after):
                continue
            current = s
            out[s] = run_stage(run, s)
    except KeyboardInterrupt:
        _safe_package(run)
        out["stopped_at"] = current
        msg = stopped_message(current)
        run.log(msg)
        if run.quiet:
            print(msg, flush=True)
        return out
    out["package"] = run_stage(run, "package")
    return out


def summary(run: Run) -> str:
    """Plain-language summary of results.json."""
    path = run.workdir / "results.json"
    if not path.exists():
        return "No results.json yet."
    r = read_json(path)
    st = r["stages"]
    lines = [f"Preset {r['preset']['name']}, config {r['config_hash']}, "
             f"GPU {r['versions'].get('gpu', 'none')}."]
    bench = r.get("benchmark") or {}
    if bench.get("decision"):
        lines.append(f"  preflight benchmark: {bench['decision']}")
    ts = r.get("test_scorings") or {}
    if ts.get("n_configurations_scored", 0) > 1:
        lines.append(f"  {ts['note']} (config hashes {', '.join(ts['config_hashes_scored'])}; "
                     "see results.json 'preregistrations')")
    for s in ("data", "train", "eval", "test", "plan"):
        o = st.get(s, {})
        lines.append(f"- {s}: {o.get('status')}"
                     + (f" ({o.get('reason') or o.get('error')})"
                        if o.get("status") != "done" else f" in {o.get('seconds')}s"))
    d = st.get("data", {})
    if d.get("status") == "done":
        for k, v in d["splits"].items():
            lines.append(f"  {k}: {v['games']} games, {v['records']} records, "
                         f"seasons {v['seasons'][0]}-{v['seasons'][-1]}")
    t = st.get("train", {})
    if t.get("status") == "done":
        h = t.get("history") or {}
        speed = h.get("train_records_per_s")
        lines.append(f"  model: {t['params']:,} parameters, trained {t['train_seconds_total']}s"
                     + (f" on {h['device']}" if h.get("device") else "")
                     + (f" ({h.get('epochs_done')} epochs, {speed:,.0f} records/s)"
                        if isinstance(speed, (int, float)) and np.isfinite(speed) else ""))
    for s, label in (("eval", "validation"), ("test", "test")):
        o = st.get(s, {})
        if o.get("status") not in ("done", "partial") or not o.get("wp", {}).get("metrics"):
            continue
        b = o["nll"]["baseline_freq_phase_down"]["total_per_transition"]
        lines.append(f"  {label} NLL per transition: model {_total_nll(o['nll']['model'])}"
                     f" vs frequency baseline {b:.3f} (lower is better)")
        for k, m in o["wp"]["metrics"].items():
            ll = m["log_loss"]
            lines.append(f"  {label} win probability, {k}: log loss {ll['mean']:.4f} "
                         f"(95% game bootstrap {ll['lo95']:.4f} to {ll['hi95']:.4f}), "
                         f"Brier {m['brier']['mean']:.4f}, {ll['n_states']} states")
    p = st.get("plan", {})
    if p.get("status") in ("done", "partial") and p.get("agreement", {}).get("n"):
        ag = p["agreement"]
        lines.append(f"  fourth downs: model agrees with the coach on "
                     f"{ag.get('agreement_all', 0):.0%} of {ag.get('n')} test decisions"
                     + (f", {ag['agreement_clear']:.0%} of the {ag['n_clear']} that are not "
                        "too close to call" if ag.get("agreement_clear") is not None else ""))
        res = p["hypothetical"]["result"]
        bits = [f"{a} {res[k][0]:.3f}" for a, k in (("go", "wp_go"), ("field goal", "wp_fg"),
                                                     ("punt", "wp_punt"))
                if res.get(k) and res[k][0] is not None]
        bits += [f"{a} not allowed (hard rule)" for a, k in (("go", "allowed_go"),
                                                           ("field goal", "allowed_fg"),
                                                           ("punt", "allowed_punt"))
                 if (res.get(k) or [True])[0] is False]
        lines.append(f"  hypothetical (up 4, 4:00 left, 4th and 2 at own 40): "
                     f"{', '.join(bits)}; best {res.get('best', ['?'])[0]}")
    if p.get("status") in ("done", "partial"):
        lines += _support_summary_lines(p)
    return "\n".join(lines)


def _fmt(v, spec: str = ".2f") -> str:
    return "none" if v is None else format(v, spec)


def _pct(a: dict) -> str:
    v = a.get("agreement_all")
    return "n/a" if v is None else f"{v:.0%}"


def _support_summary_lines(p: dict) -> list[str]:
    """The action-support part of the summary: rules, masked actions, low-support flags and
    the simulated-versus-training kicking diagnostic, in plain language."""
    sup = p.get("support") or {}
    rules = sup.get("rules")
    if not rules:
        return []
    lines = [f"  action rules (from the training seasons only): {rules['text']}"]
    ms = p.get("masked_summary") or {}
    if ms.get("planned_states"):
        lines.append(f"  masked: {ms['states_with_a_masked_action']} of {ms['planned_states']} "
                     f"sampled test fourth downs had an action ruled out (field goal on "
                     f"{ms['fg_masked']}, punt on {ms['punt_masked']}); each is listed under "
                     "stages.plan.masked_actions")
    ab = p.get("agreement_by_support") or {}
    ws, ls = ab.get("well_supported") or {}, ab.get("low_support") or {}
    n_ws, n_ls = ws.get("n", 0), ls.get("n", 0)
    if n_ws or n_ls:
        lines.append(f"  support: {n_ls} of {n_ws + n_ls} recommendations have low support (the "
                     f"coaches made that call fewer than {ab.get('min_support')} times in "
                     "comparable training fourth downs); agreement with the coach "
                     f"{_pct(ws)} on the {n_ws} well-supported, {_pct(ls)} on the {n_ls} "
                     "low-support")
    fr = p.get("fg_recommendations") or {}
    if fr.get("n"):
        bit = ""
        if fr.get("mean_sim_fg_make") is not None and fr.get("mean_emp_fg_make") is not None:
            bit = (f"; on them the simulator's make rate averages {fr['mean_sim_fg_make']:.2f} "
                   f"against {fr['mean_emp_fg_make']:.2f} for training kicks from the same "
                   "yard lines (within 2)")
        lines.append(f"  field-goal calls: {fr['n']} (coach: {fr.get('coach_choice_counts')})"
                     + bit)
    diag = sup.get("diagnostic") or {}
    if diag.get("error"):
        lines.append(f"  kicking diagnostic failed: {diag['error']}")
        return lines
    rows = [r for r in diag.get("fg") or [] if r.get("sim_make_rate") is not None]
    if rows:
        bits = [f"{r['distance_yards'][0]}-{r['distance_yards'][1]} yd {r['sim_make_rate']:.2f}"
                f" vs {_fmt(r['empirical_make_rate'])} ({r['train_attempts']})" for r in rows]
        lines.append("  field goals, simulated vs training make rate (training attempts): "
                     + "; ".join(bits))
        h = diag.get("fg_headline") or {}
        parts = []
        if h.get("weighted_mean_gap_with_data") is not None:
            a, b = h["largest_gap_band"]
            parts.append(f"where training has at least {h['min_attempts']} attempts the "
                         f"simulator is off by {h['weighted_mean_gap_with_data']:+.2f} on "
                         f"average (attempt-weighted), at most {h['largest_gap']:+.2f} at "
                         f"{a}-{b} yd")
        if h.get("sim_make_rate_thin_bands") is not None:
            parts.append(f"in thin bands it makes {h['sim_make_rate_thin_bands']:.2f}")
        if h.get("sim_make_rate_beyond_limit") is not None:
            parts.append(f"beyond the {rules['fg_max_distance_yards']}-yd limit it still makes "
                         f"{h['sim_make_rate_beyond_limit']:.2f}")
        if parts:
            lines.append("  " + "; ".join(parts))
    hf = diag.get("hypothetical_fg")
    if hf:
        lines.append(f"  the hypothetical's {hf['distance_yards']}-yard field goal: the simulator "
                     f"makes {hf['sim_make_rate']:.2f} of them (masked by the hard rule)")
    prow = [r for r in diag.get("punt") or [] if r.get("sim_net_yards") is not None]
    if prow:
        bits = [f"{r['yardline'][0]}-{r['yardline'][1]} {r['sim_net_yards']:.1f} vs "
                f"{_fmt(r['empirical_net_yards'], '.1f')} ({r['train_punts']})" for r in prow]
        lines.append("  punts, simulated vs training net yards by yard line (training punts): "
                     + "; ".join(bits))
    return lines


def main(argv: list[str] | None = None) -> None:
    import argparse

    ap = argparse.ArgumentParser(description=__doc__.splitlines()[0])
    ap.add_argument("--preset", choices=sorted(PRESETS), default=None)
    ap.add_argument("--workdir", type=Path, default=None)
    ap.add_argument("--budget-min", type=float, default=None)
    ap.add_argument("--no-benchmark", action="store_true")
    ap.add_argument("--allow-rescore", action="store_true")
    ap.add_argument("--force-new-preset", action="store_true")
    ap.add_argument("--stop-after", choices=[s for s in STAGES if s != "package"],
                    default=None)
    ap.add_argument("--force", nargs="*", default=[], choices=list(STAGES[:-1]),
                    help="stages to recompute even if done")
    a = ap.parse_args(argv)
    run = prepare(a.preset, a.workdir, benchmark=not a.no_benchmark, budget_min=a.budget_min,
                  force_new_preset=a.force_new_preset, allow_rescore=a.allow_rescore,
                  force=tuple(a.force))
    out = run_all(run, stop_after=a.stop_after)
    print(summary(run))
    if out.get("stopped_at"):
        raise SystemExit(130)


if __name__ == "__main__":
    main()
```

```
%%writefile /content/wm_code/test_amp_cpu.py
"""Mixed precision on the CPU: the GPU code paths under autocast, simulated with bfloat16.

On a GPU, `train.amp_settings` turns autocast on (bfloat16 on compute capability 8 and up,
float16 with a GradScaler on a T4). On the CPU it is off, so these tests patch it to
(True, bfloat16, False) and run a training step and validation pass under autocast, and
rollouts through the key/value cache with bfloat16 caches and cascade attention. The
float16 GradScaler path needs CUDA and is not covered here."""

from __future__ import annotations

import time

import numpy as np
import pytest
import schema as S
import simulate
import torch
import train
from model import WorldModel
from toy import make_toy_split
from train import collate, train_model

torch.set_num_threads(1)
TINY = {"d_model": 32, "n_layers": 2, "n_heads": 2, "d_ff": 64, "dec_hidden": 32,
        "dropout": 0.0, "max_len": 48, "batch_games": 8, "eval_batch_games": 16, "lr": 3e-3,
        "warmup_steps": 5, "epochs": 1, "max_steps": 150, "sim_batch": 64, "seed": 0}
FAR = time.time() + 3600.0


def _cpu_bf16_autocast() -> bool:
    try:
        with torch.autocast("cpu", dtype=torch.bfloat16):
            y = torch.nn.functional.linear(torch.randn(2, 4), torch.randn(3, 4))
        return y.dtype == torch.bfloat16
    except Exception:
        return False


pytestmark = pytest.mark.skipif(not _cpu_bf16_autocast(),
                                reason="this torch build has no CPU autocast for bfloat16")


@pytest.fixture
def bf16(monkeypatch):
    real = train.amp_settings

    def fake(device):
        if torch.device(device).type == "cpu":
            return True, torch.bfloat16, False
        return real(device)

    monkeypatch.setattr(train, "amp_settings", fake)
    monkeypatch.setattr(simulate, "amp_settings", fake)


@pytest.fixture(scope="module")
def split() -> S.Split:
    return make_toy_split(16, np.random.default_rng(0))


def test_training_step_under_autocast(split, bf16):
    torch.manual_seed(0)
    m = WorldModel(TINY)
    before = [p.detach().clone() for p in m.parameters()]
    batch = collate(split, np.arange(4), m.max_len)
    with torch.no_grad():
        ref, _ = m.loss(batch)
        with torch.autocast("cpu", dtype=torch.bfloat16):
            low, _ = m.loss(batch)
    assert low.dtype == torch.float32  # the cross-entropy is computed in float32
    assert float(low) == pytest.approx(float(ref), rel=0.03)
    res = train_model(m, split, split, TINY, "", FAR, log=lambda *_: None)
    assert res["amp"] == "torch.bfloat16" and res["grad_scaler"] is False
    assert res["optimizer"] == "AdamW"  # fused AdamW is requested only on CUDA
    h = res["history"][0]
    assert np.isfinite(h["train_loss"]) and np.isfinite(h["val_total"])
    assert all(torch.isfinite(p).all() for p in m.parameters())
    assert any(not torch.equal(a, b) for a, b in zip(before, m.parameters(), strict=True))


def test_cascade_attention_with_low_precision_caches(split):
    """One encode_step with bfloat16 caches under autocast equals the full float32 encode
    up to bfloat16 rounding."""
    torch.manual_seed(6)
    m = WorldModel(TINY).eval()
    c = m.cfg
    nl, nh, dh = c["n_layers"], c["n_heads"], c["d_model"] // c["n_heads"]
    one = collate(split, np.array([2]), m.max_len)
    L = 10
    states = one["states"][:, : L + 1].repeat(2, 1, 1)
    states[1, L - 1, S.STATE_INDEX["action"]] = S.ACTION["PUNT"]
    header = one["header"].repeat(2, 1)
    with torch.no_grad():
        full = m.encode(header, states)
        with torch.autocast("cpu", dtype=torch.bfloat16):
            _, kv = m.encode(header[:1], states[:1, : L - 1], return_kv=True)
            kp = torch.stack([k for k, _ in kv]).to(torch.bfloat16)
            vp = torch.stack([v for _, v in kv]).to(torch.bfloat16)
            ko = torch.zeros(nl, 2, nh, 2, dh, dtype=torch.bfloat16)
            vo = torch.zeros_like(ko)
            cell = torch.tensor([0, 1])
            for j in range(2):
                ok = torch.cat([torch.ones(2, L, dtype=torch.bool),
                                torch.arange(j + 1).unsqueeze(0).expand(2, -1) <= j], dim=1)
                bias = torch.zeros(ok.shape).masked_fill_(~ok, float("-inf"))
                h = m.encode_step(states[:, L - 1 + j], torch.full((2,), L + j), ko, vo, j + 1,
                                  cache_pos=torch.full((2,), j), prefix=(kp, vp, cell, 2, bias))
                assert h.dtype == torch.float32
                ref = full[:, L - 1 + j]
                err = (h - ref).norm(dim=-1) / ref.norm(dim=-1)
                assert (err < 0.05).all(), (j, err)


def test_sampler_runs_in_float32_inside_autocast():
    torch.manual_seed(2)
    m = WorldModel(TINY).eval()
    h = torch.randn(500, TINY["d_model"])
    a = m.sample_transition(h, 1.0, torch.Generator().manual_seed(1))
    with torch.autocast("cpu", dtype=torch.bfloat16):
        b = m.sample_transition(h, 1.0, torch.Generator().manual_seed(1))
    assert torch.equal(a, b)


def test_kv_cache_rollouts_under_autocast(split, bf16):
    torch.manual_seed(0)
    m = WorldModel(TINY)
    train_model(m, split, split, {**TINY, "epochs": 2}, "", FAR, log=lambda *_: None)
    m.eval()
    cfg = {**TINY, "kv_cache": True}
    idx = np.arange(3, len(split.states), 17)
    r = simulate.rollout(m, split, idx, S.ACTION["FG"], 16, cfg, FAR, np.random.default_rng(4))
    t = r["timing"]
    assert t["path"] == "kv_cache"
    assert t["cache_mib_per_rollout"] == pytest.approx(m.kv_bytes_per_row(2) / 2**20)
    assert r["done"].all() and np.isfinite(r["home_wp"]).all()
    assert ((r["home_wp"] >= 0) & (r["home_wp"] <= 1)).all()
    assert r["counters"]["transitions"] > 0
    st = split.states
    fourth = np.flatnonzero((st[:, S.STATE_INDEX["phase"]] == S.PHASE["PLAY"])
                            & (st[:, S.STATE_INDEX["down"]] == 4))[:3]
    assert len(fourth) == 3
    mask = np.ones((3, 3), dtype=bool)
    mask[0, 1] = False  # one masked field goal, as the support rules produce
    df = simulate.plan_fourth_downs(m, split, fourth, 8, cfg, FAR, np.random.default_rng(5),
                                    action_mask=mask)
    assert len(df) == 3 and df["done"].all() and df["best"].notna().all()
    assert np.isnan(df["wp_fg"].iloc[0]) and df["best"].iloc[0] != "FG"
```

```
%%writefile /content/wm_code/test_model.py
"""CPU tests for the world model, training loop and simulator, on synthetic toy games."""

from __future__ import annotations

import time

import numpy as np
import pytest
import schema as S
import simulate
import torch
import train
from model import END_FIELDS, WorldModel
from simulate import check_transitions, plan_fourth_downs, rollout, simulate_games
from toy import make_toy_split
from train import _full_counts, _micro_slices, collate, evaluate_nll, train_model

torch.set_num_threads(1)

TINY = {
    "d_model": 32,
    "n_layers": 2,
    "n_heads": 2,
    "d_ff": 64,
    "dec_hidden": 32,
    "dropout": 0.0,
    "max_len": 48,
    "batch_games": 8,
    "eval_batch_games": 16,
    "lr": 3e-3,
    "warmup_steps": 5,
    "epochs": 3,
    "max_steps": 150,
    "sim_batch": 512,
    "seed": 0,
}
FAR = time.time() + 3600.0


def tiny_model(seed: int = 0, **over) -> WorldModel:
    torch.manual_seed(seed)
    return WorldModel({**TINY, **over})


@pytest.fixture(scope="module")
def split() -> S.Split:
    return make_toy_split(24, np.random.default_rng(0))


@pytest.fixture(scope="module")
def trained(split) -> WorldModel:
    m = tiny_model()
    train_model(m, split, split, {**TINY, "epochs": 4}, "", FAR, log=lambda *_: None)
    return m.eval()


def fourth_downs(split: S.Split, poss: str) -> np.ndarray:
    st = split.states
    return np.flatnonzero(
        (st[:, S.STATE_INDEX["down"]] == 4)
        & (st[:, S.STATE_INDEX["phase"]] == S.PHASE["PLAY"])
        & (st[:, S.STATE_INDEX["poss"]] == S.POSS_CODE[poss])
    )


def test_toy_split_follows_the_contract(split):
    for g in range(len(split.offsets) - 1):
        a, b = split.offsets[g], split.offsets[g + 1]
        raw_next = S.apply_transition(split.raw[a : b - 1], split.targets[a : b - 1])
        assert np.array_equal(raw_next, split.raw[a + 1 : b])
        codes = S.next_state_codes(raw_next, split.targets[a : b - 1])
        assert np.array_equal(codes, split.states[a + 1 : b])
        assert split.targets[b - 1, S.TARGET_INDEX["phase"]] == S.PHASE["END"]
    assert check_transitions(split.targets)["inconsistent"] == 0


def test_loss_decreases_when_overfitting_four_games(split):
    m = tiny_model()
    batch = collate(split, np.arange(4), m.max_len)
    opt = torch.optim.AdamW(m.parameters(), lr=3e-3)
    first = None
    for _ in range(150):
        total, _ = m.loss(batch)
        first = float(total.detach()) if first is None else first
        opt.zero_grad()
        total.backward()
        opt.step()
    assert float(total.detach()) < 0.5 * first


def test_causal_mask(split):
    m = tiny_model().eval()
    batch = collate(split, np.array([0]), m.max_len)
    t = 10
    other = {k: v.clone() for k, v in batch.items()}
    other["states"][0, t + 1 :] = torch.randint(1, 3, other["states"][0, t + 1 :].shape)
    with torch.no_grad():
        h1 = m.encode(batch["header"], batch["states"])
        h2 = m.encode(other["header"], other["states"])
        l1 = m.decode_logits(h1, batch["targets"])
        l2 = m.decode_logits(h2, batch["targets"])
    assert torch.allclose(h1[:, : t + 1], h2[:, : t + 1], atol=1e-6)
    assert not torch.allclose(h1[:, t + 1 :], h2[:, t + 1 :])
    for a, b in zip(l1, l2, strict=True):
        assert torch.allclose(a[:, : t + 1], b[:, : t + 1], atol=1e-6)


def test_generation_stays_in_vocabulary():
    m = tiny_model().eval()
    gen = torch.Generator().manual_seed(0)
    h = 20.0 * torch.randn(4096, TINY["d_model"], generator=gen)
    for temp in (1.0, 3.0, 0.0):
        codes = m.sample_transition(h, temp, gen).numpy()
        end = codes[:, S.TARGET_INDEX["phase"]] == S.PHASE["END"]
        for j, (name, size) in enumerate(S.TARGET_FIELDS):
            assert codes[:, j].min() >= 0 and codes[:, j].max() < size, name
            if name not in ("down", "ydstogo"):
                assert codes[~end, j].min() >= 1, name  # padding code is never sampled
            if name in END_FIELDS:
                assert (codes[end, j] >= 1).all(), name
            else:
                assert (codes[end, j] == 0).all(), name  # END rows: same prefix as training


def test_forced_actions_are_respected(split, trained):
    idx = np.concatenate([fourth_downs(split, "HOME")[:2], fourth_downs(split, "AWAY")[:2]])
    rng = np.random.default_rng(3)
    fg = rollout(trained, split, idx, S.ACTION["FG"], 8, TINY, FAR, rng)
    assert fg["start_actions"] == {"FG": 8 * len(idx)}
    go = rollout(trained, split, idx, "GO", 64, TINY, FAR, rng)
    assert set(go["start_actions"]) <= {"RUN", "PASS"}
    assert sum(go["start_actions"].values()) == 64 * len(idx)
    real = rollout(trained, split, idx, None, 4, TINY, FAR, rng)
    expect = {}
    for a in split.states[idx, S.STATE_INDEX["action"]]:
        expect[S.ACTIONS[a]] = expect.get(S.ACTIONS[a], 0) + 4
    assert real["start_actions"] == expect


def test_rollouts_end_and_are_counted(split, trained):
    idx = np.arange(0, len(split.states), 37)
    r = rollout(trained, split, idx, None, 8, TINY, FAR, np.random.default_rng(0))
    c = r["counters"]
    assert r["done"].all()
    assert np.all((r["home_wp"] >= 0) & (r["home_wp"] <= 1))
    assert c["ended_by_end"] + c["ended_by_clock"] + c["truncated_max_steps"] == c["rollouts"]
    short = rollout(
        trained, split, idx[:3], None, 8, {**TINY, "max_steps": 1}, FAR, np.random.default_rng(0)
    )
    assert short["counters"]["truncated_max_steps"] > 0
    assert np.all(short["mean_steps"] <= 1)
    late = rollout(
        trained, split, idx[:3], None, 8, TINY, time.time() - 1, np.random.default_rng(0)
    )
    assert not late["done"].any() and np.isnan(late["home_wp"]).all()


def test_plan_fourth_downs_perspective(split, trained):
    home, away = fourth_downs(split, "HOME")[:1], fourth_downs(split, "AWAY")[:1]
    idx = np.concatenate([home, away])
    df = plan_fourth_downs(trained, split, idx, 32, TINY, FAR, np.random.default_rng(11))
    assert list(df["poss"]) == ["HOME", "AWAY"]
    assert set(df["best"]) <= {"GO", "FG", "PUNT"}
    rng = np.random.default_rng(11)  # same draws in the same order as plan_fourth_downs
    for key, act in (("go", "GO"), ("fg", S.ACTION["FG"]), ("punt", S.ACTION["PUNT"])):
        r = rollout(trained, split, idx, act, 32, TINY, FAR, rng)
        assert df[f"wp_{key}"].iloc[0] == pytest.approx(r["home_wp"][0])
        assert df[f"wp_{key}"].iloc[1] == pytest.approx(1.0 - r["home_wp"][1])
        assert df[f"se_{key}"].tolist() == pytest.approx(r["home_wp_se"].tolist())
    wps = df[["wp_go", "wp_fg", "wp_punt"]].to_numpy()
    assert np.allclose(df["margin"], np.sort(wps, 1)[:, -1] - np.sort(wps, 1)[:, -2])
    assert (df["too_close"] == (df["margin"] < 2 * df["margin_se"])).all()


def test_checkpoint_resume_is_identical(split, tmp_path):
    cfg = {**TINY, "dropout": 0.1, "epochs": 2}
    a = tiny_model()
    ra = train_model(a, split, split, cfg, str(tmp_path / "a.pt"), FAR, log=lambda *_: None)
    b = tiny_model()
    path = str(tmp_path / "b.pt")
    rb1 = train_model(
        b, split, split, {**cfg, "stop_after_epoch": 1}, path, FAR, log=lambda *_: None
    )
    assert rb1["epochs_done"] == 1 and rb1["stopped"] == "stop_after_epoch"
    c = tiny_model(seed=123)  # different init: everything must come from the checkpoint
    rc = train_model(c, split, split, cfg, path, FAR, log=lambda *_: None)
    assert rc["epochs_done"] == 2
    for (k, va), vc in zip(a.state_dict().items(), c.state_dict().values(), strict=True):
        assert torch.equal(va, vc), k
    assert [h["val_total"] for h in ra["history"]] == [h["val_total"] for h in rc["history"]]
    ck = torch.load(str(tmp_path / "a.pt"), weights_only=False)
    assert {"model", "optimizer", "epoch", "cfg", "history"} <= set(ck)


def test_deterministic_with_fixed_seed(split, trained):
    idx = fourth_downs(split, "HOME")[:2]
    r1 = rollout(trained, split, idx, "GO", 16, TINY, FAR, np.random.default_rng(5))
    r2 = rollout(trained, split, idx, "GO", 16, TINY, FAR, np.random.default_rng(5))
    assert np.array_equal(r1["home_wp"], r2["home_wp"]) and r1["counters"] == r2["counters"]
    g1 = simulate_games(trained, split, 6, TINY, FAR, np.random.default_rng(1))
    g2 = simulate_games(trained, split, 6, TINY, FAR, np.random.default_rng(1))
    assert g1["sim"] == g2["sim"] and g1["n_games"] == 6
    m1, m2 = tiny_model(), tiny_model()
    cfg = {**TINY, "epochs": 1, "dropout": 0.1}
    train_model(m1, split, split, cfg, "", FAR, log=lambda *_: None)
    train_model(m2, split, split, cfg, "", FAR, log=lambda *_: None)
    for v1, v2 in zip(m1.state_dict().values(), m2.state_dict().values(), strict=True):
        assert torch.equal(v1, v2)
    assert evaluate_nll(m1, split)["total"] == evaluate_nll(m2, split)["total"]


# ------------------------------------------------------------------ review additions
def test_decoder_uses_only_earlier_true_fields(split):
    m = tiny_model().eval()
    batch = collate(split, np.array([0, 1]), m.max_len)
    h = m.encode(batch["header"], batch["states"])
    base = m.decode_logits(h, batch["targets"])
    for j in range(len(S.TARGET_FIELDS)):
        tg = batch["targets"].clone()
        tg[..., j:] = torch.randint(1, 3, tg[..., j:].shape)  # field j and later changed
        other = m.decode_logits(h, tg)
        for k in range(j + 1):
            assert torch.equal(base[k], other[k]), (j, k)
        if j + 1 < len(S.TARGET_FIELDS):
            assert not torch.allclose(base[j + 1], other[j + 1])  # field j is used by j+1


def test_padding_and_other_games_do_not_leak(split):
    m = tiny_model().eval()
    one = collate(split, np.array([0]), m.max_len)
    two = collate(split, np.array([0, 5]), m.max_len)
    k = int(one["mask"].sum())
    with torch.no_grad():
        _, i1 = m.loss(one)
        h1 = m.encode(one["header"], one["states"])
        h2 = m.encode(two["header"], two["states"])
        padded = {kk: v.clone() for kk, v in two.items()}
        padded["states"][0, k:] = torch.randint(1, 3, padded["states"][0, k:].shape)
        h3 = m.encode(padded["header"], padded["states"])
    assert torch.allclose(h1[0, :k], h2[0, :k], atol=1e-5)
    assert torch.allclose(h1[0, :k], h3[0, :k], atol=1e-5)


def test_max_len_keeps_last_records(split):
    lens = np.diff(split.offsets)
    g = int(np.argmax(lens))
    ml = int(lens[g]) - 5
    b = collate(split, np.array([g]), ml)
    e = split.offsets[g + 1]
    assert torch.equal(b["states"][0], torch.as_tensor(split.states[e - ml : e].astype(np.int64)))
    assert bool(b["last"][0, -1])
    buf, ln = simulate._window_buffer(split, np.array([e - 1]), ml)
    assert ln[0] == ml and np.array_equal(buf[0], split.states[e - ml : e])


def test_last_record_loss_counts_only_end_fields(split):
    m = tiny_model().eval()
    batch = collate(split, np.array([0]), m.max_len)
    k = int(batch["mask"].sum())
    last_tg = batch["targets"][0, k - 1]
    assert last_tg[S.TARGET_INDEX["score_delta"]] > 0 and last_tg[S.TARGET_INDEX["elapsed"]] > 0
    masks = m.target_masks(batch["targets"], batch["mask"], batch["last"])
    for j, (name, _) in enumerate(S.TARGET_FIELDS):
        assert bool(masks[j][0, k - 1]) == (name in END_FIELDS), name
    # a non-END field on the last record never reaches the decoder input
    poisoned = {kk: v.clone() for kk, v in batch.items()}
    poisoned["targets"][0, k - 1, S.TARGET_INDEX["poss"]] = 2
    with torch.no_grad():
        a, _ = m.loss(batch)
        b, _ = m.loss(poisoned)
    assert torch.equal(a, b)


def test_score_on_end_is_applied(split, trained, monkeypatch):
    idx = fourth_downs(split, "HOME")[:2]
    end = np.zeros(len(S.TARGET_FIELDS), dtype=np.int64)
    end[S.TARGET_INDEX["phase"]] = S.PHASE["END"]
    end[S.TARGET_INDEX["score_delta"]] = S.SCORE_DELTA_CODE[3]
    end[S.TARGET_INDEX["elapsed"]] = 6

    def fake(h, *a, **k):
        return torch.as_tensor(np.tile(end, (h.shape[0], 1)))

    monkeypatch.setattr(trained, "sample_transition", fake)
    base = split.raw[idx, 1]
    for flag, add in ((True, 3), (False, 0)):
        cfg = {**TINY, "score_on_end": flag}
        r = rollout(trained, split, idx, None, 4, cfg, FAR, np.random.default_rng(0))
        assert np.allclose(r["home_wp"], S.outcome(base + add)), flag
        assert r["counters"]["ended_by_end"] == 8 and (r["mean_steps"] == 1).all()


def test_sorted_encoding_gives_identical_rollouts(split, trained):
    idx = np.arange(3, len(split.states), 23)
    out = {}
    for flag in (False, True):
        cfg = {**TINY, "sort_by_length": flag, "sim_batch": 16}
        out[flag] = rollout(trained, split, idx, None, 6, cfg, FAR, np.random.default_rng(4))
    assert np.array_equal(out[False]["home_wp"], out[True]["home_wp"])
    assert np.array_equal(out[False]["mean_steps"], out[True]["mean_steps"])
    assert out[False]["counters"] == out[True]["counters"]


def test_plan_is_chunked_by_state(split, trained):
    idx = np.concatenate([fourth_downs(split, "HOME")[:3], fourth_downs(split, "AWAY")[:3]])
    cfg = {**TINY, "sim_max_live": 16}  # 2 states per chunk at 8 rollouts
    df = plan_fourth_downs(trained, split, idx, 8, cfg, FAR, np.random.default_rng(2))
    assert df["done"].all() and len(df) == len(idx)
    rng = np.random.default_rng(2)
    for a in range(0, len(idx), 2):
        for key, act in (("go", "GO"), ("fg", S.ACTION["FG"]), ("punt", S.ACTION["PUNT"])):
            r = rollout(trained, split, idx[a : a + 2], act, 8, cfg, FAR, rng)
            home = df["poss"].iloc[a : a + 2].to_numpy() == "HOME"
            want = np.where(home, r["home_wp"], 1 - r["home_wp"])
            assert np.allclose(df[f"wp_{key}"].iloc[a : a + 2], want)
    assert sum(df.attrs["counters"]["GO"][k] for k in ("rollouts",)) == 8 * len(idx)


def test_micro_batches_give_the_same_gradient(split):
    m = tiny_model()
    batch = collate(split, np.arange(6), m.max_len)
    full, _ = m.loss(batch)
    full.backward()
    g_full = [p.grad.clone() for p in m.parameters()]
    m.zero_grad()
    denom = _full_counts(m, batch)
    tot = 0.0
    for part in _micro_slices(batch, 4):
        t, _ = m.loss(part, denom=denom)
        t.backward()
        tot += float(t.detach())
    assert tot == pytest.approx(float(full.detach()), rel=1e-5)
    for a, b in zip(g_full, (p.grad for p in m.parameters()), strict=True):
        assert torch.allclose(a, b, atol=1e-6, rtol=1e-4)


def test_out_of_memory_retries_the_same_step(split, monkeypatch):
    cfg = {**TINY, "epochs": 2}
    ref = tiny_model()
    train_model(ref, split, split, {**cfg}, "", FAR, log=lambda *_: None)
    real_loss = WorldModel.loss
    calls = {"oom": 0}

    def flaky(self, batch, denom=None):
        if self.training and batch["mask"].shape[0] > 2:  # evaluation is not affected
            calls["oom"] += 1
            raise torch.OutOfMemoryError("CUDA out of memory. Tried to allocate 2.00 GiB")
        return real_loss(self, batch, denom)

    monkeypatch.setattr(WorldModel, "loss", flaky)
    m = tiny_model()
    res = train_model(m, split, split, cfg, "", FAR, log=lambda *_: None)
    assert calls["oom"] >= 1 and res["micro_batches"] == 4 and res["epochs_done"] == 2
    assert res["history"][0]["batch_games"] == TINY["batch_games"]  # same batches
    for (k, a), b in zip(ref.state_dict().items(), m.state_dict().values(), strict=True):
        assert torch.allclose(a, b, atol=1e-4), k


def test_deadline_mid_epoch_resumes_exactly(split, tmp_path, monkeypatch):
    cfg = {**TINY, "dropout": 0.1, "epochs": 2}
    a = tiny_model()
    ra = train_model(a, split, split, cfg, "", FAR, log=lambda *_: None)
    path = str(tmp_path / "d.pt")
    real_loss = WorldModel.loss
    seen = {"n": 0}

    def counting(self, batch, denom=None):
        seen["n"] += int(self.training)
        return real_loss(self, batch, denom)

    real_time = time.time
    monkeypatch.setattr(WorldModel, "loss", counting)
    monkeypatch.setattr(train.time, "time", lambda: real_time() + (1e9 if seen["n"] >= 2 else 0))
    b = tiny_model()
    rb = train_model(b, split, split, cfg, path, FAR, log=lambda *_: None)  # stops in epoch 0
    assert rb["stopped"] == "deadline" and rb["epochs_done"] == 0
    ck = torch.load(path, weights_only=False)
    assert ck["batch_in_epoch"] == 2 and ck["step"] == 2
    monkeypatch.undo()
    c = tiny_model(seed=7)
    rc = train_model(c, split, split, cfg, path, FAR, log=lambda *_: None)
    assert rc["epochs_done"] == 2
    for (k, va), vc in zip(a.state_dict().items(), c.state_dict().values(), strict=True):
        assert torch.equal(va, vc), k
    assert [h["val_total"] for h in ra["history"]] == [h["val_total"] for h in rc["history"]]
```

```
%%writefile /content/wm_code/test_records.py
"""Tests for records.py: a hand-made game checked field by field, guards, validation."""

from __future__ import annotations

import copy
import os

import numpy as np
import polars as pl
import pytest
import records as R
import schema as S

H, A = "HOM", "AWY"


def _row(order, play_type, posteam, clock, tot=(0, 0), to=(3, 3), half="Half1", qtr=2,
         down=None, ytg=None, yl=None, desc="", nfl="", rush=0, pas=0, two=0):  # fmt: skip
    return {
        "game_id": "2023_01_AWY_HOM", "play_id": float(order * 10), "season": 2023,
        "season_type": "REG", "home_team": H, "away_team": A, "posteam": posteam,
        "order_sequence": float(order), "desc": desc, "qtr": float(qtr), "game_half": half,
        "down": down, "ydstogo": ytg, "yardline_100": yl,
        "game_seconds_remaining": float(clock), "total_home_score": float(tot[0]),
        "total_away_score": float(tot[1]), "home_timeouts_remaining": float(to[0]),
        "away_timeouts_remaining": float(to[1]), "play_type": play_type,
        "play_type_nfl": nfl, "rush": float(rush), "pass": float(pas),
        "two_point_attempt": float(two), "home_score": 15.0, "away_score": 11.0,
    }  # fmt: skip


def synthetic_game() -> pl.DataFrame:
    rows = [
        _row(0, None, None, 1900, qtr=1, desc="GAME", nfl="GAME_START"),
        _row(1, "kickoff", H, 1900, yl=35, desc="X kicks 65 yards from AWY 35"),
        _row(2, "run", H, 1896, tot=(6, 0), down=1, ytg=10, yl=75, rush=1),
        _row(3, "extra_point", H, 1890, tot=(7, 0), yl=15),
        _row(4, "kickoff", A, 1890, tot=(7, 0), yl=35, desc="Y kicks 65 yards"),
        _row(
            5, "no_play", A, 1885, tot=(7, 0), to=(3, 2), desc="Timeout #1 by AWY", nfl="TIMEOUT"
        ),  # fmt: skip
        _row(6, "pass", A, 1885, tot=(7, 0), to=(3, 2), down=1, ytg=10, yl=70, pas=1),
        _row(7, "punt", A, 1850, tot=(7, 0), to=(3, 2), down=4, ytg=3, yl=63),
        _row(
            8,
            "no_play",
            H,
            1842,
            tot=(7, 0),
            to=(3, 2),
            down=1,
            ytg=10,
            yl=80,
            pas=1,
            desc="PENALTY on AWY, Holding - No Play.",
            nfl="PENALTY",
        ),  # fmt: skip
        _row(9, "pass", H, 1838, tot=(13, 0), to=(2, 2), down=1, ytg=15, yl=85, pas=1),
        _row(10, "pass", H, 1838, tot=(15, 0), to=(2, 2), yl=2, pas=1, two=1),
        _row(11, "kickoff", A, 1838, tot=(15, 0), to=(2, 2), yl=35, desc="kicks 65"),
        _row(12, "qb_kneel", A, 1832, tot=(15, 0), to=(2, 2), down=1, ytg=10, yl=75),
        _row(13, None, None, 1800, tot=(15, 0), to=(2, 2), desc="END QUARTER 2"),
        _row(14, "kickoff", H, 1800, tot=(15, 0), half="Half2", qtr=3, yl=35, desc="kicks"),
        _row(
            15, "run", H, 1795, tot=(15, 2), half="Half2", qtr=3, down=1, ytg=10, yl=99, rush=1
        ),  # fmt: skip
        _row(16, "kickoff", A, 1790, tot=(15, 2), half="Half2", qtr=3, yl=20, desc="kicks"),
        _row(
            17, "field_goal", A, 1760, tot=(15, 5), half="Half2", qtr=4, down=4, ytg=5, yl=20
        ),  # fmt: skip
        _row(
            18,
            "kickoff",
            H,
            1760,
            tot=(15, 5),
            half="Half2",
            qtr=4,
            yl=35,
            desc="Z kicks onside 12 yards",
        ),  # fmt: skip
        _row(
            19, "run", A, 600, tot=(15, 11), half="Overtime", qtr=5, down=1, ytg=10, yl=75, rush=1
        ),  # fmt: skip
    ]
    return pl.DataFrame(rows, infer_schema_length=None)


SCHED = pl.DataFrame({"game_id": ["2023_01_AWY_HOM"], "spread_line": [-3.0]})

# phase, poss, down, ytg, yl, clock, diff, to_h, to_a, action
EXPECTED = [
    ("KICKOFF", "HOME", 0, 0, 35, 1900, 0, 3, 3, "KICKOFF"),
    ("PLAY", "HOME", 1, 10, 75, 1896, 0, 3, 3, "RUN"),
    ("TRY", "HOME", 0, 0, 15, 1890, 6, 3, 3, "XP"),
    ("KICKOFF", "AWAY", 0, 0, 35, 1890, 7, 3, 3, "KICKOFF"),
    ("PLAY", "AWAY", 1, 10, 70, 1885, 7, 3, 2, "PASS"),
    ("PLAY", "AWAY", 4, 3, 63, 1850, 7, 3, 2, "PUNT"),
    ("PLAY", "HOME", 1, 10, 80, 1842, 7, 3, 2, "PASS"),
    ("PLAY", "HOME", 1, 15, 85, 1838, 7, 3, 2, "PASS"),
    ("TRY", "HOME", 0, 0, 2, 1838, 13, 2, 2, "TWO_PT"),
    ("KICKOFF", "AWAY", 0, 0, 35, 1838, 15, 2, 2, "KICKOFF"),
    ("PLAY", "AWAY", 1, 10, 75, 1832, 15, 2, 2, "KNEEL"),
    ("KICKOFF", "HOME", 0, 0, 35, 1800, 15, 3, 3, "KICKOFF"),
    ("PLAY", "HOME", 1, 10, 99, 1795, 15, 3, 3, "RUN"),
    ("KICKOFF", "AWAY", 0, 0, 20, 1790, 13, 3, 3, "KICKOFF"),
    ("PLAY", "AWAY", 4, 5, 20, 1760, 13, 3, 3, "FG"),
    ("KICKOFF", "HOME", 0, 0, 35, 1760, 10, 3, 3, "ONSIDE"),
]
# score_delta, elapsed, to_used_home, to_used_away (target of record i)
EXPECTED_T = [
    (0, 4, 0, 0), (6, 6, 0, 0), (1, 0, 0, 0), (0, 5, 0, 1), (0, 35, 0, 0), (0, 8, 0, 0),
    (0, 4, 0, 0), (6, 0, 1, 0), (2, 0, 0, 0), (0, 6, 0, 0), (0, 32, 0, 0), (0, 5, 0, 0),
    (-2, 5, 0, 0), (0, 30, 0, 0), (-3, 0, 0, 0),
]  # fmt: skip


@pytest.fixture(scope="module")
def built():
    return R.build_split(synthetic_game(), SCHED, [2023])


def test_states_by_hand(built):
    split, report = built
    assert len(split.game_ids) == 1 and report["rows_kept"] == len(EXPECTED)
    for i, e in enumerate(EXPECTED):
        d = R.decode_record(split, i)
        got = (d["phase"], d["poss"], d["down"], d["ydstogo"], d["yardline_100"],
               d["clock_seconds"], d["score_diff_home"], d["timeouts_home"],
               d["timeouts_away"], d["action"])  # fmt: skip
        assert got == e, (i, got, e)


def test_targets_by_hand(built):
    split, _ = built
    for i, (dlt, el, uh, ua) in enumerate(EXPECTED_T):
        t = R.decode_record(split, i)["target"]
        nxt = EXPECTED[i + 1]
        assert (t["phase"], t["poss"], t["down"], t["ydstogo"], t["yardline"], t["action"]) == (
            nxt[0],
            nxt[1],
            nxt[2],
            nxt[3],
            nxt[4],
            nxt[9],
        ), i
        assert (t["score_delta"], t["elapsed"], t["to_used_home"], t["to_used_away"]) == (
            dlt,
            el,
            uh,
            ua,
        ), i
    last = split.targets[-1]
    assert last[0] == S.PHASE["END"]
    # the final record (onside kick at 1760 s, diff 10) scores nothing; elapsed clips at 60
    assert last[S.TARGET_INDEX["score_delta"]] == S.SCORE_DELTA_CODE[0]
    assert last[S.TARGET_INDEX["elapsed"]] == S.elapsed_code(1760)
    assert not last[list(R.END_ZERO_FIELDS)].any()


def test_header_final_and_report(built):
    split, report = built
    assert split.header[0, 0] == S.spread_code(-3.0) == 19
    assert split.header[0, 1] == S.POSS_CODE["HOME"]
    assert split.final_diff[0] == 10  # regulation 15-5; the OT score 15-11 is cut
    assert report["dropped"]["timeout_only"] == 1
    assert report["dropped"]["overtime_or_null_qtr"] == 1
    assert report["soft"]["halftime_crossings"] == 1
    assert report["overtime_games_cut"] == 1
    np.testing.assert_array_equal(R.fourth_down_index(split), [5, 14])


def test_validate_passes_and_bookkeeping(built):
    split, _ = built
    v = R.validate_split(split, min_transition_match=1.0)
    assert v["info_mismatched_pairs"] == 0


def test_first_half_snap_at_1800_repaired():
    df = synthetic_game().with_columns(
        pl.when(pl.col("order_sequence") == 12).then(1800.0)
        .otherwise(pl.col("game_seconds_remaining")).alias("game_seconds_remaining")
    )  # fmt: skip
    split, report = R.build_split(df, SCHED, [2023])
    assert report["repairs"]["first_half_clock_1800_set_to_1801"] == 1
    assert split.raw[10, 0] == 1801
    assert R.validate_split(split, 1.0)["info_mismatched_pairs"] == 0


def test_points_on_final_record_are_in_end_target():
    # walk-off: HOME kicks a FG with 3 s left in regulation, then overtime rows follow
    walk = _row(
        18.5, "field_goal", H, 3, tot=(18, 5), half="Half2", qtr=4, down=4, ytg=2, yl=10
    )  # fmt: skip
    df = pl.concat([synthetic_game(), pl.DataFrame([walk], schema=synthetic_game().schema)])
    split, report = R.build_split(df, SCHED, [2023])
    assert split.final_diff[0] == 13 and report["soft"]["games_scoring_on_last_record"] == 1
    d = R.decode_record(split, len(split.states) - 1)
    assert (d["action"], d["score_diff_home"], d["clock_seconds"]) == ("FG", 10, 3)
    assert d["target"] == {"phase": "END", "score_delta": 3, "elapsed": 3}
    v = R.validate_split(split, 0.9)  # the 1760 s -> 3 s jump is one clock mismatch
    assert v["final_diff_ne_last_score_plus_end_delta"] == 0
    assert v["info_games_scoring_on_last_record"] == 1


def test_validate_catches_final_diff_mismatch(built):
    split = copy.deepcopy(built[0])
    split.final_diff[0] += 3  # END target no longer explains the final score
    with pytest.raises(AssertionError, match="final_diff_ne_last_score_plus_end_delta"):
        R.validate_split(split, 1.0)


def test_noplay_punt_and_fg_keep_their_action():
    wiped = {
        7: ("4-J.Doe punts 45 yards to HOM 20. PENALTY on HOM, Holding, enforced - No Play.", 0),
        17: ("1-K.Kik 38 yard field goal is GOOD, NULLIFIED by Penalty - No Play.", 0),
    }  # fmt: skip
    df = synthetic_game().with_columns(
        pl.when(pl.col("order_sequence").is_in([float(k) for k in wiped]))
        .then(pl.lit("no_play")).otherwise(pl.col("play_type")).alias("play_type"),
        pl.col("order_sequence").replace_strict(
            {float(k): v[0] for k, v in wiped.items()},
            default=pl.col("desc"),
            return_dtype=pl.String,
        ).alias("desc"),
    )  # fmt: skip
    split, _ = R.build_split(df, SCHED, [2023])
    acts = [R.decode_record(split, i)["action"] for i in (5, 14)]
    assert acts == ["PUNT", "FG"]
    assert R.decode_record(split, 6)["action"] == "PASS"  # penalty row with pass flag


@pytest.mark.parametrize("bad", [[2026], [2009], [2024, 2026], []])
def test_season_guard(bad):
    with pytest.raises(ValueError):
        R.load_pbp(bad)
    with pytest.raises(ValueError):
        R.load_schedules(bad)
    with pytest.raises(ValueError):
        R.build_split(synthetic_game(), SCHED, bad)


@pytest.mark.parametrize(
    "corrupt",
    [
        lambda s: s.states.__setitem__((3, S.STATE_INDEX["down"]), 2),  # down on a kickoff
        lambda s: s.states.__setitem__((1, S.STATE_INDEX["poss"]), 0),  # pad code
        lambda s: s.targets.__setitem__((-1, 0), S.PHASE["PLAY"]),  # no END
        lambda s: s.targets.__setitem__((4, S.TARGET_INDEX["action"]), 99),  # out of vocab
        lambda s: s.offsets.__setitem__(-1, 3),  # offsets do not cover all records
        lambda s: s.raw.__setitem__((5, 1), 40),  # score bookkeeping broken
    ],
)
def test_validate_catches_corruption(built, corrupt):
    split = copy.deepcopy(built[0])
    corrupt(split)
    with pytest.raises(AssertionError):
        R.validate_split(split, min_transition_match=0.99)


def test_disjoint():
    s, _ = R.build_split(synthetic_game(), SCHED, [2023])
    with pytest.raises(AssertionError):
        R.validate_disjoint({"train": s, "val": s})


@pytest.mark.skipif(os.environ.get("WM_REAL_DATA") != "1", reason="set WM_REAL_DATA=1")
def test_real_2023():
    split, report = R.build_split(R.load_pbp([2023]), R.load_schedules([2023]), [2023])
    v = R.validate_split(split)
    assert report["games"] > 250 and v["info_apply_transition_match_fraction"] >= 0.99
```

```
%%writefile /content/wm_code/test_run.py
"""Tests for run.py with stub records/model/train/simulate modules (fast, CPU, no network)."""

from __future__ import annotations

import sys
import time
import types

import numpy as np
import pandas as pd
import pytest
import run as R
import schema as S


# ---------------------------------------------------------------- stubs
class _StubModel:
    def __init__(self, cfg=None):
        self.cfg = dict(cfg or {})

    def count_params(self):
        return 1234

    def eval(self):
        return self

    def to(self, device):
        return self


def _stubs(calls: dict, fail_train_once: bool = False, build_seed: int = 0,
           interrupt_wp_call: int | None = None) -> dict[str, types.ModuleType]:
    rec = types.ModuleType("records")

    def load_pbp(seasons):
        assert max(seasons) <= 2025
        calls.setdefault("load_pbp", []).append(list(seasons))
        return pd.DataFrame({"game_id": ["x"], "play_id": [1], "home_wp": [0.5],
                             "vegas_home_wp": [0.5]})

    def load_schedules(seasons):
        return pd.DataFrame({"game_id": ["x"]})

    def build_split(pbp, schedules, seasons):
        calls.setdefault("build_split", []).append(list(seasons))
        sp = R.synthetic_split(n_games=4, seed=int(seasons[0]) + build_seed)
        sp.game_ids = [f"{seasons[0]}_{g}" for g in sp.game_ids]
        return sp, {"games": 4}

    rec.load_pbp, rec.load_schedules, rec.build_split = load_pbp, load_schedules, build_split
    rec.validate_split = lambda split: {"ok": True}
    rec.fourth_down_index = lambda split: np.flatnonzero(
        (split.states[:, S.STATE_INDEX["phase"]] == S.PHASE["PLAY"])
        & (split.states[:, S.STATE_INDEX["down"]] == 4))

    model = types.ModuleType("model")
    model.WorldModel = _StubModel

    train = types.ModuleType("train")

    def train_model(m, tr, va, cfg, ckpt_path, deadline_ts, log):
        calls["train"] = calls.get("train", 0) + 1
        import torch

        resumed = R.Path(ckpt_path).exists()
        torch.save({"epoch": 1, "best_val": 1.1}, ckpt_path)
        if fail_train_once and calls["train"] == 1:
            raise RuntimeError("disconnect")
        log("epoch 0: train 1.0 val 1.1")
        return {"resumed": resumed, "history": [{"epoch": 0, "train_loss": 1.0,
                                                 "val_total": 1.1}],
                "train_records_per_s": 1e6}

    train.train_model = train_model
    train.load_checkpoint = lambda path, cfg_override=None, device=None: (_StubModel(), {})
    train.evaluate_nll = lambda m, split, cfg=None: {
        "total": 9.0, "fields": {n: 0.9 for n, _ in S.TARGET_FIELDS}}

    sim = types.ModuleType("simulate")

    def wp_by_simulation(m, split, idx, n, cfg, deadline_ts, rng):
        calls["wp_calls"] = calls.get("wp_calls", 0) + 1
        if interrupt_wp_call is not None and calls["wp_calls"] == interrupt_wp_call:
            raise KeyboardInterrupt
        calls["wp"] = calls.get("wp", 0) + len(idx)

        g = np.searchsorted(split.offsets, idx, side="right") - 1
        return pd.DataFrame({"idx": idx, "home_wp_sim": rng.uniform(0.1, 0.9, len(idx)),
                             "home_result": S.outcome(split.final_diff[g]),
                             "done": np.ones(len(idx), dtype=bool),
                             "mean_steps": np.full(len(idx), 10.0)})

    def plan_fourth_downs(m, split, idx, n, cfg, deadline_ts, rng, action_mask=None):
        calls["plan"] = calls.get("plan", 0) + len(idx)
        k = len(idx)
        return pd.DataFrame({"idx": idx, "wp_go": np.full(k, 0.6), "wp_fg": np.full(k, 0.5),
                             "wp_punt": np.full(k, 0.55), "se_go": np.full(k, 0.01),
                             "se_fg": np.full(k, 0.01), "se_punt": np.full(k, 0.01),
                             "best": ["GO"] * k, "margin": np.full(k, 0.05),
                             "too_close": np.zeros(k, dtype=bool),
                             "real_action": ["PUNT"] * k, "real_choice": ["PUNT"] * k,
                             "done": np.ones(k, dtype=bool)})

    sim.wp_by_simulation, sim.plan_fourth_downs = wp_by_simulation, plan_fourth_downs
    sim.simulate_games = lambda m, split, n, cfg, d, rng: {"n_games": n, "sim": {"x": 1.0},
                                                             "real": {"x": 1.1}}
    return {"records": rec, "model": model, "train": train, "simulate": sim}


@pytest.fixture
def stubbed(monkeypatch):
    calls: dict = {}
    for name, mod in _stubs(calls).items():
        monkeypatch.setitem(sys.modules, name, mod)
    return calls


def _tiny_preset(**kw):
    p = R.get_preset("smoke")
    p["seasons"] = {"train": [2019], "val": [2020], "test": [2021]}
    p.update(kw)
    return p


# ---------------------------------------------------------------- tests
def test_pick_preset():
    assert R.pick_preset("Tesla T4", 15.0) == "t4"
    assert R.pick_preset("NVIDIA L4", 22.5) == "l4"
    assert R.pick_preset("NVIDIA A100-SXM4-40GB", 40.0) == "a100"
    assert R.pick_preset("NVIDIA H100 80GB HBM3", 80.0) == "a100"
    assert R.pick_preset("Some New GPU", 48.0) == "a100"
    assert R.pick_preset("Some Old GPU", 8.0) == "t4"


def test_pick_preset_cpu(monkeypatch):
    import torch

    monkeypatch.setattr(torch.cuda, "is_available", lambda: False)
    assert R.pick_preset() == "smoke"


def test_presets_use_full_seasons_and_fit_budget():
    for name in ("t4", "l4", "a100"):
        p = R.PRESETS[name]
        assert p["seasons"]["train"] == list(range(2010, 2022))
        assert p["seasons"]["val"] == [2022, 2023]
        assert p["seasons"]["test"] == [2024, 2025]
        assert p["budget_min"] <= 55
        assert p["reserve_package_s"] > 0
    for p in R.PRESETS.values():
        assert set(p["stage_min"]) == set(R.STAGES)
        for s in p["seasons"].values():
            assert max(s) <= R.MAX_SEASON


def test_2026_guard(tmp_path, stubbed):
    with pytest.raises(AssertionError):
        R.check_seasons([2025, 2026])
    p = _tiny_preset()
    p["seasons"]["test"] = [2026]
    run = R.Run(p, tmp_path, quiet=True)
    out = R.run_stage(run, "data")
    assert out["status"] == "error" and "2025" in out["error"]
    assert all(max(s) <= 2025 for s in stubbed.get("load_pbp", []))


def test_split_roundtrip(tmp_path):
    sp = R.synthetic_split(3, seed=2)
    R.save_split(tmp_path / "a.npz", sp, w=np.arange(R.n_records(sp), dtype=float))
    back, extra = R.load_split(tmp_path / "a.npz")
    assert back.game_ids == sp.game_ids
    assert np.array_equal(back.states, sp.states)
    assert len(extra["w"]) == R.n_records(sp)
    both = R.concat_splits([sp, back])
    assert both.offsets[-1] == 2 * R.n_records(sp)
    sub = R.subset_split(both, [4])
    assert np.array_equal(sub.states, sp.states[sp.offsets[1]:sp.offsets[2]])


def test_frequency_baseline_beats_uniform():
    sp = R.synthetic_split(6, seed=3)
    tables = R.fit_frequency_baseline(sp)
    nll = R.baseline_nll(tables, sp)
    uniform = sum(np.log(size) for _, size in S.TARGET_FIELDS)
    assert np.isfinite(nll["total_per_transition"])
    assert nll["total_per_transition"] < uniform


def test_hypothetical_edits_codes():
    sp = R.synthetic_split(4, seed=4)
    hyp, doc = R.build_hypothetical(sp)
    j = R.n_records(hyp) - 1
    st = hyp.states[j]
    assert st[S.STATE_INDEX["down"]] == 4 and st[S.STATE_INDEX["ydstogo"]] == 2
    assert st[S.STATE_INDEX["yardline"]] == 60
    assert hyp.raw[j, 0] == 240 and st[S.STATE_INDEX["clock"]] == S.clock_code(240)
    sign = 1 if st[S.STATE_INDEX["poss"]] == S.POSS_CODE["HOME"] else -1
    assert hyp.raw[j, 1] == 4 * sign
    assert st[S.STATE_INDEX["score"]] == S.score_code(4 * sign)
    assert doc["base_play_id"] in sp.play_ids


def test_bootstrap_interval_contains_mean():
    rng = np.random.default_rng(0)
    v = rng.normal(size=200)
    b = R.game_bootstrap(v, np.repeat(np.arange(20), 10), 300, rng)
    assert b["lo95"] <= b["mean"] <= b["hi95"]


def test_full_run_resume_and_test_once(tmp_path, stubbed):
    run = R.Run(_tiny_preset(), tmp_path, quiet=True)
    out = R.run_all(run)
    for s in ("data", "train", "eval", "test", "plan", "package"):
        assert out[s]["status"] == "done", (s, out[s])
    h1 = run.full_config_hash()
    res = R.read_json(tmp_path / "results.json")
    assert res["stages"]["test"]["status"] == "done"
    assert res["prereg"]["config_hash"] == h1
    assert [p["config_hash"] for p in res["preregistrations"]] == [h1]
    assert res["preregistrations"][0]["test_scored"] and res["preregistrations"][0]["this_run"]
    assert res["test_scorings"]["n_configurations_scored"] == 1
    assert (tmp_path / "world_model_results.zip").exists()
    assert "Preset smoke" in R.summary(run)
    # Re-running skips every finished stage.
    before = dict(stubbed)
    run2 = R.Run(_tiny_preset(), tmp_path, quiet=True)
    R.run_all(run2)
    assert stubbed["train"] == before["train"] and stubbed["wp"] == before["wp"]
    assert stubbed["build_split"] == before["build_split"]
    # A second scoring of the SAME configuration is refused without the flag ...
    out3 = R.run_stage(R.Run(_tiny_preset(), tmp_path, quiet=True, force=("test",)), "test")
    assert out3["status"] == "error" and "refusing" in out3["error"]
    # ... and allowed, and recorded, with it; the preregistration is never rewritten.
    first = (tmp_path / "prereg" / f"{h1}.json").read_text()
    out4 = R.run_stage(R.Run(_tiny_preset(), tmp_path, quiet=True, allow_rescore=True,
                             force=("test",)), "test")
    assert out4["status"] == "done" and out4["rescore"] is True
    assert (tmp_path / "prereg" / f"{h1}.json").read_text() == first
    attempts = R.read_json(tmp_path / "prereg" / f"{h1}.attempts.json")
    assert attempts[-1] == {**attempts[-1], "stage": "test", "rescore": True}
    # A different configuration (here: a new test plan) gets its own preregistration and
    # its own scoring; both are listed in results.json.
    p = _tiny_preset()
    p["test"]["wp_states"] = 10
    run5 = R.Run(p, tmp_path, quiet=True)
    out5 = R.run_stage(run5, "test")
    assert out5["status"] == "done", out5
    h2 = run5.full_config_hash()
    assert h2 != h1 and (tmp_path / "prereg" / f"{h2}.json").exists()
    assert (tmp_path / "prereg" / f"{h1}.json").read_text() == first
    R.run_stage(run5, "package")
    res = R.read_json(tmp_path / "results.json")
    rows = {r["config_hash"]: r for r in res["preregistrations"]}
    assert set(rows) == {h1, h2} and rows[h2]["this_run"] and not rows[h1]["this_run"]
    assert rows[h1]["rescores"] == 1 and all(r["test_scored"] for r in rows.values())
    assert res["test_scorings"]["n_configurations_scored"] == 2
    assert "scored by 2 configuration(s)" in R.summary(run5)
    with R.zipfile.ZipFile(tmp_path / "world_model_results.zip") as z:
        assert f"prereg/{h1}.json" in z.namelist() and f"prereg/{h2}.json" in z.namelist()


def test_prereg_written_before_test_data_is_read(tmp_path, stubbed, monkeypatch):
    run = R.Run(_tiny_preset(), tmp_path, quiet=True)
    for s in ("data", "train"):
        R.run_stage(run, s)
    seen = []
    orig = R.Run.split

    def spy(self, name):
        if name == "test":
            seen.append(R.prereg_path(self).exists())
        return orig(self, name)

    monkeypatch.setattr(R.Run, "split", spy)
    # The plan stage reads the test seasons too: with the test stage not run, it writes
    # the preregistration itself before reading them.
    assert R.run_stage(run, "plan")["status"] == "done"
    assert seen and all(seen)
    att = R.read_json(R.prereg_path(run).with_name(f"{run.full_config_hash()}.attempts.json"))
    assert [a["stage"] for a in att] == ["plan"]
    R.run_stage(run, "test")
    assert len(seen) >= 2 and all(seen)


def test_first_notebook_prereg_does_not_block_a_new_model(tmp_path, stubbed):
    """The first Colab run left a single prereg.json and outputs/test_scored.json for its
    model; a new model must still be able to score the test seasons, and both show up."""
    run = R.Run(_tiny_preset(), tmp_path, quiet=True)
    R.write_json(tmp_path / "prereg.json", {
        "config_hash": "4bf68c37adae789a", "test_hash": "old", "written_at": 1.0,
        "attempts": [{"started_at": 1.0, "rescore": False}], "previous": None})
    R.write_json(tmp_path / "outputs" / "test_scored.json",
                 {"config_hash": "4bf68c37adae789a", "test_hash": "old", "at": 2.0})
    out = R.run_all(run)
    assert out["test"]["status"] == "done", out["test"]
    res = R.read_json(tmp_path / "results.json")
    rows = {r["config_hash"]: r for r in res["preregistrations"]}
    assert set(rows) == {"4bf68c37adae789a", run.full_config_hash()}
    assert rows["4bf68c37adae789a"]["test_scored"] and rows["4bf68c37adae789a"]["migrated_from"]
    assert res["test_scorings"]["n_configurations_scored"] == 2
    assert (tmp_path / "prereg.json").exists()  # the legacy file is left in place


def test_outputs_of_another_configuration_are_not_reported(tmp_path, stubbed):
    run = R.Run(_tiny_preset(), tmp_path, quiet=True)
    R.write_json(run.output_path("eval"), {"status": "done", "hash": "someoldhash",
                                           "wp": {"metrics": {"x": 1}}})
    assert run.status("eval") == "other config"
    R.run_stage(run, "package")
    res = R.read_json(tmp_path / "results.json")
    assert res["stages"]["eval"]["status"] == "missing"
    assert "another configuration" in res["stages"]["eval"]["reason"]


def test_saved_preset_is_reused_only_while_its_definition_matches(tmp_path, stubbed, capsys):
    run = R.prepare("smoke", tmp_path, benchmark=False)
    saved = R.read_json(tmp_path / "preset.json")
    assert saved["definition"] == R.preset_definition("smoke")
    R.write_json(tmp_path / "ledger.json", {"consumed_s": 2400.0})
    R.write_json(tmp_path / "results.json", {"from": "first run"})
    (tmp_path / "figures" / "fourth_down.png").write_bytes(b"png")
    run = R.prepare("smoke", tmp_path, benchmark=False)  # same definition: reused
    assert run.consumed_before == 2400.0 and "Reusing the saved preset" in capsys.readouterr().out
    # The first notebook's preset.json had no definition: picked afresh, old run archived.
    del saved["definition"]
    saved["plan"]["n_fourth"] = 3
    R.write_json(tmp_path / "preset.json", saved)
    run = R.prepare("smoke", tmp_path, benchmark=False)
    log = (tmp_path / "run.log").read_text()
    assert "Picking the preset afresh" in log and "no definition version" in log
    assert run.preset["plan"]["n_fourth"] == R.PRESETS["smoke"]["plan"]["n_fourth"]
    assert run.consumed_before == 0.0  # a fresh pick starts a fresh time budget
    arch = list((tmp_path / "archive").glob("run_*"))
    assert len(arch) == 1
    assert R.read_json(arch[0] / "results.json") == {"from": "first run"}
    assert R.read_json(arch[0] / "ledger.json")["consumed_s"] == 2400.0
    assert (arch[0] / "figures" / "fourth_down.png").exists()
    assert not (tmp_path / "results.json").exists()
    assert not (tmp_path / "figures" / "fourth_down.png").exists()
    # A changed PRESETS entry changes the definition hash: picked afresh again.
    old = R.read_json(tmp_path / "preset.json")["definition"]
    R.PRESETS["smoke"]["plan"]["n_fourth"] += 1
    try:
        assert R.preset_definition("smoke") != old
        run = R.prepare("smoke", tmp_path, benchmark=False)
        assert run.preset["plan"]["n_fourth"] == R.PRESETS["smoke"]["plan"]["n_fourth"]
        assert "differs from the code" in (tmp_path / "run.log").read_text()
    finally:
        R.PRESETS["smoke"]["plan"]["n_fourth"] -= 1


def test_an_earlier_ledger_without_preset_json_is_archived(tmp_path, stubbed):
    R.write_json(tmp_path / "ledger.json", {"consumed_s": 2400.0})
    run = R.prepare("smoke", tmp_path, benchmark=False)
    assert run.consumed_before == 0.0
    (arch,) = list((tmp_path / "archive").glob("run_*"))
    assert R.read_json(arch / "ledger.json")["consumed_s"] == 2400.0


def test_benchmark_scales_down_only_when_much_slower_and_never_below_floors():
    p = R.get_preset("t4")
    a = p["assumed"]
    kept, changes, decision, _, ro = R.apply_benchmark(p, a["train_rec_per_s"],
                                                      0.6 * a["rollout_units_per_s"])
    assert ro == pytest.approx(0.6) and changes == [] and decision.startswith("kept")
    assert kept["plan"] == p["plan"] and kept["eval"] == p["eval"]
    slow, changes, decision, _, _ = R.apply_benchmark(p, a["train_rec_per_s"],
                                                      0.1 * a["rollout_units_per_s"])
    assert decision.startswith("scaled down") and changes
    assert slow["plan"]["n_fourth"] == R.BENCH_FLOORS["n_fourth"]  # 300 x 0.1 = 30 < floor
    assert slow["eval"]["wp_states"] == R.BENCH_FLOORS["wp_states"]
    assert slow["plan"]["n_rollouts"] == p["plan"]["n_rollouts"]  # precision per state kept
    mid, _, _, _, _ = R.apply_benchmark(p, a["train_rec_per_s"],
                                        0.45 * a["rollout_units_per_s"])
    assert mid["plan"]["n_fourth"] == max(int(300 * 0.45), 200)
    # A preset smaller than the floors is never scaled up, and never below its own size.
    s = R.get_preset("smoke")
    small, _, _, _, _ = R.apply_benchmark(s, 1.0, 1.0)
    assert small["plan"] == s["plan"] and small["eval"] == s["eval"]
    # Slow training downgrades the model one size (not below t4).
    l4 = R.get_preset("l4")
    d, changes, decision, _, _ = R.apply_benchmark(l4, 0.2 * l4["assumed"]["train_rec_per_s"],
                                                   l4["assumed"]["rollout_units_per_s"])
    assert d["model"] == R.PRESETS["t4"]["model"] and "downgraded" in decision


def test_micro_benchmark_times_the_kv_cache_path_real_modules(monkeypatch):
    """Real model, train and simulate on the CPU: the benchmark measures the cached rollout
    path at one plan-stage call's batch and records its decision."""
    pytest.importorskip("torch")
    import model as real_model
    import simulate as real_sim
    import train as real_train

    for name, mod in (("model", real_model), ("train", real_train), ("simulate", real_sim)):
        monkeypatch.setitem(sys.modules, name, mod)
    p = R.get_preset("smoke")
    b = R.micro_benchmark(p, seconds=20.0)
    rb = b["rollout_batch"]
    assert rb["path"] == "kv_cache" and rb["timing"]["path"] == "kv_cache"
    assert rb["rows"] == rb["states"] * p["plan"]["n_rollouts"] and rb["states"] >= 4
    assert b["rollout_steps_per_s"] > 0 and b["train_records_per_s"] > 0
    assert b["decision"].startswith(("kept", "scaled down"))
    assert b["adjusted_preset"]["plan"]["n_fourth"] == p["plan"]["n_fourth"]  # below floors


def test_training_resumes_after_disconnect(tmp_path, monkeypatch):
    calls: dict = {}
    for name, mod in _stubs(calls, fail_train_once=True).items():
        monkeypatch.setitem(sys.modules, name, mod)
    run = R.Run(_tiny_preset(), tmp_path, quiet=True)
    R.run_stage(run, "data")
    assert R.run_stage(run, "train")["status"] == "error"
    assert not run.is_done("train")
    out = R.run_stage(R.Run(_tiny_preset(), tmp_path, quiet=True), "train")
    assert out["status"] == "done" and out["history"]["resumed"] is True


def test_deadline_guard_still_writes_results(tmp_path, stubbed):
    p = _tiny_preset(budget_min=0.01, reserve_package_s=1)
    run = R.Run(p, tmp_path, quiet=True)
    time.sleep(0.05)
    out = R.run_all(run)
    assert out["data"]["status"] == "skipped" and "time budget" in out["data"]["reason"]
    assert out["train"]["status"] == "skipped"
    res = R.read_json(tmp_path / "results.json")
    assert res["stages"]["eval"]["status"] == "skipped"
    assert out["package"]["status"] == "done"


def test_ledger_shrinks_budget_on_resume(tmp_path):
    p = _tiny_preset()
    R.write_json(tmp_path / "ledger.json", {"consumed_s": 600.0})
    run = R.Run(p, tmp_path, quiet=True)
    assert run.remaining() < 60 * p["budget_min"] - 590


def test_resume_survives_source_change(tmp_path, monkeypatch):
    calls: dict = {}
    for name, mod in _stubs(calls).items():
        monkeypatch.setitem(sys.modules, name, mod)
    run = R.Run(_tiny_preset(), tmp_path, quiet=True)
    R.run_stage(run, "data")
    R.run_stage(run, "train")
    ckpt = run.ckpt_path()
    # Edited model/train/records code: the train stage stays done, same checkpoint.
    orig = R._source_hash
    monkeypatch.setattr(R, "_source_hash", lambda mods: "edited" + orig(mods))
    run2 = R.Run(_tiny_preset(), tmp_path, quiet=True)
    assert run2.is_done("train") and run2.ckpt_path() == ckpt
    assert not run2.is_done("data")  # records changed: rebuild in memory and compare
    out = R.run_stage(run2, "data")
    assert set(out["seasons_rebuilt"].values()) == {"identical, kept"}
    assert run2.is_done("train") and run2.ckpt_path() == ckpt
    # Code that really changes the arrays replaces them, and training starts again.
    for name, mod in _stubs(calls, build_seed=99).items():
        monkeypatch.setitem(sys.modules, name, mod)
    monkeypatch.setattr(R, "_source_hash", lambda mods: "edited again" + orig(mods))
    run3 = R.Run(_tiny_preset(), tmp_path, quiet=True)
    out = R.run_stage(run3, "data")
    assert "replaced" in out["seasons_rebuilt"].values()
    assert not run3.is_done("train") and run3.ckpt_path() != ckpt


def test_stop_after_train(tmp_path, stubbed):
    run = R.Run(_tiny_preset(), tmp_path, quiet=True)
    out = R.run_all(run, stop_after="train")
    assert "eval" not in out and out["package"]["status"] == "done"
    res = R.read_json(tmp_path / "results.json")
    assert res["stages"]["train"]["status"] == "done"
    assert res["stages"]["eval"]["status"] == "missing"
    assert "train    done" in R.status_table(run)
    with pytest.raises(ValueError):
        R._beyond("eval", "nonsense")


def test_interrupted_eval_resumes_from_saved_chunks(tmp_path, monkeypatch, capsys):
    calls: dict = {}
    for name, mod in _stubs(calls, interrupt_wp_call=2).items():
        monkeypatch.setitem(sys.modules, name, mod)
    p = _tiny_preset()
    p["eval"].update(wp_states=6, chunk_states=2, sim_games=0)
    run = R.Run(p, tmp_path, quiet=True)
    out = R.run_all(run, stop_after="eval")
    assert out["stopped_at"] == "eval"
    assert "Stopped safely at eval" in capsys.readouterr().out
    res = R.read_json(tmp_path / "results.json")
    assert res["stages"]["eval"]["status"] == "interrupted"
    assert calls["wp"] == 2  # one chunk finished and was saved
    calls2: dict = {}
    for name, mod in _stubs(calls2).items():
        monkeypatch.setitem(sys.modules, name, mod)
    out = R.run_stage(R.Run(p, tmp_path, quiet=True), "eval")
    assert out["status"] == "done"
    assert calls2["wp"] == 4  # only the two missing chunks were simulated
    assert out["wp"]["progress"]["reused_states"] == 2 and out["wp"]["n_states"] == 6


def test_keyboard_interrupt_mid_training_resumes_real_modules(tmp_path, monkeypatch, capsys):
    """Real model.py and train.py: stop after epoch 0, then resume at epoch 1."""
    pytest.importorskip("torch")
    import model as real_model
    import train as real_train

    calls: dict = {}
    stubs = _stubs(calls)
    monkeypatch.setitem(sys.modules, "records", stubs["records"])
    monkeypatch.setitem(sys.modules, "model", real_model)
    monkeypatch.setitem(sys.modules, "train", real_train)
    p = _tiny_preset()
    p["model"] = {"d_model": 16, "n_layers": 1, "n_heads": 2, "d_ff": 32, "dec_hidden": 16,
                  "max_len": 256, "dropout": 0.0}
    p["train"].update(epochs=2, patience=5, batch_games=4, warmup_steps=2)
    orig_log = R.Run.log

    def log_then_stop(self, *msg):
        orig_log(self, *msg)
        if any("epoch 0:" in str(m) for m in msg):
            raise KeyboardInterrupt

    monkeypatch.setattr(R.Run, "log", log_then_stop)
    run = R.Run(p, tmp_path, quiet=True)
    out = R.run_all(run, stop_after="train")
    assert out["stopped_at"] == "train"
    assert "Stopped safely at train" in capsys.readouterr().out
    assert R.read_json(tmp_path / "results.json")["stages"]["train"]["status"] == "interrupted"
    assert run.ckpt_path().exists() and not run.is_done("train")
    assert R.checkpoint_info(run)["epochs_done"] == 1
    monkeypatch.setattr(R.Run, "log", orig_log)
    run2 = R.Run(p, tmp_path, quiet=True)
    out = R.run_stage(run2, "train")
    assert out["status"] == "done"
    hist = out["history"]
    assert hist["epochs_done"] == 2
    assert [h["epoch"] for h in hist["history"]] == [0, 1]
    assert R.checkpoint_info(run2)["epochs_done"] == 2


def test_mismatched_checkpoint_starts_fresh(tmp_path):
    torch = pytest.importorskip("torch")
    import model as real_model

    m = real_model.WorldModel({"d_model": 16, "n_layers": 1, "n_heads": 2, "d_ff": 32,
                               "dec_hidden": 16})
    path = tmp_path / "ck.pt"
    torch.save({"current_model": {"not_a_param": torch.zeros(3)}}, path)
    assert "missing" in R._check_checkpoint(path, m)
    torch.save({"current_model": m.state_dict()}, path)
    assert R._check_checkpoint(path, m) is None
```

```
%%writefile /content/wm_code/test_speed.py
"""Tests for the fast paths: pre-built tensor batches, the batched field decoder, the
key/value-cache rollouts and the training timing. Each fast path is checked against the
slow reference it replaces (CPU, toy games)."""

from __future__ import annotations

import time

import numpy as np
import pytest
import schema as S
import simulate
import torch
import train
from model import TARGET_SIZES, WorldModel
from toy import make_toy_split
from train import collate, prebuild_split, tensor_batch, train_model

torch.set_num_threads(1)

TINY = {
    "d_model": 32,
    "n_layers": 2,
    "n_heads": 2,
    "d_ff": 64,
    "dec_hidden": 32,
    "dropout": 0.0,
    "max_len": 48,
    "batch_games": 8,
    "eval_batch_games": 16,
    "lr": 3e-3,
    "warmup_steps": 5,
    "epochs": 3,
    "max_steps": 150,
    "sim_batch": 64,
    "seed": 0,
}
FAR = time.time() + 3600.0


@pytest.fixture(scope="module")
def split() -> S.Split:
    return make_toy_split(24, np.random.default_rng(0))


def _trained(split: S.Split, **over) -> WorldModel:
    torch.manual_seed(0)
    m = WorldModel({**TINY, **over})
    train_model(m, split, split, {**TINY, **over, "epochs": 4}, "", FAR, log=lambda *_: None)
    return m.eval()


@pytest.fixture(scope="module")
def trained(split) -> WorldModel:
    return _trained(split)


@pytest.fixture(scope="module")
def trained_short(split) -> WorldModel:
    """A 12-record window: most rollouts outgrow it, so the cache falls back to re-encoding."""
    return _trained(split, max_len=12)


# ------------------------------------------------------------------ pre-built batches
@pytest.mark.parametrize("max_len", [48, 20])
def test_tensor_batches_equal_collate(split, max_len):
    ts = prebuild_split(split, max_len)
    rng = np.random.default_rng(1)
    for _ in range(5):
        games = rng.choice(len(split.game_ids), size=7, replace=False)
        a = collate(split, games, max_len)
        b = tensor_batch(ts, games)
        assert list(a) == list(b)
        for k in a:
            assert a[k].dtype == b[k].dtype and torch.equal(a[k], b[k]), k


def test_tensor_batches_give_the_same_loss_as_collate(split):
    m = WorldModel(TINY).eval()
    ts = prebuild_split(split, m.max_len)
    order = np.random.default_rng([0, 0]).permutation(len(split.game_ids))
    old = list(train.make_batches(split, 8, m.max_len, np.random.default_rng([0, 0])))
    for bi, ref in enumerate(old):
        new = tensor_batch(ts, order[bi * 8 : (bi + 1) * 8])
        with torch.no_grad():
            la, ia = m.loss(ref)
            lb, ib = m.loss(new)
        assert torch.equal(la, lb)
        assert torch.equal(ia["counts"], ib["counts"])


def test_evaluate_nll_matches_reference_batching(split):
    m = WorldModel(TINY).eval()
    fast = train.evaluate_nll(m, split, TINY)
    sums = torch.zeros(len(S.TARGET_FIELDS), dtype=torch.float64)
    n_rec = 0
    with torch.no_grad():
        for b in train.make_batches(split, 16, m.max_len, None):
            _, info = m.loss(b, parallel=False)
            sums += info["sums"].double()
            n_rec += int(b["mask"].sum())
    assert fast["n_records"] == n_rec
    assert fast["total"] == pytest.approx(float(sums.sum()) / n_rec, rel=1e-6)


# ------------------------------------------------------------------ batched decoder
def test_parallel_decoder_gives_the_same_loss_and_gradients(split):
    torch.manual_seed(3)
    m = WorldModel(TINY)
    batch = collate(split, np.arange(6), m.max_len)
    out = {}
    for par in (False, True):
        m.zero_grad()
        total, info = m.loss(batch, parallel=par)
        total.backward()
        out[par] = (total.detach(), info, [p.grad.clone() for p in m.parameters()])
    (ta, ia, ga), (tb, ib, gb) = out[False], out[True]
    assert float(tb) == pytest.approx(float(ta), rel=1e-6)
    assert torch.equal(ia["counts"], ib["counts"])
    assert torch.allclose(ia["sums"], ib["sums"], rtol=1e-5)
    for name in ia["fields"]:
        a, b = float(ia["fields"][name].detach()), float(ib["fields"][name].detach())
        assert b == pytest.approx(a, rel=1e-5), name
    for a, b in zip(ga, gb, strict=True):
        assert torch.allclose(a, b, atol=1e-6, rtol=1e-4)
    # with explicit denominators (the micro-batch path)
    denom = train._full_counts(m, batch)
    with torch.no_grad():
        la, _ = m.loss(batch, denom=denom, parallel=False)
        lb, _ = m.loss(batch, denom=denom, parallel=True)
    assert float(lb) == pytest.approx(float(la), rel=1e-6)


def test_parallel_decoder_logits_and_masks(split):
    m = WorldModel(TINY).eval()
    batch = collate(split, np.arange(3), m.max_len)
    with torch.no_grad():
        h = m.encode(batch["header"], batch["states"])
        tg = m.end_prefix(batch["targets"], batch["last"])
        ref = m.decode_logits(h, tg)
        pad = m.decode_logits_padded(h.reshape(-1, h.shape[-1]), tg.reshape(-1, tg.shape[-1]))
    for j, n in enumerate(TARGET_SIZES):
        assert torch.allclose(pad[j, :, :n], ref[j].reshape(-1, n), atol=1e-5)
        assert torch.isinf(pad[j, :, n:]).all()
    assert torch.equal(m.end_prefix_fast(batch["targets"], batch["last"]), tg)
    masks = m.target_masks(tg, batch["mask"], batch["last"])
    fast = m.target_mask_tensor(tg, batch["mask"], batch["last"])
    for j, mk in enumerate(masks):
        assert torch.equal(fast[..., j], mk), j


# ------------------------------------------------------------------ key/value cache
def test_kv_cache_step_equals_full_encode(split):
    torch.manual_seed(5)
    m = WorldModel(TINY).eval()
    b = collate(split, np.array([0, 1, 2]), m.max_len)
    n0 = 6  # prefill this many records, then add the rest one at a time
    lens = b["mask"].sum(1)
    total = int(lens.min())
    c = m.cfg
    with torch.no_grad():
        full = m.encode(b["header"], b["states"])
        h0, kv = m.encode(b["header"], b["states"][:, :n0], return_kv=True)
        assert torch.allclose(h0, full[:, :n0], atol=1e-6)
        dh = c["d_model"] // c["n_heads"]
        kc = torch.zeros(c["n_layers"], 3, c["n_heads"], m.max_len + 1, dh)
        vc = torch.zeros_like(kc)
        for layer, (k, v) in enumerate(kv):
            kc[layer, :, :, : n0 + 1] = k
            vc[layer, :, :, : n0 + 1] = v
        for t in range(n0, total):
            pos = torch.full((3,), t + 1, dtype=torch.long)  # header is position 0
            h = m.encode_step(b["states"][:, t], pos, kc, vc, t + 2)
            assert torch.allclose(h, full[:, t], atol=1e-5), t
            la = m.decode_logits(h, b["targets"][:, t])
            lb = m.decode_logits(full[:, t], b["targets"][:, t])
            for x, y in zip(la, lb, strict=True):
                assert torch.allclose(x, y, atol=1e-5)


def test_shared_prefix_step_equals_full_encode(split):
    """Two rows share a game's first records and differ in the last one (as rollouts of one
    query with different start actions); the shared part is cached once for both."""
    torch.manual_seed(6)
    m = WorldModel(TINY).eval()
    c = m.cfg
    nl, nh, dh = c["n_layers"], c["n_heads"], c["d_model"] // c["n_heads"]
    one = collate(split, np.array([4]), m.max_len)
    L, extra = 10, 5  # 9 shared records, the 10th differs, then 5 more records per row
    states = one["states"][:, : L + extra].repeat(2, 1, 1)
    states[1, L - 1, S.STATE_INDEX["action"]] = S.ACTION["PUNT"]
    header = one["header"].repeat(2, 1)
    with torch.no_grad():
        full = m.encode(header, states)
        _, kv = m.encode(header[:1], states[:1, : L - 1], return_kv=True)
        kp = torch.stack([k for k, _ in kv])  # [layers, 1, heads, L, dh]
        vp = torch.stack([v for _, v in kv])
        ko = torch.zeros(nl, 2, nh, extra + 1, dh)
        vo = torch.zeros_like(ko)
        cell = torch.tensor([0, 1])  # one group of two rows
        for j in range(extra + 1):
            ok = torch.cat(
                [
                    torch.ones(2, L, dtype=torch.bool),
                    torch.arange(j + 1).unsqueeze(0).expand(2, -1) <= j,
                ],
                dim=1,
            )
            bias = torch.zeros(ok.shape).masked_fill_(~ok, float("-inf"))
            pos = torch.full((2,), L + j, dtype=torch.long)
            h = m.encode_step(
                states[:, L - 1 + j],
                pos,
                ko,
                vo,
                j + 1,
                cache_pos=torch.full((2,), j, dtype=torch.long),
                prefix=(kp, vp, cell, 2, bias),
            )
            assert torch.allclose(h, full[:, L - 1 + j], atol=1e-5), j


def test_sampler_draws_from_the_model_distribution():
    """Gumbel-max with every field's noise drawn at once: the first field's frequencies
    match softmax of its logits, the second field's its conditional given the first."""
    torch.manual_seed(2)
    m = WorldModel(TINY).eval()
    n = 40000
    h = 2.0 * torch.randn(1, TINY["d_model"]).repeat(n, 1)
    codes = m.sample_transition(h, 1.0, torch.Generator().manual_seed(0))
    with torch.no_grad():
        lg = m.field_logits(h[:1], codes[:1], 0)[0]
        lg[0] = -float("inf")  # the pad code is masked
        p0 = torch.softmax(lg, -1).double()
    f0 = torch.bincount(codes[:, 0], minlength=len(p0)).double() / n
    assert ((f0 - p0).abs() <= 4 * (p0 * (1 - p0) / n).sqrt() + 1e-9).all()
    top = int(torch.argmax(p0))
    assert top != S.PHASE["END"]
    rows = codes[:, 0] == top
    with torch.no_grad():
        lg1 = m.field_logits(h[:1], codes[rows][:1], 1)[0]
        lg1[0] = -float("inf")
        p1 = torch.softmax(lg1, -1).double()
    k = int(rows.sum())
    f1 = torch.bincount(codes[rows, 1], minlength=len(p1)).double() / k
    assert ((f1 - p1).abs() <= 4 * (p1 * (1 - p1) / k).sqrt() + 1e-9).all()


def _same(a: dict, b: dict) -> bool:
    return (
        np.array_equal(a["home_wp"], b["home_wp"])
        and np.array_equal(a["mean_steps"], b["mean_steps"])
        and a["counters"] == b["counters"]
        and a["start_actions"] == b["start_actions"]
    )


@pytest.mark.parametrize("which", ["trained", "trained_short"])
def test_kv_cache_rollouts_identical_to_recompute(split, which, request):
    model = request.getfixturevalue(which)
    idx = np.arange(3, len(split.states), 13)
    for act in (None, "GO", S.ACTION["FG"]):
        out = {}
        for kv in (False, True):
            cfg = {**TINY, "max_len": model.max_len, "kv_cache": kv}
            out[kv] = simulate.rollout(
                model, split, idx, act, 24, cfg, FAR, np.random.default_rng(9)
            )
        assert _same(out[False], out[True]), act
        t = out[True]["timing"]
        assert t["path"] == "kv_cache" and out[False]["timing"]["path"] == "recompute"
        assert t["compactions"] > 0
        if which == "trained_short":
            assert t["fallback_row_steps"] > 0  # windows outgrew max_len: re-encoded, exact
        else:
            assert t["fallback_row_steps"] == 0
    g = {
        kv: simulate.simulate_games(
            model,
            split,
            5,
            {**TINY, "max_len": model.max_len, "kv_cache": kv},
            FAR,
            np.random.default_rng(1),
        )
        for kv in (False, True)
    }
    assert g[False]["sim"] == g[True]["sim"] and g[False]["counters"] == g[True]["counters"]


def test_memory_limited_groups_are_identical_and_complete(split, trained):
    idx = np.arange(5, len(split.states), 29)
    tiny_budget = 40 * trained.kv_bytes_per_row(4) / 2**30  # 40 rollouts fit at once
    cfg = {**TINY, "kv_cache_cpu_gb": tiny_budget}
    assert simulate.live_rows(trained, cfg) == 40
    out = {
        kv: simulate.rollout(
            trained, split, idx, None, 16, {**cfg, "kv_cache": kv}, FAR, np.random.default_rng(2)
        )
        for kv in (False, True)
    }
    assert _same(out[False], out[True])
    assert out[True]["done"].all()
    # one query with more rollouts than fit at once is split inside _run
    one = {
        kv: simulate.rollout(
            trained,
            split,
            idx[:1],
            "GO",
            100,
            {**cfg, "kv_cache": kv},
            FAR,
            np.random.default_rng(3),
        )
        for kv in (False, True)
    }
    assert _same(one[False], one[True]) and one[True]["timing"]["rows"] == 100


def test_cache_memory_per_rollout():
    m = WorldModel({"d_model": 256, "n_layers": 6, "n_heads": 8, "d_ff": 1024, "max_len": 256})
    assert m.kv_bytes_per_row(2) == 6 * 2 * 257 * 256 * 2  # about 1.5 MiB under fp16
    rows = (4 * 2**30) // m.kv_bytes_per_row(4)  # CPU: float32 caches, 4 GiB budget
    assert simulate.cache_rows(m, {"kv_cache_cpu_gb": 4.0}) == rows - rows % 256
    assert simulate.live_rows(m, {"kv_cache_cpu_gb": 4.0, "sim_max_live": 512}) == 512


# ------------------------------------------------------------------ training timing
def test_training_history_has_timing(split, tmp_path):
    m = WorldModel(TINY)
    res = train_model(
        m, split, split, {**TINY, "epochs": 2}, str(tmp_path / "t.pt"), FAR, log=lambda *_: None
    )
    assert res["device"] == "cpu"
    keys = (
        "batch_s",
        "h2d_s",
        "forward_s",
        "backward_opt_s",
        "train_loop_s",
        "validation_s",
        "checkpoint_s",
        "epoch_s",
    )
    for h in res["history"]:
        t = h["timing"]
        for k in keys:
            assert isinstance(t[k], float) and t[k] >= 0, k
        phases = t["batch_s"] + t["h2d_s"] + t["forward_s"] + t["backward_opt_s"]
        assert phases <= t["train_loop_s"] * 1.01 + 1e-3
        assert phases >= 0.5 * t["train_loop_s"]
        assert t["steps"] == 3 and t["train_records"] == int(split.offsets[-1])
        assert t["train_records_per_s"] > 0
    assert set(res["timing_totals_s"]) == set(keys)
    assert res["timing_totals_s"]["epoch_s"] == pytest.approx(
        sum(h["timing"]["epoch_s"] for h in res["history"])
    )


def test_train_model_moves_the_model_to_the_configured_device(split, monkeypatch):
    """The first Colab run trained on the CPU: nothing moved the model to cfg["device"]."""
    m = WorldModel(TINY)
    seen = []
    real_to = torch.nn.Module.to

    def spy(self, *a, **k):
        seen.append(a[0] if a else k.get("device"))
        return real_to(self, *a, **k)

    monkeypatch.setattr(torch.nn.Module, "to", spy)
    train_model(
        m, split, split, {**TINY, "epochs": 1, "device": "cpu"}, "", FAR, log=lambda *_: None
    )
    assert torch.device("cpu") in [torch.device(s) for s in seen if s is not None]


# ------------------------------------------------------------------ GPU-only guards, on CPU
def test_rollouts_retry_a_group_in_halves_after_out_of_memory(split, trained, monkeypatch):
    real = simulate._run_cached
    calls = []

    def flaky(model, header, *a, **k):
        calls.append(len(header))
        if len(calls) == 1:
            raise torch.OutOfMemoryError("CUDA out of memory (simulated)")
        return real(model, header, *a, **k)

    monkeypatch.setattr(simulate, "_run_cached", flaky)
    idx = np.arange(5, len(split.states), 41)
    r = simulate.rollout(trained, split, idx, None, 8, TINY, FAR, np.random.default_rng(2))
    assert r["done"].all() and np.isfinite(r["home_wp"]).all()
    assert r["timing"]["oom_retries"] == 1
    assert calls[1] == calls[0] // 2 and sum(calls[1:]) == calls[0]


def test_fused_adamw_falls_back_when_the_build_rejects_it(monkeypatch):
    m = WorldModel(TINY)
    real = torch.optim.AdamW
    msgs = []

    def picky(params, **kw):
        if kw.get("fused"):
            raise RuntimeError("fused=True requires all the params to be CUDA tensors")
        return real(params, **kw)

    monkeypatch.setattr(torch.optim, "AdamW", picky)
    c = {**m.cfg, "betas": (0.9, 0.95)}
    opt, fused = train.make_optimizer(m, c, torch.device("cuda"), msgs.append)
    assert not fused and isinstance(opt, real) and "using the default AdamW" in msgs[0]
    opt, fused = train.make_optimizer(m, c, torch.device("cpu"), msgs.append)
    assert not fused and len(msgs) == 1  # fused is never requested on the CPU


def test_cuda_cache_plan_uses_total_memory_capped_by_what_is_free(monkeypatch):
    m = WorldModel({"d_model": 256, "n_layers": 6, "n_heads": 8, "d_ff": 1024, "max_len": 256})
    per_row = m.kv_bytes_per_row(2)
    total, free, idle = 15 * 2**30, 14 * 2**30, 0

    class Props:
        total_memory = total

    monkeypatch.setattr(simulate, "device_of", lambda model: torch.device("cuda", 0))
    monkeypatch.setattr(simulate, "amp_settings", lambda d: (True, torch.float16, True))
    monkeypatch.setattr(torch.cuda, "get_device_properties", lambda d: Props())
    monkeypatch.setattr(torch.cuda, "mem_get_info", lambda d: (free, total))
    monkeypatch.setattr(torch.cuda, "memory_reserved", lambda d: idle)
    monkeypatch.setattr(torch.cuda, "memory_allocated", lambda d: 0)
    rows = int(0.4 * total // per_row)
    assert simulate.cache_rows(m, {}) == rows - rows % 256  # a T4: 3840 rollouts
    assert simulate.cache_rows(m, {}) == 3840
    free = 2 * 2**30  # another process holds most of the GPU: lowered to 70% of what is free
    rows = int(0.7 * free // per_row)
    assert simulate.cache_rows(m, {}) == rows - rows % 256
    idle = 4 * 2**30  # memory PyTorch holds but does not use counts as available
    rows = int(min(0.4 * total, 0.7 * (free + idle)) // per_row)
    assert simulate.cache_rows(m, {}) == rows - rows % 256
```

```
%%writefile /content/wm_code/test_support.py
"""Tests for support.py (action support, hard rules, soft flags, the kicking diagnostic) and
its wiring into simulate.plan_fourth_downs and run.stage_plan. CPU, toy data, no network."""

from __future__ import annotations

import sys
import time

import numpy as np
import pandas as pd
import pytest
import run as R
import schema as S
import support as SP
import torch
from model import WorldModel
from simulate import plan_fourth_downs, rollout
from test_run import _stubs, _tiny_preset
from toy import make_toy_split

torch.set_num_threads(1)
SI, TI, PH, A = S.STATE_INDEX, S.TARGET_INDEX, S.PHASE, S.ACTION
FAR = time.time() + 3600.0
TINY = {"d_model": 32, "n_layers": 1, "n_heads": 2, "d_ff": 64, "dec_hidden": 32,
        "dropout": 0.0, "max_len": 48, "max_steps": 150, "sim_batch": 512, "seed": 0}


@pytest.fixture(scope="module")
def train_split() -> S.Split:
    return make_toy_split(120, np.random.default_rng(0))


@pytest.fixture(scope="module")
def big_train() -> S.Split:
    """Longer toy games, enough field-goal attempts per 5-yard band to estimate a rate."""
    return make_toy_split(300, np.random.default_rng(7), start_clock=1800)


@pytest.fixture(scope="module")
def test_split() -> S.Split:
    return make_toy_split(40, np.random.default_rng(1), first_game_id=10_000)


def toy_make_prob(yl):
    """toy.py's field-goal law: good with probability max(0.2, 1 - yardline / 50)."""
    return np.maximum(0.2, 1.0 - np.asarray(yl, dtype=float) / 50.0)


PUNT_NET = 35  # FakeKicker's punt: 35 yards net, or a touchback at the 20


class FakeKicker:
    """A stand-in world model with a known first transition after a forced kick.

    encode() exposes the yard line, action and possession of each record; sample_transition()
    makes a field goal with probability make_prob(yard line) (3 points for the kicking team),
    and a punt nets PUNT_NET yards (touchback at the receiver's 20 if it would reach the end
    zone). Any other action: an incomplete-looking change of possession with no points.
    """

    max_len = 64

    def __init__(self, make_prob=toy_make_prob):
        self.cfg = {"temperature": 1.0, "mask_pad_codes": True}
        self.make_prob = make_prob
        self.encoded_actions: list[np.ndarray] = []
        self._p = torch.zeros(1)

    def parameters(self):
        return iter([self._p])

    def eval(self):
        return self

    def count_params(self):
        return 0

    def encode(self, header, states):
        self.encoded_actions.append(states[:, :, SI["action"]].numpy().copy())
        cols = [states[..., SI[k]] for k in ("yardline", "action", "poss")]
        return torch.stack(cols, dim=-1).float()

    def sample_transition(self, h, temperature=1.0, generator=None, mask_pad=True):
        yl, act, poss = h[:, 0].long(), h[:, 1].long(), h[:, 2].long()
        r = h.shape[0]
        u = torch.rand(r, generator=generator)
        p = torch.as_tensor(self.make_prob(yl.numpy()), dtype=torch.float32)
        made = (act == A["FG"]) & (u < p)
        sign = torch.where(poss == S.POSS_CODE["HOME"], 1, -1)
        pts = torch.where(made, 3 * sign, torch.zeros_like(sign))
        after = yl - PUNT_NET  # punter's distance to the goal line after the punt
        recv_yl = torch.where(after <= 0, torch.full_like(yl, 80), 100 - after)
        out = torch.zeros(r, len(S.TARGET_FIELDS), dtype=torch.long)
        out[:, TI["phase"]] = torch.where(made, PH["KICKOFF"], PH["PLAY"])
        out[:, TI["poss"]] = 3 - poss
        out[:, TI["score_delta"]] = torch.as_tensor(
            [S.SCORE_DELTA_CODE[int(v)] for v in pts], dtype=torch.long)
        out[:, TI["elapsed"]] = 6
        out[:, TI["to_used_home"]] = 1
        out[:, TI["to_used_away"]] = 1
        out[:, TI["down"]] = torch.where(made, 0, 1)
        out[:, TI["ydstogo"]] = torch.where(made, 0, 10)
        out[:, TI["yardline"]] = torch.where(
            act == A["PUNT"], recv_yl, torch.where(made, 65, (100 - yl).clamp(1, 99)))
        out[:, TI["action"]] = torch.where(made, A["KICKOFF"], A["RUN"])
        return out


# ---------------------------------------------------------------- 1. training split only
def test_support_tables_count_the_training_split(train_split, test_split):
    sup = SP.build_support(train_split)
    st = train_split.states
    fd = SP.fourth_down_records(train_split)
    manual = np.zeros((100, S.MAX_YDSTOGO + 1, 3), dtype=np.int64)
    for i in fd:
        a = int(st[i, SI["action"]])
        j = 0 if a in S.GO_ACTIONS else (1 if a == A["FG"] else 2)
        manual[st[i, SI["yardline"]], st[i, SI["ydstogo"]], j] += 1
    assert np.array_equal(sup.counts, manual)
    assert sup.meta["split_games"] == len(train_split.game_ids)
    assert sup.meta["fourth_downs"] == len(fd)
    # Comparable states: yard line within +-3 and the same yards-to-go bucket.
    yl, ytg = 50, 10
    want = manual[47:54, 8:11].sum(axis=(0, 1))
    assert np.array_equal(sup.smooth[yl, ytg], want)
    # Field-goal makes, counted independently from the score bookkeeping (raw columns).
    last = np.zeros(len(st), dtype=bool)
    last[train_split.offsets[1:] - 1] = True
    games = R.record_game(train_split)
    att = np.zeros(100, dtype=np.int64)
    mk = np.zeros(100, dtype=np.int64)
    for i in np.flatnonzero((st[:, SI["phase"]] == PH["PLAY"]) & (st[:, SI["action"]] == A["FG"])):
        after = (int(train_split.final_diff[games[i]]) if last[i]
                 else int(train_split.raw[i + 1, 1]))
        sign = 1 if st[i, SI["poss"]] == S.POSS_CODE["HOME"] else -1
        att[st[i, SI["yardline"]]] += 1
        mk[st[i, SI["yardline"]]] += int((after - int(train_split.raw[i, 1])) * sign == 3)
    assert np.array_equal(sup.fg_attempts, att) and np.array_equal(sup.fg_makes, mk)
    # toy.py makes a field goal with probability max(0.2, 1 - yl / 50): the empirical rate
    # over all training attempts matches that law.
    yls = np.repeat(np.arange(100), att)
    expect = toy_make_prob(yls).mean()
    se = np.sqrt(expect * (1 - expect) / att.sum())
    assert abs(mk.sum() / att.sum() - expect) < 4 * se
    # A different split gives different tables: nothing from it leaked in.
    both = SP.build_support(R.concat_splits([train_split, test_split]))
    assert not np.array_equal(both.counts, sup.counts)


def test_stage_plan_builds_support_from_the_training_split_only(tmp_path, monkeypatch):
    calls: dict = {}
    stubs = _stubs(calls)
    seen_masks: list[np.ndarray] = []

    def plan_stub(m, split, idx, n, cfg, deadline_ts, rng, action_mask=None):
        """Honours the mask: best = first allowed of FG, PUNT, GO; masked wp are NaN."""
        idx = np.asarray(idx)
        seen_masks.append((split, idx.copy(), np.asarray(action_mask).copy()))
        k = len(idx)
        wp = np.where(action_mask, np.array([[0.5, 0.6, 0.55]]), np.nan)
        best = [next(c for c, j in (("FG", 1), ("PUNT", 2), ("GO", 0)) if action_mask[q, j])
                for q in range(k)]
        real = [SP.CHOICES[SP.choice_index(a)] if SP.choice_index(a) >= 0 else "GO"
                for a in split.states[idx, SI["action"]]]
        return pd.DataFrame({"idx": idx, "wp_go": wp[:, 0], "wp_fg": wp[:, 1],
                             "wp_punt": wp[:, 2], "se_go": 0.01, "se_fg": 0.01, "se_punt": 0.01,
                             "best": best, "margin": 0.05, "too_close": False,
                             "real_action": real, "real_choice": real,
                             "done": np.ones(k, dtype=bool)})

    stubs["simulate"].plan_fourth_downs = plan_stub
    stubs["train"].load_checkpoint = lambda path, cfg_override=None, device=None: (
        FakeKicker(), {})
    for name, mod in stubs.items():
        monkeypatch.setitem(sys.modules, name, mod)
    built = []
    real_build = SP.build_support

    def spy(split, cfg=None):
        built.append(list(split.game_ids))
        return real_build(split, cfg)

    monkeypatch.setattr(SP, "build_support", spy)
    p = _tiny_preset()
    run = R.Run(p, tmp_path, quiet=True)
    for s in ("data", "train"):
        assert R.run_stage(run, s)["status"] == "done"
    out = R.run_stage(run, "plan")
    assert out["status"] == "done", out
    tr, _ = run.split("train")
    assert built == [list(tr.game_ids)]  # built once, from the training split
    assert all(g.startswith(f"{p['seasons']['train'][0]}_") for g in built[0])
    tables = out["support"]["training_tables"]
    assert tables["split_seasons"] == sorted({int(x) for x in tr.seasons})
    # Every planner call got the hard-rule mask for exactly its states.
    rules = out["support"]["rules"]
    for split, idx, mask in seen_masks:
        assert np.array_equal(mask, SP.action_mask(split, idx, rules))
    # Masked actions recorded, flags and the split agreement present.
    assert out["masked_actions"] and all(m["masked"] in ("FG", "PUNT")
                                         for m in out["masked_actions"])
    assert {"well_supported", "low_support", "min_support"} <= set(out["agreement_by_support"])
    dec = pd.DataFrame(out["decisions"])
    assert {"sup_n", "sup_count_best", "sup_share_best", "low_support", "sim_fg_make",
            "emp_fg_make"} <= set(dec.columns)
    assert "error" not in out["support"]["diagnostic"], out["support"]["diagnostic"]
    pkg = R.run_stage(run, "package")
    assert "fg_support" in pkg["figures"], pkg["figures"]
    assert (tmp_path / "figures" / "fg_support.png").exists()
    text = R.summary(run)
    assert "action rules" in text and "support:" in text and "field goals, simulated" in text


# ---------------------------------------------------------------- 2. hard rules
def test_mask_removes_out_of_range_field_goals_and_deep_punts(train_split, test_split):
    sup = SP.build_support(train_split)
    rules = SP.hard_rules(sup)
    longest = int(np.flatnonzero(sup.fg_makes > 0).max())
    assert rules["longest_made_fg_yardline_train"] == longest
    assert rules["fg_max_yardline"] == longest + 3
    assert rules["fg_max_distance_yards"] == longest + 3 + SP.KICK_DISTANCE_OFFSET
    assert rules["punt_min_yardline"] == 30
    sp = R.subset_split(test_split, np.arange(len(test_split.game_ids)))
    idx = SP.fourth_down_records(sp)[:7]
    yls = [1, 29, 30, rules["fg_max_yardline"], rules["fg_max_yardline"] + 1, 75, 99]
    sp.states[idx, SI["yardline"]] = yls
    m = SP.action_mask(sp, idx, rules)
    assert m[:, 0].all()  # going for it is always allowed
    assert m[:, 1].tolist() == [y <= rules["fg_max_yardline"] for y in yls]
    assert m[:, 2].tolist() == [y >= 30 for y in yls]
    masked = SP.masked_actions(sp, idx, rules)
    assert len(masked) == int((~m).sum())
    assert {(e["idx"], e["masked"]) for e in masked} == {
        (int(idx[q]), SP.CHOICES[j]) for q, j in zip(*np.nonzero(~m), strict=True)}
    assert all(e["rule"] for e in masked)
    # Changing the margin moves the limit.
    rules5 = SP.hard_rules(SP.build_support(train_split, {"fg_margin_yards": 5}))
    assert rules5["fg_max_yardline"] == longest + 5
    with pytest.raises(ValueError):
        SP.config({"support": {"no_such_setting": 1}})


def test_low_support_flag(train_split, test_split):
    sup = SP.build_support(train_split)
    rules = SP.hard_rules(sup)
    idx = SP.fourth_down_records(test_split)[:30]
    df = pd.DataFrame({"idx": idx, "best": ["FG"] * 15 + ["GO"] * 14 + [None],
                       "real_choice": ["PUNT"] * 30})
    out = SP.add_support_columns(df, sup, test_split, rules)
    lk = SP.lookup(sup, test_split, idx)
    want = np.r_[lk["sup_fg"][:15], lk["sup_go"][15:29]]
    assert np.array_equal(out["sup_count_best"][:29].to_numpy(), want)
    assert out["low_support"][:29].tolist() == (want < 10).tolist()
    assert not out["low_support"].iloc[29]  # no recommendation, no flag
    ok = lk["sup_n"][:29].to_numpy() > 0
    share = out["sup_share_best"][:29].to_numpy()
    assert np.allclose(share[ok], want[ok] / lk["sup_n"][:29].to_numpy()[ok])
    assert (lk["ctx_n"] <= lk["sup_n"]).all()  # context cells are subsets


# ---------------------------------------------------------------- 3. plan with a mask
def _fourth(split: S.Split, n: int) -> np.ndarray:
    st = split.states
    home = np.flatnonzero((st[:, SI["down"]] == 4) & (st[:, SI["phase"]] == PH["PLAY"])
                          & (st[:, SI["poss"]] == S.POSS_CODE["HOME"]))
    away = np.flatnonzero((st[:, SI["down"]] == 4) & (st[:, SI["phase"]] == PH["PLAY"])
                          & (st[:, SI["poss"]] == S.POSS_CODE["AWAY"]))
    return np.sort(np.r_[home[: n // 2], away[: n - n // 2]])


def test_plan_with_mask_never_recommends_a_masked_action(test_split):
    torch.manual_seed(0)
    model = WorldModel(TINY).eval()
    idx = _fourth(test_split, 8)
    mask = np.ones((len(idx), 3), dtype=bool)
    mask[0, 1] = mask[1, 2] = False
    mask[2, 1:] = False  # only GO allowed
    mask[3, 0] = False  # GO masked (support never does this; the planner must still obey)
    mask[4:, 1] = [True, False, True, False]
    df = plan_fourth_downs(model, test_split, idx, 6, TINY, FAR, np.random.default_rng(3),
                           action_mask=mask)
    assert df["done"].all()
    for j, ch in enumerate(("go", "fg", "punt")):
        assert df[f"wp_{ch}"][~mask[:, j]].isna().all()
        assert df[f"se_{ch}"][~mask[:, j]].isna().all()
        assert df[f"wp_{ch}"][mask[:, j]].notna().all()
        assert df[f"allowed_{ch}"].tolist() == mask[:, j].tolist()
    for q, b in enumerate(df["best"]):
        assert mask[q, ("GO", "FG", "PUNT").index(b)], (q, b)
    assert df["best"].iloc[2] == "GO" and not df["too_close"].iloc[2]
    c = df.attrs["counters"]
    for ch, j in (("GO", 0), ("FG", 1), ("PUNT", 2)):
        assert c[ch]["rollouts"] == 6 * int(mask[:, j].sum())
    # An all-True mask is the same computation as no mask (same draws, same numbers).
    a = plan_fourth_downs(model, test_split, idx, 6, TINY, FAR, np.random.default_rng(5))
    b = plan_fourth_downs(model, test_split, idx, 6, TINY, FAR, np.random.default_rng(5),
                          action_mask=np.ones((len(idx), 3), dtype=bool))
    cols = ["wp_go", "wp_fg", "wp_punt", "se_go", "se_fg", "se_punt", "best", "margin"]
    pd.testing.assert_frame_equal(a[cols], b[cols])
    with pytest.raises(ValueError):
        plan_fourth_downs(model, test_split, idx, 6, TINY, FAR, np.random.default_rng(5),
                          action_mask=np.ones((len(idx) + 1, 3), dtype=bool))


def test_mask_rows_align_with_idx_as_passed(test_split):
    """Rows the planner drops (not fourth-down PLAY records) take their mask row with them."""
    torch.manual_seed(0)
    model = WorldModel(TINY).eval()
    fourth = _fourth(test_split, 2)
    other = np.flatnonzero(test_split.states[:, SI["down"]] == 1)[:1]
    idx = np.r_[other, fourth]
    mask = np.array([[True, False, False], [True, False, True], [True, True, False]])
    df = plan_fourth_downs(model, test_split, idx, 4, TINY, FAR, np.random.default_rng(1),
                           action_mask=mask)
    assert df["idx"].tolist() == fourth.tolist()
    assert df["allowed_fg"].tolist() == [False, True]
    assert df["allowed_punt"].tolist() == [True, False]


# ---------------------------------------------------------------- 4. the diagnostic
def test_first_transitions_force_the_action_on_the_last_record(test_split):
    model = FakeKicker()
    idx = _fourth(test_split, 4)
    SP.first_transitions(model, test_split, idx, "FG", 3, None, np.random.default_rng(0))
    acts = model.encoded_actions[-1]
    games = R.record_game(test_split)[idx]
    for r, (i, g) in enumerate(zip(idx, games, strict=True)):
        n = int(i) - max(int(test_split.offsets[g]), int(i) + 1 - model.max_len) + 1
        assert acts[r, n - 1] == A["FG"]
        start = int(i) - n + 1
        assert np.array_equal(acts[r, : n - 1], test_split.states[start:int(i), SI["action"]])


def test_first_transitions_sample_the_rollout_first_step(test_split):
    """The one-step sampler conditions on the same hidden state as the first step of a real
    rollout with the action forced, so it samples the same first transition."""
    torch.manual_seed(0)
    model = WorldModel(TINY).eval()
    idx = _fourth(test_split, 4)
    seen: list[torch.Tensor] = []
    real = model.sample_transition

    def spy(h, *a, **k):
        seen.append(h.detach().float().clone())
        return real(h, *a, **k)

    model.sample_transition = spy
    rollout(model, test_split, idx, A["FG"], 3, {**TINY, "max_steps": 1}, FAR,
            np.random.default_rng(0))
    first_step = seen[0]
    seen.clear()
    SP.first_transitions(model, test_split, idx, "FG", 3, TINY, np.random.default_rng(0))
    assert first_step.shape == seen[0].shape
    assert torch.allclose(first_step, seen[0], atol=1e-5)


def test_simulated_make_rate_matches_a_toy_model_with_a_known_rate(big_train, test_split):
    sup = SP.build_support(big_train, {"diag_states_per_band": 6, "diag_draws": 400})
    rules = SP.hard_rules(sup)
    model = FakeKicker()  # makes field goals with the same law as toy.py
    idx = _fourth(test_split, 10)
    sim = SP.sim_fg_make(model, test_split, idx, 400, None, np.random.default_rng(2))
    p = toy_make_prob(test_split.states[idx, SI["yardline"]])
    assert np.all(np.abs(sim - p) < 4 * np.sqrt(p * (1 - p) / 400) + 1e-9)  # HOME and AWAY

    d = SP.sim_vs_empirical(model, test_split, sup, rules, None, np.random.default_rng(4))
    n_draws = d["draws_per_state"]
    checked = 0
    for row in d["fg"]:
        lo, hi = row["yardline"]
        assert row["train_attempts"] == int(sup.fg_attempts[lo:hi + 1].sum())
        assert row["train_makes"] == int(sup.fg_makes[lo:hi + 1].sum())
        if not row["state_idx"]:
            assert row["sim_make_rate"] is None
            continue
        yl = test_split.states[row["state_idx"], SI["yardline"]]
        assert np.all((yl >= lo) & (yl <= hi))
        want = toy_make_prob(yl).mean()
        se = np.sqrt(want * (1 - want) / (len(yl) * n_draws))
        assert abs(row["sim_make_rate"] - want) < 4 * se + 1e-9
        assert row["beyond_fg_limit"] == (lo > rules["fg_max_yardline"])
        checked += 1
    assert checked >= 3
    for row in d["punt"]:
        if not row["state_idx"]:
            continue
        yl = test_split.states[row["state_idx"], SI["yardline"]].astype(float)
        want = np.where(yl - PUNT_NET <= 0, yl - 20, PUNT_NET).mean()
        assert row["sim_net_yards"] == pytest.approx(want)
        assert row["sim_other_share"] == 0.0
    # The same law as the data: no systematic gap where the training data is thick. An
    # over-optimistic simulator (every field goal good) shows a large positive gap.
    h = SP.fg_gap_summary(d, min_attempts=30)
    assert h["bands_with_data"] >= 1 and abs(h["weighted_mean_gap_with_data"]) < 0.1
    rosy = SP.sim_vs_empirical(FakeKicker(lambda y: np.ones(len(y))), test_split, sup, rules,
                               None, np.random.default_rng(4))
    hr = SP.fg_gap_summary(rosy, min_attempts=30)
    assert hr["weighted_mean_gap_with_data"] > 0.1
    assert hr["sim_make_rate_beyond_limit"] is None or hr["sim_make_rate_beyond_limit"] == 1.0


def test_punt_net_yards_definition():
    tg = np.zeros((4, len(S.TARGET_FIELDS)), dtype=np.int16)
    tg[:, TI["phase"]] = [PH["PLAY"], PH["PLAY"], PH["PLAY"], PH["TRY"]]
    tg[:, TI["poss"]] = [2, 2, 1, 2]  # punter is HOME (1)
    tg[:, TI["yardline"]] = [75, 80, 60, 15]
    net = SP.punt_net_yards(np.array([70, 30, 70, 70]), np.array([1, 1, 1, 1]), tg)
    # receiver at its own 25 from the punter's 30: 45 net; touchback from the opponent's
    # 30: 10; muff recovered by the punter at the opponent's 40 (yardline 60): 10; a return
    # touchdown (next record is a try): not a net-yards outcome.
    assert net[:3].tolist() == [45.0, 10.0, 10.0] and np.isnan(net[3])
```
